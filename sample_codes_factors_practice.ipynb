## Example 9: Reusable function

Change the input and run the cell to see how the factors change.

```python
from math import isqrt

def positive_factors(n):
    if n == 0:
        raise ValueError("Zero has infinitely many positive divisors")
    n = abs(n)
    found = set()
    for i in range(1, isqrt(n) + 1):
        if n % i == 0:
            found.add(i)
            found.add(n // i)
    return sorted(found)

print(positive_factors(28))
print(positive_factors(125))
```

## Example 10: 1, negative numbers and zero

Change the input and run the cell to see how the factors change.

```python
from math import isqrt

def positive_factors(n):
    if n == 0:
        raise ValueError("Zero has infinitely many positive divisors")
    n = abs(n)
    result = set()
    for i in range(1, isqrt(n) + 1):
        if n % i == 0:
            result.update((i, n // i))
    return sorted(result)

print("1:", positive_factors(1))
print("-12:", positive_factors(-12))
try:
    print(positive_factors(0))
except ValueError as error:
    print("0:", error)
```

## Example 15: Handle invalid input

Change the input and run the cell to see how the factors change.

```python
from math import isqrt

def positive_factors(n):
    if n == 0:
        raise ValueError("Enter a nonzero integer")
    n = abs(n)
    factors = set()
    for i in range(1, isqrt(n) + 1):
        if n % i == 0:
            factors.update((i, n // i))
    return sorted(factors)

for text in ("24", "-10", "0", "hello", "3.8"):
    try:
        number = int(text)
        print(text, "->", positive_factors(number))
    except ValueError as error:
        print(text, "->", error)
```

## Exercise 6: Handle 1, -36 and 0

**Challenge:** Predict the answer, change the number and use the final `assert` lines to check your work.

```python
from math import isqrt

def factors_of_nonzero(n):
    if n == 0:
        raise ValueError('Zero has infinitely many positive factors')
    n = abs(n)
    return sorted({d for i in range(1, isqrt(n) + 1)
                   if n % i == 0 for d in (i, n // i)})

print(factors_of_nonzero(1))
print(factors_of_nonzero(-36))
assert factors_of_nonzero(1) == [1]
assert factors_of_nonzero(-36) == [1, 2, 3, 4, 6, 9, 12, 18, 36]
try:
    factors_of_nonzero(0)
except ValueError:
    print('Zero case handled')
else:
    raise AssertionError('Zero must raise ValueError')
```

## Exercise 8: Make a reusable factor counter

**Challenge:** Predict the answer, change the number and use the final `assert` lines to check your work.

```python
from math import isqrt

def factor_count(n):
    if n <= 0:
        raise ValueError('Use a positive integer')
    result = set()
    for i in range(1, isqrt(n) + 1):
        if n % i == 0:
            result.update((i, n // i))
    return len(result)

print(factor_count(28), factor_count(100))
assert factor_count(28) == 6
assert factor_count(100) == 9
```
