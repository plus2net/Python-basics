### 1. Basic Invocations & Multiple Parameters

```python
add = lambda x, y: x + y
multiply = lambda a, b, c=1: a * b * c

print("3 + 5 =", add(3, 5))
print("3 * 4 * 2 =", multiply(3, 4, 2))
print("3 * 4 (default c=1) =", multiply(3, 4))
```

### 2. Stream Pipelines with map() and filter()

```python
nums = [1, 2, 3, 4, 5, 6]
squares = list(map(lambda x: x ** 2, nums))
evens = list(filter(lambda x: x % 2 == 0, nums))

print("Squares:", squares)
print("Evens:  ", evens)
```

### 3. Custom Sorting Keys for Complex Structures

```python
students = [
    {"name": "Alice", "score": 88},
    {"name": "Bob", "score": 95},
    {"name": "Charlie", "score": 78}
]

ranked = sorted(students, key=lambda s: s["score"], reverse=True)
for s in ranked:
    print(f"{s['name']}: {s['score']}")
```

### 4. Avoiding Late-Binding Loop Closures

```python
# Default argument binds value at definition time
funcs = [lambda val=i: val * 10 for i in range(4)]
results = [f() for f in funcs]
print("Fixed results:", results)
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Arithmetic
f1 = lambda x: x * 3
assert f1(10) == 30

# Test 2: Conditional
f2 = lambda x: 'even' if x % 2 == 0 else 'odd'
assert f2(4) == 'even'
assert f2(5) == 'odd'

# Test 3: Length key with distinct lengths
words = ['banana', 'fig', 'apple']
assert sorted(words, key=lambda s: len(s)) == ['fig', 'apple', 'banana']

print("All Python lambda unit assertions successfully verified! 100% passed.")
```
