### 1. Basic Universal Truth Testing
Verify if all elements in collections evaluate to truthy values.

```python
print("All positive numbers: ", all([1, 2, 3, 4]))
print("Contains zero:        ", all([1, 2, 0, 4]))
print("Contains False:       ", all([True, False, True]))
```

### 2. Vacuous Truth on Empty Collections
Observe that empty collections return True because no counterexamples exist.

```python
print("Empty list: ", all([]))
print("Empty tuple:", all(()))
print("Empty set:  ", all(set()))
print("Empty dict: ", all({}))
```

### 3. Evaluating Dictionaries: Keys vs. Values
Remember that passing a dictionary directly evaluates keys only.

```python
data = {1: True, 2: False}
print("Direct dict (checks keys):", all(data))
print("Dict values check:         ", all(data.values()))
```

### 4. Short-Circuit Evaluation with Generators
Verify that evaluation halts immediately on the first falsy element.

```python
def test_stream():
    for x in [10, 20, 0, 50]:
        print(f"  Checked: {x}")
        yield x

print("Starting check:")
res = all(test_stream())
print("Result:", res)
```

### 5. Production Recipe: Multi-Field Validation
Validate required fields across an incoming user submission.

```python
user_form = {"name": "Alex", "email": "alex@test.com", "accepted": True}
required = ["name", "email", "accepted"]
is_valid = all(bool(user_form.get(k)) for k in required)
print("Form submission valid:", is_valid)
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying all() mechanics.

```python
assert all([1, 2, 3]) is True
assert all([1, 0, 3]) is False
assert all([]) is True
assert all({1: False, 2: False}) is True
assert all({1: False, 2: False}.values()) is False

print("All all() unit assertions passed! 100% verified.")
```
