### 1. Inspecting the Global Symbol Table

```python
site_title = "plus2net Python Learning"
topic = "globals() built-in"

g_dict = globals()
print("Module __name__: ", g_dict['__name__'])
print("site_title:      ", g_dict['site_title'])
print("topic:           ", g_dict['topic'])
```

### 2. Runtime Mutation via globals()

```python
# Update existing variable
globals()['topic'] = "globals() Advanced Mastery"
print("Updated topic:          ", topic)

# Create new variable dynamically
globals()['runtime_created_var'] = 42
print("Dynamically created var:", runtime_created_var)
```

### 3. LEGB Scoping Rule & Shadowing

```python
level = 'Global Level'

def test_scoping():
    level = 'Local Function Level'
    print("Inside function (Local): ", level)
    print("From globals() (Global):   ", globals()['level'])

test_scoping()
print("Outside function:          ", level)
```

### 4. Dynamic Action Dispatcher

```python
def action_create():
    return "Resource created successfully"

def action_delete():
    return "Resource deleted cleanly"

def dispatch(event_name):
    handler_name = f"action_{event_name}"
    handlers = globals()
    if handler_name in handlers and callable(handlers[handler_name]):
        return handlers[handler_name]()
    return f"Handler for {event_name} not found"

print("Dispatch create:", dispatch("create"))
print("Dispatch delete:", dispatch("delete"))
print("Dispatch update:", dispatch("update"))
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: globals() returns dict
assert isinstance(globals(), dict)

# Test 2: Standard dunder exists
assert "__name__" in globals()

# Test 3: Runtime mutation updates module scope
globals()["unit_test_var"] = 99
assert unit_test_var == 99

# Test 4: Dynamic callable dispatch
def ping(): return "pong"
assert globals()["ping"]() == "pong"

# Test 5: Cleanup dynamic variable
del globals()["unit_test_var"]
assert "unit_test_var" not in globals()

print("All globals() unit assertions successfully verified! 100% passed.")
```
