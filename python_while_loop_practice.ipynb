### 1. Basic while Loop

```python
i = 0
while i < 5:
    print("i =", i)
    i = i + 1
```

### 2. while-else with break and continue

```python
# With continue (else executes)
x = 0
while x < 4:
    x = x + 1
    if x == 2:
        continue
    print("x =", x)
else:
    print("Natural loop completion!")

# With break (else skipped)
y = 0
while y < 4:
    y = y + 1
    if y == 2:
        break
    print("y =", y)
else:
    print("This won't print!")
```

### 3. Prime Number Testing Algorithm

```python
def check_prime(n):
    if n < 2:
        return False
    d = 2
    while d <= n // 2:
        if n % d == 0:
            return False
        d = d + 1
    else:
        return True

print("Is 17 prime?", check_prime(17))
print("Is 24 prime?", check_prime(24))
```

### 4. Emulating do-while Loop

```python
count = 0
while True:
    print("Step:", count)
    count = count + 1
    if count >= 3:
        break
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Iterations
c, k = 0, 0
while k < 5:
    c += 1
    k += 1
assert c == 5 and k == 5

# Test 2: Factor testing
def is_p(v):
    if v < 2: return False
    d = 2
    while d <= v // 2:
        if v % d == 0: return False
        d += 1
    return True
primes = [p for p in range(2, 20) if is_p(p)]
assert primes == [2, 3, 5, 7, 11, 13, 17, 19]

print("All while loop unit assertions successfully verified! 100% passed.")
```
