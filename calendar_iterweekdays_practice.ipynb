## 1. First example — Monday through Sunday
In the weekday numbering system, Monday is `0` and Sunday is `6`.

```python
import calendar
my_cal = calendar.Calendar()
for x in my_cal.iterweekdays():
    print(x)
```

```python
week = list(calendar.Calendar().iterweekdays())
assert week == [0, 1, 2, 3, 4, 5, 6]
print("Check passed: Monday-first week")
```

## 2. Start the week on Friday
The same seven weekday numbers are returned in rotated order.

```python
my_cal = calendar.Calendar(firstweekday=4)
for x in my_cal.iterweekdays():
    print(x)
assert list(my_cal.iterweekdays()) == [4, 5, 6, 0, 1, 2, 3]
```

## 3. Display full and abbreviated weekday names

```python
my_cal = calendar.Calendar(firstweekday=4)
for day in my_cal.iterweekdays():
    print(day, calendar.day_name[day])
assert calendar.day_name[4] == "Friday"
```

```python
my_cal = calendar.Calendar(firstweekday=0)
for day in my_cal.iterweekdays():
    print(calendar.day_abbr[day], end=" ")
print()
```

## 4. Start on Wednesday

```python
my_cal = calendar.Calendar(firstweekday=2)
for day in my_cal.iterweekdays():
    print(day)
assert list(my_cal.iterweekdays()) == [2, 3, 4, 5, 6, 0, 1]
```

## 5. Recurring tasks for all seven days

```python
my_cal = calendar.Calendar(firstweekday=0)
for day in my_cal.iterweekdays():
    print("Assign task for day:", day)
```

## 6. Working-day schedule (Monday to Friday)

```python
my_cal = calendar.Calendar(firstweekday=6)
working_days = []
for day in my_cal.iterweekdays():
    if day < 5:
        working_days.append(day)
        print(calendar.day_name[day], "- working day")
assert working_days == [0, 1, 2, 3, 4]
```

## 7. Convert to a list

```python
my_cal = calendar.Calendar(firstweekday=6)
weekdays = list(my_cal.iterweekdays())
print(weekdays)
print("First:", weekdays[0])
print("Total:", len(weekdays))
assert len(weekdays) == 7
```

## 8. Change the starting weekday on an existing object

```python
my_cal = calendar.Calendar()
print(list(my_cal.iterweekdays()))
my_cal.setfirstweekday(calendar.SUNDAY)
print(list(my_cal.iterweekdays()))
print("First day number:", my_cal.getfirstweekday())
assert my_cal.getfirstweekday() == 6
```

## 9. Sunday-first display names

```python
my_cal = calendar.Calendar(firstweekday=calendar.SUNDAY)
print([calendar.day_name[d] for d in my_cal.iterweekdays()])
```

## 10. Match each day with a specific task

```python
tasks = {
    calendar.MONDAY: "Review work",
    calendar.TUESDAY: "Write code",
    calendar.WEDNESDAY: "Test changes",
    calendar.THURSDAY: "Update notes",
    calendar.FRIDAY: "Share report",
    calendar.SATURDAY: "Rest",
    calendar.SUNDAY: "Plan next week",
}
my_cal = calendar.Calendar(firstweekday=calendar.MONDAY)
for day in my_cal.iterweekdays():
    print(f"{calendar.day_name[day]}: {tasks[day]}")
assert tasks[calendar.FRIDAY] == "Share report"
```

## 11. Why a saved iterator becomes empty

```python
my_cal = calendar.Calendar()
days = my_cal.iterweekdays()
print(next(days))
print(list(days))
print(list(days))
print(list(my_cal.iterweekdays()))
```

## 12. Week order vs the weekday of a date

```python
from datetime import date
my_cal = calendar.Calendar(firstweekday=calendar.SUNDAY)
print("Week order:", list(my_cal.iterweekdays()))
sample_date = date(2026, 1, 1)
print("Date weekday:", calendar.weekday(sample_date.year, sample_date.month, sample_date.day))
print("Date name:", calendar.day_name[sample_date.weekday()])
assert sample_date.weekday() == 3
```

## 13. Validate the chosen weekday

```python
start_day = 8
if not isinstance(start_day, int) or not 0 <= start_day <= 6:
    print("Choose a weekday number from 0 to 6")
else:
    my_cal = calendar.Calendar(firstweekday=start_day)
    print(list(my_cal.iterweekdays()))
```

## Solutions and self-checking answers
Open these after trying the exercises independently.

```python
# Solution 1
weekdays = list(calendar.Calendar(firstweekday=calendar.TUESDAY).iterweekdays())
assert weekdays == [1, 2, 3, 4, 5, 6, 0]
print("Exercise 1 passed:", weekdays)
```

```python
# Solution 2
abbr_names = [calendar.day_abbr[d] for d in calendar.Calendar(firstweekday=calendar.SUNDAY).iterweekdays()]
assert abbr_names == ['Sun', 'Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat']
print("Exercise 2 passed:", abbr_names)
```

```python
# Solution 3
my_cal = calendar.Calendar(firstweekday=calendar.FRIDAY)
working_days = [day for day in my_cal.iterweekdays() if day < 5]
assert working_days == [4, 0, 1, 2, 3]
print("Exercise 3 passed:", working_days)
```

```python
# Solution 4
for first in range(7):
    days = list(calendar.Calendar(firstweekday=first).iterweekdays())
    assert len(days) == 7
    assert sorted(days) == list(range(7))
    assert days[0] == first
print("Exercise 4 passed: all 7 starting weekdays")
```

```python
# Solution 5
my_cal = calendar.Calendar(firstweekday=calendar.THURSDAY)
schedule = [(calendar.day_name[day], day < 5) for day in my_cal.iterweekdays()]
assert schedule[0] == ('Thursday', True)
assert schedule[2] == ('Saturday', False)
print(schedule)
print("Exercise 5 passed")
```
