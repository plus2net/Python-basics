Run cells in order in Google Colab. This notebook demonstrates how to convert XML files to JSON using both Pandas (`read_xml` + `to_json`) and Python's standard `xml.etree.ElementTree` and `json` modules. Save a copy in Google Drive to keep your edits.

### 1. Convert XML to JSON using Pandas

```python
import pandas as pd

# Create sample XML file
xml_data = """<data>
    <record>
        <id>1</id>
        <name>John Doe</name>
        <age>30</age>
        <city>New York</city>
    </record>
    <record>
        <id>2</id>
        <name>Jane Smith</name>
        <age>25</age>
        <city>Los Angeles</city>
    </record>
</data>"""

with open("data.xml", "w") as f:
    f.write(xml_data)

# Read XML and export to JSON
df = pd.read_xml("data.xml")
df.to_json("data.json", orient='records', indent=4)
print("JSON file generated successfully using Pandas:")
print(df)
```

### 2. Convert XML to JSON using ElementTree and JSON module

```python
import xml.etree.ElementTree as ET
import json

tree = ET.parse("data.xml")
root = tree.getroot()

records = []
for record in root.findall('record'):
    data = {}
    for element in record:
        data[element.tag] = element.text
    records.append(data)

with open("data_et.json", "w", encoding="utf-8") as json_file:
    json.dump(records, json_file, indent=4)

print("JSON file generated successfully using ElementTree!")
```
