## Example 2: Recursive GCD: the original approach

Read the code, predict the output and run this cell.

```python
def gcd_recursive(a, b):
    if b == 0:
        return abs(a)
    return gcd_recursive(b, a % b)
print(gcd_recursive(85, 20))
```

## Example 3: Iterative Euclidean algorithm

Read the code, predict the output and run this cell.

```python
def get_gcd(a, b):
    while b:
        a, b = b, a % b
    return abs(a)
for pair in ((85, 20), (54, 24), (-12, 18)):
    print(pair, get_gcd(*pair))
```

## Example 5: LCM using GCD

Read the code, predict the output and run this cell.

```python
from math import gcd
def get_lcm(a, b):
    if a == 0 or b == 0:
        return 0
    return abs(a * b) // gcd(a, b)
for a, b in ((20, 12), (0, 12), (-6, 8)):
    print(a, b, get_lcm(a, b))
```

## Example 6: Use math.gcd() and math.lcm()

Read the code, predict the output and run this cell.

```python
from math import gcd, lcm
for a, b in ((0, 0), (0, 12), (-6, 8), (18, 24)):
    print((a, b), "gcd", gcd(a, b), "lcm", lcm(a, b))
```
