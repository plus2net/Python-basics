### 1. Basic Generator Function with yield

```python
def count_up_to(max_val):
    count = 1
    while count <= max_val:
        yield count
        count += 1

counter = count_up_to(3)
print("Item 1:", next(counter))
print("Item 2:", next(counter))
print("Item 3:", next(counter))
```

### 2. Memory Footprint Benchmark: List vs. Generator

```python
import sys

# List allocated in heap memory
full_list = [n * 2 for n in range(500_000)]
print("List RAM usage:     ", sys.getsizeof(full_list), "bytes")

# Generator allocated lazily
lazy_gen = (n * 2 for n in range(500_000))
print("Generator RAM usage:", sys.getsizeof(lazy_gen), "bytes")
```

### 3. Two-Way Communication with send()

```python
def running_total():
    total = 0
    while True:
        val = yield total
        if val is not None:
            total += val

acc = running_total()
print("Primed:  ", next(acc))
print("Add 5:   ", acc.send(5))
print("Add 20:  ", acc.send(20))
acc.close()
```

### 4. Delegation with yield from

```python
def odds():
    yield 1
    yield 3

def evens():
    yield 2
    yield 4

def combined():
    yield from odds()
    yield from evens()

print("Combined output:", list(combined()))
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Fibonacci generator check
def fib(limit):
    a, b = 0, 1
    for _ in range(limit):
        yield b
        a, b = b, a + b

assert list(fib(5)) == [1, 1, 2, 3, 5]

# Test 2: Generator expression
squares = (x * x for x in [1, 2, 3])
assert next(squares) == 1
assert next(squares) == 4
assert next(squares) == 9

# Test 3: StopIteration on exhaust
try:
    next(squares)
    assert False
except StopIteration:
    pass

print("All generator unit assertions successfully verified! 100% passed.")
```
