## 2. Define Sub-Components (Bar Chart, Line Chart, Table, Paragraphs, Circles)

```python
from reportlab.graphics.shapes import Drawing, Rect, String, Circle
from reportlab.graphics.charts.barcharts import VerticalBarChart
from reportlab.graphics.charts.lineplots import LinePlot
from reportlab.platypus import Table, TableStyle, Paragraph
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib import colors

# 1. Vertical Bar Chart
d = Drawing(200, 150)
bc = VerticalBarChart()
bc.x = 20
bc.y = 20
bc.height = 100
bc.width = 160
bc.data = [[10, 20, 30, 40]]
bc.categoryAxis.categoryNames = ['Q1', 'Q2', 'Q3', 'Q4']
d.add(bc)

# 2. Table
table_data = [
    ['Quarter', 'Sales', 'Profit'],
    ['Q1', '$10K', '$2K'],
    ['Q2', '$20K', '$5K'],
    ['Q3', '$30K', '$8K'],
    ['Q4', '$40K', '$12K']
]
t = Table(table_data, colWidths=[60, 60, 60])
t.setStyle(TableStyle([
    ('BACKGROUND', (0,0), (-1,0), colors.navy),
    ('TEXTCOLOR', (0,0), (-1,0), colors.whitesmoke),
    ('ALIGN', (0,0), (-1,-1), 'CENTER'),
    ('GRID', (0,0), (-1,-1), 1, colors.grey)
]))

# 3. Line Chart
line_chart = Drawing(300, 120)
lp = LinePlot()
lp.x = 20
lp.y = 20
lp.height = 80
lp.width = 260
lp.data = [[(1, 5), (2, 15), (3, 25), (4, 40)]]
line_chart.add(lp)

# 4. Paragraphs
styles = getSampleStyleSheet()
style_para = ParagraphStyle('ReportStyle', parent=styles['Normal'], fontSize=9, leading=12)
p1 = Paragraph('<b>Executive Summary:</b><br>Overall quarterly report performance overview.', style_para)
p2 = Paragraph('<b>Financial Breakdown:</b><br>Quarterly breakdown with growth metrics.', style_para)
p3 = Paragraph('<b>Projections:</b><br>Future growth trends and forecast projections.', style_para)

# 5. Circles / Shapes
d_circle = Drawing(80, 80)
d_circle.add(Circle(40, 40, 30, fillColor=colors.lightskyblue, strokeColor=colors.navy))

print("All modular sub-components initialized!")
```

## 3. Assemble Full PDF Report in main.py Script

```python
from pathlib import Path
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import letter
from reportlab.lib.units import inch

output_pdf = Path('output/my_pdf_report.pdf')
c = canvas.Canvas(str(output_pdf), pagesize=letter)
width, height = letter

# Draw Header Image
c.drawImage('images/top2.jpg', 0.1*inch, 10*inch, width=3*inch, preserveAspectRatio=True)

# Draw Bar Chart
d.wrapOn(c, 400, 400)
d.drawOn(c, 3*inch, 7.5*inch)

# Draw Table
t.wrapOn(c, width, height)
t.drawOn(c, 3.6*inch, 5.6*inch)

# Draw Paragraphs
p1.wrapOn(c, 200, 200)
p1.drawOn(c, 0.6*inch, 9.4*inch)

p2.wrapOn(c, 200, 200)
p2.drawOn(c, 0.6*inch, 6.4*inch)

p3.wrapOn(c, 200, 200)
p3.drawOn(c, 0.6*inch, 4.4*inch)

# Draw Shapes
d_circle.wrapOn(c, 100, 100)
d_circle.drawOn(c, 4*inch, 2.4*inch)

# Draw Line Chart
line_chart.wrapOn(c, 810, 200)
line_chart.drawOn(c, 0.6*inch, 0.1*inch)

c.save()
print(f"PDF Report generated successfully at: {output_pdf}")
```
