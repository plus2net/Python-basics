## 2. Define Card Template (`tk_id_card_temp.py`)

Define `my_temp(c)` function to draw card header labels, layout lines, copyright info, and a 35-degree rotated watermark.

```python
from reportlab.lib.units import inch

def my_temp(c):
    c.translate(inch, inch)
    c.setFont("Helvetica", 14)
    c.setStrokeColorRGB(0.1, 0.8, 0.1)
    c.setFillColorRGB(0, 0, 1)
    
    # Watermark
    c.rotate(35)
    c.setFillColorCMYK(0, 0, 0, 0.08)
    c.setFont("Helvetica", 100)
    c.drawString(-1.1 * inch, -0.5 * inch, "SAMPLE")
    c.rotate(-35)
    
    c.setFillColorRGB(1, 0, 0)
    c.setFont("Helvetica", 25)
    c.drawRightString(1.7 * inch, 2.3 * inch, 'Identity Card')
    
    c.setFillColorRGB(0, 0, 0)
    c.setFont("Helvetica", 24)
    c.drawRightString(0.3 * inch, 1.7 * inch, 'ID:')
    c.drawRightString(0.3 * inch, 1.3 * inch, 'Name:')
    c.drawRightString(0.3 * inch, 0.9 * inch, 'Class:')
    c.drawRightString(0.3 * inch, 0.5 * inch, 'Gender:')
    c.drawRightString(4.0 * inch, -0.5 * inch, 'Signature')
    
    c.line(-1.1, -0.7 * inch, 5 * inch, -0.7 * inch)
    c.setFont("Helvetica", 8)
    c.setFillColorRGB(1, 0, 0)
    c.drawString(0, -0.9 * inch, "\u00A9 plus2net.com")
    
    return c

print("ID card template function initialized!")
```

## 3. Simulated Tkinter Form Data to PDF Generation

In Google Colab, desktop Tkinter GUI windows cannot open directly. We simulate user input fields (`StringVar` objects) collected from GUI forms and build the PDF ID card.

```python
from pathlib import Path
from reportlab.pdfgen import canvas

# Simulated Tkinter StringVar containers
class SimulatedStringVar:
    def __init__(self, value):
        self._value = value
    def get(self):
        return self._value

s_name = SimulatedStringVar('Kalu')
s_id = SimulatedStringVar('16')
s_class = SimulatedStringVar('Four')
s_gender = SimulatedStringVar('Female')
s_filename = 'photos/sample_photo.jpg'

output_dir = Path('output')
pdf_id = output_dir / 'tkinter_id_card.pdf'

c = canvas.Canvas(str(pdf_id), pagesize=(400, 300))
c = my_temp(c)

if s_filename:
    c.drawImage(s_filename, 2.2 * inch, 0.7 * inch, width=1.5*inch, height=1.5*inch)

c.setFillColorRGB(0, 0, 1)
c.setFont("Helvetica", 20)
c.drawString(0.5 * inch, 1.7 * inch, str(s_id.get()))
c.drawString(0.5 * inch, 1.3 * inch, s_name.get())
c.drawString(0.5 * inch, 0.9 * inch, s_class.get())
c.drawString(0.5 * inch, 0.5 * inch, s_gender.get())

c.showPage()
c.save()
print(f"ID Card PDF generated successfully: {pdf_id}")
```
