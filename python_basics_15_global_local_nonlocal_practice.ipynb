## 1. Local and Global Variables

A variable created outside a function is available in the module-level scope. A variable assigned inside a function is local to that function unless we declare it as `global` or `nonlocal`.


````python
n = 5  # global

def my_fun():
    n = 6  # local variable
    print("Inside function:", n)

my_fun()
print("Outside function:", n)
````

**Expected output**

```text
Inside function: 6
Outside function: 5
```

The local variable inside `my_fun()` does not change the module-level variable.


## 2. Changing a Local Variable

The assignment inside the function works only with the local variable.


````python
n = 5

def my_fun():
    n = 10
    n = n + 1
    print("Inside function:", n)

my_fun()
print("Outside function:", n)
````

## 3. Modifying a Global Variable with `global`

We can read a global variable without using `global`. We use `global` when an assignment inside the function should update the module-level variable.


````python
n = 5

def my_fun():
    global n
    n = n + 1
    print("Inside function:", n)

my_fun()
n = n + 1
print("Outside function:", n)
````

## 4. LEGB Scope Resolution

Python searches for a name in this order: **Local → Enclosing → Global → Built-in**.


````python
name = "Global"

def outer():
    name = "Enclosing"

    def inner():
        name = "Local"
        print(name)

    inner()

outer()
````

## 5. Using `nonlocal` in Nested Functions

Without `nonlocal`, the inner function creates its own local variable.


````python
n = 5

def my_fun1():
    n = 10

    def my_fun2():
        n = 20
        print("Inside n:", n)

    my_fun2()
    print("Outside n:", n)

my_fun1()
print("Main n:", n)
````

Now use `nonlocal` so the inner function updates the variable from the enclosing function.


````python
n = 5

def my_fun1():
    n = 10

    def my_fun2():
        nonlocal n
        n = 20
        print("Inside n:", n)

    my_fun2()
    print("Outside n:", n)

my_fun1()
print("Main n:", n)
````

## 6. Difference Between `global` and `nonlocal`

- `global` modifies a **module-level** variable.
- `nonlocal` modifies a variable in the **nearest enclosing function**.


````python
x = 10

def change_value():
    global x
    x = 20

change_value()
print(x)
````

````python
def outer():
    x = 10

    def inner():
        nonlocal x
        x = 20

    inner()
    print(x)

outer()
````

## 7. Reading a Global Variable Without `global`

If the function only reads the value, `global` is not required.


````python
tax_rate = 18

def show_tax_rate():
    print(tax_rate)

show_tax_rate()
````

## 8. Why Assignment Can Cause `UnboundLocalError`

Because `score = 60` appears inside the function, Python treats `score` as a local name throughout the function.


````python
score = 50

def update_score():
    print(score)
    score = 60

# Uncomment the next line to see UnboundLocalError
# update_score()
````

Fix the same example with `global`.


````python
score = 50

def update_score():
    global score
    print(score)
    score = 60

update_score()
print(score)
````

## 9. Augmented Assignment and Scope

Statements such as `+=` read and assign a value, so they can also cause `UnboundLocalError`.


````python
visits = 10

def add_visit():
    global visits
    visits += 1

add_visit()
print(visits)
````

## 10. Changing a Mutable Global Object

We can modify the contents of a global list without declaring the list name as `global`.


````python
students = ["Alex", "Ron"]

def add_student():
    students.append("Ravi")

add_student()
print(students)
````

## 11. Using `nonlocal` to Keep State Between Calls

A nested function can remember and update a value from its enclosing function.


````python
def create_counter():
    count = 0

    def counter():
        nonlocal count
        count += 1
        return count

    return counter

my_counter = create_counter()

print(my_counter())
print(my_counter())
print(my_counter())
````
