## 3. Database Connection using SQL Magic Extension

```python
%load_ext sql
%sql sqlite:///my_db.db
```

```python
import prettytable
prettytable.__dict__['ASCII'] = prettytable.MARKDOWN
%config SqlMagic.style = 'ASCII'
%sql SELECT name FROM sqlite_master WHERE type='table';
```

```python
%sql PRAGMA table_info([student]);
```

```python
%%sql
SELECT * FROM student LIMIT 0,5;
```

```python
result = %sql SELECT * FROM student LIMIT 0,5;
print(result.loc[:, 'name'])
```

## 4. Connecting using SQLAlchemy (v2.0+ Standard)

```python
from sqlalchemy import create_engine, text

# Create connection engine
my_conn = create_engine("sqlite:///my_db.db")

# Context-managed query execution
with my_conn.connect() as connection:
    query = text("SELECT name FROM sqlite_master WHERE type='table'")
    r_set = connection.execute(query)
    for row in r_set:
        print(row)
```

```python
with my_conn.connect() as connection:
    query = text("SELECT * FROM student LIMIT 0,5")
    r_set = connection.execute(query)
    for row in r_set:
        print(row)
```

## 5. Connecting using sqlite3

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
query = "SELECT * FROM student LIMIT 0,5"
my_data = list(my_conn.execute(query))
print(my_data)
my_conn.close()
```
