### 1. Basic Float Conversions
Convert integers, formatted numeric strings, and scientific notation.

```python
print("float():        ", float())
print("float(5):       ", float(5))
print("float('  34.5 '):", float('  34.5 '))
print("float('1e-3'):   ", float('1e-3'))
```

### 2. Special Values: inf and nan
Inspect infinity and NaN (Not a Number).

```python
import math
inf = float('inf')
nan = float('nan')
print("math.isinf(inf):", math.isinf(inf))
print("math.isnan(nan):", math.isnan(nan))
print("nan == nan:     ", nan == nan)  # False!
```

### 3. Custom Class __float__() Implementation
Implement the float conversion protocol on custom classes.

```python
class Gauge:
    def __init__(self, val):
        self.val = val
    def __float__(self):
        return float(self.val)

g = Gauge(42.5)
print("float(g):", float(g))
assert float(g) == 42.5
```

### 4. Production Recipe: Safe Float Parser
Build a resilient parser that handles commas, currency, and defaults.

```python
def parse_price(text, default=0.0):
    try:
        clean = text.replace('$', '').replace(',', '').strip()
        return float(clean)
    except (ValueError, TypeError, AttributeError):
        return default

print("Parse '$1,500.25':", parse_price('$1,500.25'))
print("Parse 'invalid':  ", parse_price('invalid', -1.0))
assert parse_price('$1,500.25') == 1500.25
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying float() mechanics.

```python
assert float() == 0.0
assert float(10) == 10.0
assert float('3.14') == 3.14
assert float('  2.5  ') == 2.5
assert float('1e-2') == 0.01

print("All float() unit assertions passed! 100% verified.")
```
