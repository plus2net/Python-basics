### 1. Execute XPath Queries with lxml

```python
# !pip install lxml
from lxml import etree

xml_data = """
<students>
    <student>
        <id>1</id>
        <name>John Deo</name>
        <mark>75</mark>
    </student>
    <student>
        <id>2</id>
        <name>Jane Smith</name>
        <mark>85</mark>
    </student>
</students>
"""

root = etree.fromstring(xml_data)

# Extract all student names
names = root.xpath("//student/name/text()")
print("All Names:", names)

# Extract student with ID 2
student = root.xpath("//student[id='2']/name/text()")
print("Student ID 2:", student)

# Extract high scorers (> 80)
high_scorers = root.xpath("//student[mark > 80]/name/text()")
print("High Scorers (>80):", high_scorers)
```
