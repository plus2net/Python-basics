### 1. Basic str.isalnum() Validation
`str.isalnum()` returns `True` if all characters are letters or numbers, and length > 0.

```python
print("'plus2net'.isalnum():  ", 'plus2net'.isalnum())
print("'Python311'.isalnum(): ", 'Python311'.isalnum())
print("'2026'.isalnum():      ", '2026'.isalnum())

# False triggers: spaces, underscores, symbols
print("'plus 2net'.isalnum(): ", 'plus 2net'.isalnum())
print("'user_name'.isalnum(): ", 'user_name'.isalnum())
print("''.isalnum():          ", ''.isalnum())

assert 'plus2net'.isalnum() is True
assert 'plus 2net'.isalnum() is False
assert 'user_name'.isalnum() is False
assert ''.isalnum() is False
```

### 2. Unicode and International Characters
`str.isalnum()` supports global alphabetic and numeric systems (Cyrillic, Greek, Japanese, Devanagari digits).

```python
samples = ['Café101', 'München', 'Москва2026', 'Αθήνα', '日本語', '१२३', 'x²']

for s in samples:
    print(f"{s:<12}: {s.isalnum()}")
    assert s.isalnum() is True

# Note: Indic words with nonspacing combining marks (category Mn) evaluate to False
print("'नमस्ते'.isalnum():", 'नमस्ते'.isalnum())
assert 'नमस्ते'.isalnum() is False
```

### 3. Comparison with isalpha(), isdigit(), and isidentifier()
Compare behavior across common string formats.

```python
test_cases = ['hello', 'hello123', '123', 'hello_world', '_var1', 'hello world']
print(f"{'String':<15} | {'isalnum':<8} | {'isalpha':<8} | {'isdigit':<8} | {'isidentifier':<12}")
print("-" * 60)
for t in test_cases:
    print(f"{t:<15} | {str(t.isalnum()):<8} | {str(t.isalpha()):<8} | {str(t.isdigit()):<8} | {str(t.isidentifier()):<12}")

assert 'hello123'.isalnum() and not 'hello123'.isalpha()
assert 'hello_world'.isidentifier() and not 'hello_world'.isalnum()
```

### 4. Text Sanitization and Punctuation Stripping
Filter out non-alphanumeric noise using list comprehensions.

```python
raw_str = "Item #492-B (Discount: 15%!)"
clean_token = "".join(ch for ch in raw_str if ch.isalnum())
print("Raw string:  ", raw_str)
print("Clean token: ", clean_token)
assert clean_token == "Item492BDiscount15"
```

### 5. Practice Exercise: Account Handle & Serial Validator
Run the complete validator testing usernames and serial number normalization.

```python
class AccountValidator:
    def __init__(self, min_len=4, max_len=15):
        self.min_len = min_len
        self.max_len = max_len

    def is_valid_handle(self, handle):
        if not isinstance(handle, str):
            return False
        if not (self.min_len <= len(handle) <= self.max_len):
            return False
        return handle.isalnum()

    def sanitize_serial(self, raw_serial):
        if not isinstance(raw_serial, str):
            return ""
        return "".join(ch.upper() for ch in raw_serial if ch.isalnum())

# Unit Assertions
validator = AccountValidator(min_len=4, max_len=12)
assert validator.is_valid_handle("Alex2026") is True
assert validator.is_valid_handle("Dev_10") is False
assert validator.is_valid_handle("Sam 24") is False
assert validator.is_valid_handle("Bob") is False
assert validator.is_valid_handle("VeryLongUsernameIndeed") is False
assert validator.is_valid_handle("") is False
assert validator.is_valid_handle("München99") is True

assert validator.sanitize_serial("SN-8942-X10/US") == "SN8942X10US"
assert validator.sanitize_serial("### 99-AB-cd ###") == "99ABCD"

print("Validation tests passed successfully!")
```
