### 1. Basic str.rfind() Usage & Rightmost vs Leftmost Offsets
`str.rfind()` returns the highest (rightmost) 0-based index where the target substring is found.

```python
banner = "Welcome to plus2net.com Python section"

# Search for rightmost occurrence of "co"
# "co" appears in "Welcome" (index 3) and "plus2net.com" (index 20)
right_pos = banner.rfind("co")
left_pos = banner.find("co")

print("Full string:", banner)
print("First occurrence  (str.find): ", left_pos)   # 3
print("Last occurrence   (str.rfind):", right_pos)  # 20
```

### 2. Handling Missing Targets: Return Value -1 vs ValueError
`rfind()` returns -1 when a substring is missing, avoiding exceptions. In contrast, `rindex()` raises a ValueError.

```python
missing = banner.rfind("Java")
print("Missing target index:", missing)  # Output: -1

# Safe existence verification
if banner.rfind("Python") != -1:
    print("Target found via rfind()")
else:
    print("Target missing")
```

### 3. Constraining Search Windows with start and end Offsets
You can limit the search interval using start and end bounds, including negative slice indices.

```python
phrase = "alpha_beta_gamma_delta"

# Search across the entire string
print("Full string rightmost '_':", phrase.rfind("_"))

# Search restricted to index range [0:10]
print("In slice [0:10]:           ", phrase.rfind("_", 0, 10))

# Search using negative start offset (last 10 characters)
print("In last 10 characters:     ", phrase.rfind("_", -10))
```

### 4. Real-World Recipe: File Extension Extraction
Extract the file base name and extension by finding the rightmost period separator.

```python
def extract_extension(filename: str) -> tuple:
    dot_pos = filename.rfind(".")
    if dot_pos <= 0:
        return filename, ""
    return filename[:dot_pos], filename[dot_pos + 1:]

files = ["report.final.pdf", "archive.tar.gz", ".gitignore", "README"]
for f in files:
    stem, ext = extract_extension(f)
    print(f"{f:20} -> Stem: {stem:15} Ext: {ext}")
```

### 5. Real-World Recipe: Word-Boundary Text Truncation
Truncate long text cleanly at word boundaries instead of cutting words in half.

```python
def truncate_at_word_boundary(text: str, max_chars: int = 50) -> str:
    if len(text) <= max_chars:
        return text
    cutoff = text.rfind(" ", 0, max_chars)
    if cutoff == -1:
        return text[:max_chars]
    return text[:cutoff]

article_teaser = "The quick brown fox jumps gracefully over the lazy sleeping dog."
print("Original:", article_teaser)
print("Max 25:  ", truncate_at_word_boundary(article_teaser, 25))
print("Max 40:  ", truncate_at_word_boundary(article_teaser, 40))
```

### 6. Real-World Recipe: Directory Path Splitting
Split a path into parent directory and filename without extra imports.

```python
def split_directory_path(path: str) -> tuple:
    slash_pos = path.rfind("/")
    if slash_pos == -1:
        return "", path
    return path[:slash_pos], path[slash_pos + 1:]

paths = ["/var/log/nginx/access.log", "docs/guide.txt", "server.py"]
for p in paths:
    dir_part, file_part = split_directory_path(p)
    print(f"{p:30} -> Dir: {dir_part:20} File: {file_part}")
```

### 7. Self-Checking Verification Suite
Run complete unit assertions verifying all rfind behavior and edge cases.

```python
class RightmostFinder:
    @staticmethod
    def last_occurrence(text: str, sub: str) -> int:
        return text.rfind(sub)

    @staticmethod
    def get_extension(path: str) -> str:
        dot = path.rfind(".")
        return path[dot + 1:] if dot > 0 else ""

# Verification assertions
finder = RightmostFinder()
assert finder.last_occurrence("banana", "a") == 5
assert finder.last_occurrence("banana", "na") == 4
assert finder.last_occurrence("banana", "z") == -1
assert finder.get_extension("archive.tar.gz") == "gz"
assert finder.get_extension(".hidden") == ""
assert finder.get_extension("no_ext") == ""
assert truncate_at_word_boundary("one two three four", 11) == "one two"
assert split_directory_path("/etc/hosts") == ("/etc", "hosts")

print("All assertions passed successfully! 100% verified.")
```
