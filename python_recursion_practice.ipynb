## Example 1: Countdown with a base case

Read the code, predict the output and run this cell.

```python
def countdown(n):
    if n < 0:
        print("Blastoff!")
        return
    print(n, end=", ")
    countdown(n - 1)
countdown(5)
```

## Example 2: Factorial recursively

Read the code, predict the output and run this cell.

```python
def factorial(n):
    if n < 0:
        raise ValueError("Use n >= 0")
    if n <= 1:
        return 1
    return n * factorial(n - 1)
print(factorial(0), factorial(4), factorial(6))
```

## Example 3: Catch RecursionError

Read the code, predict the output and run this cell.

```python
def recursive_forever():
    return recursive_forever()
try:
    recursive_forever()
except RecursionError:
    print("Recursion depth exceeded")
```

## Example 4: Memoized Fibonacci

Read the code, predict the output and run this cell.

```python
from functools import lru_cache
@lru_cache(None)
def fib(n):
    if n < 2:
        return n
    return fib(n-1) + fib(n-2)
print(fib(10), fib(35))
```

## Example 5: Recursive palindrome check

Read the code, predict the output and run this cell.

```python
def is_palindrome(s):
    if len(s) < 2:
        return True
    return s[0] == s[-1] and is_palindrome(s[1:-1])
print(is_palindrome("radar"), is_palindrome("python"))
```

## Example 6: Recursive list flattening

Read the code, predict the output and run this cell.

```python
def flatten(items):
    result = []
    for item in items:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result
print(flatten([1, [2, [3, 4]], 5]))
```
