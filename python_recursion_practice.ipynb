## Example 2: Factorial recursively

Read the code, predict the output and run this cell.

````python
def factorial(n):
    if n < 0:
        raise ValueError("Use n >= 0")
    if n <= 1:
        return 1
    return n * factorial(n - 1)
print(factorial(0), factorial(4), factorial(6))
````

````python
def recursive_forever():
    return recursive_forever()
try:
    recursive_forever()
except RecursionError:
    print("Recursion depth exceeded")
````

## Example 4: Memoized Fibonacci

Read the code, predict the output and run this cell.

````python
from functools import lru_cache
@lru_cache(None)
def fib(n):
    if n < 2:
        return n
    return fib(n-1) + fib(n-2)
print(fib(10), fib(35))
````

## Example 5: Recursive palindrome check

Read the code, predict the output and run this cell.

````python
def is_palindrome(s):
    if len(s) < 2:
        return True
    return s[0] == s[-1] and is_palindrome(s[1:-1])
print(is_palindrome("radar"), is_palindrome("python"))
````

## Example 6: Recursive list flattening

Read the code, predict the output and run this cell.

````python
def flatten(items):
    result = []
    for item in items:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result
print(flatten([1, [2, [3, 4]], 5]))
````

### Additional tutorial example 6

**Run this example from the main tutorial:**

Tip: If the example requests input, run the cell and type a value.

````python
# 5. Fibonacci: Naive O(2^N) vs. Memoized O(N)
import functools

@functools.lru_cache(maxsize=None)
def fib(n: int) -> int:
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    return fib(n - 1) + fib(n - 2)

print("fib(10):", fib(10))
print("fib(35):", fib(35))
````

### Additional tutorial example 7

**Run this example from the main tutorial:**

Tip: If the example requests input, run the cell and type a value.

````python
# 6. Production Recipe: Deep Nested List Flattener
def flatten(items: list) -> list:
    """Recursively flattens an arbitrary nested list."""
    flat = []
    for x in items:
        if isinstance(x, list):
            flat.extend(flatten(x))
        else:
            flat.append(x)
    return flat

nested = [1, [2, [3, 4], 5], [6, 7]]
print("Flattened list:", flatten(nested))
````

### Additional tutorial example 8

**Run this example from the main tutorial:**

````python
# 7. Production Recipe: Recursive Palindrome Verifier
def is_palindrome(s: str) -> bool:
    s = s.lower().replace(" ", "")
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return is_palindrome(s[1:-1])

print("radar:   ", is_palindrome("radar"))
print("plus2net:", is_palindrome("plus2net"))
````

### Additional tutorial example 9

**Run this example from the main tutorial:**

Tip: If the example requests input, run the cell and type a value.

````python
# 8. Avoiding common recursion traps
# Trap 1: Forgetting the base case triggers RecursionError
# Trap 2: Using mutable default arguments across recursive frames
def bad_collect(item, bucket=[]):  # Persists across recursive calls!
    bucket.append(item)
    return bucket

def safe_collect(item, bucket=None):  # Fresh instance per invocation
    if bucket is None:
        bucket = []
    bucket.append(item)
    return bucket

print("Safe collect:", safe_collect("A"))
````
