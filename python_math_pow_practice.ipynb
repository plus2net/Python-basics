### 1. Basic Exponentiation with math.pow()
`math.pow(x, y)` converts both arguments to float and calculates `x` raised to the power `y`.


```python
import math

print("pow(2, 3):    ", math.pow(2, 3))       # 8.0
print("pow(2, -3):   ", math.pow(2, -3))      # 0.125
print("pow(27, 1/3): ", math.pow(27, 1/3))   # 3.0
print("pow(-2, 3):   ", math.pow(-2, 3))      # -8.0
print("pow(-2, 4):   ", math.pow(-2, 4))      # 16.0
print("pow(2.5, 2.0):", math.pow(2.5, 2.0))    # 6.25
```

### 2. Triad Comparison: math.pow() vs. Built-in pow() vs. **
Observe how return types, precision, and complex root support differ between the three power mechanisms.


```python
import math

# Large integer exponentiation
print("2 ** 100:     ", 2 ** 100)           # Exact arbitrary-precision integer
print("math.pow(2,100):", math.pow(2, 100))    # Approximate float (53 bits)

# Complex roots
print("(-4) ** 0.5:   ", (-4) ** 0.5)       # Complex number
try:
    math.pow(-4, 0.5)
except ValueError as err:
    print("math.pow(-4, 0.5):", err)          # ValueError
```

### 3. Domain Constraints & Error Handling
Test exception handling on fractional powers of negative numbers, division by zero, and float overflow.


```python
import math

# 1. ValueError on fractional power of negative base
try:
    math.pow(-2, 0.5)
except ValueError as e:
    print("Caught ValueError:        ", e)

# 2. ZeroDivisionError on 0.0 ** negative
try:
    math.pow(0.0, -2)
except ZeroDivisionError as e:
    print("Caught ZeroDivisionError: ", e)

# 3. OverflowError
try:
    math.pow(10.0, 309)
except OverflowError as e:
    print("Caught OverflowError:     ", e)
```

### 4. IEEE 754 Special Edge Cases (The 1.0 Exponent Rule)
In IEEE 754-2008, 1.0 raised to any power (including NaN and Inf) is strictly 1.0.


```python
import math

print("pow(1.0, nan):", math.pow(1.0, float("nan")))
print("pow(1.0, inf):", math.pow(1.0, float("inf")))
print("pow(nan, 0.0):", math.pow(float("nan"), 0.0))
print("pow(inf, 0.0):", math.pow(float("inf"), 0.0))
```

### 5. Hands-on Practice Exercise: Loan Mortgage & Amortization Calculator
Calculate fixed monthly mortgage payments: `M = P * (r * (1 + r)^n) / ((1 + r)^n - 1)`.
Implement `calculate_monthly_mortgage(principal, annual_interest_rate, years)` and verify with self-checking assertions.


```python
import math

def calculate_monthly_mortgage(principal, annual_interest_rate, years):
    """
    Calculates fixed monthly mortgage payment and total interest paid:
    M = P * (r * (1 + r)^n) / ((1 + r)^n - 1)
    """
    if principal <= 0 or years <= 0:
        raise ValueError("Principal and loan duration must be positive.")
    if annual_interest_rate < 0:
        raise ValueError("Interest rate cannot be negative.")
        
    total_payments = years * 12
    
    # Special case: 0% interest loan
    if annual_interest_rate == 0.0:
        monthly_payment = principal / total_payments
        return (round(monthly_payment, 2), 0.0)
        
    monthly_r = (annual_interest_rate / 100.0) / 12.0
    growth_factor = math.pow(1.0 + monthly_r, total_payments)
    
    monthly_payment = principal * (monthly_r * growth_factor) / (growth_factor - 1.0)
    total_paid = monthly_payment * total_payments
    total_interest = total_paid - principal
    
    return (round(monthly_payment, 2), round(total_interest, 2))

# Unit Assertions
# Test Case 1: $200,000 loan at 6.5% for 30 years
m1, i1 = calculate_monthly_mortgage(200000.0, 6.5, 30)
assert m1 == 1264.14
assert i1 == 255088.98

# Test Case 2: $100,000 loan at 5.0% for 15 years
m2, i2 = calculate_monthly_mortgage(100000.0, 5.0, 15)
assert m2 == 790.79
assert i2 == 42342.85

# Test Case 3: 0% interest loan
m3, i3 = calculate_monthly_mortgage(12000.0, 0.0, 1)
assert m3 == 1000.0
assert i3 == 0.0

print(f"$200k mortgage (6.5%, 30yr): Monthly = ${m1} | Total Interest = ${i1}")
print("All loan amortization assertions passed successfully!")
```
