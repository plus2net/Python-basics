### 1. Basic Finite Verification with math.isfinite()
`math.isfinite(x)` returns `True` if `x` is neither an infinity nor a NaN.


```python
import math

print("isfinite(42):       ", math.isfinite(42))          # True
print("isfinite(-19.85):   ", math.isfinite(-19.85))       # True
print("isfinite(0.0):      ", math.isfinite(0.0))         # True
print("isfinite(-0.0):     ", math.isfinite(-0.0))        # True
print("isfinite(10**1000): ", math.isfinite(10**1000))     # True (arbitrary-precision integer)
print("isfinite(inf):      ", math.isfinite(float("inf")))  # False
print("isfinite(-inf):     ", math.isfinite(float("-inf"))) # False
print("isfinite(nan):      ", math.isfinite(float("nan")))  # False
```

### 2. The IEEE 754 Classification Triad
Compare `math.isfinite()`, `math.isinf()`, and `math.isnan()` across the full spectrum of floating-point values.


```python
import math

test_samples = [42.5, 0.0, -0.0, 1e-315, float("inf"), float("-inf"), float("nan")]
print(f"{'Value':>12} | {'isfinite':>8} | {'isinf':>8} | {'isnan':>8}")
print("-" * 44)
for val in test_samples:
    fin = math.isfinite(val)
    inf = math.isinf(val)
    nan = math.isnan(val)
    # Logical identity: isfinite == not (isinf or isnan)
    assert fin == (not (inf or nan))
    print(f"{str(val):>12} | {str(fin):>8} | {str(inf):>8} | {str(nan):>8}")
```

### 3. Data Cleaning: Filtering Non-Finite Elements
Filter lists and sequences using `math.isfinite()` in list comprehensions.


```python
import math

raw_measurements = [12.4, float("nan"), 15.8, float("inf"), -9.2, float("-inf"), 0.0]
clean_data = [val for val in raw_measurements if math.isfinite(val)]
print("Raw data:  ", raw_measurements)
print("Clean data:", clean_data)
```

### 4. Strict JSON Serialization Compliance (RFC 8259)
JSON does not support `inf` or `nan`. Setting `allow_nan=False` in `json.dumps()` enforces strict compliance.


```python
import json
import math

payload = {"status": "ok", "reading": float("inf")}
try:
    json.dumps(payload, allow_nan=False)
except ValueError as err:
    print("Caught expected JSON error:", err)

# Sanitizing before dumping
sanitized_payload = {k: (v if math.isfinite(v) else None) for k, v in payload.items() if isinstance(v, (int, float))}
print("Sanitized JSON:", json.dumps(sanitized_payload, allow_nan=False))
```

### 5. Hands-on Practice Exercise: IoT Sensor Telemetry Sanitizer
Implement `sanitize_telemetry(readings, fallback=None)` to validate dictionaries of metrics and replace non-finite floats with safe defaults.


```python
import math

def sanitize_telemetry(readings, fallback=None):
    """
    Sanitizes a dictionary of sensor readings.
    Any non-finite float (inf, -inf, nan) or non-numeric reading is replaced with fallback.
    """
    if not isinstance(readings, dict):
        raise TypeError("Readings must be provided as a dictionary.")
    
    sanitized = {}
    for sensor, value in readings.items():
        if isinstance(value, (int, float)) and math.isfinite(value):
            sanitized[sensor] = float(value)
        else:
            sanitized[sensor] = fallback
    return sanitized

# Test Data
raw_telemetry = {
    "temp_celsius": 24.5,
    "humidity_pct": float("nan"),
    "pressure_hpa": 1013.25,
    "voltage_v": float("inf"),
    "battery_pct": 85,
    "status_code": "ERR_BUS"
}

clean_telemetry = sanitize_telemetry(raw_telemetry, fallback=0.0)

# Unit Assertions
assert clean_telemetry["temp_celsius"] == 24.5
assert clean_telemetry["humidity_pct"] == 0.0     # NaN sanitized
assert clean_telemetry["pressure_hpa"] == 1013.25
assert clean_telemetry["voltage_v"] == 0.0        # Inf sanitized
assert clean_telemetry["battery_pct"] == 85.0
assert clean_telemetry["status_code"] == 0.0      # String sanitized

print("Sanitized Telemetry:", clean_telemetry)
print("All telemetry sanitization assertions passed successfully!")
```
