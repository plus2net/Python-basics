# Python math.tan(): Calculate Tangent Ratios

### 1. Basic Tangent Calculations in Radians
`math.tan(x)` returns the tangent of `x`, where `x` is measured strictly in radians.

```python
import math

print("tan(0):        ", math.tan(0))
print("tan(pi/4):     ", math.tan(math.pi / 4))
print("tan(1 radian): ", math.tan(1))
print("tan(pi):       ", math.tan(math.pi))
```

### 2. Converting Degrees to Radians
Convert degree angles using `math.radians()` before evaluating `math.tan()`.

```python
import math

degrees = [0, 30, 45, 60]
for deg in degrees:
    rad = math.radians(deg)
    print(f"tan({deg:2d} deg) = {math.tan(rad):.4f}")
```

### 3. The 90-Degree Asymptote Floating-Point Trap
At $90^\circ$ ($rac{\pi}{2}$), tangent is mathematically undefined. Because $\pi$ is represented as a finite float, Python returns a huge finite number rather than an error.

```python
import math

rad_90 = math.radians(90)
print("tan(90 deg):", math.tan(rad_90))
```

### 4. Trigonometric Symmetry & Period of pi
Verify odd symmetry $	an(-x) = -	an(x)$ and periodicity $	an(x + \pi) = 	an(x)$.

```python
import math

theta = 0.85
print("Odd symmetry: ", math.isclose(math.tan(-theta), -math.tan(theta)))
print("Period of pi: ", math.isclose(math.tan(theta + math.pi), math.tan(theta)))
```

### 5. Hands-on Practice Exercise: Structure Height Estimator
Run the self-checking unit assertions below to verify height calculations using trigonometric leveling: $h = d \cdot 	an(	heta) + h_{\text{eye}}$.

```python
def estimate_structure_height(distance_meters, elevation_degrees, eye_level_meters=1.6):
    if elevation_degrees >= 90 or elevation_degrees <= 0:
        raise ValueError("Elevation angle must be between 0 and 90 degrees.")
    theta_rad = math.radians(elevation_degrees)
    height = distance_meters * math.tan(theta_rad) + eye_level_meters
    return round(height, 2)

# Unit Assertions
assert estimate_structure_height(50, 45, 1.6) == 51.60
assert estimate_structure_height(50, 30, 1.6) == 30.47
assert estimate_structure_height(100, 60, 2.0) == 175.21

print("Building Height (50m away, 45 deg):", estimate_structure_height(50, 45), "meters")
print("Tower Height (100m away, 60 deg):   ", estimate_structure_height(100, 60, 2.0), "meters")
print("All structure height unit assertions passed successfully!")
```
