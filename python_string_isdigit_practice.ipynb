### 1. Basic str.isdigit() Validation
`str.isdigit()` returns `True` if all characters in the string are digits, and length > 0.

```python
print("'012345'.isdigit(): ", '012345'.isdigit())
print("'2026'.isdigit():   ", '2026'.isdigit())
print("'-42'.isdigit():    ", '-42'.isdigit())
print("'12.34'.isdigit():  ", '12.34'.isdigit())
print("''.isdigit():       ", ''.isdigit())

assert '012345'.isdigit() is True
assert '-42'.isdigit() is False
assert '12.34'.isdigit() is False
assert ''.isdigit() is False
```

### 2. The Numeric Hierarchy: isdecimal() vs. isdigit() vs. isnumeric()
Compare the three tiers of Python numeric string methods.

```python
samples = ['123', '१२३', '²', '①', '½', 'Ⅳ', '一千']
print(f"{'String':<8} | {'isdecimal':<10} | {'isdigit':<10} | {'isnumeric':<10}")
print("-" * 45)
for s in samples:
    print(f"{s:<8} | {str(s.isdecimal()):<10} | {str(s.isdigit()):<10} | {str(s.isnumeric()):<10}")

assert '²'.isdigit() and not '²'.isdecimal()
assert '½'.isnumeric() and not '½'.isdigit()
```

### 3. Parsing Signed Integers and Float Strings
Use string prefix checking and splitting to validate signed numbers and decimals.

```python
def is_signed_int(s):
    s = s.strip()
    if s.startswith(('+', '-')):
        s = s[1:]
    return s.isdigit()

print("'-42' signed int:", is_signed_int('-42'))
print("'+100' signed int:", is_signed_int('+100'))
assert is_signed_int('-42') is True
assert is_signed_int('+100') is True
assert is_signed_int('-') is False
```

### 4. Credit Card Digit Stream Extraction
Strip formatting delimiters and verify digit stream.

```python
card = "4532-1488-2940-1284"
clean_digits = card.replace('-', '').replace(' ', '')
print("Clean digits: ", clean_digits)
print("Is pure digits:", clean_digits.isdigit())
assert clean_digits == "4532148829401284"
assert clean_digits.isdigit() is True
```

### 5. Practice Exercise: Numeric Token & Authentication PIN Validator
Run the complete validator testing OTP codes and currency token parsing.

```python
class NumericValidator:
    def is_valid_pin(self, pin_code, allowed_lengths=(4, 6)):
        if not isinstance(pin_code, str):
            return False
        trimmed = pin_code.strip()
        return len(trimmed) in allowed_lengths and trimmed.isdecimal()

    def clean_integer_token(self, raw_token):
        if not isinstance(raw_token, str):
            return None
        cleaned = raw_token.strip().replace(',', '').replace('$', '')
        if cleaned.startswith(('+', '-')):
            sign = -1 if cleaned[0] == '-' else 1
            digits = cleaned[1:]
            return sign * int(digits) if digits.isdigit() else None
        return int(cleaned) if cleaned.isdigit() else None

# Unit Assertions
val = NumericValidator()
assert val.is_valid_pin('1234') is True
assert val.is_valid_pin('984201') is True
assert val.is_valid_pin('123') is False
assert val.is_valid_pin('123a') is False
assert val.is_valid_pin('²²²²') is False

assert val.clean_integer_token('$1,000,000') == 1000000
assert val.clean_integer_token('-500') == -500
assert val.clean_integer_token('12.34') is None

print("All PIN and integer token assertions passed successfully!")
```
