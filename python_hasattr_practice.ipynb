### 1. Basic Attribute Existence Check
`hasattr(object, name)` returns `True` if the attribute exists; otherwise `False`.

```python
class Animal:
    species = "carnivores"
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

tiger = Animal("Ronald", 5)

print("Has age:", hasattr(tiger, "age"))        # True
print("Has species:", hasattr(tiger, "species"))  # True
print("Has height:", hasattr(tiger, "height"))    # False
```

### 2. Duck Typing & Protocol Verification
Check if an unknown object supports required methods before invoking them.

```python
def read_stream(source):
    if hasattr(source, "read") and callable(source.read):
        return f"Stream read: {source.read()}"
    return f"Plain data: {source}"

import io
buffer = io.StringIO("In-memory stream text")
print(read_stream(buffer))
print(read_stream("Normal string"))
```

### 3. Inheritance & Method Resolution
`hasattr()` checks the entire inheritance hierarchy.

```python
class Base:
    def ping(self):
        return "pong"

class Child(Base):
    pass

c = Child()
print("Has inherited ping:", hasattr(c, "ping"))
print("Has dunder __class__:", hasattr(c, "__class__"))
```

### 4. The Property Getter Side-Effect Nuance
Calling `hasattr()` on a `@property` actually invokes the getter code.

```python
class Sensor:
    @property
    def reading(self):
        print("Sensor hardware queried!")
        return 98.6

s = Sensor()
print("Running hasattr check:")
has_reading = hasattr(s, "reading")
print("hasattr result:", has_reading)
```

### 5. Production Recipe: Resource Cleanup Guard
Safely close resource handles dynamically.

```python
def close_handle(handle):
    if hasattr(handle, "close") and callable(handle.close):
        handle.close()
        return "Closed successfully."
    return "No close method present."

class DummySocket:
    def close(self):
        pass

print(close_handle(DummySocket()))
print(close_handle(999))
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying hasattr() mechanics.

```python
class HasattrVerifier:
    @staticmethod
    def test_all():
        class Gadget:
            color = "silver"
            def __init__(self, model):
                self.model = model
                
        g = Gadget("G-100")
        assert hasattr(g, "model") is True
        assert hasattr(g, "color") is True
        assert hasattr(g, "price") is False
        
        setattr(g, "price", 299)
        assert hasattr(g, "price") is True
        
        # Non-string identifier raises TypeError
        try:
            hasattr(g, 123)
            assert False
        except TypeError:
            pass
            
        print("All hasattr() unit assertions passed successfully! 100% verified.")

HasattrVerifier.test_all()
```
