### 1. Basic Base-10 Calculations with math.log10()
`math.log10(x)` computes the common (base-10) logarithm of `x`. For exact powers of ten, it returns exact integral floating-point numbers.


```python
import math

print("log10(1):        ", math.log10(1))          # 0.0
print("log10(10):       ", math.log10(10))         # 1.0
print("log10(100):      ", math.log10(100))        # 2.0
print("log10(1000):     ", math.log10(1000))       # 3.0
print("log10(1_000_000):", math.log10(1_000_000))  # 6.0
print("log10(0.1):      ", math.log10(0.1))        # -1.0
print("log10(0.01):     ", math.log10(0.01))       # -2.0
print("log10(2):        ", round(math.log10(2), 6))     # 0.301030
```

### 2. Precision Advantage: math.log10(x) vs. math.log(x, 10)
`math.log(x, 10)` calculates `ln(x) / ln(10)`, which suffers from floating-point division rounding errors. `math.log10(x)` calls C99 hardware instructions directly, ensuring bit-exact results.


```python
import math

powers = [10, 100, 1000, 10000, 100000]
print(f"{'x':>8} | {'math.log10(x)':>15} | {'math.log(x, 10)':>18} | Difference")
print("-" * 58)
for p in powers:
    exact_log = math.log10(p)
    div_log = math.log(p, 10)
    diff = abs(exact_log - div_log)
    print(f"{p:8d} | {exact_log:15.1f} | {div_log:18.15f} | {diff:.2e}")
```

### 3. Domain Constraints (x > 0)
Common logarithms are undefined for zero and negative real numbers, raising `ValueError: math domain error`.


```python
import math

for bad in [0.0, -10.0]:
    try:
        math.log10(bad)
    except ValueError as err:
        print(f"Caught expected error on {bad}: {err}")

print("log10(inf):", math.log10(float("inf")))
print("log10(nan):", math.log10(float("nan")))
```

### 4. Counting Decimal Digits in Integers
The number of decimal digits in any positive integer `n` is `math.floor(math.log10(n)) + 1`.


```python
import math

test_numbers = [1, 9, 10, 99, 100, 987654321]
for num in test_numbers:
    calculated_digits = math.floor(math.log10(num)) + 1
    actual_digits = len(str(num))
    assert calculated_digits == actual_digits
    print(f"Number: {num:10d} -> Digits: {calculated_digits} (matches len: {actual_digits})")
```

### 5. Hands-on Practice Exercise: Sound Decibel & Chemistry pH Calculator
Acoustic sound pressure level (SPL): `20 * log10(p / p_0)`. Solution pH: `-log10([H+])`.
Implement `calculate_sound_db()` and `calculate_ph()` and verify with self-checking assertions.


```python
import math

AUDITORY_THRESHOLD_PA = 2e-5  # 20 uPa

def calculate_sound_db(pressure_pascals, reference_pa=AUDITORY_THRESHOLD_PA):
    """Calculates Sound Pressure Level in decibels (dB): 20 * log10(p / p_0)."""
    if pressure_pascals <= 0 or reference_pa <= 0:
        raise ValueError("Sound pressures must be strictly positive.")
    spl_db = 20.0 * math.log10(pressure_pascals / reference_pa)
    return round(spl_db, 2)

def calculate_ph(hydrogen_ion_conc):
    """Calculates solution pH: -log10([H+])."""
    if hydrogen_ion_conc <= 0:
        raise ValueError("Hydrogen ion concentration must be strictly positive.")
    return round(-math.log10(hydrogen_ion_conc), 2)

# Unit Assertions
assert calculate_sound_db(2e-5) == 0.0      # Auditory threshold: 0 dB
assert calculate_sound_db(0.02) == 60.0     # Normal speech: 60 dB
assert calculate_sound_db(20.0) == 120.0    # Threshold of pain: 120 dB

assert calculate_ph(1e-7) == 7.0   # Pure water: pH 7.0
assert calculate_ph(0.01) == 2.0   # Strong acid: pH 2.0
assert calculate_ph(1e-13) == 13.0 # Strong base: pH 13.0

print("Sound level of 0.063 Pa:     ", calculate_sound_db(0.063), "dB")
print("pH of black coffee (1e-5 M):  ", calculate_ph(1e-5))
print("All decibel and pH assertions passed successfully!")
```
