### 1. Basic if-else Branching

```python
mark = 65
if mark >= 50:
    status = 'Pass'
else:
    status = 'Fail'

print('Result:', status)
```

### 2. Multi-Tier elif Grading Chain

```python
score = 82
if score >= 80:
    grade = 'A'
elif score >= 60:
    grade = 'B'
elif score >= 40:
    grade = 'C'
else:
    grade = 'Fail'

print(f'Score {score} -> Grade {grade}')
```

### 3. Ternary Conditional Expression

```python
age = 20
category = 'Adult' if age >= 18 else 'Minor'
print('Category:', category)
```

### 4. Chained Range Comparisons

```python
temp = 24
# Pythonic chained comparison
if 18 <= temp <= 26:
    print('Comfortable room temperature.')
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Ternary pass check
m = 75
assert ('Pass' if m >= 50 else 'Fail') == 'Pass'

# Test 2: Chained comparison
x = 50
assert (20 <= x <= 80) is True

# Test 3: Short-circuiting evaluation
assert (True or 1/0) is True

# Test 4: Truthy and Falsy evaluations
assert bool(None) is False
assert bool([]) is False
assert bool([1, 2]) is True

print('All if-else condition unit assertions successfully verified! 100% passed.')
```
