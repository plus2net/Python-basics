# Python math.isnan(): Detect NaN Values

**Tutorial & Full Documentation:** [https://www.plus2net.com/python/math-isnan.php](https://www.plus2net.com/python/math-isnan.php)

### 1. The IEEE 754 NaN Equality Paradox
NaN is defined as not equal to anything, including itself. Testing `x == float('nan')` always yields `False`.

```python
import math

val = float("nan")

print("val == val:          ", val == val)           # False
print("val == float('nan'): ", val == float("nan"))  # False
print("math.isnan(val):     ", math.isnan(val))      # True
```

### 2. Indeterminate Mathematical Operations
Operations involving infinity with no single defined real answer produce NaN.

```python
import math

inf = float("inf")

print("inf / inf: ", inf / inf)
print("isnan?     ", math.isnan(inf / inf))

print("inf - inf: ", inf - inf)
print("isnan?     ", math.isnan(inf - inf))

print("0 * inf:   ", 0 * inf)
print("isnan?     ", math.isnan(0 * inf))
```

### 3. The Float Inspection Trio: isnan vs. isinf vs. isfinite
Inspect whether values are regular finite numbers, infinities, or NaN.

```python
import math

samples = [42.5, float("inf"), float("-inf"), float("nan")]

print(f"{'Value':<12} | {'isnan':<7} | {'isinf':<7} | {'isfinite':<8}")
print("-" * 42)
for x in samples:
    print(f"{str(x):<12} | {str(math.isnan(x)):<7} | {str(math.isinf(x)):<7} | {str(math.isfinite(x)):<8}")
```

### 4. TypeError on Non-Numeric Arguments
`math.isnan()` requires real numbers (`int` or `float`). Passing strings or collections raises `TypeError`.

```python
import math

try:
    math.isnan("nan")
except TypeError as err:
    print("Caught expected TypeError:", err)
```

### 5. Hands-on Practice Exercise: Telemetry Sensor Data Sanitizer
Run the self-checking unit assertions below to verify that NaNs, infinities, and invalid strings are filtered out cleanly.

```python
def sanitize_telemetry(readings):
    clean = []
    for item in readings:
        try:
            val = float(item)
            if not math.isnan(val) and math.isfinite(val):
                clean.append(val)
        except (ValueError, TypeError):
            continue
    return clean

sensor_stream = [22.4, float("nan"), 23.1, "DEVICE_OFFLINE", float("inf"), 22.9, float("nan")]
sanitized = sanitize_telemetry(sensor_stream)

# Unit Assertions
assert sanitized == [22.4, 23.1, 22.9]
assert len(sanitized) == 3
assert round(sum(sanitized) / len(sanitized), 2) == 22.8

print("Sanitized readings:", sanitized)
print("All sensor telemetry unit assertions passed successfully!")
```
