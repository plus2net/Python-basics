## 2. Basic ORDER BY ASC & DESC (`sqlite3`)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT id, name, class, mark, sex FROM student ORDER BY class"
my_cursor = my_conn.execute(q)
for row in my_cursor:
    print(row)
my_conn.close()
```

## 3. Multiple Columns Ordering with Error Handling

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT * FROM student ORDER BY class, mark"
try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 4. Modern SQLAlchemy 2.0+ Context Manager with ORDER BY

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

my_conn = create_engine("sqlite:///my_db.db")
q = text("SELECT id, name, class, mark, sex FROM student ORDER BY class DESC, mark ASC")

try:
    with my_conn.connect() as connection:
        r_set = connection.execute(q)
        for row in r_set:
            print(row)
except SQLAlchemyError as e:
    print("SQLAlchemy error:", e)
```
