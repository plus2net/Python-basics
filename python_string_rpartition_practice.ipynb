## 1. Right-First 3-Tuple Return Mechanics

`str.rpartition(sep)` searches for the **last occurrence** of `sep` from right to left.
If found, `head` contains everything before the last separator, `sep` contains the separator itself, and `tail` contains everything after.

```python
filepath = "home/spark/projects/python/analysis.py"

# Splits on the rightmost slash '/'
directory, sep, filename = filepath.rpartition("/")
print("Directory: ", repr(directory))
print("Separator: ", repr(sep))
print("Filename:  ", repr(filename))

assert directory == "home/spark/projects/python"
assert sep == "/"
assert filename == "analysis.py"

print("\nAll basic rpartition mechanics assertions passed!")
```

## 2. Crucial Asymmetry: `rpartition()` vs. `partition()`

When the separator is absent, Python's placement of the original string differs:
- `partition(sep)` returns `(original, "", "")` &mdash; in position 0 (`head`).
- `rpartition(sep)` returns `("", "", original)` &mdash; in position 2 (`tail`).

```python
text = "alpha.beta.gamma"

# 1. Delimiter present
print("partition('.'):  ", text.partition("."))
print("rpartition('.'): ", text.rpartition("."))
assert text.partition(".") == ("alpha", ".", "beta.gamma")
assert text.rpartition(".") == ("alpha.beta", ".", "gamma")

# 2. Delimiter absent: asymmetry check
plain = "standalone_token"
p_res = plain.partition(":")
rp_res = plain.rpartition(":")

print("\npartition missing:  ", p_res)
print("rpartition missing: ", rp_res)
assert p_res == ("standalone_token", "", "")
assert rp_res == ("", "", "standalone_token")

print("\nAsymmetry verification assertions passed!")
```

## 3. Real-World Engineering Recipes

Explore three common software engineering patterns utilizing `str.rpartition()`:
1. File Stem and Extension Extraction
2. Fully Qualified Python Module and Class Decomposition
3. Network Host:Port Address Splitter

```python
# Recipe 1: Safe file extension extractor
def extract_extension(filename: str) -> tuple[str, str]:
    stem, sep, ext = filename.rpartition(".")
    if not sep or not stem:
        return filename, ""
    return stem, ext

assert extract_extension("report.pdf") == ("report", "pdf")
assert extract_extension("archive.tar.gz") == ("archive.tar", "gz")
assert extract_extension(".gitignore") == (".gitignore", "")

# Recipe 2: Python qualified name decomposition
def split_module(qualname: str) -> tuple[str, str]:
    module, sep, target = qualname.rpartition(".")
    return (module, target) if sep else ("", qualname)

assert split_module("urllib.request.urlopen") == ("urllib.request", "urlopen")
assert split_module("CustomClass") == ("", "CustomClass")

# Recipe 3: Host:Port address splitter
def parse_host_port(addr: str, default_port: int = 80) -> tuple[str, int]:
    host, sep, port_str = addr.rpartition(":")
    if sep and port_str.isdigit():
        return host, int(port_str)
    return addr, default_port

assert parse_host_port("192.168.1.100:8080") == ("192.168.1.100", 8080)
assert parse_host_port("api.plus2net.com", 443) == ("api.plus2net.com", 443)

print("All engineering recipes passed!")
```

## 4. Boundary Pitfalls & Error Handling

Passing an empty separator `""` raises a `ValueError: empty separator`.

```python
# Empty separator raises ValueError
try:
    "test".rpartition("")
    assert False, "Expected ValueError not raised"
except ValueError as e:
    assert "empty separator" in str(e)

print("Boundary error validation passed!")
```

## 5. Production Implementation & Self-Checking Unit Assertions

This production-grade `PathPartitioner` class encapsulates directory splitting, extension extraction, and trailing segment resolution.

```python
class PathPartitioner:
    @staticmethod
    def split_path(filepath: str) -> tuple[str, str]:
        """Splits into directory and filename."""
        dirname, sep, basename = filepath.rpartition("/")
        return (dirname, basename) if sep else ("", filepath)

    @staticmethod
    def split_extension(filename: str) -> tuple[str, str]:
        """Splits into stem and extension."""
        stem, sep, ext = filename.rpartition(".")
        return (stem, ext) if sep else (filename, "")

    @staticmethod
    def get_terminal_segment(text: str, sep: str) -> str:
        """Returns only the trailing portion after the last sep."""
        _, _, tail = text.rpartition(sep)
        return tail


# ==========================================
# UNIT ASSERTION SUITE
# ==========================================
partitioner = PathPartitioner()

# 1. Path parsing
d, f = partitioner.split_path("/var/log/nginx/access.log")
assert d == "/var/log/nginx"
assert f == "access.log"

d2, f2 = partitioner.split_path("readme.txt")
assert d2 == ""
assert f2 == "readme.txt"

# 2. Extension parsing
stem, ext = partitioner.split_extension("backup.2026.tar.gz")
assert stem == "backup.2026.tar"
assert ext == "gz"

# 3. Terminal segment extraction
assert partitioner.get_terminal_segment("pkg.module.Class", ".") == "Class"

print("All str.rpartition() production unit assertions passed successfully!")
```
