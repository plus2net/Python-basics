## 1. Simple Slicing Palindrome Check

Comparing `s == s[::-1]` is the fastest, most idiomatic check for short to moderate strings.

```python
def is_simple(s: str) -> bool:
    return s == s[::-1]

print("'radar': ", is_simple("radar"))
print("'level': ", is_simple("level"))
print("'python':", is_simple("python"))

assert is_simple("radar") is True
assert is_simple("level") is True
assert is_simple("python") is False

print("\nBasic slicing assertions passed!")
```

## 2. Two-Pointer Algorithm with Early Exit ($O(1)$ Space)

Compares characters from opposite ends, terminating on the first mismatch without allocating reversed copies.

```python
def two_pointer(s: str) -> bool:
    l, r = 0, len(s) - 1
    while l < r:
        if s[l] != s[r]:
            return False
        l += 1
        r -= 1
    return True

assert two_pointer("racecar") is True
assert two_pointer("mismatch") is False

print("Two-pointer assertions passed!")
```

## 3. Natural Language Phrase Palindromes

Real phrases contain spaces and punctuation. We filter with `isalnum()` and normalize with `casefold()`.

```python
def is_phrase(text: str) -> bool:
    cleaned = [c.casefold() for c in text if c.isalnum()]
    return cleaned == cleaned[::-1]

phrases = [
    ("A man, a plan, a canal: Panama", True),
    ("Was it a car or a cat I saw?", True),
    ("Madam, in Eden, I'm Adam", True),
    ("Hello World", False)
]

for p, expected in phrases:
    res = is_phrase(p)
    print(f"{p:<32} -> {res}")
    assert res == expected

print("\nPhrase palindrome assertions passed!")
```

## 4. Mathematical Integer Palindromes (Zero String Allocations)

Reverses the numerical digits algebraically without calling `str()`.

```python
def is_int_palindrome(n: int) -> bool:
    if n < 0 or (n % 10 == 0 and n != 0):
        return False
    rev = 0
    temp = n
    while temp > rev:
        rev = (rev * 10) + (temp % 10)
        temp //= 10
    return temp == rev or temp == rev // 10

assert is_int_palindrome(12321) is True
assert is_int_palindrome(1221) is True
assert is_int_palindrome(0) is True
assert is_int_palindrome(-121) is False
assert is_int_palindrome(10) is False

print("Integer math palindrome assertions passed!")
```

## 5. Palindrome Permutations via `collections.Counter`

A string can be rearranged into a palindrome if at most one character occurs an odd number of times.

```python
from collections import Counter

def can_permute_palindrome(s: str) -> bool:
    clean = [c.casefold() for c in s if c.isalnum()]
    counts = Counter(clean)
    return sum(1 for v in counts.values() if v % 2 != 0) <= 1

assert can_permute_palindrome("civic") is True
assert can_permute_palindrome("ivicc") is True
assert can_permute_palindrome("tact coa") is True
assert can_permute_palindrome("random") is False

print("Palindrome permutation assertions passed!")
```

## 6. Production Implementation & Self-Checking Unit Assertions

This production-grade `PalindromeChecker` class encapsulates exact checks, streaming phrase checks, and math integer validations.

```python
class PalindromeChecker:
    @staticmethod
    def is_exact(word: str) -> bool:
        """Exact case-sensitive palindrome check."""
        return word == word[::-1]

    @staticmethod
    def is_phrase(text: str) -> bool:
        """Sanitized phrase palindrome using two-pointer stream."""
        l, r = 0, len(text) - 1
        while l < r:
            while l < r and not text[l].isalnum():
                l += 1
            while l < r and not text[r].isalnum():
                r -= 1
            if text[l].casefold() != text[r].casefold():
                return False
            l += 1
            r -= 1
        return True

    @staticmethod
    def is_integer(n: int) -> bool:
        """Mathematical palindrome check without string conversion."""
        if n < 0 or (n % 10 == 0 and n != 0):
            return False
        rev = 0
        temp = n
        while temp > rev:
            rev = (rev * 10) + (temp % 10)
            temp //= 10
        return temp == rev or temp == rev // 10


# ==========================================
# UNIT ASSERTION SUITE
# ==========================================
checker = PalindromeChecker()

# 1. Exact checks
assert checker.is_exact("racecar") is True
assert checker.is_exact("Racecar") is False
assert checker.is_exact("") is True

# 2. Phrase checks
assert checker.is_phrase("A man, a plan, a canal: Panama") is True
assert checker.is_phrase("Was it a car or a cat I saw?") is True
assert checker.is_phrase("Hello World") is False

# 3. Integer checks
assert checker.is_integer(12321) is True
assert checker.is_integer(1221) is True
assert checker.is_integer(0) is True
assert checker.is_integer(-121) is False
assert checker.is_integer(10) is False

print("All palindrome production unit assertions passed successfully!")
```
