## 2. Exporting Data from SQLite to Excel Sheet (`read_sql` & `to_excel`)

```python
from sqlalchemy import create_engine
from sqlalchemy.exc import SQLAlchemyError
import pandas as pd

my_path = 'my_db.db'
my_conn = create_engine("sqlite:///" + my_path)

try:
    q = "SELECT * FROM student"
    with my_conn.connect() as connection:
        df = pd.read_sql(q, connection, index_col='id')
    print(df.head())
    df.to_excel('student.xlsx')
    print("DataFrame exported to Excel successfully!")
except SQLAlchemyError as e:
    print("SQLAlchemy Error:", e)
```

## 3. Importing Data from Excel Sheet to SQLite Table (`read_excel` & `to_sql`)

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError
import pandas as pd

my_path = 'my_db.db'
my_conn = create_engine("sqlite:///" + my_path)

try:
    df = pd.read_excel('student.xlsx')
    print(df.head())
    with my_conn.begin() as connection:
        df.to_sql(con=connection, name='student3', if_exists='append', index=False)
        r_set = connection.execute(text("SELECT * FROM student3 LIMIT 5"))
        for row in r_set:
            print(row)
except SQLAlchemyError as e:
    print("SQLAlchemy Error:", e)
```
