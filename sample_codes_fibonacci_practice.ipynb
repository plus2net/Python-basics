```python
a, b = 0, 1
for _ in range(10):
    print(a, end=" ")
    a, b = b, a + b
```

## Example 2: Original while-loop series

Read the code, predict the output and run this cell.

```python
n = 15
i = 0
a, b = 0, 1
while i < n:
    print(a, end=", ")
    a, b = b, a + b
    i += 1
```

## Example 3: Original for-loop approach

Read the code, predict the output and run this cell.

```python
n = 15
a, b = 0, 1
for i in range(n):
    print(a, end=", ")
    a, b = b, a + b
```

## Example 4: Recursive Fibonacci function

Read the code, predict the output and run this cell.

```python
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)
print([fib(i) for i in range(10)])
```

## Example 5: Use memoization for recursion

Read the code, predict the output and run this cell.

```python
from functools import lru_cache
@lru_cache(None)
def fib(n):
    if n < 2:
        return n
    return fib(n-1) + fib(n-2)
print(fib(30))
```

## Example 6: Generator: values up to a limit

Read the code, predict the output and run this cell.

```python
def fibonacci(limit):
    a, b = 0, 1
    while a < limit:
        yield a
        a, b = b, a + b
print(list(fibonacci(100)))
```

## Example 7: Return a fixed number of terms

Read the code, predict the output and run this cell.

```python
def fibonacci_terms(count):
    a, b = 0, 1
    result = []
    for _ in range(count):
        result.append(a)
        a, b = b, a + b
    return result
print(fibonacci_terms(8))
```
