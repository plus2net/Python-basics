### 1. Basic if-elif-else Grade Assignment

```python
def get_grade(m):
    if m >= 80:
        return 'A'
    elif m >= 60:
        return 'B'
    elif m >= 40:
        return 'C'
    else:
        return 'Fail'

scores = [85, 72, 55, 30]
for s in scores:
    print(f"Score: {s} -> Grade: {get_grade(s)}")
```

### 2. Input Validation with Boundary Checking

```python
def safe_grade_eval(raw_val):
    try:
        val = float(raw_val)
    except ValueError:
        return "Invalid input: must be numeric"
    if val < 0 or val > 100:
        return "Error: Marks must be within 0-100 range"
    return 'Pass' if val >= 40 else 'Fail'

print(safe_grade_eval("88"))
print(safe_grade_eval("120"))
print(safe_grade_eval("abc"))
```

### 3. Calculating Average Marks and Class Letter Grade

```python
class_marks = [85, 78, 92, 68, 74, 88, 90]
avg = sum(class_marks) / len(class_marks)

if avg >= 90:
    cg = 'A'
elif avg >= 80:
    cg = 'B'
elif avg >= 70:
    cg = 'C'
else:
    cg = 'D'

print(f"Class Average: {avg:.2f} | Overall Grade: {cg}")
```

### 4. Self-Checking Unit Verification Suite

```python
# Verify grade boundaries
def calc_grade(m):
    if m >= 90: return 'A+'
    if m >= 80: return 'A'
    if m >= 70: return 'B'
    if m >= 60: return 'C'
    if m >= 50: return 'D'
    return 'F'

assert calc_grade(100) == 'A+'
assert calc_grade(90) == 'A+'
assert calc_grade(89) == 'A'
assert calc_grade(80) == 'A'
assert calc_grade(70) == 'B'
assert calc_grade(60) == 'C'
assert calc_grade(50) == 'D'
assert calc_grade(45) == 'F'
assert calc_grade(0) == 'F'

print("All grading logic assertions successfully verified! 100% passed.")
```
