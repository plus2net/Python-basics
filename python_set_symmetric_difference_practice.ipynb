### 1. Basic Two-Set Symmetric Difference (XOR)
`A.symmetric_difference(B)` returns a new set containing elements present in either `A` or `B`, but not both.

```python
A = {1, 2, 3}
B = {3, 4, 5}

# Using ^ operator
result_op = A ^ B
print("A ^ B:", result_op)  # {1, 2, 4, 5}

# Using symmetric_difference() method
result_meth = A.symmetric_difference(B)
print("A.symmetric_difference(B):", result_meth)  # {1, 2, 4, 5}

# Verify originals remain unmodified
print("Original A:", A)  # {1, 2, 3}
print("Original B:", B)  # {3, 4, 5}
```

### 2. Method vs Operator Flexibility
The `symmetric_difference()` method accepts any iterable (list, tuple, string), whereas the `^` operator strictly requires set operands.

```python
base_set = {"a", "b", "c"}
filter_list = ["a", "x", "y"]

# Method accepts a list cleanly
clean_diff = base_set.symmetric_difference(filter_list)
print("symmetric_difference() with list:", clean_diff)  # {'b', 'c', 'x', 'y'}

# Operator with a list raises TypeError
try:
    invalid = base_set ^ filter_list
except TypeError as err:
    print("Caught expected error:", err)
```

### 3. Multi-Set Chaining & Odd Parity
In multi-set XOR, an element appears in the result if it is present in an odd number of sets.

```python
A = {"a", "b", "c"}
B = {"a", "y", "z"}
C = {"a", "k", "l"}

# 'a' is present in 3 sets (odd count) -> included!
result_chain = A ^ B ^ C
print("Multi-set XOR:", sorted(result_chain))
```

### 4. String Character Unpacking Trap
Strings are iterables of characters. When comparing whole words, wrap them in a list or set.

```python
usernames = {"Alex", "Maria", "David"}

# Trap: bare string decomposes into 'A', 'l', 'e', 'x'
trap_res = usernames.symmetric_difference("Alex")
print("Trap (bare string):", trap_res)

# Correct: wrap word in list
correct_res = usernames.symmetric_difference(["Alex"])
print("Correct (list of words):", correct_res)  # {'Maria', 'David'}
```

### 5. Production Recipe: Inventory SKU Diffing
Compare local and remote catalog entries to find out-of-sync SKU items.

```python
def find_inventory_diff(local: set[str], remote: list[str]) -> set[str]:
    """Returns SKUs that are out-of-sync between stores."""
    return local.symmetric_difference(remote)

local_catalog = {"SKU-100", "SKU-200", "SKU-300"}
remote_catalog = ["SKU-200", "SKU-300", "SKU-400"]

diff = find_inventory_diff(local_catalog, remote_catalog)
print("Catalog discrepancies:", sorted(diff))  # ['SKU-100', 'SKU-400']
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying symmetric difference properties and immutability.

```python
class SetSymmetricDifferenceVerifier:
    @staticmethod
    def test_all():
        s1 = {1, 2, 3}
        s2 = {3, 4, 5}
        
        # Basic XOR
        assert s1.symmetric_difference(s2) == {1, 2, 4, 5}
        assert s1 ^ s2 == {1, 2, 4, 5}
        
        # Caller unmodified
        assert s1 == {1, 2, 3}
        
        # Heterogeneous iterables
        assert s1.symmetric_difference([2, 3, 9]) == {1, 9}
        
        # Self XOR yields empty set
        assert s1 ^ s1 == set()
        
        # Disjoint XOR equals union
        assert s1 ^ {8, 9} == s1 | {8, 9}
        
        # Empty set XOR returns copy
        copy_s = s1.symmetric_difference(set())
        assert copy_s == s1
        assert copy_s is not s1
        
        print("All set.symmetric_difference() unit assertions passed successfully! 100% verified.")

SetSymmetricDifferenceVerifier.test_all()
```
