### 1. Basic Numerical Filtering with Named Functions
Filter even numbers using a dedicated predicate function.

```python
def is_even(num: int) -> bool:
    return num % 2 == 0

numbers = [5, 82, 40, 37, 19, 94]
even_filter = filter(is_even, numbers)

print("Filter object type:", type(even_filter))
print("Filtered numbers:", list(even_filter))
```

### 2. String Filtering & Lambda Predicates
Extract matching characters using both named functions and inline lambdas.

```python
sample_text = "Plus2Net Portal"
lower_chars = list(filter(lambda c: c.islower(), sample_text))
print("Lowercase characters:", lower_chars)
print("Rejoined string:", "".join(lower_chars))
```

### 3. Truthy Filtering with filter(None, iterable)
Remove all falsy values (`0`, empty strings, `None`, empty containers) in compiled C speed.

```python
raw_records = [0, "", None, "Python", [], '0', {}, False, 42, "Active"]
truthy_records = list(filter(None, raw_records))
print("Cleaned truthy records:", truthy_records)
```

### 4. Memory Footprint: filter() vs. List Comprehension
Compare memory overhead between lazy filter iterators and eager lists across 1,000,000 elements.

```python
import sys

large_range = range(1_000_000)
lazy_iter = filter(lambda x: x % 100 == 0, large_range)
eager_list = [x for x in large_range if x % 100 == 0]

print("Filter iterator memory:", sys.getsizeof(lazy_iter), "bytes")
print("Materialized list memory:", sys.getsizeof(eager_list), "bytes")
print("First three values via next():", [next(lazy_iter), next(lazy_iter), next(lazy_iter)])
```

### 5. Production Recipe: User Payload Sanitization
Filter list of user dictionaries using multiple criteria.

```python
user_payloads = [
    {"user_id": 101, "username": "alice", "is_active": True, "age": 28},
    {"user_id": 102, "username": "", "is_active": False, "age": 17},
    {"user_id": 103, "username": "charlie", "is_active": True, "age": 34},
    {"user_id": 104, "username": "david", "is_active": False, "age": 42},
    {"user_id": 105, "username": "elena", "is_active": True, "age": 22},
]

def is_eligible(u):
    return u["is_active"] and u["age"] >= 18 and bool(u["username"])

for valid_user in filter(is_eligible, user_payloads):
    print(f"Valid: {valid_user['username']} (ID: {valid_user['user_id']})")
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying filter() mechanics and iterator exhaustion.

```python
# 1. Positives
assert list(filter(lambda x: x > 0, [-2, 0, 5, 8])) == [5, 8]

# 2. Truthiness
assert list(filter(None, [0, 1, "", "text", None])) == [1, "text"]

# 3. Exhaustion
it = filter(lambda x: x > 0, [1, 2])
assert list(it) == [1, 2]
assert list(it) == []

print("All filter() unit assertions passed! 100% verified.")
```
