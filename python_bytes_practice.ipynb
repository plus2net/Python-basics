### 1. Basic Byte Buffer Allocation
Construct byte buffers from integer sizes, strings, and integer iterables.

```python
print("bytes(4):           ", bytes(4))
print("bytes('hi', 'utf8'):", bytes('hi', 'utf-8'))
print("bytes([65, 66, 67]):", bytes([65, 66, 67]))
```

### 2. Error Handling in Encodings
Handle non-ASCII characters using strict, ignore, and replace policies.

```python
text = "plus2net.cÖm"
print("ignore: ", bytes(text, encoding='ascii', errors='ignore'))
print("replace:", bytes(text, encoding='ascii', errors='replace'))
```

### 3. Hexadecimal Conversions: fromhex() and .hex()
Translate between hex strings and binary byte buffers.

```python
b = bytes.fromhex('48656c6c6f21')
print("Decoded bytes: ", b)
print("Exported to hex:", b.hex())
assert b == b'Hello!'
```

### 4. Production Recipe: File Header Validation
Validate magic numbers on incoming binary files.

```python
PNG_MAGIC = bytes.fromhex('89504E470D0A1A0A')
sample = bytes.fromhex('89504E470D0A1A0A0000000D')
print("Is valid PNG:", sample.startswith(PNG_MAGIC))
assert sample.startswith(PNG_MAGIC) is True
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying bytes() mechanics.

```python
assert bytes() == b''
assert bytes(2) == b'\x00\x00'
assert bytes([65, 66]) == b'AB'
assert bytes.fromhex('ff00') == b'\xff\x00'

print("All bytes() unit assertions passed! 100% verified.")
```
