### 1. Basic Attribute Deletion
`delattr(object, name)` removes an attribute dynamically from an object.

```python
class Animal:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

tiger = Animal("Ronald", 5)
print("Before deletion, age =", tiger.age)

delattr(tiger, "age")
print("Has age after delattr:", hasattr(tiger, "age"))
```

### 2. del vs delattr() on Dynamic & Non-Identifier Keys
Delete attributes created with hyphens that are illegal under dot notation.

```python
class Bag:
    pass

b = Bag()
setattr(b, "content-type", "application/json")
setattr(b, "api-key", "secret_token")
print("Initial keys:", list(b.__dict__.keys()))

# Delete hyphenated key
delattr(b, "api-key")
print("After deletion:", list(b.__dict__.keys()))
```

### 3. Deleting Class Attributes
Deleting on the class removes the attribute for all instances.

```python
class Vehicle:
    wheels = 4

car1 = Vehicle()
car2 = Vehicle()
print("Initial wheels:", car1.wheels, car2.wheels)

delattr(Vehicle, "wheels")
print("car1 has wheels:", hasattr(car1, "wheels"))
print("car2 has wheels:", hasattr(car2, "wheels"))
```

### 4. Slotted Classes & Deletion Constraints
Deleting assigned slots removes the attribute value.

```python
class SlottedPoint:
    __slots__ = ("x", "y")

pt = SlottedPoint()
pt.x = 10
print("pt has x:", hasattr(pt, "x"))

delattr(pt, "x")
print("pt has x after delattr:", hasattr(pt, "x"))
```

### 5. Production Recipe: Dynamic PII Scrubbing
Sanitize objects by stripping out sensitive fields before transmission.

```python
def scrub_pii(record, sensitive_fields):
    for field in sensitive_fields:
        if hasattr(record, field):
            delattr(record, field)

class UserRecord:
    def __init__(self):
        self.username = "alice"
        self.token = "secret_123"
        self.ssn = "000-00-0000"

u = UserRecord()
scrub_pii(u, ["token", "ssn"])
print("Clean record fields:", u.__dict__)
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying delattr() mechanics.

```python
class DelattrVerifier:
    @staticmethod
    def test_all():
        class Box:
            def __init__(self):
                self.item = "book"
                
        b = Box()
        assert hasattr(b, "item") is True
        delattr(b, "item")
        assert hasattr(b, "item") is False
        
        try:
            delattr(b, "item")
            assert False
        except AttributeError:
            pass
            
        print("All delattr() unit assertions passed successfully! 100% verified.")

DelattrVerifier.test_all()
```
