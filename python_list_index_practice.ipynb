### 1. Basic Positional Search

```python
fruits = ['apple', 'banana', 'cherry', 'date']
print("Index of 'banana':", fruits.index('banana'))
print("Index of 'date':  ", fruits.index('date'))
```

### 2. Searching with start & end Bounds

```python
items = ['x', 'y', 'z', 'y', 'x']
# Start search at index 2
print("Second 'y' at:", items.index('y', 2))
```

### 3. Graceful ValueError Handling

```python
names = ['Alice', 'Bob']
try:
    pos = names.index('Charlie')
except ValueError as err:
    print("Safely caught error:", err)
```

### 4. Collecting All Duplicate Indices

```python
chars = ['a', 'b', 'c', 'b', 'a', 'b']
all_b = [i for i, c in enumerate(chars) if c == 'b']
print("All 'b' indices:", all_b)
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Exact index returns
seq = [10, 20, 30]
assert seq.index(20) == 1

# Test 2: Bounded search
arr = ['a', 'b', 'a', 'c']
assert arr.index('a', 1) == 2

# Test 3: Missing raises ValueError
try:
    arr.index('z')
    assert False
except ValueError:
    pass

print("All list.index() unit assertions successfully verified! 100% passed.")
```
