### 1. Basic dict.fromkeys() Initialization
`dict.fromkeys(keys, value=None)` creates a dictionary with all keys mapped to the specified value.

```python
fields = ["name", "email", "phone"]

# Default value None
empty_record = dict.fromkeys(fields)
print("Default None: ", empty_record)

# Immutable custom default (0)
score_tracker = dict.fromkeys(fields, 0)
print("Zero defaults:", score_tracker)
```

### 2. The Shared Mutable Reference Bug Exposed
When passing a mutable object like `[]`, every key references the exact same list in memory!

```python
# BUG: Single shared list instance
buggy = dict.fromkeys(["team_a", "team_b"], [])
buggy["team_a"].append("Alice")

print("team_a:", buggy["team_a"])  # ['Alice']
print("team_b:", buggy["team_b"])  # ['Alice'] (unintended mutation!)
print("Shared memory identity?", buggy["team_a"] is buggy["team_b"])  # True
```

### 3. The Solution: Dictionary Comprehensions
A comprehension evaluates a fresh `[]` allocation for each key.

```python
# Clean: Independent list for every key
clean = {k: [] for k in ["team_a", "team_b"]}
clean["team_a"].append("Alice")

print("team_a:", clean["team_a"])  # ['Alice']
print("team_b:", clean["team_b"])  # [] (independent!)
print("Shared identity?       ", clean["team_a"] is clean["team_b"])  # False
```

### 4. Real-World Recipe: Order-Preserving Fast Deduplication
`dict.fromkeys()` removes duplicates from lists in O(N) time while preserving insertion order.

```python
def deduplicate_preserve_order(items: list) -> list:
    return list(dict.fromkeys(items))

raw_tags = ["python", "fastapi", "python", "docker", "fastapi"]
unique_tags = deduplicate_preserve_order(raw_tags)
print("Deduplicated tags:", unique_tags)  # ['python', 'fastapi', 'docker']
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying `dict.fromkeys()` defaults, deduplication, and comprehension independence.

```python
class FromkeysVerifier:
    @staticmethod
    def make_schema(fields: list, val=None) -> dict:
        return dict.fromkeys(fields, val)

v = FromkeysVerifier()

# Test 1: Schema creation
assert v.make_schema(["a", "b"]) == {"a": None, "b": None}
assert v.make_schema(["x", "y"], 0) == {"x": 0, "y": 0}

# Test 2: Order-preserving deduplication
seq = [10, 20, 10, 30, 20]
assert list(dict.fromkeys(seq)) == [10, 20, 30]

# Test 3: Range as keys
r_dict = dict.fromkeys(range(3), True)
assert r_dict == {0: True, 1: True, 2: True}

print("All dict.fromkeys() unit assertions passed successfully! 100% verified.")
```
