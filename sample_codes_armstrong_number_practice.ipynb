## Example 1: First working example

Read the code, predict the output and run this cell.

```python
n = 153
digits = str(n)
total = sum(int(d) ** len(digits) for d in digits)
print("Sum of powered digits:", total)
print("Armstrong number:", total == n)
```

## Example 2: Original three-digit example

Read the code, predict the output and run this cell.

```python
from unittest.mock import patch
with patch("builtins.input", side_effect=['153']):
    raw = input("Enter a three-digit number: ").strip()
    if not (raw.isdecimal() and len(raw) == 3):
        print("Enter exactly three digits")
    else:
        n = int(raw)
        total = sum(int(d) ** 3 for d in raw)
        print("Sum of cubes:", total)
        print("Armstrong:", total == n)
```

## Example 3: Check an Armstrong number of any length

Read the code, predict the output and run this cell.

```python
def is_armstrong(n):
    if n < 0:
        return False
    digits = str(n)
    return n == sum(int(d) ** len(digits) for d in digits)

for n in (0, 1, 153, 371, 1634, 9474, 100):
    print(n, is_armstrong(n))
```

## Example 4: List Armstrong numbers within a range

Read the code, predict the output and run this cell.

```python
def is_armstrong(n):
    digits = str(n)
    return n == sum(int(d) ** len(digits) for d in digits)
print([n for n in range(10, 10000) if is_armstrong(n)])
```

## Example 7: Compare to the digit-cubes rule

Read the code, predict the output and run this cell.

```python
n = 1634
cubes = sum(int(d) ** 3 for d in str(n))
powers = sum(int(d) ** len(str(n)) for d in str(n))
print(cubes, powers)
```
