### 1. Basic Cosine Calculations in Radians
`math.cos(x)` returns the cosine of `x`, where `x` is measured strictly in radians.

```python
import math

print("cos(0):       ", math.cos(0))
print("cos(pi):      ", math.cos(math.pi))
print("cos(2*pi):    ", math.cos(2 * math.pi))
print("cos(1 radian):", math.cos(1))
```

### 2. Working with Angles in Degrees (math.radians)
Convert degree angles using `math.radians()` before passing them to `math.cos()`.

```python
import math

angles_deg = [0, 60, 90, 180, 270, 360]
for d in angles_deg:
    rad = math.radians(d)
    print(f"cos({d:3d} deg) = {math.cos(rad):.4f}")
```

### 3. Floating-Point Precision at 90 Degrees
Because `pi` is represented as a finite float, $\cos(90^\circ)$ evaluates to a tiny non-zero residue (`6.12e-17`). Use `math.isclose()` for comparisons.

```python
import math

rad_90 = math.radians(90)
res = math.cos(rad_90)

print("Raw cos(90 deg):", res)
print("Is close to 0.0?", math.isclose(res, 0.0, abs_tol=1e-15))
```

### 4. Trigonometric Symmetry & Pythagorean Identity
Verify even symmetry $\cos(-x) = \cos(x)$ and $\sin^2(x) + \cos^2(x) = 1.0$.

```python
import math

theta = 1.25

print("Symmetry holds?       ", math.cos(-theta) == math.cos(theta))

pythagoras = math.sin(theta)**2 + math.cos(theta)**2
print("Pythagoras == 1.0?    ", math.isclose(pythagoras, 1.0))
```

### 5. Hands-on Practice Exercise: 2D Polar-to-Cartesian Vector Decomposition
Run the self-checking unit assertions below to verify vector conversion from polar $(r, 	heta)$ to Cartesian $(x, y)$.

```python
def polar_to_cartesian(radius, angle_degrees):
    theta_rad = math.radians(angle_degrees)
    x = radius * math.cos(theta_rad)
    y = radius * math.sin(theta_rad)
    
    # Clamp near-zero floating point residues
    x = 0.0 if math.isclose(x, 0.0, abs_tol=1e-12) else x
    y = 0.0 if math.isclose(y, 0.0, abs_tol=1e-12) else y
    return round(x, 4), round(y, 4)

# Unit Assertions
assert polar_to_cartesian(10, 0) == (10.0, 0.0)
assert polar_to_cartesian(10, 90) == (0.0, 10.0)
assert polar_to_cartesian(10, 180) == (-10.0, 0.0)
assert polar_to_cartesian(10, 270) == (0.0, -10.0)
assert polar_to_cartesian(10, 45) == (7.0711, 7.0711)

print("Vector (10, 45 deg)  -> (x, y):", polar_to_cartesian(10, 45))
print("Vector (10, 180 deg) -> (x, y):", polar_to_cartesian(10, 180))
print("All vector decomposition assertions passed successfully!")
```
