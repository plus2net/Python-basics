### 1. Basic str.index() and ValueError Handling
`str.index()` returns the 0-based offset or raises `ValueError` if missing.

```python
msg = "Welcome to plus2net.com"
print("Index of 'Welcome':", msg.index("Welcome"))
print("Index of 'plus':   ", msg.index("plus"))

assert msg.index("Welcome") == 0
assert msg.index("plus") == 11

try:
    msg.index("Python")
    assert False, "Should raise ValueError"
except ValueError as e:
    print("Caught expected error:", e)
```

### 2. Searching within Slice Windows (start & end)
Restrict search range to specific intervals.

```python
phrase = "alpha beta gamma delta alpha beta"
print("First 'alpha': ", phrase.index("alpha"))
print("Second 'alpha':", phrase.index("alpha", 5))
print("Beta in [5:20]:", phrase.index("beta", 5, 20))

assert phrase.index("alpha") == 0
assert phrase.index("alpha", 5) == 23
assert phrase.index("beta", 5, 20) == 6
```

### 3. Forward index() vs. Reverse rindex()
Locate the first vs. last occurrence of a delimiter.

```python
path = "src/controllers/api/v1/auth.py"
first_slash = path.index("/")
last_slash = path.rindex("/")

print("First slash:", first_slash)
print("Last slash: ", last_slash)
print("Filename:   ", path[last_slash + 1:])

assert first_slash == 3
assert last_slash == 22
assert path[last_slash + 1:] == "auth.py"
```

### 4. Strict Protocol Parsing
Parse configuration lines, failing fast if the delimiter is missing.

```python
def parse_header(line):
    try:
        delim = line.index(": ")
        return line[:delim], line[delim + 2:]
    except ValueError:
        return None

valid = parse_header("Content-Type: application/json")
invalid = parse_header("MalformedHeader")
print("Valid header:  ", valid)
print("Invalid header:", invalid)

assert valid == ("Content-Type", "application/json")
assert invalid is None
```

### 5. Practice Exercise: Strict Record Parser Suite
Run the complete parser testing valid Extractions and ValueError verification.

```python
class StrictRecordParser:
    @staticmethod
    def parse_key_value(line):
        idx = line.index("=")
        k = line[:idx].strip()
        v = line[idx + 1:].strip()
        if not k:
            raise ValueError("Empty key")
        return k, v

# Unit Assertions
parser = StrictRecordParser()
assert parser.parse_key_value("HOST=127.0.0.1") == ("HOST", "127.0.0.1")
assert parser.parse_key_value("PORT = 8080") == ("PORT", "8080")

try:
    parser.parse_key_value("NO_DELIMITER_HERE")
    assert False, "Should raise ValueError"
except ValueError:
    pass

print("All StrictRecordParser assertions passed successfully!")
```
