## 2. Basic DISTINCT Query (`sqlite3`)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT DISTINCT(class) FROM student ORDER BY class"
try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row[0])
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 3. Using COUNT(DISTINCT(...))

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT COUNT(DISTINCT(class)) FROM student"
my_cursor = my_conn.execute(q)
print("Total unique classes:", my_cursor.fetchone()[0])
my_conn.close()
```

## 4. Modern SQLAlchemy 2.0+ Context Manager with DISTINCT

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

my_conn = create_engine("sqlite:///my_db.db")
q = text("SELECT DISTINCT(class) FROM student WHERE mark >= 80 ORDER BY class")

try:
    with my_conn.connect() as connection:
        r_set = connection.execute(q)
        data = r_set.fetchall()
        for row in data:
            print(row[0])
        print("No of records displayed : ", len(data))
except SQLAlchemyError as e:
    print("SQLAlchemy error:", e)
```

## 5. Using Multiple Columns with DISTINCT

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT DISTINCT class, gender FROM student"
try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row[0], row[1])
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 6. Handling NULL Data with NOT NULL

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT DISTINCT class FROM student WHERE class IS NOT NULL"
try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row[0])
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```
