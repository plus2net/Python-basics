### 1. Calculate Time Left for New Year

```python
from dateutil.relativedelta import relativedelta
from datetime import datetime

dt = datetime.today()
dt2 = datetime(dt.year + 1, 1, 1, 0, 0, 0)
dt3 = relativedelta(dt2, dt)

print("Time left for New Year:")
print(f"Months: {dt3.months}, Days: {dt3.days}, Hours: {dt3.hours}, Minutes: {dt3.minutes}, Seconds: {dt3.seconds}")
assert dt3.years == 0
print("New Year delta calculation test passed!")
```
