### 1. Direct Inspection of Built-in Functions & Methods

```python
import pydoc

# In notebooks, pydoc.render_doc provides clean string documentation without terminal paging
doc_len = pydoc.render_doc(len, title="Documentation: %s")
print(doc_len)
```

### 2. Inspecting Language Keywords & Concepts

```python
# Headless inspection of keyword topics via pydoc
import keyword
print("Total Python keywords:", len(keyword.kwlist))
print("First 10 keywords:    ", keyword.kwlist[:10])
```

### 3. Custom Class with PEP 257 Docstrings

```python
class StringCleaner:
    """Utility class for sanitizing user-provided text inputs."""
    
    @staticmethod
    def clean(text: str) -> str:
        """Strips leading/trailing whitespace and normalizes internal spacing."""
        return " ".join(text.strip().split())

# Extract class documentation
print("Class docstring: ", StringCleaner.__doc__)
print("Method docstring:", StringCleaner.clean.__doc__)
```

### 4. Automated Method Docstring Catalog

```python
import inspect

catalog = {}
for name, func in inspect.getmembers(str, predicate=inspect.isroutine):
    if not name.startswith("__"):
        doc = inspect.getdoc(func) or ""
        catalog[name] = doc.split("\n")[0]

print("Total string methods cataloged:", len(catalog))
for m in ["split", "join", "replace", "upper", "strip"]:
    print(f"  {m:<10}: {catalog.get(m, 'N/A')}")
```

### 5. Self-Checking Unit Verification Suite

```python
import pydoc
import inspect

class TestTarget:
    """Utility test class docstring."""
    def execute(self):
        """Execute test method."""
        return True

# Test 1: pydoc documentation rendering
doc = pydoc.render_doc(print)
assert "Prints the values to a stream" in doc

# Test 2: Custom docstring verification
assert "Utility test class docstring." in inspect.getdoc(TestTarget)

# Test 3: Method docstring verification
assert "Execute test method." in inspect.getdoc(TestTarget.execute)

print("All help() unit assertions successfully verified! 100% passed.")
```
