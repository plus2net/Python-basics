## 1. Creating a List from a String

The string method `split()` separates a string and returns the parts as a list. By default, it splits at whitespace.

````python
message = "Welcome to Python"
words = message.split()

print(words)
````

**Expected output**

```text
['Welcome', 'to', 'Python']
```

We can also supply a separator when the text uses commas or another delimiter.

````python
colors_text = "red,green,blue"
colors = colors_text.split(",")

print(colors)
````

**Expected output**

```text
['red', 'green', 'blue']
```

## 2. Creating Lists with List Comprehension

A list comprehension creates a new list by applying an expression to each item in an iterable.

```python
new_list = [expression for item in iterable]
```

````python
squares = [number ** 2 for number in range(5)]

print(squares)
````

**Expected output**

```text
[0, 1, 4, 9, 16]
```

A condition can be added to keep only selected values.

````python
even_squares = [number ** 2 for number in range(10) if number % 2 == 0]

print(even_squares)
````

**Expected output**

```text
[0, 4, 16, 36, 64]
```

Only even numbers pass the condition before they are squared.

## 3. Adding Items to a List

Three useful methods add values in different ways:

- `append(x)` adds one item at the end.
- `insert(i, x)` adds one item at a chosen index.
- `extend(iterable)` adds every item from another iterable.

````python
languages = ["Python", "PHP"]

languages.append("JavaScript")
languages.insert(1, "Java")
languages.extend(["C", "Ruby"])

print(languages)
````

**Expected output**

```text
['Python', 'Java', 'PHP', 'JavaScript', 'C', 'Ruby']
```

### `append()` Compared with `extend()`

`append()` adds its argument as one item. `extend()` reads an iterable and adds its items separately.

````python
list_1 = [1, 2]
list_2 = [1, 2]

list_1.append([3, 4])
list_2.extend([3, 4])

print("Using append:", list_1)
print("Using extend:", list_2)
````

**Expected output**

```text
Using append: [1, 2, [3, 4]]
Using extend: [1, 2, 3, 4]
```

## 4. Finding and Counting List Items

- `index(x)` returns the index of the first matching value.
- `count(x)` returns how many times a value appears.

````python
colors = ["red", "blue", "green", "blue", "yellow"]

print("First blue index:", colors.index("blue"))
print("Number of blue items:", colors.count("blue"))
````

**Expected output**

```text
First blue index: 1
Number of blue items: 2
```

Use `in` before `index()` when a value may be missing. Otherwise, `index()` raises `ValueError`.

````python
search_value = "purple"

if search_value in colors:
    print(colors.index(search_value))
else:
    print(search_value, "is not in the list")
````

**Expected output**

```text
purple is not in the list
```

## 5. Removing List Items

- `remove(x)` removes the first matching value.
- `pop(i)` removes and returns the item at index `i`.
- `pop()` without an index removes and returns the last item.
- `clear()` removes every item.

````python
students = ["Alex", "Ronald", "John", "Ravi"]

students.remove("Ronald")
removed_student = students.pop(1)
last_student = students.pop()

print("Removed by index:", removed_student)
print("Removed last:", last_student)
print("Remaining list:", students)
````

**Expected output**

```text
Removed by index: John
Removed last: Ravi
Remaining list: ['Alex']
```

````python
students.clear()

print(students)
````

**Expected output**

```text
[]
```

The list variable still exists, but it is now empty.

## 6. Copying a List

The `copy()` method creates a new list. This matters because assigning one list variable to another does not create an independent list.

````python
original = [10, 20, 30]
copied = original.copy()

copied.append(40)

print("Original:", original)
print("Copy:", copied)
````

**Expected output**

```text
Original: [10, 20, 30]
Copy: [10, 20, 30, 40]
```

`copy()` makes a **shallow copy**. For a simple one-dimensional list, changes such as appending or replacing an item do not change the original list.

## 7. Sorting and Reversing Lists

The `sort()` method arranges list items in place. The `reverse()` method reverses their current order.

````python
numbers = [40, 10, 30, 20]

numbers.sort()
print("Ascending:", numbers)

numbers.sort(reverse=True)
print("Descending:", numbers)
````

**Expected output**

```text
Ascending: [10, 20, 30, 40]
Descending: [40, 30, 20, 10]
```

````python
names = ["Alex", "John", "Ravi", "Maya"]
names.reverse()

print(names)
````

**Expected output**

```text
['Maya', 'Ravi', 'John', 'Alex']
```

`reverse()` does not sort. It only flips the existing order.

## 8. Finding the Sum of List Values

The built-in `sum()` function returns the total of numeric list items.

````python
numbers = [1, 2, 5, 6]

print("Using sum():", sum(numbers))

total = 0
for number in numbers:
    total = total + number

print("Using a loop:", total)
````

**Expected output**

```text
Using sum(): 14
Using a loop: 14
```

Both approaches produce the same total. `sum()` is shorter, while the loop shows how the total is built one value at a time.

## 9. Collecting Matching Items by Filtering

The `filter()` function keeps items for which a function returns `True`.

````python
def contains_xy(value):
    return "xy" in value


source = ["abcd.php", "xyabcd", "pqrxy", "dataxy"]
matching = list(filter(contains_xy, source))

print(matching)
````

**Expected output**

```text
['xyabcd', 'pqrxy', 'dataxy']
```

The same task can be written using a list comprehension.

````python
matching = [value for value in source if "xy" in value]

print(matching)
````

**Expected output**

```text
['xyabcd', 'pqrxy', 'dataxy']
```

## 10. Creating a List with Unique Items

Converting a list to a set removes duplicate values. Convert it back to a list when a list is required.

````python
numbers = [1, 2, 3, 4, 4, 5, 5, 6]
unique_numbers = list(set(numbers))

print(sorted(unique_numbers))
````

**Expected output**

```text
[1, 2, 3, 4, 5, 6]
```

A set does not guarantee the original order. If order matters, use a loop.

````python
numbers = [3, 1, 3, 2, 1, 4]
unique_in_order = []

for number in numbers:
    if number not in unique_in_order:
        unique_in_order.append(number)

print(unique_in_order)
````

**Expected output**

```text
[3, 1, 2, 4]
```

## 11. Finding Duplicate Items

`collections.Counter` counts each value. Items with a count greater than `1` are duplicates.

````python
from collections import Counter

numbers = [1, 2, 3, 2, 1, 5, 6, 5, 5, 5]
counts = Counter(numbers)
duplicates = [item for item, count in counts.items() if count > 1]

print(duplicates)
````

**Expected output**

```text
[1, 2, 5]
```

## 12. Counting the Frequency of Every Item

A dictionary can store each list item as a key and its frequency as the value.

````python
letters = ["a", "z", "c", "z", "z", "c", "b", "a"]
frequency = {}

for letter in letters:
    if letter in frequency:
        frequency[letter] = frequency[letter] + 1
    else:
        frequency[letter] = 1

print(frequency)
````

**Expected output**

```text
{'a': 2, 'z': 3, 'c': 2, 'b': 1}
```

`Counter` provides a shorter alternative.

````python
print(dict(Counter(letters)))
````

**Expected output**

```text
{'a': 2, 'z': 3, 'c': 2, 'b': 1}
```

## 13. Finding Common Items Between Two Lists

Convert both lists to sets and use the intersection operator `&` to get values found in both.

````python
list_1 = [3, 2, 5, 8]
list_2 = [5, 2, 8, 9]

common = set(list_1) & set(list_2)

print(sorted(common))
print("Highest common value:", max(common))
print("Lowest common value:", min(common))
````

**Expected output**

```text
[2, 5, 8]
Highest common value: 8
Lowest common value: 2
```

## 14. Finding the Difference Between Two Lists

Set difference returns values present in the first list but not the second. Duplicate values and the original order are not preserved.

````python
list_1 = [5, 10, 20, 25, 15, 10, 5]
list_2 = [20, 10]

difference = set(list_1) - set(list_2)

print(sorted(difference))
````

**Expected output**

```text
[5, 15, 25]
```

Use a list comprehension to preserve order and duplicates.

````python
difference_in_order = [item for item in list_1 if item not in list_2]

print(difference_in_order)
````

**Expected output**

```text
[5, 25, 15, 5]
```

## 15. Converting a List to a String

The string method `join()` combines string items using a chosen separator. If a list also contains numbers, convert every item to a string first.

````python
items = ["One", 2, "Three"]
result = ",".join(map(str, items))

print(result)
````

**Expected output**

```text
One,2,Three
```

If all items are already strings, `map(str, ...)` is not required.

## 16. Selecting a Random Item from a List

The `choice()` function in Python's `random` module returns one random list item.

````python
import random

students = ["Alex", "Ron", "Ravi", "Maya", "John"]
selected_student = random.choice(students)

print(selected_student)
````

**Sample output**

```text

```

Your output may be different each time because the selection is random.

## 17. Working with a Two-Dimensional List

A two-dimensional or 2D list contains inner lists. Use two indexes to access an individual value: the first selects the row and the second selects the item in that row.

````python
data = [
    ["abc", "def", "ghi", "jkl"],
    ["mno", "pkr", "frt", "qwr"],
    ["asd", "air", "abc", "zpq"],
    ["zae", "vbg", "qir", "zab"]
]

print(data[1])
print(data[1][1])
````

**Expected output**

```text
['mno', 'pkr', 'frt', 'qwr']
pkr
```

A list comprehension can collect the item at index `1` from every row.

````python
second_items = [row[1] for row in data]

print(second_items)
````

**Expected output**

```text
['def', 'pkr', 'air', 'vbg']
```

## 18. Searching Inside a 2D List

Loop through the inner lists and check each row with the `in` operator.

````python
for row in data:
    if "pkr" in row:
        print(row)
````

**Expected output**

```text
['mno', 'pkr', 'frt', 'qwr']
```

The `next()` function can return the first matching row, or `None` when there is no match.

````python
matching_row = next((row for row in data if "pkr" in row), None)
missing_row = next((row for row in data if "Python" in row), None)

print(matching_row)
print(missing_row)
````

**Expected output**

```text
['mno', 'pkr', 'frt', 'qwr']
None
```

## 19. Creating a List from JSON Data

JSON often stores records as a list of dictionaries. After loading the JSON text, a list comprehension can collect one field from every record.

````python
import json

json_text = """
[
  {"name": "Alex", "mark": 82},
  {"name": "Ravi", "mark": 91},
  {"name": "Maya", "mark": 88}
]
"""

student_data = json.loads(json_text)
student_names = [student["name"] for student in student_data]

print(student_names)
````

**Expected output**

```text
['Alex', 'Ravi', 'Maya']
```

`json.loads()` reads JSON from a string. When reading JSON from a file, use `json.load()` instead.

## 20. Creating a List from Database Query Results

Database queries return rows. We can extract one value from each row to create a Python list. This example uses an in-memory SQLite database, so it runs directly in Colab without downloading a database file.

````python
import sqlite3

connection = sqlite3.connect(":memory:")
cursor = connection.cursor()

cursor.execute("CREATE TABLE students (name TEXT, mark INTEGER)")
cursor.executemany(
    "INSERT INTO students VALUES (?, ?)",
    [("Alex", 82), ("Ravi", 91), ("Maya", 88)]
)

cursor.execute("SELECT name FROM students ORDER BY name")
rows = cursor.fetchall()
student_names = [row[0] for row in rows]

print(student_names)

connection.close()
````

**Expected output**

```text
['Alex', 'Maya', 'Ravi']
```

Each returned row is a tuple. `row[0]` selects the name from every row.

## Mini Project: Student Score Summary

Use several Part 2 techniques together to clean and summarize a list of student results.

````python
results = [
    ["Alex", 82],
    ["Ravi", 91],
    ["Maya", 88],
    ["Alex", 82],
    ["John", 74]
]

# Keep unique rows while preserving their order
unique_results = []
for row in results:
    if row not in unique_results:
        unique_results.append(row)

names = [row[0] for row in unique_results]
scores = [row[1] for row in unique_results]
passed = [row for row in unique_results if row[1] >= 80]

print("Students:", ", ".join(names))
print("Total score:", sum(scores))
print("Average score:", sum(scores) / len(scores))
print("Scores from highest to lowest:", sorted(scores, reverse=True))
print("Scores of 80 or more:", passed)
````

**Expected output**

```text
Students: Alex, Ravi, Maya, John
Total score: 335
Average score: 83.75
Scores from highest to lowest: [91, 88, 82, 74]
Scores of 80 or more: [['Alex', 82], ['Ravi', 91], ['Maya', 88]]
```
