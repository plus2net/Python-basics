### 1. Module Scope vs. Function Scope

```python
# At module level, locals() is identical to globals()
print("Module level: locals() is globals():", locals() is globals())

def sample_function(x, y=10):
    z = x + y
    local_dict = locals()
    print("Inside function locals():", local_dict)
    return z

res = sample_function(5)
```

### 2. The Immutability Trap Inside Functions

```python
def test_mutation_behavior():
    greeting = "original"
    print("Before mutation: ", greeting)
    
    # Mutating the returned dictionary does NOT change the local variable
    locals()["greeting"] = "modified"
    print("After mutation:  ", greeting)

test_mutation_behavior()
```

### 3. Class Scope Introspection

```python
class Course:
    name = "Python Basics"
    lessons = 24
    body_locals = locals()
    print("Class body attributes:", [k for k in body_locals if not k.startswith("__")])

    def __init__(self, code, students):
        self.code = code
        self.students = students
        print("Init params:", {k: v for k, v in locals().items() if k != 'self'})

c = Course("PY-101", 45)
```

### 4. Dynamic String Template Formatting

```python
def format_summary(title, rating, votes):
    formatted = "Topic: {title} | Rating: {rating}/5 ({votes} votes)".format(**locals())
    return formatted

print(format_summary("Python locals() Guide", 4.9, 142))
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Module level locals is globals
assert locals() is globals()

# Test 2: Function local extraction
def calc(a, b=5):
    c = a * b
    return locals()

res_dict = calc(4)
assert res_dict['a'] == 4 and res_dict['b'] == 5 and res_dict['c'] == 20

# Test 3: Immutability inside functions
def test_var():
    x = 'stable'
    locals()['x'] = 'changed'
    return x

assert test_var() == 'stable'

# Test 4: Template formatting
user, status = 'Admin', 'Active'
assert '{user}: {status}'.format(**locals()) == 'Admin: Active'

print("All locals() unit assertions successfully verified! 100% passed.")
```
