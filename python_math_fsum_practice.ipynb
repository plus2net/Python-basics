### 1. Basic Summation with math.fsum()
`math.fsum(iterable)` takes any iterable of numbers and returns their exact sum as a 64-bit IEEE 754 float.


```python
import math

print("List sum:     ", math.fsum([10, 20.5, 30, 40.25]))
print("Tuple sum:    ", math.fsum((1.1, 2.2, 3.3)))
print("Set sum:      ", math.fsum({5, 10, 15}))
print("Range sum:    ", math.fsum(range(1, 11)))
print("Generator sum:", math.fsum(x**2 for x in range(1, 6)))
print("Empty sum:    ", math.fsum([]))
```

### 2. Why math.fsum() Beats sum(): Precision & Catastrophic Cancellation
Python's built-in `sum()` suffers from sequential intermediate rounding errors. `math.fsum()` tracks multiple precision expansions and rounds only once at the end.


```python
import math

# Test 1: Repeated decimal addition (0.1 x 10)
tenths = [0.1] * 10
print("Built-in sum([0.1]*10):", sum(tenths))        # 0.9999999999999999
print("math.fsum([0.1]*10):   ", math.fsum(tenths))   # 1.0 (Exact!)

# Test 2: Catastrophic cancellation
values = [1e100, 1.0, -1e100]
print("Built-in sum([1e100, 1.0, -1e100]):", sum(values))       # 0.0 (catastrophic cancellation!)
print("math.fsum([1e100, 1.0, -1e100]):   ", math.fsum(values))  # 1.0 (Accurate!)
```

### 3. Special Values: Infinities and NaNs
Observe how `math.fsum()` handles IEEE 754 special values.


```python
import math

print("fsum([1.0, inf]):    ", math.fsum([1.0, float("inf")]))
print("fsum([inf, -inf]):   ", math.fsum([float("inf"), float("-inf")]))
print("fsum([10.0, nan, 5]):", math.fsum([10.0, float("nan"), 5.0]))
```

### 4. Hands-on Practice Exercise: High-Volume Financial Ledger Auditor
In financial clearinghouses, transactions combine massive institutional balances and retail micro-transactions. Implement `audit_ledger_balance(transactions)` to verify accounting reconciliation without cumulative rounding errors.


```python
import math

def audit_ledger_balance(transactions, tolerance=1e-12):
    """
    Audits a stream of debit and credit transactions.
    Returns (is_balanced, fsum_total, naive_sum_total).
    """
    if not isinstance(transactions, (list, tuple)):
        raise TypeError("Transactions must be a list or tuple.")
    exact_total = math.fsum(transactions)
    naive_total = sum(transactions)
    balanced = math.isclose(exact_total, 0.0, abs_tol=tolerance)
    return (balanced, exact_total, naive_total)

# Unit Assertions
# Batch 1: Micro-transactions balancing to 0
batch1 = [0.1] * 10 + [-1.0]
is_bal, f_tot, n_tot = audit_ledger_balance(batch1)
assert is_bal is True
assert f_tot == 0.0
assert n_tot != 0.0  # Naive sum fails due to rounding noise

# Batch 2: Large transfers with small residual
batch2 = [50000000.0, 0.07, -50000000.0, -0.07]
is_bal2, f_tot2, _ = audit_ledger_balance(batch2)
assert is_bal2 is True
assert f_tot2 == 0.0

# Type checking validation
try:
    audit_ledger_balance("invalid")
    assert False, "Should raise TypeError"
except TypeError:
    pass

print("Batch 1 reconciled with fsum():", f_tot)
print("All financial ledger audit assertions passed successfully!")
```
