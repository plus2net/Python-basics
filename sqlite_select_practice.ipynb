## 2. Connecting using SQLAlchemy (2.0+ Standard)

```python
from sqlalchemy import create_engine, text

my_conn = create_engine("sqlite:///my_db.db")

with my_conn.connect() as connection:
    query = text("SELECT id, name, class, mark, gender FROM student LIMIT 0,5")
    r_set = connection.execute(query)
    for row in r_set:
        print(row)
```

## 3. Basic SELECT Query Iteration (`sqlite3`)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT id, name, class, mark, gender FROM student LIMIT 0,5"
my_cursor = my_conn.execute(q)
for row in my_cursor:
    print(row[0], row[1], row[2], row[3], row[4])
my_conn.close()
```

## 4. Fetching Records using `fetchall()` and `fetchone()`

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')

# fetchall()
q = "SELECT id, name, class, mark, gender FROM student LIMIT 0,5"
my_cursor = my_conn.execute(q)
data_rows = my_cursor.fetchall()
print("Type of data_rows:", type(data_rows))
for row in data_rows:
    print(row)

# fetchone()
q = "SELECT id, name, class, mark, gender FROM student WHERE id = 8"
my_cursor = my_conn.execute(q)
data_row = my_cursor.fetchone()
print("Type of data_row:", type(data_row))
print(data_row)
my_conn.close()
```

## 5. Using Error Handling with `try-except`

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT * FROM student WHERE class = 'Four'"
try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row[0], row[1], row[2], row[3], row[4])
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 6. Parameterized SELECT Queries (Preventing SQL Injection)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = ('Seven', 'Six')
q = "SELECT * FROM student WHERE class = ? OR class = ?"
try:
    my_cursor = my_conn.execute(q, my_data)
    for row in my_cursor:
        print(row[0], row[1], row[2], row[3], row[4])
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```
