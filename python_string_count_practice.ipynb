## 1. Basic Occurrence Counting & Case Sensitivity

`str.count(sub)` returns the number of non-overlapping occurrences of `sub` in the string.
Matching is strictly case-sensitive.

```python
text = "Welcome to plus2net.com. Enjoy learning Python at plus2net!"

site_count = text.count("plus2net")
letter_o = text.count("o")

print("Occurrences of 'plus2net':", site_count)
print("Occurrences of 'o':       ", letter_o)

assert site_count == 2
assert letter_o == 5

# Case sensitivity
assert text.count("PLUS2NET") == 0

print("\nAll basic counting assertions passed!")
```

## 2. The Non-Overlapping Invariant vs. Regex Lookaheads

`str.count()` advances past matched characters and does not double-count overlapping sequences.
To capture overlapping matches, use regular expression lookahead assertions `(?=(...))`.

```python
import re

seq = "aaaa"

# 1. str.count(): Non-overlapping
non_overlap = seq.count("aa")
print("str.count('aa'):          ", non_overlap)
assert non_overlap == 2

# 2. Regex lookahead: Overlapping
overlap = len(re.findall(r"(?=(aa))", seq))
print("Overlapping count of 'aa':", overlap)
assert overlap == 3

# 'banana' example: 'ana' appears at index 1 and index 3
fruit = "banana"
assert fruit.count("ana") == 1
assert len(re.findall(r"(?=(ana))", fruit)) == 2

print("\nNon-overlapping vs overlapping assertions passed!")
```

## 3. Slice Boundaries: `start` and `end` Offsets

`str.count(sub, start, end)` restricts the search to index range `[start:end]` without creating temporary substring slices.

```python
phrase = "ab_ab_ab_ab"

assert phrase.count("ab") == 4
assert phrase.count("ab", 3) == 3
assert phrase.count("ab", 3, 8) == 2
assert phrase.count("ab", -5) == 2

print("Slice boundary assertions passed!")
```

## 4. The Empty String Occurrence Formula ($N + 1$)

For any string of length $N$, `s.count("")` returns $N + 1$.

```python
word = "code"
print("Length:    ", len(word))
print("count(''): ", word.count(""))

assert word.count("") == len(word) + 1
assert "".count("") == 1

print("Empty string occurrence assertions passed!")
```

## 5. Real-World Engineering Recipes

Explore three common software engineering patterns utilizing `str.count()`:
1. Bioinformatics GC-Content Ratio Calculator
2. Server Log Error Severity Auditor
3. Syntax Bracket Parity Validator

```python
# Recipe 1: GC Content
def gc_content(dna: str) -> float:
    seq = dna.upper()
    if not seq:
        return 0.0
    return ((seq.count("G") + seq.count("C")) / len(seq)) * 100.0

gc = gc_content("ATGCGATCGATCGATCGATCGATCGATCGATAGCTAGCTAGCTA")
print(f"Recipe 1 GC: {gc:.2f}%")
assert round(gc, 2) == 47.73

# Recipe 2: Log severity auditor
def audit_logs(log_text: str) -> dict[str, int]:
    return {
        "CRITICAL": log_text.count("CRITICAL:"),
        "ERROR": log_text.count("ERROR:"),
        "WARN": log_text.count("WARN:")
    }

sample_log = "INFO: Start\nERROR: Conn fail\nWARN: Slow disk\nERROR: Retry fail"
stats = audit_logs(sample_log)
print("Recipe 2 Stats:", stats)
assert stats["ERROR"] == 2 and stats["WARN"] == 1

# Recipe 3: Bracket parity
def is_balanced(expr: str) -> bool:
    return expr.count("(") == expr.count(")") and expr.count("[") == expr.count("]")

assert is_balanced("((a + b) * [c - d])") is True
assert is_balanced("((a + b)") is False

print("\nAll engineering recipes passed!")
```

## 6. Production Implementation & Self-Checking Unit Assertions

This production-grade `SubstringCounter` class encapsulates case-insensitive counting, GC-content ratios, and bracket balance verification.

```python
class SubstringCounter:
    @staticmethod
    def count_case_insensitive(text: str, sub: str) -> int:
        """Counts substring occurrences ignoring case."""
        return text.lower().count(sub.lower())

    @staticmethod
    def gc_percentage(dna: str) -> float:
        """Computes GC content percentage."""
        s = dna.upper()
        if not s:
            return 0.0
        return ((s.count("G") + s.count("C")) / len(s)) * 100.0

    @staticmethod
    def is_balanced_parens(expr: str) -> bool:
        """Checks parentheses parity via count."""
        return expr.count("(") == expr.count(")")


# ==========================================
# UNIT ASSERTION SUITE
# ==========================================
counter = SubstringCounter()

# 1. Non-overlapping counting
assert "aaaa".count("aa") == 2
assert "banana".count("ana") == 1

# 2. Slice bounds [start:end]
assert "abc_abc_abc".count("abc", 4) == 2
assert "abc_abc_abc".count("abc", 4, 7) == 1

# 3. Empty string occurrence rule (N + 1)
assert "hello".count("") == 6
assert "".count("") == 1

# 4. Case-insensitive counting
assert counter.count_case_insensitive("Python python PYTHON", "python") == 3

# 5. GC content calculation
assert round(counter.gc_percentage("GCGCAT"), 2) == 66.67
assert counter.gc_percentage("") == 0.0

# 6. Parentheses parity
assert counter.is_balanced_parens("(a + (b * c))") is True
assert counter.is_balanced_parens("(a + b") is False

print("All str.count() production unit assertions passed successfully!")
```
