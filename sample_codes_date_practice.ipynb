## Example 1: First working example

Read the code, predict the output and run this cell.

```python
from datetime import date, timedelta
today = date.today()
print("Today:", today)
print("Yesterday:", today - timedelta(days=1))
print("Tomorrow:", today + timedelta(days=1))
```

## Example 2: Display the next five days

Read the code, predict the output and run this cell.

```python
from datetime import date, timedelta
start = date(2026, 1, 29)
for i in range(5):
    print(start + timedelta(days=i))
```

## Example 3: Check leap years correctly

Read the code, predict the output and run this cell.

```python
import calendar
for year in (1900, 2000, 2024, 2025):
    print(year, calendar.isleap(year))
```

## Example 4: Find weekends within a date range

Read the code, predict the output and run this cell.

```python
from datetime import datetime, timedelta
start = datetime.strptime("1 Dec 2021", "%d %b %Y").date()
end = datetime.strptime("12 Dec 2021", "%d %b %Y").date()
days = (end - start).days + 1
for offset in range(days):
    current = start + timedelta(days=offset)
    if current.weekday() >= 5:
        print(current.strftime("%Y-%m-%d (%A)"))
```

## Example 5: Find weekday from user input

Read the code, predict the output and run this cell.

```python
from unittest.mock import patch
with patch("builtins.input", side_effect=['31/12/2022']):
    from datetime import datetime
    text = input("Enter date dd/mm/YYYY: ")
    value = datetime.strptime(text, "%d/%m/%Y")
    print(text, value.strftime("%A"))
```

## Example 6: First Monday of a month

Read the code, predict the output and run this cell.

```python
from datetime import date
year, month = 2026, 11
first_day = date(year, month, 1)
first_monday = first_day.replace(day=1 + (-first_day.weekday()) % 7)
print(first_monday)
```

## Example 7: First day of next month

Read the code, predict the output and run this cell.

```python
from datetime import date
d = date(2019, 12, 15)
first_next = date(d.year + (d.month == 12), d.month % 12 + 1, 1)
print(first_next)
```

## Example 8: Find the last day of the current month

Read the code, predict the output and run this cell.

```python
from datetime import date, timedelta
d = date(2024, 2, 10)
next_month = date(d.year + (d.month == 12), d.month % 12 + 1, 1)
print(next_month - timedelta(days=1))
```

## Example 9: Add hours and minutes

Read the code, predict the output and run this cell.

```python
from datetime import datetime, timedelta
start = datetime(2026, 5, 1, 11, 25)
print(start + timedelta(hours=2, minutes=35))
```

## Example 10: Validate a date string

Read the code, predict the output and run this cell.

```python
from datetime import datetime
try:
    value = datetime.strptime("31/02/2025", "%d/%m/%Y")
    print(value.date())
except ValueError:
    print("Invalid date")
```
