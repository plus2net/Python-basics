## 1. Installation of Database & Connection

```python
!wget https://www.plus2net.com/python/download/my_db.db
```

```python
import sqlite3
my_conn = sqlite3.connect('my_db.db')
query = "SELECT name FROM sqlite_master WHERE type='table'"
my_data = list(my_conn.execute(query))
print("Tables available:", my_data)
```

## 2. From SQLite Table to Excel (`read_sql` & `to_excel`)

```python
import pandas as pd
query = 'SELECT * FROM student'
df = pd.read_sql(query, my_conn, index_col='id')
print(df.head())
df.to_excel('my_student.xlsx')
print("Excel file created successfully!")
```

## 3. From Excel to SQLite Table (`read_excel` & `to_sql`)

```python
try:
    r_set = my_conn.execute('DROP TABLE student')
    my_conn.commit()
    print("Old student table dropped.")
except sqlite3.Error as my_error:
    print("error:", my_error)

df_excel = pd.read_excel('my_student.xlsx', index_col='id')
df_excel.to_sql(con=my_conn, name='student2', if_exists='append')

query = "SELECT * FROM student2 LIMIT 5"
for row in my_conn.execute(query):
    print(row)
```

## 4. Modern SQLAlchemy 2.0+ Context Manager Alternative

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

my_engine = create_engine("sqlite:///my_db.db")
try:
    with my_engine.connect() as connection:
        df_sa = pd.read_sql(text("SELECT * FROM student2"), connection, index_col='id')
        print(df_sa.head(3))
except SQLAlchemyError as e:
    print("SQLAlchemy error:", e)
```
