### 1. Basic Truncation with math.trunc()
`math.trunc(x)` returns the integer part of `x` by stripping all decimal digits (rounding strictly towards zero).


```python
import math

print("trunc(3.14): ", math.trunc(3.14))   # 3
print("trunc(3.99): ", math.trunc(3.99))   # 3
print("trunc(-3.14):", math.trunc(-3.14))  # -3
print("trunc(-3.99):", math.trunc(-3.99))  # -3
print("trunc(42):   ", math.trunc(42))     # 42
print("trunc(0.0):  ", math.trunc(0.0))    # 0
```

### 2. The __trunc__() Dunder Protocol for Custom Classes
`math.trunc()` delegates to the `__trunc__()` special method. Implementing it on custom classes enables clean integer truncation.


```python
import math

class FinancialAmount:
    def __init__(self, dollars, cents):
        self.total = dollars + cents / 100.0
        
    def __trunc__(self):
        return math.trunc(self.total)

wallet = FinancialAmount(149, 85)
print("math.trunc(wallet):", math.trunc(wallet))
print("int(wallet):       ", int(wallet))
assert math.trunc(wallet) == 149
```

### 3. Comparison: trunc() vs. floor() vs. ceil() vs. round()
Compare rounding directions across positive and negative floats.


```python
import math

samples = [4.2, 4.8, -4.2, -4.8]
print(f"{'Value':>6} | {'trunc':>6} | {'floor':>6} | {'ceil':>6} | {'round':>6}")
print("-" * 38)
for v in samples:
    print(f"{v:6.1f} | {math.trunc(v):6d} | {math.floor(v):6d} | {math.ceil(v):6d} | {round(v):6d}")
```

### 4. Handling Non-Finite Values (Infinities and NaNs)
Because infinities and NaNs cannot be converted to integers, `math.trunc()` raises explicit exceptions.


```python
import math

try:
    math.trunc(float("inf"))
except OverflowError as err:
    print("Caught expected OverflowError:", err)

try:
    math.trunc(float("nan"))
except ValueError as err:
    print("Caught expected ValueError:   ", err)
```

### 5. Hands-on Practice Exercise: Video SMPTE Timecode Formatter
In video editing, playback elapsed seconds must be converted to discrete timecode `HH:MM:SS:FF` (Hours, Minutes, Seconds, Frames) without rounding drift.

Implement `seconds_to_smpte(total_seconds, fps)` and verify with self-checking assertions.


```python
import math

def seconds_to_smpte(total_seconds, fps=30):
    """
    Converts elapsed seconds into SMPTE timecode string: HH:MM:SS:FF.
    Uses math.trunc() to ensure clean, non-rounding component extraction.
    """
    if not isinstance(total_seconds, (int, float)) or total_seconds < 0:
        raise ValueError("Total seconds must be a non-negative number.")
    if not isinstance(fps, int) or fps <= 0:
        raise ValueError("Frames per second (fps) must be a positive integer.")
        
    int_sec = math.trunc(total_seconds)
    sub_sec = total_seconds - int_sec
    
    hours = math.trunc(int_sec / 3600)
    rem_sec = int_sec % 3600
    minutes = math.trunc(rem_sec / 60)
    seconds = rem_sec % 60
    frames = math.trunc(sub_sec * fps)
    
    return f"{hours:02d}:{minutes:02d}:{seconds:02d}:{frames:02d}"

# Unit Assertions
assert seconds_to_smpte(0.0) == "00:00:00:00"
assert seconds_to_smpte(65.5, fps=30) == "00:01:05:15"
assert seconds_to_smpte(3661.1, fps=30) == "01:01:01:03"
assert seconds_to_smpte(7200.0, fps=24) == "02:00:00:00"

# Error check
try:
    seconds_to_smpte(-5.0)
    assert False, "Should raise ValueError"
except ValueError:
    pass

print("Sample video timecode (3661.1s at 30fps):", seconds_to_smpte(3661.1, 30))
print("All SMPTE timecode assertions passed successfully!")
```
