### 1. Basic Date Arithmetic

```python
from datetime import date, timedelta

today = date.today()
next_week = today + timedelta(weeks=1)
ten_days_ago = today - timedelta(days=10)

print("Today:        ", today)
print("Next week:    ", next_week)
print("Ten days ago: ", ten_days_ago)
```

### 2. Normalization & Total Seconds

```python
from datetime import timedelta

delta = timedelta(days=2, hours=3, minutes=15)
print("Normalized days:       ", delta.days)
print("Partial sec (.seconds):", delta.seconds)
print("Total seconds:         ", delta.total_seconds())
```

### 3. Duration Scaling & Division

```python
from datetime import timedelta

shift = timedelta(hours=8)
half_shift = shift / 2
total_work = shift * 5

print("Half shift:  ", half_shift)
print("5-day total: ", total_work)
print("Shift ratio: ", total_work / shift)
```

### 4. Token TTL Expiration Recipe

```python
from datetime import datetime, timedelta

created = datetime(2026, 10, 18, 12, 0, 0)
ttl = timedelta(hours=1)

check1 = datetime(2026, 10, 18, 12, 45, 0)
check2 = datetime(2026, 10, 18, 13, 15, 0)

print("Check 1 expired?", (check1 - created) > ttl)
print("Check 2 expired?", (check2 - created) > ttl)
```

### 5. Self-Checking Unit Verification Suite

```python
from datetime import date, datetime, timedelta

# Test 1: Date arithmetic
d = date(2026, 10, 18)
assert d + timedelta(days=5) == date(2026, 10, 23)

# Test 2: Normalization
t = timedelta(hours=48)
assert t.days == 2 and t.seconds == 0

# Test 3: total_seconds
sec_test = timedelta(hours=1)
assert sec_test.total_seconds() == 3600.0

# Test 4: Division between timedeltas
assert timedelta(days=1) / timedelta(hours=12) == 2.0

print("All datetime.timedelta unit assertions successfully verified! 100% passed.")
```
