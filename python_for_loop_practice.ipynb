### 1. Basic range() Iterations

```python
# Iterating with range(stop), range(start, stop), and range(start, stop, step)
print("Range 0 to 4:")
for i in range(5):
    print(i, end=" ")
print()

print("\nRange 5 to 25 step 5:")
for i in range(5, 25, 5):
    print(i, end=" ")
print()
```

### 2. Flow Controls: break, continue & for...else

```python
# Demonstration of continue vs break and for...else
print("Testing continue (skipping 15):")
for x in range(5, 25, 5):
    if x == 15:
        continue
    print(x, end=" ")
else:
    print("\n-> Normal loop completion without break!")

print("\nTesting break (halting at 15):")
for x in range(5, 25, 5):
    if x == 15:
        print("\n-> Break hit at 15!")
        break
    print(x, end=" ")
else:
    print("This else block will NOT execute!")
```

### 3. Prime Number Scanner (for...else pattern)

```python
primes = []
for n in range(2, 30):
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            break
    else:
        primes.append(n)
print("Primes under 30:", primes)
```

### 4. Advanced Iteration: enumerate() and zip()

```python
names = ["Alice", "Bob", "Charlie"]
scores = [95, 88, 92]

# enumerate
for idx, name in enumerate(names, start=1):
    print(f"{idx}. {name}")

# zip
for name, score in zip(names, scores):
    print(f"{name} got {score}")
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Sum range
assert sum(range(10)) == 45

# Test 2: for...else normal completion
flag = False
for i in range(3):
    pass
else:
    flag = True
assert flag is True

# Test 3: enumerate indexing
res = list(enumerate(['x', 'y'], start=1))
assert res == [(1, 'x'), (2, 'y')]

# Test 4: zip pairing
pair = list(zip([1, 2], ['a', 'b']))
assert pair == [(1, 'a'), (2, 'b')]

print("All Python for loop unit assertions successfully verified! 100% passed.")
```
