### 1. Comparing split() vs. rsplit() with maxsplit=1
Directional difference only manifests when `maxsplit` is constrained. `split()` isolates leading tokens, while `rsplit()` isolates trailing tokens.

```python
path = "usr/local/bin/python"

# split() consumes the first delimiter from the left
left_split = path.split("/", 1)
print("split(maxsplit=1): ", left_split)   # ['usr', 'local/bin/python']

# rsplit() consumes the first delimiter from the right
right_split = path.rsplit("/", 1)
print("rsplit(maxsplit=1):", right_split)  # ['usr/local/bin', 'python']
```

### 2. Default Whitespace Algorithm (sep=None) vs. Explicit Delimiters
When `sep=None`, consecutive whitespace characters are consolidated. When `maxsplit=1`, trailing whitespace is stripped before the rightmost split.

```python
record = "  apple    banana   cherry  "

# Default whitespace grouping from right (maxsplit=1)
clean_tokens = record.rsplit(None, 1)
print("sep=None: ", clean_tokens)

# Explicit single space matching from right (maxsplit=1)
raw_tokens = record.rsplit(" ", 1)
print("sep=' ':  ", raw_tokens)
```

### 3. Engineering Recipe 1: Splitting Path into Directory and Filename
Use a single right-hand split to separate directories from filenames without third-party libraries.

```python
def split_path(filepath: str) -> tuple:
    parts = filepath.rsplit("/", 1)
    if len(parts) == 1:
        return "", parts[0]
    return parts[0], parts[1]

paths = ["/var/log/nginx/access.log", "docs/guide.txt", "server.py"]
for p in paths:
    d, f = split_path(p)
    print(f"{p:30} -> Dir: {d:18} File: {f}")
```

### 4. Engineering Recipe 2: Isolating Usernames and Email Domains
Isolate the true domain host by splitting strictly at the rightmost `@` symbol.

```python
def parse_email(email: str) -> tuple:
    tokens = email.rsplit("@", 1)
    if len(tokens) != 2:
        raise ValueError(f"Invalid email format: {email}")
    return tokens[0], tokens[1]

user, domain = parse_email("developer.support@plus2net.com")
print("Username:", user)
print("Domain:  ", domain)
```

### 5. Engineering Recipe 3: Parsing Key-Value Configuration Lines
Split trailing parameters or connection string queries without disturbing earlier tokens.

```python
config_entry = "database_url=postgres://user:pass@host:5432/db?ssl=true"

# Separate key from val
key, val = config_entry.split("=", 1)

# Separate trailing query params from base URL
base, last_param = val.rsplit("?", 1)
print("Key:       ", key)
print("Base URL:  ", base)
print("Parameters:", last_param)
```

### 6. Common Pitfalls: Empty Separator & Unconstrained Parity
Passing an empty string `""` as separator raises `ValueError`. When `maxsplit` is omitted, `rsplit()` produces identical output to `split()`.

```python
try:
    "sample".rsplit("")
except ValueError as err:
    print("Handled expected error:", err)

text = "red,green,blue"
print("split: ", text.split(","))
print("rsplit:", text.rsplit(","))
print("Identical lists?", text.split(",") == text.rsplit(","))
```

### 7. Self-Checking Verification Suite
Run complete unit assertions verifying all `rsplit` behavior and edge cases.

```python
class RightSplitVerifier:
    @staticmethod
    def extract_suffix(target: str, delim: str) -> str:
        tokens = target.rsplit(delim, 1)
        return tokens[-1] if len(tokens) > 1 else ""

    @staticmethod
    def tail_n_tokens(text: str, delim: str, count: int) -> list:
        return text.rsplit(delim, count)

verifier = RightSplitVerifier()
assert verifier.extract_suffix("archive.tar.gz", ".") == "gz"
assert verifier.extract_suffix("README", ".") == ""

sample = "one:two:three:four"
assert sample.rsplit(":", 1) == ["one:two:three", "four"]
assert sample.split(":", 1) == ["one", "two:three:four"]
assert sample.rsplit(":") == sample.split(":")

ws_test = "alpha   beta   gamma"
assert ws_test.rsplit(None, 1) == ["alpha   beta", "gamma"]
ws_padded = "  alpha   beta   gamma  "
assert ws_padded.rsplit(None, 1) == ["  alpha   beta", "gamma"]
assert ws_padded.rsplit(None) == ["alpha", "beta", "gamma"]

print("All str.rsplit() unit assertions passed successfully! 100% verified.")
```
