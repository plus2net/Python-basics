## 1. Print the July 2020 week grid
`monthdayscalendar()` returns complete seven-day rows; days outside July appear as zero.

```python
import calendar
weeks = calendar.Calendar().monthdayscalendar(2020, 7)
for week in weeks:
    print(week)
assert weeks[0] == [0, 0, 1, 2, 3, 4, 5]
assert weeks[-1] == [27, 28, 29, 30, 31, 0, 0]
```

## 2. Inspect the return type
Both the month result and each week are Python lists.

```python
import calendar
weeks = calendar.Calendar().monthdayscalendar(2020, 7)
print(type(weeks).__name__, type(weeks[0]).__name__)
assert isinstance(weeks, list)
assert all(isinstance(week, list) and len(week) == 7 for week in weeks)
```

## 3. Original Thursday-first example
Change `firstweekday` to 3. What happens to the number of displayed rows?

```python
import calendar
weeks = calendar.Calendar(firstweekday=3).monthdayscalendar(2020, 7)
for week in weeks:
    print(week)
assert all(len(week) == 7 for week in weeks)
assert sum(day != 0 for week in weeks for day in week) == 31
```

## 4. Count weeks and slots
An output row always has seven entries.

```python
import calendar
weeks = calendar.Calendar(firstweekday=0).monthdayscalendar(2020, 7)
print("Number of weeks",len(weeks))
print("Number of days in week",len(weeks[0]))
assert len(weeks) == 5
assert len(weeks[0]) == 7
```

## 7. Change to Sunday-first
Sunday has weekday number 6. The day count must still be 31.

```python
import calendar
cal = calendar.Calendar(firstweekday=calendar.SUNDAY)
weeks = cal.monthdayscalendar(2020,7)
print(weeks[0])
print([calendar.day_abbr[d] for d in cal.iterweekdays()])
assert sum(d>0 for week in weeks for d in week) == 31
assert list(cal.iterweekdays())[0] == calendar.SUNDAY
```

## 9. Count weekdays and weekend dates
Do not assume index zero is Monday when you use a Sunday-first calendar.

```python
import calendar
cal = calendar.Calendar(firstweekday=6)
weekdays = list(cal.iterweekdays())
working = [d for week in cal.monthdayscalendar(2020,7) for wd,d in zip(weekdays,week) if d and wd < 5]
weekend = [d for week in cal.monthdayscalendar(2020,7) for wd,d in zip(weekdays,week) if d and wd >= 5]
print("Weekdays:",len(working),"Weekend days:",len(weekend))
assert len(working)==23 and len(weekend)==8
```

## 10. Find the week containing day 15

```python
import calendar
weeks=calendar.Calendar().monthdayscalendar(2020,7)
found=[(i,week) for i,week in enumerate(weeks, start=1) if 15 in week]
print(found)
assert len(found)==1
assert found[0][0]==3
```

## 11. Count February days in a leap year

```python
import calendar
weeks=calendar.Calendar().monthdayscalendar(2024,2)
days=[d for week in weeks for d in week if d]
print("February 2024 has",len(days),"days")
assert len(days)==29
assert max(days)==29
```

## 13. Compare three output representations
`monthdays2calendar()` returns tuples, while `monthdatescalendar()` returns real `datetime.date` objects.

```python
import calendar
from datetime import date
cal=calendar.Calendar()
nums=cal.monthdayscalendar(2020,7)
pairs=cal.monthdays2calendar(2020,7)
dates=cal.monthdatescalendar(2020,7)
print(nums[0],pairs[0][:3],dates[0][:3])
assert nums[0][0]==0
assert pairs[0][0] == (0,0)
assert isinstance(dates[0][0],date)
assert dates[0][0].isoformat() == '2020-06-29'
```

## 14. Build a simple weekly task plan

```python
import calendar
rows=calendar.Calendar().monthdayscalendar(2020,7)
plan={f"Week {i}":[d for d in row if d] for i,row in enumerate(rows,start=1)}
for label,days in plan.items():
    print(label,days)
assert len(plan)==5
assert sum(len(v) for v in plan.values())==31
```

## 16. Catch an invalid month safely

```python
import calendar
try:
    calendar.Calendar().monthdayscalendar(2020,13)
except calendar.IllegalMonthError as error:
    print("Expected invalid month:",error)
else:
    raise AssertionError("Expected an IllegalMonthError")
```
