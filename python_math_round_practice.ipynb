### 1. Basic Rounding with round(number, ndigits)
Omitting `ndigits` returns an integer. Positive `ndigits` rounds to decimal places, while negative `ndigits` rounds to powers of 10.


```python
print("round(12.34):     ", round(12.34), type(round(12.34)))      # 12 <class 'int'>
print("round(12.89):     ", round(12.89), type(round(12.89)))      # 13 <class 'int'>
print("round(12.34567, 2):", round(12.34567, 2))                   # 12.35 <class 'float'>
print("round(1245, -1):   ", round(1245, -1))                     # 1250 (tens)
print("round(1245, -2):   ", round(1245, -2))                     # 1200 (hundreds)
print("round(1745, -3):   ", round(1745, -3))                     # 2000 (thousands)
```

### 2. Banker's Rounding (Round-Half-to-Even)
When a number is halfway between two possibilities, Python rounds to the nearest *even* number to eliminate statistical upward bias.


```python
print("round(0.5):", round(0.5))  # 0 is even -> 0
print("round(1.5):", round(1.5))  # 2 is even -> 2
print("round(2.5):", round(2.5))  # 2 is even -> 2
print("round(3.5):", round(3.5))  # 4 is even -> 4
print("round(4.5):", round(4.5))  # 4 is even -> 4
print("round(5.5):", round(5.5))  # 6 is even -> 6
```

### 3. The IEEE 754 Float Precision Trap & the Decimal Module
Why does `round(2.675, 2)` yield `2.67`? Because `2.675` is stored internally as `2.6749999999999998...`, strictly less than halfway!


```python
from decimal import Decimal, ROUND_HALF_UP, ROUND_HALF_EVEN

print("Float round(2.675, 2):  ", round(2.675, 2))
print(f"Exact float representation: {2.675:.25f}")

val = Decimal("2.675")
print("Decimal ROUND_HALF_UP:   ", val.quantize(Decimal("0.01"), rounding=ROUND_HALF_UP))   # 2.68
print("Decimal ROUND_HALF_EVEN: ", val.quantize(Decimal("0.01"), rounding=ROUND_HALF_EVEN)) # 2.68
```

### 4. Master Comparison: round() vs. floor() vs. ceil() vs. trunc()
Observe how rounding behavior differs across positive, negative, and halfway numbers.


```python
import math

samples = [2.2, 2.5, 2.8, 3.5, -2.5, -2.8]
print(f"{'Value':>6} | {'round':>6} | {'floor':>6} | {'ceil':>6} | {'trunc':>6}")
print("-" * 38)
for v in samples:
    print(f"{v:6.1f} | {round(v):6d} | {math.floor(v):6d} | {math.ceil(v):6d} | {math.trunc(v):6d}")
```

### 5. Hands-on Practice Exercise: Sales Tax & Cash POS Rounder
Calculate exact regulatory sales tax with `ROUND_HALF_UP` and Swedish cash register denomination rounding (nearest $0.05).


```python
from decimal import Decimal, ROUND_HALF_UP

def calculate_sales_tax(subtotal_dollars, tax_rate_percent):
    """Computes exact sales tax with traditional ROUND_HALF_UP rounding."""
    if subtotal_dollars < 0 or tax_rate_percent < 0:
        raise ValueError("Subtotal and tax rate must be non-negative.")
    sub_dec = Decimal(str(subtotal_dollars))
    rate_dec = Decimal(str(tax_rate_percent)) / Decimal("100")
    raw_tax = sub_dec * rate_dec
    tax_cents = raw_tax.quantize(Decimal("0.01"), rounding=ROUND_HALF_UP)
    return float(tax_cents)

def swedish_cash_round(amount, step=0.05):
    """Rounds a monetary total to the nearest cash denomination (e.g. 0.05)."""
    if amount < 0 or step <= 0:
        raise ValueError("Amount and step must be positive.")
    units = round(amount / step)
    return round(units * step, 2)

# Unit Assertions
assert calculate_sales_tax(35.45, 8.25) == 2.92
assert calculate_sales_tax(49.99, 7.5) == 3.75
assert swedish_cash_round(10.02) == 10.00
assert swedish_cash_round(10.03) == 10.05
assert swedish_cash_round(10.07) == 10.05
assert swedish_cash_round(10.08) == 10.10

print("Sales tax on $100 at 8.875%:", calculate_sales_tax(100.0, 8.875))
print("Cash total for $15.43:       ", swedish_cash_round(15.43))
print("All sales tax & POS rounding assertions passed successfully!")
```
