## 2. Basic BETWEEN Query (`sqlite3`)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT * FROM student WHERE mark BETWEEN 65 AND 70"
try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error: ", my_error)
finally:
    my_conn.close()
```

## 3. Using Parameterized BETWEEN Query

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = (65, 70)
q = "SELECT * FROM student WHERE mark BETWEEN ? AND ?"
try:
    my_cursor = my_conn.execute(q, my_data)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error: ", my_error)
finally:
    my_conn.close()
```

## 4. Negating Ranges using NOT BETWEEN

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = (25, 90)
q = "SELECT * FROM student WHERE mark NOT BETWEEN ? AND ?"
try:
    my_cursor = my_conn.execute(q, my_data)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error: ", my_error)
finally:
    my_conn.close()
```

## 5. Modern SQLAlchemy 2.0+ Context Manager with BETWEEN

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

my_conn = create_engine("sqlite:///my_db.db")

try:
    with my_conn.connect() as connection:
        q = text("SELECT * FROM student WHERE mark BETWEEN :low AND :high")
        r_set = connection.execute(q, {"low": 60, "high": 75})
        r_list = r_set.fetchall()
        for row in r_list:
            print(row)
        print("Total Number of rows : ", len(r_list))
except SQLAlchemyError as e:
    print("SQLAlchemy error:", e)
```

## 6. Filtering Date Ranges (`dot` Column)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = ('2020-05-01', '2020-07-20')
q = "SELECT * FROM student WHERE dot BETWEEN ? AND ?"
try:
    my_cursor = my_conn.execute(q, my_data)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error: ", my_error)
finally:
    my_conn.close()
```
