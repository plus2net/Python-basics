## 2. Generate PDF Table from DataFrame (Dictionary Source)

Construct a Pandas DataFrame from a Python dictionary, convert values to a list, prepend column headers, and build a ReportLab Platypus table.

```python
from pathlib import Path
import pandas as pd
from reportlab.lib.units import inch
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate
from reportlab.platypus.tables import Table, TableStyle, colors

# Create dictionary source
my_dict = {
    'NAME': ['Ravi', 'Raju', 'Alex', 'Ron', 'King', 'Jack'],
    'ID': [1, 2, 3, 4, 5, 6],
    'MATH': [80, 40, 70, 70, 60, 30],
    'ENGLISH': [80, 70, 40, 50, 60, 30]
}
df = pd.DataFrame(data=my_dict)
my_data = df.values.tolist()
my_data.insert(0, list(df.columns))

output_dir = Path('output')
output_dir.mkdir(exist_ok=True)
pdf_path = output_dir / 'dataframe_table.pdf'

my_doc = SimpleDocTemplate(str(pdf_path), pagesize=letter)
c_width = [1.5 * inch, 1.0 * inch, 1.2 * inch, 1.2 * inch]

t = Table(my_data, colWidths=c_width, repeatRows=1)
t.setStyle(TableStyle([
    ('FONTSIZE', (0, 0), (-1, -1), 12),
    ('BACKGROUND', (0, 0), (-1, 0), colors.lightgreen),
    ('VALIGN', (0, 0), (-1, 0), 'TOP'),
    ('GRID', (0, 0), (-1, -1), 0.5, colors.grey)
]))

elements = [t]
my_doc.build(elements)
print(f"PDF table generated successfully: {pdf_path}")
```

## 3. Generate PDF Table from SQLite Database via SQLAlchemy

Query an SQLite database table using Pandas `read_sql` and render the result as a ReportLab PDF table.

```python
from sqlalchemy import create_engine
from sqlalchemy.exc import SQLAlchemyError

db_file = 'data/my_db.db'
file1 = 'sqlite:///' + db_file
my_conn = create_engine(file1)

try:
    # Create a sample table if it doesn't exist
    df_sample = pd.DataFrame({
        'student_id': [101, 102, 103],
        'student_name': ['Alice', 'Bob', 'Charlie'],
        'score': [92, 85, 88]
    })
    df_sample.to_sql('student', my_conn, if_exists='replace', index=False)
    
    df = pd.read_sql("SELECT * FROM student", my_conn)
    my_data_db = df.values.tolist()
    my_data_db.insert(0, list(df.columns))
    
    pdf_path_db = output_dir / 'sqlite_dataframe_table.pdf'
    my_doc_db = SimpleDocTemplate(str(pdf_path_db), pagesize=letter)
    t_db = Table(my_data_db, repeatRows=1)
    t_db.setStyle(TableStyle([
        ('FONTSIZE', (0, 0), (-1, -1), 11),
        ('BACKGROUND', (0, 0), (-1, 0), colors.lightblue),
        ('GRID', (0, 0), (-1, -1), 0.5, colors.grey)
    ]))
    my_doc_db.build([t_db])
    print(f"SQLite DataFrame PDF table generated: {pdf_path_db}")
except SQLAlchemyError as e:
    print("Database Error:", e)
```
