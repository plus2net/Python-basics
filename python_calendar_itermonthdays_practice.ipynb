### 1. Basic itermonthdays() Iteration
`itermonthdays(year, month)` yields integers representing the days of the month. Days outside the month (used as padding to complete full 7-day rows) are yielded as `0`.

```python
import calendar

# Default Monday-first calendar
cal = calendar.Calendar()
days = list(cal.itermonthdays(2026, 10))

print("Total days yielded (full weeks):", len(days))
print("All values:", days)
```

### 2. Changing the First Day of the Week (firstweekday)
Configure whether weeks start on Monday (`calendar.MONDAY` = 0) or Sunday (`calendar.SUNDAY` = 6).

```python
cal_sunday = calendar.Calendar(firstweekday=calendar.SUNDAY)
days_sunday = list(cal_sunday.itermonthdays(2026, 10))

print("Sunday-first first week:", days_sunday[:7])
print("Sunday-first last week: ", days_sunday[-7:])
```

### 3. Chunking Days into Weekly Rows
Divide the flat list of days into 7-day chunks to print a formatted console calendar.

```python
cal = calendar.Calendar(firstweekday=calendar.SUNDAY)
days = list(cal.itermonthdays(2026, 10))

weeks = [days[i:i + 7] for i in range(0, len(days), 7)]

print("Su Mo Tu We Th Fr Sa")
for week in weeks:
    print(" ".join(f"{d:2d}" if d > 0 else "  " for d in week))
```

### 4. Comparing Sibling Iterators (itermonthdays2 & itermonthdates)
Explore alternative representations like `(day, weekday)` tuples and `datetime.date` objects.

```python
cal = calendar.Calendar()

# itermonthdays2: yields (day, weekday)
sample_days2 = list(cal.itermonthdays2(2026, 10))[:4]
print("itermonthdays2 sample:", sample_days2)

# itermonthdates: yields actual datetime.date instances
sample_dates = list(cal.itermonthdates(2026, 10))[:4]
print("itermonthdates sample:", sample_dates)
```

### 5. Hands-on Practice Exercise: Calendar Matrix Generator
Run the self-checking unit assertions below to verify that calendar matrices correctly align days and pad external dates with 0.

```python
def generate_calendar_matrix(year, month, sunday_start=True):
    start_weekday = calendar.SUNDAY if sunday_start else calendar.MONDAY
    c = calendar.Calendar(firstweekday=start_weekday)
    
    days_list = list(c.itermonthdays(year, month))
    return [days_list[i:i + 7] for i in range(0, len(days_list), 7)]

# Test October 2026
matrix = generate_calendar_matrix(2026, 10, sunday_start=True)

# Unit Assertions
assert len(matrix) == 5
assert len(matrix[0]) == 7
assert matrix[0] == [0, 0, 0, 0, 1, 2, 3]  # Oct 1 is Thursday
assert matrix[-1][0] == 25
assert matrix[-1][6] == 31

print("Generated weeks count:", len(matrix))
print("All calendar matrix assertions passed successfully!")
```
