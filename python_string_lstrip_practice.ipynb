## 1. Leading Whitespace Trimming & Right Preservation

`str.lstrip()` removes all leading spaces, tabs, and newlines from the left edge of a string, leaving the right edge untouched.

```python
text = "   \t\n  Hello World!     \n"
clean_left = text.lstrip()

print("Original:   ", repr(text))
print("clean_left: ", repr(clean_left))

assert clean_left == "Hello World!     \n"
assert "".lstrip() == ""

print("\nAll leading whitespace assertions passed!")
```

## 2. Custom Leading Character Sets

Passing a string to `chars` strips any character in that set from the left until an unmatched character is encountered.

```python
entry = "###>>>--- System Status: OK ---<<<###"
cleaned = entry.lstrip("#>- ")

print("Cleaned:", repr(cleaned))
assert cleaned == "System Status: OK ---<<<###"

# Order of characters in set does not matter
assert entry.lstrip("-># ") == "System Status: OK ---<<<###"

print("\nCharacter set assertions passed!")
```

## 3. The Prefix Removal Bug: `lstrip()` vs. `removeprefix()` (PEP 616)

Calling `url.lstrip("https://")` on `"https://help.com"` strips `'h'` from `'help'`, producing `'elp.com'`.
Always use `removeprefix()` for exact prefix substrings.

```python
url = "https://help.com"

# Flawed: lstrip() treats 'https://' as a character set
flawed = url.lstrip("https://")
print("lstrip('https://'):       ", repr(flawed))  # 'elp.com'
assert flawed == "elp.com"

# Correct: removeprefix() matches literal prefix
correct = url.removeprefix("https://")
print("removeprefix('https://'): ", repr(correct)) # 'help.com'
assert correct == "help.com"

print("\nPrefix trap contrast passed!")
```

## 4. Measuring Indentation Width with `lstrip()`

A clean Python idiom to measure leading whitespace depth: `len(line) - len(line.lstrip(' '))`.

```python
code_line = "        def calculate_total(items):"
indent_spaces = len(code_line) - len(code_line.lstrip(" "))

print("Indentation width:", indent_spaces, "spaces")
assert indent_spaces == 8

print("Indentation measurement assertion passed!")
```

## 5. Real-World Engineering Recipes

Explore three common software engineering patterns utilizing `str.lstrip()`:
1. Unpadding Leading Numeric Zeros Safely
2. Markdown List Bullet Marker Cleaner
3. Log Comment / Hash Prefix Remover

```python
# Recipe 1: Unpad zeros
def unpad_zeros(val: str) -> str:
    return val.lstrip("0") or "0"

assert unpad_zeros("000452") == "452"
assert unpad_zeros("0000") == "0"
assert unpad_zeros("102") == "102"

# Recipe 2: Markdown bullet cleaner
def clean_bullet(line: str) -> str:
    return line.lstrip(" \t-*+1234567890.")

assert clean_bullet("  * First item") == "First item"
assert clean_bullet("  1. Numbered item") == "Numbered item"

# Recipe 3: Comment prefix remover
def clean_comment(line: str) -> str:
    return line.lstrip("#; \t")

assert clean_comment("### Config setting") == "Config setting"
assert clean_comment("; Timeout = 30") == "Timeout = 30"

print("All engineering recipes passed!")
```

## 6. Production Implementation & Self-Checking Unit Assertions

This production-grade `LeadingTrimmer` class encapsulates indentation calculation, zero unpadding, and bullet marker stripping.

```python
class LeadingTrimmer:
    @staticmethod
    def measure_indent(line: str) -> int:
        """Calculates the number of leading spaces."""
        return len(line) - len(line.lstrip(" "))

    @staticmethod
    def unpad_number(num_str: str) -> str:
        """Safely strips leading zeros, preserving single zero."""
        return num_str.lstrip("0") or "0"

    @staticmethod
    def strip_bullet(line: str) -> str:
        """Strips markdown bullet points from the left."""
        return line.lstrip(" \t-*+")


# ==========================================
# UNIT ASSERTION SUITE
# ==========================================
trimmer = LeadingTrimmer()

# 1. Leading whitespace
sample = "   \t\nHello World!  "
assert sample.lstrip() == "Hello World!  "
assert "".lstrip() == ""

# 2. Indentation
code = "        return True"
assert trimmer.measure_indent(code) == 8

# 3. Unpadding
assert trimmer.unpad_number("000452") == "452"
assert trimmer.unpad_number("0000") == "0"
assert trimmer.unpad_number("789") == "789"

# 4. Bullet stripping
assert trimmer.strip_bullet("  * Section 1") == "Section 1"
assert trimmer.strip_bullet("--- Header") == "Header"

# 5. URL prefix trap
bad_url = "https://help.com"
assert bad_url.lstrip("https://") == "elp.com"
assert bad_url.removeprefix("https://") == "help.com"

# 6. TypeError validation
try:
    "test".lstrip(99)
    assert False, "Expected TypeError not raised"
except TypeError as e:
    assert "strip arg must be None or str" in str(e)

print("All str.lstrip() production unit assertions passed successfully!")
```
