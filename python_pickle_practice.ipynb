### 1. File-Based Object Serialization (`dump` and `load`)

```python
import pickle

records = {
    'names': ['Ravi', 'Raju', 'Alex'],
    'ids': [101, 102, 103],
    'scores': {'math': [88, 92, 79]}
}

# Save to file in binary mode
with open('student_cache.pkl', 'wb') as f:
    pickle.dump(records, f, protocol=pickle.HIGHEST_PROTOCOL)
print("Pickled to student_cache.pkl")

# Load from file
with open('student_cache.pkl', 'rb') as f:
    restored = pickle.load(f)

print("Restored:", restored)
print("Match original?", restored == records)
```

### 2. In-Memory Byte Serialization (`dumps` and `loads`)

```python
import pickle

session = {
    'user_id': 9481,
    'roles': {'admin', 'editor'},
    'active': True
}

byte_stream = pickle.dumps(session)
print("Byte stream size:", len(byte_stream), "bytes")

unpickled = pickle.loads(byte_stream)
print("Unpickled user ID:", unpickled['user_id'])
print("Unpickled roles:  ", unpickled['roles'])
```

### 3. Restricted Safe Unpickler

```python
import pickle
import io

class SafeUnpickler(pickle.Unpickler):
    SAFE_TYPES = {'dict', 'list', 'set', 'tuple', 'int', 'float', 'str', 'bool'}
    def find_class(self, module, name):
        if module == 'builtins' and name in self.SAFE_TYPES:
            return getattr(__builtins__, name) if isinstance(__builtins__, dict) else getattr(__builtins__, name)
        raise pickle.UnpicklingError(f"Blocked unauthorized object: {module}.{name}")

safe_data = pickle.dumps({'status': 'ok', 'code': 200})
print("Safe loads:", SafeUnpickler(io.BytesIO(safe_data)).load())
```

### 4. Self-Checking Unit Verification Suite

```python
import pickle

# Test 1: Round-trip equality
sample = {"nums": [1, 2, 3], "set": {10, 20}}
b = pickle.dumps(sample)
assert pickle.loads(b) == sample

# Test 2: Set type preservation
assert isinstance(pickle.loads(b)['set'], set)

# Test 3: Highest protocol
b_high = pickle.dumps([42], protocol=pickle.HIGHEST_PROTOCOL)
assert pickle.loads(b_high) == [42]

print("All pickle unit assertions successfully verified! 100% passed.")
```
