## 2. Searching Tags and Extracting Links

Extract headers, anchor texts, and `href` targets from parsed HTML content.

```python
import re
from bs4 import BeautifulSoup

html_doc = '''<div>
  <h1>Main Heading</h1>
  <h2>Sub Heading 1</h2>
  <h2>Sub Heading 2</h2>
  <a href="https://www.plus2net.com">Home</a>
  <a href="https://www.plus2net.com/python/">Python Tutorials</a>
</div>'''

soup = BeautifulSoup(html_doc, 'html.parser')

print("Headers:", [h.get_text() for h in soup.find_all(re.compile('^(h1|h2)$'))])
print("Links:", [a['href'] for a in soup.find_all('a', href=True)])
```
