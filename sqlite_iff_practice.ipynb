## 2. Basic IIF Query for Pass/Fail Evaluation

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT id, name, mark, IIF(mark >= 60, 'PASS', 'FAIL') as Result FROM student LIMIT 0, 10"

try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 3. Complex IIF with SUM and Aggregate Grouping

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = """SELECT id, name, SUM(social + math + science) as Total,
IIF(SUM(social + math + science) > 200, 'Pass', 'Fail') as Result
FROM student3 GROUP BY id LIMIT 0, 10"""

try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 4. Grid View Pivoting with SUM(IIF(...))

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = """SELECT
SUM(IIF(class='Three', 1, 0)) as THREE,
SUM(IIF(class='Four', 1, 0)) as FOUR,
SUM(IIF(class='Five', 1, 0)) as FIVE,
SUM(IIF(class='Six', 1, 0)) as SIX,
SUM(IIF(class='Seven', 1, 0)) as SEVEN,
SUM(IIF(class='Eight', 1, 0)) as EIGHT,
SUM(IIF(class='Nine', 1, 0)) as NINE
FROM student"""

try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 5. Modern SQLAlchemy 2.0+ Context Manager with IIF Query

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

my_conn = create_engine("sqlite:///my_db.db")
q = text("SELECT class, SUM(IIF(mark < 50, 1, 0)) AS grade_C, SUM(IIF(mark BETWEEN 50 AND 69, 1, 0)) AS grade_B, SUM(IIF(mark >= 70, 1, 0)) AS grade_A FROM student GROUP BY class")

try:
    with my_conn.connect() as connection:
        r_set = connection.execute(q)
        for row in r_set:
            print(row)
except SQLAlchemyError as e:
    print("SQLAlchemy error:", e)
```
