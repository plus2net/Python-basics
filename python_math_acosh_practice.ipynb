### 1. Basic Inverse Hyperbolic Cosine Calculations
`math.acosh(x)` returns the inverse hyperbolic cosine of `x`. Because `cosh(0) == 1.0`, `acosh(1.0)` is exactly `0.0`. For any value `x >= 1.0`, it returns a non-negative float in `[0.0, +inf)`.


```python
import math

print("acosh(1.0): ", math.acosh(1.0))
print("acosh(1.5): ", round(math.acosh(1.5), 6))
print("acosh(2.0): ", round(math.acosh(2.0), 6))
print("acosh(5.0): ", round(math.acosh(5.0), 6))
print("acosh(10):  ", round(math.acosh(10), 6))
```

### 2. Logarithmic Definition & Equivalence
Mathematically, inverse hyperbolic cosine is defined as:
`acosh(x) == ln(x + sqrt(x^2 - 1))` for `x >= 1.0`

We can verify this equivalence numerically using `math.isclose()`.


```python
import math

def manual_acosh(x):
    return math.log(x + math.sqrt(x**2 - 1))

sample_inputs = [1.0, 1.25, 2.0, 7.5, 25.0, 100.0]
print(f"{'x':>7} | {'math.acosh(x)':>15} | {'manual_acosh(x)':>16} | Match?")
print("-" * 52)
for val in sample_inputs:
    built_in = math.acosh(val)
    formula = manual_acosh(val)
    matches = math.isclose(built_in, formula, rel_tol=1e-12)
    print(f"{val:7.2f} | {built_in:15.8f} | {formula:16.8f} | {matches}")
```

### 3. Domain Boundary & ValueError Handling (x >= 1.0)
Because the forward hyperbolic cosine function never falls below 1.0 for real inputs, passing any value `x < 1.0` raises `ValueError: math domain error`.


```python
import math

invalid_inputs = [0.99, 0.5, 0.0, -2.0]
for val in invalid_inputs:
    try:
        math.acosh(val)
    except ValueError as err:
        print(f"Caught expected domain error for {val:5.2f}: {err}")
```

### 4. Real vs. Complex: math.acosh vs. cmath.acosh
While `math.acosh` is restricted to real numbers >= 1.0, Python's `cmath` module evaluates inverse hyperbolic cosine across the complex plane.
For `x = 0.5`, `cmath.acosh(0.5)` produces `0 + 1.04719755j` (which is `1j * pi / 3`).


```python
import math
import cmath

val = 0.5
try:
    math.acosh(val)
except ValueError as e:
    print("math.acosh(0.5) raises:", e)

c_result = cmath.acosh(val)
print("cmath.acosh(0.5) returns:", c_result)
print("Imaginary part equals pi / 3?", math.isclose(c_result.imag, math.pi / 3))
```

### 5. Hands-on Practice Exercise: Catenary Hanging Cable Solver
A hanging power line follows a catenary curve: `y = a * cosh(x / a)`.
Solving for horizontal distance `x` from the vertex yields: `x = a * acosh(y / a)`.

Implement `catenary_horizontal_distance(cable_height, catenary_param_a)` and verify with self-checking assertions.


```python
import math

def catenary_horizontal_distance(cable_height, catenary_param_a):
    """
    Calculates horizontal distance x from vertex of a catenary cable:
    x = a * acosh(y / a)
    """
    if catenary_param_a <= 0:
        raise ValueError("Catenary parameter 'a' must be strictly positive.")
    ratio = cable_height / catenary_param_a
    if ratio < 1.0:
        raise ValueError("Cable height cannot be less than catenary vertex parameter 'a'.")
    return round(catenary_param_a * math.acosh(ratio), 4)

# Unit Assertions
assert catenary_horizontal_distance(50.0, 50.0) == 0.0  # Vertex
assert catenary_horizontal_distance(75.0, 50.0) == 48.1212
assert catenary_horizontal_distance(100.0, 50.0) == 65.8479

# Test boundary error
try:
    catenary_horizontal_distance(40.0, 50.0)
    assert False, "Should have raised ValueError"
except ValueError:
    pass

print("Catenary calculation (h=85m, a=60m):", catenary_horizontal_distance(85.0, 60.0), "meters")
print("All unit assertions passed successfully!")
```
