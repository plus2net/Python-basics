Run cells in order in Google Colab. This notebook demonstrates how to convert XML files to CSV using both Pandas (`read_xml`) and Python's standard `xml.etree.ElementTree` and `csv` modules. Save a copy in Google Drive to keep your edits.

### 1. Convert XML to CSV using Pandas

```python
import pandas as pd

# Create sample XML file
xml_data = """<data>
    <record>
        <id>1</id>
        <name>John Doe</name>
        <age>30</age>
        <city>New York</city>
    </record>
    <record>
        <id>2</id>
        <name>Jane Smith</name>
        <age>25</age>
        <city>Los Angeles</city>
    </record>
</data>"""

with open("data.xml", "w") as f:
    f.write(xml_data)

# Read XML and export to CSV
df = pd.read_xml("data.xml")
df.to_csv("data.csv", index=False)
print("CSV file generated successfully using Pandas:")
print(df)
```

### 2. Convert XML to CSV using ElementTree and CSV module

```python
import xml.etree.ElementTree as ET
import csv

tree = ET.parse("data.xml")
root = tree.getroot()

rows = []
headers = []

for record in root.findall('record'):
    row = {}
    for element in record:
        if element.tag not in headers:
            headers.append(element.tag)
        row[element.tag] = element.text
    rows.append(row)

with open("data_et.csv", "w", newline="", encoding="utf-8") as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames=headers)
    writer.writeheader()
    writer.writerows(rows)

print("CSV file generated successfully using ElementTree!")
```
