## 2. Register a TrueType Font with TTFont

Create a `TTFont` object and register it with `pdfmetrics`.

```python
from pathlib import Path
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont

font_path = Path('fonts/DejaVuSans.ttf')
pdfmetrics.registerFont(TTFont('MyFont', str(font_path)))
print("Registered TTFont 'MyFont' successfully!")
```

```python
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import letter

c = canvas.Canvas('custom-font.pdf', pagesize=letter)
c.setFont('MyFont', 18)
c.drawString(72, 720, 'Welcome to plus2net')
c.setFont('MyFont', 12)
c.drawString(72, 690, '12 point custom font')
c.setFont('MyFont', 24)
c.drawString(72, 650, '24 point custom font')
c.save()
print("Canvas custom font PDF generated!")
```

## 3. Use a Custom Font with ReportLab Paragraph

Connect custom registered font to Platypus `ParagraphStyle`.

```python
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import ParagraphStyle
from reportlab.platypus import SimpleDocTemplate, Paragraph

style = ParagraphStyle(
    'CustomFont',
    fontName='MyFont',
    fontSize=12,
    leading=16,
    backColor='#F1F1F1',
    borderColor='#CCCCCC',
    borderWidth=1,
    borderPadding=10
)

doc = SimpleDocTemplate('paragraph-font.pdf', pagesize=A4)
story = [
    Paragraph('This paragraph uses a custom TrueType font registered with ReportLab.', style)
]
doc.build(story)
print("Paragraph custom font PDF generated!")
```

## 4. Register Regular, Bold and Italic Fonts as a Family

Register variants and map them as a family using `registerFontFamily`.

```python
pdfmetrics.registerFont(TTFont('MyFont-Regular', 'fonts/DejaVuSans.ttf'))
pdfmetrics.registerFont(TTFont('MyFont-Bold', 'fonts/DejaVuSans-Bold.ttf'))

pdfmetrics.registerFontFamily(
    'MyFamily',
    normal='MyFont-Regular',
    bold='MyFont-Bold'
)

family_style = ParagraphStyle('FamilyStyle', fontName='MyFamily', fontSize=14, leading=18)
doc = SimpleDocTemplate('family-font.pdf', pagesize=A4)
story = [
    Paragraph('Regular style text.', family_style),
    Paragraph('<b>Bold style text using font family!</b>', family_style)
]
doc.build(story)
print("Font family PDF generated!")
```

## 5. Handle a Missing Font File Safely

Safe registration wrapper function checking file existence.

```python
def register_font_safely(name, filename):
    font_path = Path(filename)
    if not font_path.is_file():
        raise FileNotFoundError(f'Font file not found: {font_path}')
    pdfmetrics.registerFont(TTFont(name, str(font_path)))
    print(f"Font '{name}' safely registered!")

register_font_safely('SafeFont', 'fonts/DejaVuSans.ttf')
```
