## 1. Set Definition and Duplicate Removal

A Python set is a collection of **unique** values. If the same value appears more than once when a set is created, only one copy is kept.

Sets are also unordered, so we should not rely on the position in which items are displayed.


````python
numbers = [1, 2, 2, 3, 4, 4, 5]
unique_numbers = set(numbers)

print("Original list:", numbers)
print("Unique values:", sorted(unique_numbers))
````

**Expected output**

```text
Original list: [1, 2, 2, 3, 4, 4, 5]
Unique values: [1, 2, 3, 4, 5]
```

The list contains repeated values, but the set keeps only one copy of each value.


````python
names = {"Alex", "Ronald", "Alex", "John", "Ronald"}

print("Number of unique names:", len(names))
print("Unique names:", sorted(names))
````

**Expected output**

```text
Number of unique names: 3
Unique names: ['Alex', 'John', 'Ronald']
```

Using `sorted()` here is only for a predictable display. The set itself is still unordered.


## 2. Set Construction and Empty Sets

A set can be created with curly braces when values are available.

```python
my_set = {value1, value2, value3}
```

The `set()` constructor can also create a set from an iterable.


````python
colors = {"red", "green", "blue"}
numbers = set([10, 20, 30, 20])

print(sorted(colors))
print(sorted(numbers))
````

**Expected output**

```text
['blue', 'green', 'red']
[10, 20, 30]
```

### Creating an Empty Set

Use `set()` for an empty set. Empty curly braces `{}` create an empty dictionary, not a set.


````python
empty_set = set()
empty_braces = {}

print(type(empty_set))
print(type(empty_braces))
print(empty_set)
````

**Expected output**

```text
<class 'set'>
<class 'dict'>
set()
```

## 3. List, Tuple and String Conversion

The `set()` constructor can accept iterable objects such as lists, tuples, and strings.

When a list or tuple is converted, duplicate elements are removed. When a string is converted, the set contains its unique characters.


````python
my_list = ["a", "b", "c", "a", "b"]
my_tuple = (5, 2, 5, 8, 2)
my_string = "plus2net"

set_from_list = set(my_list)
set_from_tuple = set(my_tuple)
set_from_string = set(my_string)

print("From list:", sorted(set_from_list))
print("From tuple:", sorted(set_from_tuple))
print("From string:", sorted(set_from_string))
````

**Expected output**

```text
From list: ['a', 'b', 'c']
From tuple: [2, 5, 8]
From string: ['2', 'e', 'l', 'n', 'p', 's', 't', 'u']
```

A string is iterable one character at a time, so repeated characters are removed when it is converted to a set.


## 4. Looping and Membership Checking

A `for` loop can read every item in a set. Because sets are unordered, the loop order should not be treated as fixed.

The `in` operator checks whether a value is present. `not in` checks whether it is absent.


````python
languages = {"Python", "PHP", "Java"}

for language in sorted(languages):
    print(language)
````

**Expected output**

```text
Java
PHP
Python
```

We used `sorted()` only to make the demonstration output consistent.


````python
languages = {"Python", "PHP", "Java"}

print("Python" in languages)
print("Ruby" in languages)
print("Ruby" not in languages)
````

**Expected output**

```text
True
False
True
```

Membership checking is useful when you need to avoid duplicate work or test whether an item has already been seen.


## 5. `add()` and `update()`

Use `add()` to add **one element** to a set.

Use `update()` to add **multiple elements** from an iterable such as another set, a list, a tuple, or a string.


````python
students = {"Alex", "Ronald", "John"}

students.add("Lewis")
students.add("Alex")  # Duplicate: the set does not add a second copy.

print(sorted(students))
````

**Expected output**

```text
['Alex', 'John', 'Lewis', 'Ronald']
```

Adding a value that already exists does not create a duplicate.


````python
letters = {"a", "b", "c"}

letters.update(["c", "d", "e"])
letters.update(("e", "f"))
letters.update("xy")

print(sorted(letters))
````

**Expected output**

```text
['a', 'b', 'c', 'd', 'e', 'f', 'x', 'y']
```

`update()` reads the incoming iterable and adds its elements individually.


## 6. `remove()`, `discard()`, `pop()` and `clear()`

Python provides several ways to remove set items:

- `remove(x)` removes `x`, but raises `KeyError` if `x` is missing.
- `discard(x)` removes `x` if present and does not raise an error if it is missing.
- `pop()` removes and returns an arbitrary item.
- `clear()` removes every item but keeps the set object.


````python
items = {"a", "b", "c", "x"}

items.remove("x")
print("After remove:", sorted(items))

items.discard("y")  # No error even though "y" is missing.
print("After discard of missing item:", sorted(items))
````

**Expected output**

```text
After remove: ['a', 'b', 'c']
After discard of missing item: ['a', 'b', 'c']
```

The important difference is what happens when the requested item does not exist.


````python
items = {"a", "b", "c"}

try:
    items.remove("z")
except KeyError:
    print("remove() raised KeyError because 'z' was not present")

items.discard("z")
print("discard() completed without an error")
````

**Expected output**

```text
remove() raised KeyError because 'z' was not present
discard() completed without an error
```


````python
numbers = {10, 20, 30}
removed_item = numbers.pop()

print("Removed item:", removed_item)
print("Items left:", numbers)
````

**Sample output**

```text
Removed item: 10
Items left: {20, 30}
```

Your result may differ. `pop()` removes an arbitrary set element because sets do not have a fixed positional order.


````python
numbers = {10, 20, 30}
numbers.clear()

print(numbers)
print(type(numbers))
````

**Expected output**

```text
set()
<class 'set'>
```

`clear()` empties the set, but the variable still refers to a set object.


## 7. `len()`, `min()`, `max()` and `sum()`

Built-in functions can work directly with a numeric set:

- `len()` returns the number of elements.
- `min()` returns the smallest value.
- `max()` returns the largest value.
- `sum()` returns the total of numeric values.


````python
marks = {72, 88, 91, 65, 88}

print("Unique count:", len(marks))
print("Minimum:", min(marks))
print("Maximum:", max(marks))
print("Total:", sum(marks))
````

**Expected output**

```text
Unique count: 4
Minimum: 65
Maximum: 91
Total: 316
```

The repeated value `88` is stored only once, so it contributes once to the length and the sum.


## 8. Set Copying and Comprehension

The `copy()` method creates a separate shallow copy of a set. Changing the copied set does not change the original set.

A **set comprehension** creates a set using a compact expression inside curly braces.

```python
{expression for item in iterable if condition}
```


````python
original = {1, 2, 3}
copied = original.copy()

copied.add(4)

print("Original:", sorted(original))
print("Copy:", sorted(copied))
````

**Expected output**

```text
Original: [1, 2, 3]
Copy: [1, 2, 3, 4]
```


````python
even_squares = {number ** 2 for number in range(1, 11) if number % 2 == 0}

print(sorted(even_squares))
````

**Expected output**

```text
[4, 16, 36, 64, 100]
```

A set comprehension automatically keeps only unique results.


````python
words = ["python", "php", "python", "java", "php"]
lengths = {len(word) for word in words}

print(sorted(lengths))
````

**Expected output**

```text
[3, 4, 6]
```

Different words can produce the same length, but a set stores each resulting length once.


## 9. Valid Set Element Types

Set elements must be **hashable**. Common immutable values such as integers, floats, strings, tuples containing hashable items, and `frozenset` objects can be set elements.

Mutable objects such as lists, dictionaries, and ordinary sets cannot be elements of another set.


````python
valid_items = {
    25,
    3.5,
    "Python",
    (1, 2),
    frozenset({3, 4})
}

print("Number of valid elements:", len(valid_items))

for item in valid_items:
    print(type(item).__name__, "->", item)
````

**Sample output**

```text
Number of valid elements: 5
str -> Python
float -> 3.5
int -> 25
tuple -> (1, 2)
frozenset -> frozenset({3, 4})
```

The display order may differ because the set is unordered.


````python
my_set = {"Python"}

try:
    my_set.add([1, 2])
except TypeError as error:
    print("List cannot be a set element:", error)
````

**Expected output**

```text
List cannot be a set element: unhashable type: 'list'
```

A tuple can be used only when all of its own elements are hashable.


## 10. Unique Values from Database-Style Rows

Database query results are often represented in Python as rows stored in tuples. Sets are useful when one column contains repeated values and only the unique values are needed.


````python
rows = [
    (1, "Alex", "Python"),
    (2, "Ravi", "PHP"),
    (3, "Maya", "Python"),
    (4, "John", "Java"),
    (5, "Sara", "PHP")
]

unique_courses = {row[2] for row in rows}

print(sorted(unique_courses))
````

**Expected output**

```text
['Java', 'PHP', 'Python']
```

The set comprehension reads the course column (`row[2]`) and keeps each course only once.


````python
rows = [
    ("Mumbai", "Python"),
    ("Delhi", "PHP"),
    ("Mumbai", "Java"),
    ("Chennai", "Python"),
    ("Delhi", "Python")
]

unique_cities = {row[0] for row in rows}

print("Unique cities:", sorted(unique_cities))
print("Number of cities:", len(unique_cities))
````

**Expected output**

```text
Unique cities: ['Chennai', 'Delhi', 'Mumbai']
Number of cities: 3
```

This pattern is useful after fetching database rows when you need distinct values in Python.

## 12. Mini Project: Unique Website Visitor Tracker

A website can record the same visitor more than once during a day. A set is a simple way to keep one copy of each visitor ID and calculate the number of unique visitors.

This mini-project also identifies whether each visit comes from a new or returning visitor.


````python
visits = [
    "user101",
    "user205",
    "user101",
    "user309",
    "user205",
    "user410",
    "user309",
    "user512"
]

unique_visitors = set(visits)

print("Total visits:", len(visits))
print("Unique visitors:", len(unique_visitors))
print("Visitor IDs:", sorted(unique_visitors))
````

**Expected output**

```text
Total visits: 8
Unique visitors: 5
Visitor IDs: ['user101', 'user205', 'user309', 'user410', 'user512']
```

The list records every visit, while the set gives the unique visitor count.


````python
visits = [
    "user101",
    "user205",
    "user101",
    "user309",
    "user205"
]

seen_visitors = set()

for visitor in visits:
    if visitor in seen_visitors:
        print(visitor, "-> returning visitor")
    else:
        print(visitor, "-> new visitor")
        seen_visitors.add(visitor)

print("Unique visitor count:", len(seen_visitors))
````

**Expected output**

```text
user101 -> new visitor
user205 -> new visitor
user101 -> returning visitor
user309 -> new visitor
user205 -> returning visitor
Unique visitor count: 3
```

This shows a practical reason to combine membership checking with `add()`: the set remembers which visitor IDs have already appeared.
