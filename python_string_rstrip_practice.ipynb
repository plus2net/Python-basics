## 1. Trailing Whitespace Trimming

`str.rstrip()` strips spaces, tabs, and newlines from the right edge, while leaving leading indentation completely untouched.

```python
raw_line = "    def compute(val):    \t\r\n"
clean_line = raw_line.rstrip()

print("Original:   ", repr(raw_line))
print("clean_line: ", repr(clean_line))

assert clean_line == "    def compute(val):"
assert "".rstrip() == ""

print("\nAll trailing whitespace assertions passed!")
```

## 2. Custom Trailing Character Sets

Passing a string to `chars` strips any character in that set from the right until a non-matching character is encountered.

```python
headline = "=== BREAKING NEWS !!!***..."
cleaned = headline.rstrip("!*. ")

print("Cleaned:", repr(cleaned))
assert cleaned == "=== BREAKING NEWS"

# Order of characters in set does not matter
assert headline.rstrip(".*! ") == "=== BREAKING NEWS"

print("\nCharacter set assertions passed!")
```

## 3. The Suffix Pitfall: `rstrip()` vs. `removesuffix()` (PEP 616)

Calling `filename.rstrip(".txt")` on `"test.txt"` strips the character set `{'.', 't', 'x'}`, leaving `'tes'`.
Similarly, `"banana".rstrip("an")` eats both `'a'` and `'n'`, leaving `'b'`.
Use `removesuffix()` for exact substring matching.

```python
filename = "test.txt"

# Flawed: rstrip() treats '.txt' as a character set
broken = filename.rstrip(".txt")
print("rstrip('.txt') result:      ", repr(broken))  # 'tes'
assert broken == "tes"

# Another classic trap: 'banana'
fruit = "banana".rstrip("an")
print("banana.rstrip('an'):       ", repr(fruit))   # 'b'
assert fruit == "b"

# Correct: removesuffix() matches literal suffix
correct = filename.removesuffix(".txt")
print("removesuffix('.txt') result:", repr(correct)) # 'test'
assert correct == "test"

print("\nSuffix trap contrast passed!")
```

## 4. Preserving Indentation When Processing Lines

While `strip()` destroys leading indentation spaces, `rstrip("\r\n")` cleans trailing line breaks while keeping block hierarchy intact.

```python
raw_lines = [
    "def calculate():\n",
    "    if ready:\n",
    "        return 100\n"
]

preserved = [ln.rstrip("\r\n") for ln in raw_lines]
print("Preserved lines:", preserved)

assert preserved[0] == "def calculate():"
assert preserved[1] == "    if ready:"
assert preserved[2] == "        return 100"

print("\nIndentation preservation verified!")
```

## 5. Real-World Engineering Recipes

Explore three common software engineering patterns utilizing `str.rstrip()`:
1. URL Trailing Slash Normalization
2. Safe Decimal Trailing Zeros Cleaner
3. Headline Punctuation Trimmer

```python
# Recipe 1: URL slash normalization
def normalize_endpoint(url: str, path: str) -> str:
    return f"{url.rstrip('/')}/{path.strip('/')}"

assert normalize_endpoint("https://api.plus2net.com///", "/users/") == "https://api.plus2net.com/users"

# Recipe 2: Safe decimal formatting
def clean_decimal(val_str: str) -> str:
    if "." in val_str:
        return val_str.rstrip("0").rstrip(".")
    return val_str

assert clean_decimal("145.250000") == "145.25"
assert clean_decimal("300.000000") == "300"
assert clean_decimal("1000") == "1000"  # Integer unharmed!

# Recipe 3: Headline punctuation trimmer
def clean_headline(title: str) -> str:
    return title.rstrip(".!?,:; ")

assert clean_headline("Breaking: New Record Hits All-Time High!!!") == "Breaking: New Record Hits All-Time High"
assert clean_headline("What is Python?....") == "What is Python"

print("All engineering recipes passed!")
```

## 6. Production Implementation & Self-Checking Unit Assertions

This production-grade `TrailingTrimmer` class encapsulates line cleaning, URL slash normalization, and safe decimal formatting.

```python
class TrailingTrimmer:
    @staticmethod
    def strip_trailing_newlines(line: str) -> str:
        """Removes trailing newline sequences preserving left indentation."""
        return line.rstrip("\r\n")

    @staticmethod
    def normalize_url_slash(url: str) -> str:
        """Removes trailing slashes from URLs."""
        return url.rstrip("/")

    @staticmethod
    def clean_decimal_zeros(val_str: str) -> str:
        """Safely strips trailing decimal zeros."""
        if "." in val_str:
            return val_str.rstrip("0").rstrip(".")
        return val_str


# ==========================================
# UNIT ASSERTION SUITE
# ==========================================
trimmer = TrailingTrimmer()

# 1. Trailing whitespace with indentation preservation
code = "    return True\t\r\n"
assert trimmer.strip_trailing_newlines(code) == "    return True\t"
assert code.rstrip() == "    return True"

# 2. URL slash normalization
assert trimmer.normalize_url_slash("https://plus2net.com///") == "https://plus2net.com"
assert trimmer.normalize_url_slash("https://plus2net.com") == "https://plus2net.com"

# 3. Decimal zeros cleaning
assert trimmer.clean_decimal_zeros("12.500") == "12.5"
assert trimmer.clean_decimal_zeros("40.000") == "40"
assert trimmer.clean_decimal_zeros("1000") == "1000"

# 4. Suffix trap verification
sample = "test.txt"
assert sample.rstrip(".txt") == "tes"
assert sample.removesuffix(".txt") == "test"

# 5. TypeError on non-string
try:
    "sample".rstrip(42)
    assert False, "Expected TypeError not raised"
except TypeError as e:
    assert "strip arg must be None or str" in str(e)

print("All str.rstrip() production unit assertions passed successfully!")
```
