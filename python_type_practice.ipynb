### 1. Basic Type Inspection

```python
print("42 is:        ", type(42))
print("3.14 is:      ", type(3.14))
print("'hello' is:   ", type('hello'))
print("[1, 2, 3] is: ", type([1, 2, 3]))
print("None is:      ", type(None))
```

### 2. type() vs. isinstance() Subclass Demonstration

```python
class Animal: pass
class Dog(Animal): pass

my_dog = Dog()
print("Exact type is Dog:       ", type(my_dog) is Dog)
print("Exact type is Animal:    ", type(my_dog) is Animal)
print("isinstance of Dog:       ", isinstance(my_dog, Dog))
print("isinstance of Animal:    ", isinstance(my_dog, Animal))
```

### 3. Dynamic Class Creation via 3-Argument type()

```python
# type(name, bases, attributes_dict)
def calculate_total(self):
    return self.qty * self.unit_price

Invoice = type("Invoice", (object,), {
    "currency": "USD",
    "total": calculate_total
})

inv = Invoice()
inv.qty = 5
inv.unit_price = 20

print("Class Name:   ", Invoice.__name__)
print("Total amount: ", inv.total(), inv.currency)
```

### 4. Dynamic Model Factory Recipe

```python
def make_record_type(name, field_names):
    def init(self, **kwargs):
        for f in field_names:
            setattr(self, f, kwargs.get(f))
    return type(name, (), {"__init__": init, "_fields": field_names})

Customer = make_record_type("Customer", ["cid", "email"])
cust = Customer(cid=1001, email="test@example.com")
print("Customer instance:", cust.cid, cust.email)
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Basic types
assert type(100) is int
assert type("str") is str

# Test 2: Metaclass properties
assert type(type) is type
assert type(object) is type
assert issubclass(type, object)

# Test 3: 3-argument class creation
DynamicClass = type("DynamicClass", (), {"num": 42})
assert DynamicClass.num == 42
assert type(DynamicClass()) is DynamicClass

# Test 4: Subclass hierarchy check
assert isinstance(True, int)
assert type(True) is bool

print("All type() unit assertions successfully verified! 100% passed.")
```
