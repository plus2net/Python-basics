### 1. Basic readable() Capability Check Across Modes
`stream.readable()` returns `True` for read modes and `False` for write-only/append-only modes.

```python
# Create demo file
with open("demo_readable.txt", "w", encoding="utf-8") as f:
    f.write("Sample payload for testing readability.\n")

# Check readability across standard modes
with open("demo_readable.txt", "r", encoding="utf-8") as f_r:
    print("Mode 'r' readable: ", f_r.readable())   # True

with open("demo_readable.txt", "a", encoding="utf-8") as f_a:
    print("Mode 'a' readable: ", f_a.readable())   # False

with open("demo_readable.txt", "r+", encoding="utf-8") as f_rw:
    print("Mode 'r+' readable:", f_rw.readable())  # True
```

### 2. Defensive Pre-Flight Guard to Avoid UnsupportedOperation
Prevent runtime crashes when accepting unknown file streams by inspecting `readable()` first.

```python
def safe_read_stream(stream):
    if stream.closed:
        raise ValueError("Stream is already closed.")
    if not stream.readable():
        print("Stream does not permit read operations.")
        return None
    return stream.read()

with open("demo_readable.txt", "r", encoding="utf-8") as f_in:
    print("Safe read result:", repr(safe_read_stream(f_in)))

with open("demo_readable.txt", "a", encoding="utf-8") as f_out:
    print("Append stream:   ", safe_read_stream(f_out))  # None
```

### 3. Polymorphic In-Memory Streams (io.StringIO and io.BytesIO)
In-memory streams implement the standard `readable()` protocol.

```python
import io

str_buf = io.StringIO("In-memory text stream")
print("io.StringIO readable:", str_buf.readable())  # True

bin_buf = io.BytesIO(b"\x01\x02\x03")
print("io.BytesIO readable: ", bin_buf.readable())  # True
```

### 4. Handling Closed Streams Safely
Calling `readable()` on closed streams raises `ValueError: I/O operation on closed file`.

```python
f = open("demo_readable.txt", "r", encoding="utf-8")
f.close()

try:
    f.readable()
except ValueError as err:
    print("Caught expected error on closed stream:", err)

# Safe idiom: verify stream.closed first
is_usable = not f.closed and f.readable()
print("Is stream safely usable?", is_usable)  # False
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying readability checks across modes.

```python
class ReadableVerifier:
    @staticmethod
    def is_usable_reader(stream) -> bool:
        try:
            return not stream.closed and stream.readable()
        except (ValueError, AttributeError):
            return False

v = ReadableVerifier()
with open("demo_readable.txt", "r", encoding="utf-8") as f:
    assert v.is_usable_reader(f) is True

with open("demo_readable.txt", "a", encoding="utf-8") as f:
    assert v.is_usable_reader(f) is False

closed_f = open("demo_readable.txt", "r", encoding="utf-8")
closed_f.close()
assert v.is_usable_reader(closed_f) is False

print("All file.readable() unit assertions passed successfully! 100% verified.")
```
