# Python math.asin(): Calculate Inverse Sine (Arc Sine)

### 1. Basic Arc Sine Calculations
`math.asin(x)` takes a numeric ratio in $[-1.0, 1.0]$ and returns the corresponding angle in radians $[-rac{\pi}{2}, rac{\pi}{2}]$.

```python
import math

print("asin(0):   ", math.asin(0))
print("asin(0.5): ", math.asin(0.5))
print("asin(1.0): ", math.asin(1.0))
print("asin(-1.0):", math.asin(-1.0))
```

### 2. Converting Output Radians to Degrees
Use `math.degrees()` to convert output angles from radians to degrees ($0^\circ$ to $90^\circ$).

```python
import math

ratios = [0.0, 0.5, 0.70710678, 0.8660254, 1.0]
for r in ratios:
    rad = math.asin(r)
    deg = math.degrees(rad)
    print(f"asin({r:4.2f}) -> {rad:6.4f} rad == {deg:5.1f} deg")
```

### 3. Handling Domain Errors Outside [-1.0, 1.0]
Passing values outside $[-1.0, 1.0]$ raises `ValueError: math domain error`.

```python
import math

for invalid_ratio in [-1.5, 1.05]:
    try:
        math.asin(invalid_ratio)
    except ValueError as err:
        print(f"Caught expected domain error for {invalid_ratio}: {err}")
```

### 4. Complementary Identity: asin(x) + acos(x) == pi / 2
The sum of the arc sine and arc cosine of any ratio is always $rac{\pi}{2}$ ($90^\circ$).

```python
import math

ratio = 0.65
sum_angles = math.asin(ratio) + math.acos(ratio)
print("asin(x) + acos(x):", sum_angles)
print("Equals pi / 2?   ", math.isclose(sum_angles, math.pi / 2))
```

### 5. Hands-on Practice Exercise: Ramp Incline Angle Calculator
Run the self-checking unit assertions below to verify ramp incline calculations: $	heta = rcsin(rac{\text{rise}}{\text{length}})$.

```python
def calculate_incline_angle(vertical_rise, ramp_length):
    if ramp_length <= 0:
        raise ValueError("Ramp length must be strictly positive.")
    if vertical_rise > ramp_length:
        raise ValueError("Rise cannot exceed ramp diagonal length.")
    ratio = vertical_rise / ramp_length
    clamped_ratio = max(-1.0, min(1.0, ratio))
    theta_rad = math.asin(clamped_ratio)
    theta_deg = math.degrees(theta_rad)
    return round(theta_deg, 2)

# Unit Assertions
assert calculate_incline_angle(1.0, 2.0) == 30.0
assert calculate_incline_angle(0.0, 5.0) == 0.0
assert calculate_incline_angle(0.75, 9.0) == 4.78
assert calculate_incline_angle(5.0, 5.0) == 90.0

print("Ramp (0.75m rise, 9.0m length):", calculate_incline_angle(0.75, 9.0), "degrees")
print("All incline angle assertions passed successfully!")
```
