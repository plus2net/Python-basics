## Example 1: First working example

Read the code, predict the output and run this cell.

```python
n = 6
result = 1
for i in range(1, n + 1):
    result *= i
print("Factorial:", result)
```

## Example 2: Original input-based factorial loop

Read the code, predict the output and run this cell.

```python
from unittest.mock import patch
with patch("builtins.input", side_effect=['6']):
    n = int(input("Enter a number: "))
    if n < 0:
        print("Factorial is undefined for negative integers")
    else:
        total = 1
        for i in range(1, n + 1):
            total *= i
        print("Factorial:", total)
```

## Example 3: Recursive factorial function

Read the code, predict the output and run this cell.

```python
def fact(n):
    if n < 0:
        raise ValueError("n must be nonnegative")
    if n <= 1:
        return 1
    return n * fact(n - 1)

print(fact(4))
print(fact(0))
```

## Example 4: Use math.factorial()

Read the code, predict the output and run this cell.

```python
from math import factorial
for value in (0, 1, 5, 10):
    print(value, factorial(value))
```

## Example 5: Compare loop and recursion

Read the code, predict the output and run this cell.

```python
from math import factorial
def factorial_loop(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
for n in range(7):
    print(n, factorial_loop(n) == factorial(n))
```

## Example 6: Factorial sequence

Read the code, predict the output and run this cell.

```python
from math import factorial
print([factorial(n) for n in range(6)])
```

## Example 7: Validate an invalid value

Read the code, predict the output and run this cell.

```python
from math import factorial
try:
    print(factorial(-3))
except ValueError:
    print("Use a nonnegative integer")
```
