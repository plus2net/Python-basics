## 2. Alternating Two-Color Concentric Circles

Draw concentric circles centered at $(300, 300)$ using alternating red (`[1, 0, 0]`) and green (`[0, 1, 0]`) color fills.

```python
from pathlib import Path
from reportlab.graphics.shapes import Drawing, Circle
from reportlab.graphics import renderPDF

output_dir = Path('output')
output_dir.mkdir(exist_ok=True)
pdf_two_color = output_dir / 'two_color_circles.pdf'

d = Drawing(600, 600)
for rd in range(0, 300, 10):
    if rd % 20 == 0:
        d.add(Circle(300, 300, 300 - rd, fillColor=[1, 0, 0]))
    else:
        d.add(Circle(300, 300, 300 - rd, fillColor=[0, 1, 0]))

renderPDF.drawToFile(d, str(pdf_two_color), '')
print(f"PDF created: {pdf_two_color}")
```

## 3. Random Multi-Color Concentric Circles

Use `random.random()` to generate float RGB values for colorful concentric circle fills.

```python
import random
pdf_random_circles = output_dir / 'random_color_circles.pdf'

d = Drawing(1200, 1200)
for rd in range(0, 600, 20):
    r = random.random()
    g = random.random()
    b = random.random()
    d.add(Circle(600, 600, 600 - rd, fillColor=[r, g, b]))

renderPDF.drawToFile(d, str(pdf_random_circles), '')
print(f"PDF created: {pdf_random_circles}")
```

## 4. Random Color Nested Rectangles

Draw nested rectangles using `random.uniform(0, 1)` for randomized color fills.

```python
from reportlab.graphics.shapes import Rect
pdf_rectangles = output_dir / 'random_color_rectangles.pdf'

w, h = 620, 620
d = Drawing(w, h)

for rd in range(0, int(w / 2), 20):
    r = random.uniform(0, 1)
    g = random.uniform(0, 1)
    b = random.uniform(0, 1)
    d.add(Rect(rd, rd, w - (rd * 2), h - (rd * 2), fillColor=[r, g, b]))

renderPDF.drawToFile(d, str(pdf_rectangles), '')
print(f"PDF created: {pdf_rectangles}")
```
