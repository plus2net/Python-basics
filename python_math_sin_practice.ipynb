### 1. Basic Sine Calculations in Radians
`math.sin(x)` returns the sine of `x`, where `x` is measured strictly in radians.

```python
import math

print("sin(0):       ", math.sin(0))
print("sin(pi / 2):  ", math.sin(math.pi / 2))
print("sin(1 radian):", math.sin(1))
print("sin(3*pi / 2):", math.sin(3 * math.pi / 2))
```

### 2. Working with Angles in Degrees (math.radians)
Always convert angles from degrees to radians using `math.radians()` before evaluating `math.sin()`.

```python
import math

angles_deg = [0, 30, 45, 60, 90, 180, 270, 360]
for deg in angles_deg:
    rad = math.radians(deg)
    print(f"sin({deg:3d} deg) = {math.sin(rad):6.3f}")
```

### 3. Floating-Point Precision at Multiples of pi
Because `pi` is represented as a finite float, $\sin(\pi)$ yields a near-zero residue (`1.22e-16`). Use `math.isclose()` for safe comparisons.

```python
import math

raw_val = math.sin(math.pi)
print("sin(pi) raw:    ", raw_val)
print("Is close to 0.0?", math.isclose(raw_val, 0.0, abs_tol=1e-15))
```

### 4. Trigonometric Symmetry & Pythagorean Identity
Verify odd symmetry $\sin(-x) = -\sin(x)$ and the Pythagorean identity $\sin^2(x) + \cos^2(x) = 1.0$.

```python
import math

theta = 1.45
print("Odd symmetry: ", math.isclose(math.sin(-theta), -math.sin(theta)))
print("Pythagorean:  ", math.isclose(math.sin(theta)**2 + math.cos(theta)**2, 1.0))
```

### 5. Hands-on Practice Exercise: Simple Harmonic Motion Simulator
Run the self-checking unit assertions below to verify oscillation displacement over time: $y(t) = A \cdot \sin(2\pi f t + \phi)$.

```python
def calculate_shm_displacement(amplitude, frequency_hz, time_seconds, phase_radians=0.0):
    omega = 2.0 * math.pi * frequency_hz
    y = amplitude * math.sin(omega * time_seconds + phase_radians)
    return round(y, 4)

# 1 Hz oscillator with amplitude 5.0m
assert calculate_shm_displacement(5.0, 1.0, 0.0) == 0.0    # equilibrium
assert calculate_shm_displacement(5.0, 1.0, 0.25) == 5.0   # peak (+A)
assert calculate_shm_displacement(5.0, 1.0, 0.50) == 0.0   # equilibrium
assert calculate_shm_displacement(5.0, 1.0, 0.75) == -5.0  # trough (-A)
assert calculate_shm_displacement(5.0, 1.0, 1.00) == 0.0   # full period

print("Displacement at t=0.25s:", calculate_shm_displacement(5.0, 1.0, 0.25), "m")
print("Displacement at t=0.75s:", calculate_shm_displacement(5.0, 1.0, 0.75), "m")
print("All harmonic motion assertions passed successfully!")
```
