### 1. Basic Integer Conversions
Convert strings and floats to integers.

```python
print("int():     ", int())
print("int('45'): ", int('45'))
print("int(4.99): ", int(4.99))   # Truncates to 4
print("int(-4.99):", int(-4.99))  # Truncates to -4
```

### 2. Multi-Base Parsing (Bases 2, 8, 16, 36)
Parse strings with explicit radixes.

```python
print("Base 2  '1011':  ", int('1011', 2))
print("Base 8  '71':    ", int('71', 8))
print("Base 16 'cd':    ", int('cd', 16))
print("Base 36 'python':", int('python', 36))
```

### 3. Custom Class __int__() Protocol
Implement the integer conversion dunder method on custom objects.

```python
class Money:
    def __init__(self, dollars, cents):
        self.dollars = dollars
        self.cents = cents
    def __int__(self):
        return self.dollars * 100 + self.cents

m = Money(15, 75)
print("Total cents via int(m):", int(m))
assert int(m) == 1575
```

### 4. Production Recipe: Safe Integer Parser
Safely parse string inputs including decimal floats.

```python
def safe_int(raw, default=0):
    try:
        return int(float(str(raw).strip()))
    except (ValueError, TypeError):
        return default

print("safe_int('45'):     ", safe_int('45'))
print("safe_int(' 45.8 '): ", safe_int(' 45.8 '))
print("safe_int(None):     ", safe_int(None))
print("safe_int('invalid'):", safe_int('invalid', -1))
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying int() mechanics.

```python
assert int() == 0
assert int(4.9) == 4
assert int(-4.9) == -4
assert int("1011", 2) == 11
assert int("71", 8) == 57
assert int("cd", 16) == 205
assert int("z", 36) == 35

print("All int() unit assertions passed! 100% verified.")
```
