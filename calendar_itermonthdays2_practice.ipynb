# `calendar.itermonthdays2()` — Interactive Python Practice
**Tutorial:** [Plus2Net Python itermonthdays2()](https://www.plus2net.com/python/calendar-itermonthdays2.php)

Run the examples in order or independently. Each exercise uses Python's standard library. `assert` checks confirm expected values, and you can change the input months to explore.

## 1. Original July 2020 example
Print the complete `(day_number, weekday_number)` sequence.

```python
import calendar
slots = list(calendar.Calendar().itermonthdays2(2020, 7))
for item in slots:
    print(item)
assert len(slots) == 35
assert slots[:3] == [(0, 0), (0, 1), (1, 2)]
assert slots[-2:] == [(0, 5), (0, 6)]
```

## 2. Inspect the iterator and tuple
Each element is a two-item tuple; the overall result is an iterator.

```python
import calendar
it = calendar.Calendar().itermonthdays2(2020, 7)
print(type(it).__name__)
first = next(it)
print(first, type(first).__name__)
assert first == (0, 0)
assert isinstance(first, tuple) and len(first) == 2
```

## 3. The original Thursday-first example
Change the first day to Thursday (3); July 1 remains Wednesday (2).

```python
import calendar
slots = list(calendar.Calendar(firstweekday=3).itermonthdays2(2020, 7))
print('Slots:', len(slots))
print('First week:', slots[:7])
assert len(slots) == 42
assert slots[0] == (0, 3)
assert (1, 2) in slots
```

## 4. Read items using tuple indexes
Use `[0]` to read the date number and `[1]` for the weekday.

```python
import calendar
slots = list(calendar.Calendar(firstweekday=3).itermonthdays2(2020, 7))
for x in slots[:8]:
    print('day:', x[0], ', weekday : ', x[1])
assert slots[6] == (1, 2)
```

## 5. Unpack `(day, weekday)`
Unpacking is often clearer than using indexes.

```python
import calendar
real = []
for day, weekday in calendar.Calendar().itermonthdays2(2020, 7):
    if day:
        real.append((day, weekday))
print(real[:4])
assert len(real) == 31
assert real[0] == (1, 2)
```

## 6. Remove zero placeholders
A zero is not a July date.

```python
import calendar
real = [pair for pair in calendar.Calendar().itermonthdays2(2020, 7) if pair[0]]
print('First date:', real[0])
print('Last date:', real[-1])
assert real[-1] == (31, 4)
```

## 7. Convert weekday numbers into names
Monday is always 0, even when the calendar starts on Thursday.

```python
import calendar
names = {day: calendar.day_name[wd]
         for day, wd in calendar.Calendar(firstweekday=3).itermonthdays2(2020, 7)
         if day}
print('1:', names[1], '31:', names[31])
assert names[1] == 'Wednesday'
assert names[31] == 'Friday'
```

## 8. Find weekends
Saturday is 5, Sunday is 6.

```python
import calendar
weekends = [(day, wd) for day, wd in calendar.Calendar().itermonthdays2(2020, 7)
            if day and wd in (5, 6)]
print(weekends)
assert len(weekends) == 8
assert weekends[0] == (4, 5)
```

## 9. Count working days
This counts Monday through Friday and does not subtract public holidays.

```python
import calendar
workdays = [day for day, wd in calendar.Calendar(firstweekday=6).itermonthdays2(2020, 7)
            if day and wd < 5]
print('Monday-Friday days:', len(workdays))
assert len(workdays) == 23
```

## 10. Group the result into weeks
Use slices of seven tuples.

```python
import calendar
slots = list(calendar.Calendar().itermonthdays2(2020, 7))
weeks = [slots[i:i+7] for i in range(0, len(slots), 7)]
for i, week in enumerate(weeks, 1):
    print(i, week)
assert len(weeks) == 5
assert all(len(week) == 7 for week in weeks)
```

## 11. A Thursday-first calendar can use six rows
Changing the first weekday may change the number of rows.

```python
import calendar
monday = list(calendar.Calendar(firstweekday=0).itermonthdays2(2020, 7))
thursday = list(calendar.Calendar(firstweekday=3).itermonthdays2(2020, 7))
print('Monday-first rows:', len(monday) // 7)
print('Thursday-first rows:', len(thursday) // 7)
assert len(monday) // 7 == 5
assert len(thursday) // 7 == 6
```

## 12. Sunday-first order
Reorder calendar slots without changing weekday numbering.

```python
import calendar
slots = list(calendar.Calendar(firstweekday=calendar.SUNDAY).itermonthdays2(2020, 7))
print(slots[:7])
assert slots[0][1] == 6
assert next(wd for day, wd in slots if day == 1) == 2
```

## 13. Watch an iterator get exhausted
Create it again if you need a second pass.

```python
import calendar
it = calendar.Calendar().itermonthdays2(2020, 7)
first = list(it)
second = list(it)
print(len(first), len(second))
assert len(first) == 35 and len(second) == 0
```

## 14. Compare the related methods
`itermonthdates()` provides full dates; `itermonthdays4()` provides four-item tuples.

```python
import calendar
cal = calendar.Calendar()
print(next(cal.itermonthdays(2020, 7)))
print(next(cal.itermonthdays2(2020, 7)))
print(next(cal.itermonthdates(2020, 7)))
print(next(cal.itermonthdays4(2020, 7)))
assert next(cal.itermonthdays4(2020, 7)) == (2020, 6, 29, 0)
```

## 15. Compare flat and nested week data
The same tuples are returned by both methods.

```python
import calendar
cal = calendar.Calendar(firstweekday=3)
flat = list(cal.itermonthdays2(2020, 7))
nested = cal.monthdays2calendar(2020, 7)
print('Same tuples:', flat == [item for row in nested for item in row])
assert flat == [item for row in nested for item in row]
```

## 16. Leap-year practice
Check February 2024.

```python
import calendar
real = [(day, wd) for day, wd in calendar.Calendar().itermonthdays2(2024, 2)
        if day != 0]
print('Days in February 2024:', len(real))
print('Last:', real[-1])
assert len(real) == 29
assert real[-1] == (29, 3)
```

## 17. Make a weekday task list
Only include the first 10 days of July that are Monday–Friday.

```python
import calendar
tasks = [f'July {day}: Review tasks' for day, wd in
         calendar.Calendar().itermonthdays2(2020, 7) if 1 <= day <= 10 and wd < 5]
for task in tasks:
    print(task)
assert len(tasks) == 8
```

## 18. Handle invalid months
The month must be between 1 and 12.

```python
import calendar
try:
    result = list(calendar.Calendar().itermonthdays2(2020, 13))
except calendar.IllegalMonthError:
    print('Invalid month; please use 1 to 12')
else:
    raise AssertionError('Expected IllegalMonthError')
```

## 21. Challenge — a custom weekday header
Generate weekday abbreviations in the same order as the calendar.

```python
import calendar
cal = calendar.Calendar(firstweekday=3)
headers = [calendar.day_abbr[i] for i in cal.iterweekdays()]
print(headers)
assert headers == ['Thu', 'Fri', 'Sat', 'Sun', 'Mon', 'Tue', 'Wed']
```

Confirm the essential behaviors learned in this tutorial.
