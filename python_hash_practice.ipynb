### 1. Hashing Immutable Built-in Types

```python
num = 100
float_num = 100.0
text = "plus2net"
data_tuple = (1, 2, 3)

print("hash(100):        ", hash(num))
print("hash(100.0):      ", hash(float_num))
print("Numeric equality: ", hash(num) == hash(float_num))
print("hash('plus2net'): ", type(hash(text)).__name__)
print("hash((1, 2, 3)):  ", type(hash(data_tuple)).__name__)
```

### 2. Unhashable Mutable Types

```python
try:
    hash([1, 2, 3])
except TypeError as err:
    print("List error: ", err)

try:
    hash({'a': 1})
except TypeError as err:
    print("Dict error: ", err)

try:
    hash((1, [2, 3]))  # Tuple with mutable child
except TypeError as err:
    print("Tuple error:", err)
```

### 3. Custom Hashable Class Implementation

```python
class Product:
    def __init__(self, sku: str, price: float):
        self.sku = sku
        self.price = price

    def __eq__(self, other):
        if isinstance(other, Product):
            return self.sku == other.sku
        return False

    def __hash__(self):
        return hash(self.sku)

p1 = Product("SKU-1001", 19.99)
p2 = Product("SKU-1001", 19.99)
p3 = Product("SKU-1002", 29.99)

unique_products = {p1, p2, p3}
print("Unique SKUs in set:", len(unique_products))
print("p1 and p2 hashes equal:", hash(p1) == hash(p2))
```

### 4. Fast Stream Record Deduplicator

```python
seen_hashes = set()

def process_record(item_id, category):
    record_key = (item_id, category)
    h = hash(record_key)
    if h in seen_hashes:
        return f"Duplicate detected for ID {item_id}"
    seen_hashes.add(h)
    return f"Processed ID {item_id}"

print(process_record(101, "Electronics"))
print(process_record(102, "Books"))
print(process_record(101, "Electronics"))
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Int hash identity
assert hash(50) == 50

# Test 2: Cross-type numerical equality
assert hash(1) == hash(1.0) == hash(True) == 1

# Test 3: Tuple hashing
assert isinstance(hash(('a', 'b')), int)

# Test 4: Custom class hash contract
class Node:
    def __init__(self, key): self.key = key
    def __eq__(self, o): return isinstance(o, Node) and self.key == o.key
    def __hash__(self): return hash(self.key)

assert hash(Node(42)) == hash(Node(42))
assert len({Node(1), Node(1)}) == 1

print("All hash() unit assertions successfully verified! 100% passed.")
```
