### 1. Basic Single Inheritance Delegation
`super().__init__()` initializes the parent class cleanly.

```python
class Animal:
    def __init__(self, name: str, height: float):
        self.name = name
        self.height = height

class Wild(Animal):
    def __init__(self, name: str, weight: int):
        super().__init__(name, 5.0)
        self.weight = weight

tiger = Wild("Tom", 300)
print("Name:  ", tiger.name)
print("Height:", tiger.height)
print("Weight:", tiger.weight)
```

### 2. Inspecting Method Resolution Order (MRO)
View the linear search order calculated via C3 linearization.

```python
class A: pass
class B(A): pass
class C(A): pass
class D(B, C): pass

print("Class D MRO:", [cls.__name__ for cls in D.__mro__])
```

### 3. Cooperative Diamond Multiple Inheritance
Ensure every class initializes exactly once.

```python
class Root:
    def __init__(self):
        print("Root init")

class Left(Root):
    def __init__(self):
        print("Left start")
        super().__init__()
        print("Left end")

class Right(Root):
    def __init__(self):
        print("Right start")
        super().__init__()
        print("Right end")

class Diamond(Left, Right):
    def __init__(self):
        print("Diamond start")
        super().__init__()
        print("Diamond end")

d = Diamond()
```

### 4. Delegating Non-Init Methods
Extend business logic in parent methods cleanly.

```python
class Service:
    def execute(self, task):
        return f"Processed {task}"

class AuditedService(Service):
    def execute(self, task):
        res = super().execute(task)
        return f"[AUDIT] {res}"

s = AuditedService()
print(s.execute("Order #1234"))
```

### 5. Production Recipe: Cooperative Mixin
Build reusable mixins that serialize class state.

```python
class DictMixin:
    def to_dict(self):
        data = super().to_dict() if hasattr(super(), "to_dict") else {}
        data.update({k: v for k, v in self.__dict__.items() if not k.startswith("_")})
        return data

class Account(DictMixin):
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

acc = Account("Bob", 500)
print("Serialized:", acc.to_dict())
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying super() mechanics.

```python
class SuperVerifier:
    @staticmethod
    def test_all():
        class P:
            def __init__(self, val):
                self.val = val
                
        class C(P):
            def __init__(self, val, extra):
                super().__init__(val)
                self.extra = extra
                
        child = C(10, "sub")
        assert child.val == 10
        assert child.extra == "sub"
        assert isinstance(child, P)
        assert isinstance(child, C)
        
        print("All super() unit assertions passed successfully! 100% verified.")

SuperVerifier.test_all()
```
