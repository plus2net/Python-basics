### 1. Basic Complex Number Construction
Construct complex numbers and inspect `.real`, `.imag`, and `.conjugate()`.

```python
z = complex(4, 3)
print("z:        ", z)
print("z.real:   ", z.real)
print("z.imag:   ", z.imag)
print("conjugate:", z.conjugate())
```

### 2. String Parsing (Strict Whitespace Rule)
Parse complex numbers from string literals without surrounding spaces.

```python
z_parsed = complex("4+3j")
print("Parsed complex:", z_parsed)
assert z_parsed == 4 + 3j
```

### 3. Modulus & Polar Coordinates with cmath
Compute magnitude and convert to polar coordinates.

```python
import cmath
z = 3 + 4j
r, theta = cmath.polar(z)
print(f"Modulus r: {r}, Phase theta: {theta:.4f} rad")
assert r == 5.0
```

### 4. Production Recipe: Vector Rotation
Rotate complex numbers counterclockwise by multiplying with `cmath.exp(1j * angle)`.

```python
import math
z = 3 + 4j
angle = math.radians(90)  # 90 degrees counterclockwise
rotated = z * cmath.exp(1j * angle)
print("Rotated by 90 deg:", rotated)
# (3+4j)*1j = -4+3j
assert round(rotated.real, 2) == -4.0
assert round(rotated.imag, 2) == 3.0
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying complex() mechanics.

```python
assert complex() == 0j
assert complex(4, 3) == 4 + 3j
assert complex("4+3j") == 4 + 3j
assert abs(complex(3, 4)) == 5.0

print("All complex() unit assertions passed! 100% verified.")
```
