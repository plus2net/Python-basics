### 1. Standard Deviation from Scratch with math.fsum() and math.sqrt()
Compute Population Standard Deviation (`ddof=0`, divided by `N`) and Sample Standard Deviation (`ddof=1`, divided by `N - 1`).


```python
import math

def calculate_std_dev(data, ddof=0):
    n = len(data)
    if n <= ddof:
        raise ValueError("Dataset size must exceed ddof.")
    mean = math.fsum(data) / n
    squared_diffs = [(x - mean)**2 for x in data]
    variance = math.fsum(squared_diffs) / (n - ddof)
    return math.sqrt(variance)

dataset = [12, 13, 15, 11, 9, 12, 13, 10, 11, 12, 13, 7, 8]
pop_sd = calculate_std_dev(dataset, ddof=0)
sample_sd = calculate_std_dev(dataset, ddof=1)

print(f"Population Std Dev (ddof=0): {pop_sd:.6f}")
print(f"Sample Std Dev     (ddof=1): {sample_sd:.6f}")
```

### 2. Standard Library: statistics.pstdev vs. statistics.stdev
Python's `statistics` module explicitly splits population and sample calculations.


```python
import statistics

print("statistics.pstdev (Population):", statistics.pstdev(dataset))
print("statistics.stdev  (Sample):    ", statistics.stdev(dataset))
```

### 3. The NumPy vs. Pandas Default Mismatch
NumPy defaults to `ddof=0` (population), while Pandas defaults to `ddof=1` (sample). Always specify `ddof` explicitly.


```python
import numpy as np
import pandas as pd

arr = np.array(dataset)
series = pd.Series(dataset)

print("np.std() default:     ", arr.std())     # 2.153846 (ddof=0)
print("series.std() default: ", series.std())  # 2.241794 (ddof=1)
print("np.std(ddof=1):       ", arr.std(ddof=1))
```

### 4. Numerical Two-Pass Stability vs. Catastrophic Cancellation
Avoid shortcut formulas like `E[X^2] - (E[X])^2` on shifted data with large means.


```python
import math

shifted_data = [1e9 + 1, 1e9 + 2, 1e9 + 3]
mean_val = sum(shifted_data) / 3
naive_var = sum(x**2 for x in shifted_data) / 3 - mean_val**2
two_pass_var = math.fsum((x - mean_val)**2 for x in shifted_data) / 3

print("Naive formula variance:   ", naive_var)    # Fails due to cancellation!
print("Two-pass exact variance:  ", two_pass_var) # Exact 0.6666666666666666
```

### 5. Hands-on Practice Exercise: Asset Volatility & Outlier Auditor
Calculate daily sample standard deviation, annualized volatility `sigma * sqrt(252)`, and detect 3-sigma outliers.


```python
import math

def financial_volatility_and_outliers(daily_returns, trading_days=252):
    """Computes daily sample std dev, annualized volatility, and detects 3-sigma outliers."""
    n = len(daily_returns)
    if n < 2:
        raise ValueError("Sample standard deviation requires at least 2 data points.")
    mean_ret = math.fsum(daily_returns) / n
    var_sample = math.fsum((r - mean_ret)**2 for r in daily_returns) / (n - 1)
    daily_stdev = math.sqrt(var_sample)
    annual_vol = daily_stdev * math.sqrt(trading_days)
    upper_bound = mean_ret + 3.0 * daily_stdev
    lower_bound = mean_ret - 3.0 * daily_stdev
    outliers = [r for r in daily_returns if r > upper_bound or r < lower_bound]
    return {
        "mean": round(mean_ret, 6),
        "daily_stdev": round(daily_stdev, 6),
        "annualized_volatility": round(annual_vol, 4),
        "outliers": outliers
    }

# Test Data
sample_returns = [0.001] * 10 + [-0.001] * 10 + [-0.08]
audit_result = financial_volatility_and_outliers(sample_returns)

# Unit Assertions
assert audit_result["outliers"] == [-0.08]
assert audit_result["daily_stdev"] > 0.01

print("Audit Result:", audit_result)
print("All standard deviation assertions passed successfully!")
```
