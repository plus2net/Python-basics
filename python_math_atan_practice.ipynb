### 1. Basic Arc Tangent Calculations
`math.atan(x)` takes any real slope ratio $x \in (-\infty, +\infty)$ and returns the corresponding angle in radians $(-rac{\pi}{2}, rac{\pi}{2})$.

```python
import math

print("atan(0):   ", math.atan(0))
print("atan(1.0): ", math.atan(1.0))
print("atan(-1.0):", math.atan(-1.0))
print("atan(10.0):", math.atan(10.0))
```

### 2. Converting Output Radians to Degrees
Use `math.degrees()` to convert output angles from radians to degrees ($-90^\circ$ to $+90^\circ$).

```python
import math

ratios = [0.0, 0.57735, 1.0, 1.73205, 5.0]
for r in ratios:
    rad = math.atan(r)
    deg = math.degrees(rad)
    print(f"atan({r:7.4f}) -> {rad:6.4f} rad == {deg:5.1f} deg")
```

### 3. Asymptotic Limits at Infinity
As $x 	o \pm\infty$, `math.atan(x)` approaches $\pmrac{\pi}{2}$.

```python
import math

print("atan(1,000,000): ", math.atan(1_000_000))
print("atan(-1,000,000):", math.atan(-1_000_000))
print("pi / 2 reference: ", math.pi / 2)
```

### 4. math.atan() vs. math.atan2()
`math.atan2(y, x)` resolves the true 4-quadrant angle without ambiguity or division-by-zero crashes.

```python
import math

x, y = -1.0, 1.0  # Quadrant 2 point (-1, 1)

print("atan(y / x): ", math.degrees(math.atan(y / x)))   # -45.0 degrees (Ambiguous!)
print("atan2(y, x): ", math.degrees(math.atan2(y, x)))  # 135.0 degrees (Exact!)
```

### 5. Hands-on Practice Exercise: Road Slope Grade Angle Calculator
Run the self-checking unit assertions below to verify civil engineering road slope conversions: $\theta = \arctan(\frac{\text{grade}}{100})$.

```python
def calculate_road_grade_angle(grade_percent):
    ratio = grade_percent / 100.0
    angle_rad = math.atan(ratio)
    angle_deg = math.degrees(angle_rad)
    return round(angle_deg, 2)

# Unit Assertions
assert calculate_road_grade_angle(0) == 0.0
assert calculate_road_grade_angle(6) == 3.43
assert calculate_road_grade_angle(10) == 5.71
assert calculate_road_grade_angle(100) == 45.0
assert calculate_road_grade_angle(-10) == -5.71

print("6%  Highway Grade: ", calculate_road_grade_angle(6), "degrees")
print("10% Mountain Grade:", calculate_road_grade_angle(10), "degrees")
print("All road grade assertions passed successfully!")
```
