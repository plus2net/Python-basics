### 1. Basic Key Removal and Value Extraction
`dict.pop(key)` removes the key and returns its value in O(1) constant time.

```python
user_session = {"id": 101, "username": "john_doe", "status": "active"}

# Remove 'status' and capture its value
current_status = user_session.pop("status")
print("Extracted status:", current_status)  # 'active'
print("Remaining dict:   ", user_session)   # {'id': 101, 'username': 'john_doe'}
```

### 2. Strict Mode: Catching KeyError on Missing Keys
Without a default fallback, attempting to pop a non-existent key raises `KeyError`.

```python
record = {"name": "Alice", "role": "Admin"}

try:
    secret = record.pop("token")
except KeyError as err:
    print(f"Caught expected error for missing key: {err}")
```

### 3. Safe Eviction with Default Fallback Values
Passing a second argument prevents exceptions and returns the fallback value.

```python
config = {"host": "localhost", "port": 8080}

debug_flag = config.pop("debug", False)
timeout = config.pop("timeout", 30)

print("Debug setting:", debug_flag)  # False
print("Timeout:      ", timeout)     # 30
print("Config state: ", config)
```

### 4. Real-World Recipe: Sanitizing Sensitive Data Before Logging
Pop credentials such as passwords or API tokens out of dictionaries before serializing to logs.

```python
def sanitize_payload(payload: dict) -> tuple:
    clean = dict(payload)
    token = clean.pop("token", None)
    clean.pop("password", None)
    return clean, token

incoming = {"user": "ravi", "token": "jwt_sec_948", "password": "hunter2", "action": "login"}
safe_log, extracted_token = sanitize_payload(incoming)

print("Safe log dict:", safe_log)
print("Token held:   ", extracted_token)
```

### 5. Real-World Recipe: Batch Key Eviction
Remove multiple keys safely without failing if some keys are absent.

```python
def batch_evict(data: dict, keys_to_remove: list) -> dict:
    evicted = {}
    for k in keys_to_remove:
        val = data.pop(k, None)
        if val is not None:
            evicted[k] = val
    return evicted

state = {"x": 10, "y": 20, "z": 30}
removed = batch_evict(state, ["x", "z", "nonexistent"])

print("Removed items:", removed)  # {'x': 10, 'z': 30}
print("Remaining:    ", state)    # {'y': 20}
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying `dict.pop()` behavior, KeyError raising, and default fallbacks.

```python
class PopVerifier:
    @staticmethod
    def safe_pop(source: dict, key: str, default=None):
        return source.pop(key, default)

v = PopVerifier()
d = {"a": 1, "b": 2, "c": 3}

# Test 1: Value return and removal
assert v.safe_pop(d, "b") == 2
assert "b" not in d
assert d == {"a": 1, "c": 3}

# Test 2: Fallback default
assert v.safe_pop(d, "missing", "default_val") == "default_val"

# Test 3: KeyError verification
try:
    d.pop("missing")
    assert False, "KeyError should have been raised"
except KeyError:
    pass

print("All dict.pop() unit assertions passed successfully! 100% verified.")
```
