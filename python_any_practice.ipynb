### 1. Basic Truth Value Testing
Evaluate truthiness across empty lists, zeros, and non-empty collections.

```python
print("Empty list:            ", any([]))
print("All zeros:             ", any([0, 0, 0]))
print("At least one truthy:   ", any([0, False, 1]))
print("Non-empty text string: ", any("hello"))
```

### 2. Dictionary Keys vs. Values Caveat
Observe that passing a dictionary directly checks keys, not values.

```python
status = {0: True, 0: False}
print("Direct dictionary check (keys):", any(status))  # False because key is 0
print("Explicit values check:          ", any(status.values()))
```

### 3. Short-Circuit Evaluation in Generators
Verify that evaluation halts as soon as the first truthy value is encountered.

```python
def test_stream():
    for item in [0, False, 99, 200]:
        print(f"  Checked: {item}")
        yield item

print("Starting check:")
result = any(test_stream())
print("Final result:", result)
```

### 4. Production Recipe: Multi-Format File Extension Matching
Validate whether an incoming file matches accepted extensions.

```python
valid_exts = [".csv", ".xlsx", ".parquet"]
filename = "export_dataset.xlsx"
is_valid = any(filename.endswith(ext) for ext in valid_exts)
print(f"Is '{filename}' supported? {is_valid}")
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying any() mechanics.

```python
assert any([]) is False
assert any([0, 1]) is True
assert any({0: True}) is False
assert any({0: True}.values()) is True
assert any("test") is True

print("All any() unit assertions passed! 100% verified.")
```
