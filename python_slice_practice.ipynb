### 1. Constructing First-Class Slice Objects

```python
# Create slices with different argument counts
s_stop = slice(5)
s_range = slice(2, 7)
s_step = slice(1, 10, 2)

print("Stop only:   ", s_stop, "-> start:", s_stop.start, "stop:", s_stop.stop, "step:", s_stop.step)
print("Start & stop:", s_range, "-> start:", s_range.start, "stop:", s_range.stop, "step:", s_range.step)
print("With step:   ", s_step, "-> start:", s_step.start, "stop:", s_step.stop, "step:", s_step.step)
```

### 2. Slicing Sequences & Negative Stepping

```python
letters = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h']
text = "plus2net_tutorials"

# Reusable slice objects
take_first_four = slice(4)
reverse_all = slice(None, None, -1)
every_second = slice(0, None, 2)

print("First four letters: ", letters[take_first_four])
print("Reversed text:      ", text[reverse_all])
print("Every second letter:", letters[every_second])
```

### 3. Normalizing with slice.indices(length)

```python
complex_slice = slice(-5, None, 2)
seq_length = 12

# Calculates concrete (start, stop, step) bounds
start, stop, step = complex_slice.indices(seq_length)
print(f"For sequence length {seq_length}:")
print(f"  start={start}, stop={stop}, step={step}")
print("  Generated index range:", list(range(start, stop, step)))
```

### 4. Fixed-Width Columnar Parsing Recipe

```python
ID_COL = slice(0, 5)
NAME_COL = slice(5, 15)
SCORE_COL = slice(15, 20)

raw_lines = [
    "10001Alexander 94.5",
    "10002Beatrice  88.0",
    "10003Catherine 91.2"
]

parsed = []
for line in raw_lines:
    parsed.append({
        "id": line[ID_COL].strip(),
        "name": line[NAME_COL].strip(),
        "score": float(line[SCORE_COL].strip())
    })

for row in parsed:
    print(f"{row['id']} | {row['name']:<10} | {row['score']}")
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Slice attributes
s = slice(2, 8, 3)
assert (s.start, s.stop, s.step) == (2, 8, 3)

# Test 2: List slice
nums = [10, 20, 30, 40, 50]
assert nums[slice(1, 4)] == [20, 30, 40]

# Test 3: String reversal
assert "python"[slice(None, None, -1)] == "nohtyp"

# Test 4: indices() calculation
assert slice(-4, -1).indices(10) == (6, 9, 1)

# Test 5: List mutation via slice assignment
data = [1, 2, 3, 4]
data[slice(1, 3)] = [99]
assert data == [1, 99, 4]

print("All slice() unit assertions successfully verified! 100% passed.")
```
