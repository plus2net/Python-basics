### 1. Basic Class Method Definition
`@classmethod` binds the method to the class `cls` rather than an instance.

```python
class Animal:
    home = "Jungle"
    
    @classmethod
    def get_home(cls):
        return f"Habitat: {cls.home}"

# Called on class directly
print(Animal.get_home())
```

### 2. Alternative Factory Constructors
Construct objects from diverse formats like strings or dictionaries.

```python
class Date:
    def __init__(self, year: int, month: int, day: int):
        self.year = year
        self.month = month
        self.day = day
        
    @classmethod
    def from_dash_string(cls, date_str: str):
        y, m, d = map(int, date_str.split("-"))
        return cls(y, m, d)

d = Date.from_dash_string("2026-10-14")
print(f"Constructed Date: {d.year}/{d.month}/{d.day}")
```

### 3. Polymorphic Subclass Factory Creation
`cls(...)` ensures that calling the factory on a subclass produces an instance of that subclass.

```python
class Document:
    def __init__(self, name):
        self.name = name
    @classmethod
    def create(cls, name):
        return cls(name)

class Invoice(Document):
    pass

inv = Invoice.create("INV-001")
print("Instance type:", type(inv))  # <class '__main__.Invoice'>
print("Is Invoice:   ", isinstance(inv, Invoice))
```

### 4. Mutating Class State Across Instances
Track global instances using class-level variables.

```python
class Server:
    active_connections = 0
    
    def __init__(self):
        Server.active_connections += 1
        
    @classmethod
    def get_connections(cls):
        return cls.active_connections

s1 = Server()
s2 = Server()
print("Active connections:", Server.get_connections())
```

### 5. Production Recipe: Deserialization Model
Instantiate models cleanly from dictionary payloads.

```python
class User:
    def __init__(self, uid, username):
        self.uid = uid
        self.username = username
    @classmethod
    def from_dict(cls, data):
        return cls(data["id"], data["username"])

raw_data = {"id": 101, "username": "alice"}
user = User.from_dict(raw_data)
print("User object:", user.uid, user.username)
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying @classmethod mechanics.

```python
class ClassMethodVerifier:
    @staticmethod
    def test_all():
        class Base:
            tag = "base"
            def __init__(self, val):
                self.val = val
            @classmethod
            def make_default(cls):
                return cls(10)
                
        class Sub(Base):
            tag = "sub"
            
        b = Base.make_default()
        s = Sub.make_default()
        
        assert b.val == 10
        assert s.val == 10
        assert type(b) is Base
        assert type(s) is Sub
        assert b.tag == "base"
        assert s.tag == "sub"
        
        print("All @classmethod unit assertions passed successfully! 100% verified.")

ClassMethodVerifier.test_all()
```
