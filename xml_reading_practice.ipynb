### 1. Read XML with ElementTree

```python
import xml.etree.ElementTree as ET

# Create sample XML file for demonstration
xml_data = """<students>
    <student>
        <id>1</id>
        <name>John Deo</name>
        <mark>75</mark>
    </student>
    <student>
        <id>2</id>
        <name>Max Ruin</name>
        <mark>85</mark>
    </student>
</students>"""

with open("students.xml", "w") as f:
    f.write(xml_data)

tree = ET.parse("students.xml")
root = tree.getroot()

for student in root.findall("student"):
    sid = student.find("id").text
    name = student.find("name").text
    mark = student.find("mark").text
    print(f"ID: {sid}, Name: {name}, Mark: {mark}")
```

### 2. Read XML with Pandas

```python
import pandas as pd

# Read XML file into DataFrame
df = pd.read_xml("students.xml", parser="etree")
print(df)
```
