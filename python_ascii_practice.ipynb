### 1. Basic Non-ASCII Character Escaping

```python
# Latin-1 character (\x escape)
print("Accented letter: ", ascii('Café'))

# BMP character (\u escape)
print("Euro currency:   ", ascii('€'))

# Supplementary plane emoji (\U escape)
print("Rocket emoji:    ", ascii('🚀'))
```

### 2. Comparative Analysis: str() vs. repr() vs. ascii()

```python
phrase = "Pythön 🐍"

print("str():  ", str(phrase))
print("repr(): ", repr(phrase))
print("ascii():", ascii(phrase))
```

### 3. Escaping Nested Collections

```python
catalog = [
    {"product": "Café Latte", "price": "3.50€"},
    {"product": "Jalapeño Chips", "rating": "⭐⭐⭐⭐⭐"}
]

print("Escaped collection:")
print(ascii(catalog))
```

### 4. Reversible Round-Trip with eval()

```python
original_string = "Überraschung! München"
escaped_rep = ascii(original_string)
print("Escaped representation: ", escaped_rep)

restored_string = eval(escaped_rep)
print("Restored unicode string:", restored_string)
print("Strings match exactly:   ", restored_string == original_string)
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Latin-1 escape
assert ascii('é') == "'\\xe9'"

# Test 2: Euro symbol escape
assert ascii('€') == "'\\u20ac'"

# Test 3: Emoji escape
assert ascii('✨') == "'\\u2728'"

# Test 4: Plain ASCII unchanged
assert ascii('plus2net') == "'plus2net'"

# Test 5: Round trip
data = "Niño"
assert eval(ascii(data)) == data

print("All ascii() unit assertions successfully verified! 100% passed.")
```
