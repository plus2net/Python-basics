### 1. Basic Button Invocation

```python
import tkinter as tk

root = tk.Tk()
res = []

def on_click():
    res.append("Clicked!")

btn = tk.Button(root, text="Test", command=on_click)
btn.invoke()
print("Result list after invoke():", res)
root.destroy()
```

### 2. Capturing Return Values via invoke()

```python
import tkinter as tk

root = tk.Tk()

def compute():
    return 42 * 2

btn_calc = tk.Button(root, command=compute)
val = btn_calc.invoke()
print("Captured return value:", val)
root.destroy()
```

### 3. Self-Checking Unit Verification Suite

```python
import tkinter as tk

root = tk.Tk()

# Test 1: Callback execution and return value
def calc():
    return 50 + 50

b1 = tk.Button(root, command=calc)
assert b1.invoke() == 100

# Test 2: Disabled button state prevents execution
ran = False
def disabled_cb():
    global ran
    ran = True

b_dis = tk.Button(root, command=disabled_cb, state=tk.DISABLED)
b_dis.invoke()
assert ran is False

print("All Button.invoke() unit assertions verified successfully! 100% passed.")
root.destroy()
```
