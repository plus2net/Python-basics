### 1. Basic In-Place Set Resetting
`set.clear()` removes all elements from the set in-place and leaves it as `set()`.

```python
users = {"alex", "ronald", "ron"}
print("Initial set:", users)

ret = users.clear()
print("Cleared set:", users)      # set()
print("Return val: ", ret)        # None
print("Type:       ", type(users))  # <class 'set'>
```

### 2. The Shared Reference Difference: clear() vs. s = set()
`clear()` updates all shared aliases, while reassignment only alters the local variable binding.

```python
# Case A: clear() affects all aliases
perms_a = {"read", "write"}
alias_a = perms_a
perms_a.clear()
print("Case A - alias_a:", alias_a)  # set() (also emptied!)

# Case B: Reassignment breaks the alias
perms_b = {"read", "write"}
alias_b = perms_b
perms_b = set()
print("Case B - alias_b:", alias_b)  # {'read', 'write'} (intact!)
```

### 3. Preserving Memory Identity
`set.clear()` retains the underlying memory address (`id(s)` unchanged).

```python
pool = {f"token_{i}" for i in range(100)}
orig_id = id(pool)

pool.clear()
print("Memory ID preserved?", id(pool) == orig_id)  # True
print("Length after clear: ", len(pool))            # 0
```

### 4. Real-World Recipe: Buffer Pooling in Batch Loops
Reuse a single set buffer across multiple iterations to minimize allocation churn.

```python
accumulator = set()
batches = [[1, 2, 1], [3, 4, 3, 2]]

for i, b in enumerate(batches):
    accumulator.update(b)
    print(f"Batch {i} unique elements:", accumulator)
    accumulator.clear()

print("Accumulator after processing:", accumulator)  # set()
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying in-place clearing, memory reuse, and alias propagation.

```python
class SetClearVerifier:
    @staticmethod
    def verify_reset(target: set) -> bool:
        saved = id(target)
        target.clear()
        return len(target) == 0 and id(target) == saved

v = SetClearVerifier()
s = {"a", "b"}
ref = s

# Test 1: In-place clearing & address preservation
assert v.verify_reset(s) is True
assert len(ref) == 0
assert ref is s

# Test 2: Double clearing is safe
assert s.clear() is None

print("All set.clear() unit assertions passed successfully! 100% verified.")
```
