# Python math.log1p(): Accurate Natural Logarithm ln(1 + x)

### 1. Basic Calculations & Scope
`math.log1p(x)` calculates `ln(1 + x)`. It is defined for all real numbers strictly greater than negative one (`x > -1.0`).


```python
import math

print("log1p(0.0): ", math.log1p(0.0))          # 0.0
print("log1p(1.0): ", round(math.log1p(1.0), 6))  # 0.693147
print("log1p(9.0): ", round(math.log1p(9.0), 6))  # 2.302585
print("log1p(-0.5):", round(math.log1p(-0.5), 6)) # -0.693147
print("log1p(-0.1):", round(math.log1p(-0.1), 6)) # -0.105361
```

### 2. Catastrophic Cancellation Crisis: log1p(x) vs. log(1 + x)
When `x` is very small, evaluating `1.0 + x` rounds away the lower bits of `x`, causing `log(1 + x)` to lose accuracy or return 0.0. `math.log1p(x)` retains full precision.


```python
import math

small_x = 1e-16
print("1.0 + 1e-16 == 1.0?     ", (1.0 + small_x) == 1.0)   # True (Precision wiped out)
print("math.log(1.0 + 1e-16):  ", math.log(1.0 + small_x))  # 0.0 (100% data loss!)
print("math.log1p(1e-16):      ", math.log1p(small_x))      # 1e-16 (Exact!)
```

### 3. Inverse Relationship with math.expm1()
`math.log1p(x)` and `math.expm1(x)` form an exact inverse pair for small floating-point numbers.


```python
import math

test_vals = [1e-15, 1e-8, 0.05, 2.0]
for v in test_vals:
    round_trip = math.expm1(math.log1p(v))
    assert math.isclose(v, round_trip)
    print(f"Original: {v:.1e} -> expm1(log1p): {round_trip:.1e} | Match: True")
```

### 4. Machine Learning: Zero-Handling Log1p Transforms
Feature values with zeroes (e.g. visit counts, revenues) transform cleanly without divide-by-zero or negative infinity errors.


```python
import math

raw_counts = [0, 1, 5, 25, 1000, 50000]
transformed = [round(math.log1p(x), 3) for x in raw_counts]
print("Raw counts:  ", raw_counts)
print("Transformed: ", transformed)
```

### 5. Hands-on Practice Exercise: Quantitative Finance Log-Return Engine
Convert daily simple percentage returns `R` to continuously compounded log returns `r = log1p(R)` and compute cumulative multi-day growth using `expm1`.


```python
import math

def simple_to_log_return(simple_return):
    """Converts simple decimal return R into continuous log return r = log1p(R)."""
    if simple_return <= -1.0:
        raise ValueError("Total loss (-100% or worse) has undefined log return.")
    return math.log1p(simple_return)

def cumulative_growth(daily_simple_returns):
    """Computes total cumulative return from daily simple returns via log1p and expm1."""
    log_returns = [simple_to_log_return(r) for r in daily_simple_returns]
    total_log_return = math.fsum(log_returns)
    return math.expm1(total_log_return)

# Unit Assertions
tiny_r = 1e-15
assert simple_to_log_return(tiny_r) == tiny_r
assert simple_to_log_return(0.0) == 0.0

# Compounding +10% and -10% yields -1% overall: (1.1 * 0.9) - 1 = -0.01
cum_return = cumulative_growth([0.10, -0.10])
assert math.isclose(cum_return, -0.01)

try:
    simple_to_log_return(-1.0)
    assert False, "Should raise ValueError"
except ValueError:
    pass

sample_days = [0.005, -0.002, 0.008, -0.001]
print(f"Cumulative return over 4 trading days: {cumulative_growth(sample_days)*100:.3f}%")
print("All financial log-return assertions passed successfully!")
```
