### 1. Inspecting Memory Identity & Hex Formatting

```python
score = 98
print("Decimal id:    ", id(score))
print("Hex address:   ", hex(id(score)))

alias = score
print("Alias shares id:", id(score) == id(alias))
print("score is alias: ", score is alias)
```

### 2. Small Integer Caching (-5 to 256)

```python
# Integers in range -5 to 256 are singletons in CPython
small_1 = 150
small_2 = 150
print("150 is cached singleton:   ", small_1 is small_2)

# Integers outside cache allocate distinct objects
large_1 = 1000
large_2 = int("1000")
print("1000 has distinct addresses:", large_1 is not large_2)
print("Values still compare equal: ", large_1 == large_2)
```

### 3. Mutable vs. Immutable Identity Shifts

```python
# Mutable list preserves memory address on modification
items = [1, 2, 3]
initial_id = id(items)
items.append(4)
items[0] = 99
print("List ID unchanged: ", id(items) == initial_id)

# Immutable string creates a new object on change
text = "hello"
text_id = id(text)
text += " world"
print("String ID modified:", id(text) != text_id)
```

### 4. Graph Cycle Detection with id()

```python
class Node:
    def __init__(self, name):
        self.name = name
        self.edges = []

def check_cycle(start_node):
    seen = set()
    stack = [start_node]
    while stack:
        curr = stack.pop()
        if id(curr) in seen:
            return True
        seen.add(id(curr))
        stack.extend(curr.edges)
    return False

a, b, c = Node('A'), Node('B'), Node('C')
a.edges.append(b)
b.edges.append(c)
print("Linear graph has cycle? ", check_cycle(a))
c.edges.append(a)
print("Circular graph has cycle?", check_cycle(a))
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: id returns an integer
assert isinstance(id(42), int)

# Test 2: None singleton identity
assert id(None) == id(None)
assert None is None

# Test 3: Small integer caching (-5 to 256)
x1, x2 = 100, 100
assert x1 is x2

# Test 4: Distinct list objects
a_list, b_list = [1, 2], [1, 2]
assert a_list == b_list
assert a_list is not b_list

# Test 5: In-place list identity
saved_id = id(a_list)
a_list.append(3)
assert id(a_list) == saved_id

print("All id() unit assertions successfully verified! 100% passed.")
```
