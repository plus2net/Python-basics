### 1. Basic Integer divmod()
Calculate quotient and remainder simultaneously in one call.

```python
a, b = 29, 5
q, r = divmod(a, b)
print(f"{a} divided by {b}: Quotient = {q}, Remainder = {r}")
assert a == (q * b) + r
```

### 2. Floating-Point divmod()
Observe float quotient and remainder calculations.

```python
x, y = 53.67, 7.2
q_f, r_f = divmod(x, y)
print(f"divmod({x}, {y}) = ({q_f}, {r_f})")
```

### 3. Negative Number Floor Division
Understand how Python's floor rounding toward negative infinity affects remainders.

```python
print("divmod(-20, 3): ", divmod(-20, 3))
print("divmod(20, -3): ", divmod(20, -3))
print("divmod(-20, -3):", divmod(-20, -3))
```

### 4. Production Recipe: Duration Time Formatting
Convert raw elapsed seconds into HH:MM:SS format.

```python
total_seconds = 7384
minutes, seconds = divmod(total_seconds, 60)
hours, minutes = divmod(minutes, 60)
formatted = f"{hours:02d}:{minutes:02d}:{seconds:02d}"
print("Formatted time:", formatted)
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying divmod() mechanics.

```python
assert divmod(29, 5) == (5, 4)
assert divmod(10, 2) == (5, 0)
assert divmod(-20, 3) == (-7, 1)
assert divmod(20, -3) == (-7, -1)

print("All divmod() unit assertions passed! 100% verified.")
```
