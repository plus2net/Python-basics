## 1. Connecting to SQLite and Creating Tables

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
print("Connected to database successfully")

try:
    my_conn.execute('''
        CREATE TABLE IF NOT EXISTS student(
            id INTEGER PRIMARY KEY,
            name TEXT,
            class TEXT,
            mark INTEGER,
            gender TEXT
        );
    ''')
    my_conn.commit()
    print("Student table created successfully")
except sqlite3.Error as my_error:
    print("error:", my_error)
```

## 2. Inserting Sample Records

```python
try:
    r_set = my_conn.execute('''
        INSERT OR IGNORE INTO student (id, name, class, mark, gender) VALUES
        (1, 'John Deo', 'Four', 75, 'female'),
        (2, 'Max Ruin', 'Three', 85, 'male'),
        (3, 'Arnold', 'Three', 55, 'male'),
        (4, 'Krish Star', 'Four', 60, 'female'),
        (5, 'John Mike', 'Four', 60, 'female');
    ''')
    my_conn.commit()
    print("Records inserted successfully, count:", r_set.rowcount)
except sqlite3.Error as my_error:
    print("error:", my_error)
```

## 3. Querying Records

```python
cursor = my_conn.execute('SELECT * FROM student LIMIT 5')
for row in cursor:
    print(row)
my_conn.close()
```

## 4. Modern SQLAlchemy 2.0+ Context Manager Workflow

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

engine = create_engine("sqlite:///my_db.db")
try:
    with engine.begin() as connection:
        connection.execute(text("""CREATE TABLE IF NOT EXISTS student_sa (
            id INTEGER PRIMARY KEY,
            name TEXT,
            mark INTEGER
)"""))
        print("SQLAlchemy table created successfully via context manager!")
except SQLAlchemyError as e:
    print("SQLAlchemy error:", e)
```
