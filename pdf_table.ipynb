## 2. Basic Styled Table in PDF

Create a formatted student table with custom column widths, row heights, and table styling using ReportLab Platypus.

```python
from pathlib import Path
from reportlab.lib.units import inch
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate
from reportlab.platypus.tables import Table, TableStyle, colors

output_dir = Path('output')
output_dir.mkdir(exist_ok=True)
my_path = output_dir / 'student_table.pdf'

my_data = [
    ['ID', 'Name', 'Class', 'Mark', 'Gender'],
    (1, 'John Deo', 'Four', 75, 'female'),
    (2, 'Max Ruin', 'Three', 85, 'male'),
    (3, 'Arnold', 'Three', 55, 'male')
]

my_doc = SimpleDocTemplate(str(my_path), pagesize=letter)
c_width = [0.4 * inch, 1.5 * inch, 1 * inch, 1 * inch, 1 * inch]
t = Table(my_data, rowHeights=20, repeatRows=1, colWidths=c_width)
t.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.lightgreen),
    ('FONTSIZE', (0, 0), (-1, -1), 10)
]))
elements = [t]
my_doc.build(elements)
print(f"PDF created: {my_path}")
```

## 3. Dynamic Table Generation from SQLite Database

Query an in-memory SQLite database using SQLAlchemy and populate a ReportLab table dynamically.

```python
from sqlalchemy import create_engine, Table, Column, Integer, String, MetaData

# Create an in-memory SQLite database and sample student table
engine = create_engine('sqlite:///:memory:')
metadata = MetaData()
student_table = Table('student', metadata,
    Column('id', Integer, primary_key=True),
    Column('name', String),
    Column('class', String),
    Column('mark', Integer),
    Column('gender', String)
)
metadata.create_all(engine)

# Insert sample records
with engine.connect() as conn:
    conn.execute(student_table.insert(), [
        {'name': 'John Deo', 'class': 'Four', 'mark': 75, 'gender': 'female'},
        {'name': 'Max Ruin', 'class': 'Three', 'mark': 85, 'gender': 'male'},
        {'name': 'Arnold', 'class': 'Three', 'mark': 55, 'gender': 'male'},
        {'name': 'Krish Star', 'class': 'Four', 'mark': 60, 'gender': 'female'}
    ])
    conn.commit()

    # Query database records
    r_set = conn.execute(student_table.select())
    db_data = [list(r_set.keys())]
    for row in r_set.fetchall():
        db_data.append(list(row))

pdf_db_table = output_dir / 'db_student_table.pdf'
my_doc = SimpleDocTemplate(str(pdf_db_table), pagesize=letter)
t = Table(db_data, rowHeights=20, repeatRows=1, colWidths=c_width)
t.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.lightblue),
    ('FONTSIZE', (0, 0), (-1, -1), 10),
    ('GRID', (0, 0), (-1, -1), 0.5, colors.grey)
]))
my_doc.build([t])
print(f"Database table PDF created: {pdf_db_table}")
```
