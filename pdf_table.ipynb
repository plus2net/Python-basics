## 2. Basic Styled Table in PDF

Part 1: Fundamental table creation, setting column widths (`colWidths`), row heights (`rowHeights`), header repetition (`repeatRows=1`), and cell styling with `TableStyle`.

```python
from pathlib import Path
from reportlab.lib.units import inch
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate
from reportlab.platypus.tables import Table, TableStyle, colors

output_dir = Path('output')
output_dir.mkdir(exist_ok=True)
my_path = output_dir / 'student_table_basic.pdf'

my_data = [
    ['ID', 'Name', 'Class', 'Mark', 'Gender'],
    (1, 'John Deo', 'Four', 75, 'female'),
    (2, 'Max Ruin', 'Three', 85, 'male'),
    (3, 'Arnold', 'Three', 55, 'male')
]

my_doc = SimpleDocTemplate(str(my_path), pagesize=letter)
c_width = [0.4 * inch, 1.5 * inch, 1 * inch, 1 * inch, 1 * inch]
t = Table(my_data, rowHeights=20, repeatRows=1, colWidths=c_width)
t.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.lightgreen),
    ('FONTSIZE', (0, 0), (-1, -1), 10)
]))
elements = [t]
my_doc.build(elements)
print(f"Basic table PDF created: {my_path}")
```

## 3. Multi-Source Ingestion: CSV Data to PDF Table

Read `student.csv` using Python's `csv` module and populate a ReportLab PDF table.

```python
import csv

csv_data = []
with open('data/student.csv', mode='r', encoding='utf-8') as f:
    reader = csv.reader(f)
    for row in reader:
        csv_data.append(row)

pdf_csv = output_dir / 'table_from_csv.pdf'
doc = SimpleDocTemplate(str(pdf_csv), pagesize=letter)
t = Table(csv_data, rowHeights=18, repeatRows=1, colWidths=[0.5*inch, 1.5*inch, 1*inch, 1*inch, 1*inch])
t.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor('#007bff')),
    ('TEXTCOLOR', (0, 0), (-1, 0), colors.white),
    ('FONTSIZE', (0, 0), (-1, -1), 9),
    ('GRID', (0, 0), (-1, -1), 0.5, colors.lightgrey)
]))
doc.build([t])
print(f"CSV PDF Table created: {pdf_csv}")
```

## 4. Multi-Source Ingestion: Excel Data (`student.xlsx`) to PDF Table

Read `student.xlsx` using `openpyxl` and create a ReportLab table.

```python
import openpyxl

wb = openpyxl.load_workbook('data/student.xlsx')
sheet = wb.active
excel_data = []
for row in sheet.iter_rows(values_only=True):
    excel_data.append(list(row))

pdf_excel = output_dir / 'table_from_excel.pdf'
doc = SimpleDocTemplate(str(pdf_excel), pagesize=letter)
t = Table(excel_data, rowHeights=18, repeatRows=1, colWidths=[0.5*inch, 1.5*inch, 1*inch, 1*inch, 1*inch])
t.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor('#28a745')),
    ('TEXTCOLOR', (0, 0), (-1, 0), colors.white),
    ('FONTSIZE', (0, 0), (-1, -1), 9),
    ('GRID', (0, 0), (-1, -1), 0.5, colors.grey)
]))
doc.build([t])
print(f"Excel PDF Table created: {pdf_excel}")
```

## 5. Multi-Source Ingestion: JSON & XML Data to PDF Table

Parse structured `student.json` and `student.xml` files into nested lists for PDF generation.

```python
import json
import xml.etree.ElementTree as ET

# 1. Load JSON
with open('data/student.json', 'r', encoding='utf-8') as f:
    json_raw = json.load(f)

headers = list(json_raw[0].keys())
json_table_data = [headers]
for item in json_raw:
    json_table_data.append([item[h] for h in headers])

pdf_json = output_dir / 'table_from_json.pdf'
doc_json = SimpleDocTemplate(str(pdf_json), pagesize=letter)
t_json = Table(json_table_data, rowHeights=18, repeatRows=1)
t_json.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor('#17a2b8')),
    ('TEXTCOLOR', (0, 0), (-1, 0), colors.white),
    ('FONTSIZE', (0, 0), (-1, -1), 9)
]))
doc_json.build([t_json])
print(f"JSON PDF Table created: {pdf_json}")

# 2. Load XML
tree = ET.parse('data/student.xml')
root = tree.getroot()

xml_table_data = [['id', 'name', 'class', 'mark', 'gender']]
for st in root.findall('student'):
    xml_table_data.append([
        st.find('id').text if st.find('id') is not None else '',
        st.find('name').text if st.find('name') is not None else '',
        st.find('class').text if st.find('class') is not None else '',
        st.find('mark').text if st.find('mark') is not None else '',
        st.find('gender').text if st.find('gender') is not None else ''
    ])

pdf_xml = output_dir / 'table_from_xml.pdf'
doc_xml = SimpleDocTemplate(str(pdf_xml), pagesize=letter)
t_xml = Table(xml_table_data, rowHeights=18, repeatRows=1)
t_xml.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor('#6c757d')),
    ('TEXTCOLOR', (0, 0), (-1, 0), colors.white),
    ('FONTSIZE', (0, 0), (-1, -1), 9)
]))
doc_xml.build([t_xml])
print(f"XML PDF Table created: {pdf_xml}")
```

## 6. Multi-Source Ingestion: SQL Dump File into In-Memory SQLite

Execute `student.sql` inside an in-memory SQLite database, fetch records, and render into a multi-page PDF table.

```python
import sqlite3

conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

with open('data/student.sql', 'r', encoding='utf-8') as f:
    sql_script = f.read()

cursor.executescript(sql_script)
cursor.execute("SELECT * FROM student")

headers = [desc[0] for desc in cursor.description]
sql_data = [headers] + cursor.fetchall()

pdf_sql = output_dir / 'table_from_sql_dump.pdf'
doc_sql = SimpleDocTemplate(str(pdf_sql), pagesize=letter)
t_sql = Table(sql_data, rowHeights=18, repeatRows=1, colWidths=[0.5*inch, 1.5*inch, 1*inch, 1*inch, 1*inch])
t_sql.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor('#fd7e14')),
    ('TEXTCOLOR', (0, 0), (-1, 0), colors.white),
    ('FONTSIZE', (0, 0), (-1, -1), 9),
    ('GRID', (0, 0), (-1, -1), 0.5, colors.grey)
]))
doc_sql.build([t_sql])
print(f"SQL Dump PDF Table created: {pdf_sql}")
```
