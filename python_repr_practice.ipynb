### 1. Basic Official Representation
Inspect string escapes and collection formatting with repr().

```python
text = "Hello\nWorld"
print("str(): ", str(text))
print("repr():", repr(text))
```

### 2. The eval() Round-Trip
Reconstruct original objects using `eval(repr(obj))`.

```python
data = [1, "apple", {"valid": True}]
reconstructed = eval(repr(data))
print("Equal to original:", reconstructed == data)
assert reconstructed == data
```

### 3. Date & Time Representation: str() vs. repr()
Compare human-facing date strings against unambiguous constructor calls.

```python
import datetime
dt = datetime.datetime(2026, 10, 15, 8, 30)
print("str(dt): ", str(dt))
print("repr(dt):", repr(dt))
```

### 4. Custom Class __repr__() Implementation
Define constructor-like representations for custom objects.

```python
class Coordinate:
    def __init__(self, lat, lon):
        self.lat = lat
        self.lon = lon
    def __repr__(self):
        return f"Coordinate(lat={self.lat}, lon={self.lon})"

c = Coordinate(17.3850, 78.4867)
print("repr(c):", repr(c))
assert repr(c) == "Coordinate(lat=17.385, lon=78.4867)"
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying repr() mechanics.

```python
assert repr('test') == "'test'"
assert repr(100) == "100"
assert eval(repr({1, 2, 3})) == {1, 2, 3}

print("All repr() unit assertions passed! 100% verified.")
```
