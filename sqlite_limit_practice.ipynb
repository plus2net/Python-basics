## 2. Basic LIMIT Query (`sqlite3`)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT id, name, class, mark, sex FROM student LIMIT 0, 5"
my_cursor = my_conn.execute(q)
for row in my_cursor:
    print(row)
my_conn.close()
```

## 3. Using LIMIT with Exception Handling

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT * FROM student LIMIT 10, 5"
try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 4. Parameterized LIMIT Queries

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = (15, 5)
q = "SELECT * FROM student LIMIT ?, ?"
try:
    my_cursor = my_conn.execute(q, my_data)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 5. Modern SQLAlchemy 2.0+ Context Manager with LIMIT

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

my_conn = create_engine("sqlite:///my_db.db")
q = text("SELECT id, name, class, mark, sex FROM student LIMIT :start, :limit")

try:
    with my_conn.connect() as connection:
        r_set = connection.execute(q, {"start": 0, "limit": 5})
        for row in r_set:
            print(row)
except SQLAlchemyError as e:
    print("SQLAlchemy error:", e)
```
