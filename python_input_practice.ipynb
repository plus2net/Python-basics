### 1. Basic String Input & Type Verification

```python
import io, sys
orig_stdin = sys.stdin
sys.stdin = io.StringIO('Alex\n')
user_val = input('Enter username: ')
print('Received:', user_val)
print('Type:    ', type(user_val).__name__)
sys.stdin = orig_stdin
```

### 2. Numerical Type Casting: int() & float()

```python
import io, sys
orig_stdin = sys.stdin
sys.stdin = io.StringIO('5\n19.99\n')
qty = int(input('Enter quantity: '))
price = float(input('Enter unit price: '))
total = qty * price
print(f'Total: USD {total:.2f}')
sys.stdin = orig_stdin
```

### 3. Reading Multiple Space-Separated Values

```python
import io, sys
orig_stdin = sys.stdin
sys.stdin = io.StringIO('10 20 30 40\n')
nums = list(map(int, input('Enter numbers: ').split()))
print('Parsed list:', nums)
print('Sum:        ', sum(nums))
sys.stdin = orig_stdin
```

### 4. Robust Validation Loop Pattern

```python
import io, sys
orig_stdin = sys.stdin
sys.stdin = io.StringIO('bad\n-5\n25\n')
def get_positive_int():
    while True:
        raw = input('Enter a positive integer: ').strip()
        try:
            val = int(raw)
            if val > 0: return val
            print('Must be positive.')
        except ValueError:
            print('Invalid integer.')
valid_num = get_positive_int()
print('Validated positive integer:', valid_num)
sys.stdin = orig_stdin
```

### 5. Self-Checking Unit Verification Suite

```python
import io, sys
orig_stdin = sys.stdin
sys.stdin = io.StringIO('hello\n100\n')
t1 = input()
t2 = input()
assert isinstance(t1, str) and t1 == 'hello'
assert isinstance(t2, str) and t2 == '100'
sys.stdin = io.StringIO('42\n')
assert int(input()) == 42
sys.stdin = io.StringIO('1 2 3\n')
assert list(map(int, input().split())) == [1, 2, 3]
sys.stdin = orig_stdin
print('All input() unit assertions successfully verified! 100% passed.')
```
