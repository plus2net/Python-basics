## 2. Basic LIKE Query with Wildcards (`sqlite3`)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = "SELECT * FROM student WHERE name LIKE '%John'"
try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 3. Parameterized LIKE Queries

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = ('%john%',)
q = "SELECT * FROM student WHERE name LIKE ?"
try:
    my_cursor = my_conn.execute(q, my_data)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 4. Matching Single Characters using Underscore (`_`)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = ('9_',)
q = "SELECT * FROM student WHERE mark LIKE ?"
try:
    my_cursor = my_conn.execute(q, my_data)
    for row in my_cursor:
        print(row)
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 5. Case Sensitive LIKE via PRAGMA

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = ('b%n',)
q = "SELECT * FROM student WHERE name LIKE ?"
try:
    my_conn.execute('PRAGMA case_sensitive_like = ON')
    my_cursor = my_conn.execute(q, my_data)
    print("Rows with case sensitive ON:", len(my_cursor.fetchall()))
    my_conn.execute('PRAGMA case_sensitive_like = OFF')
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 6. Modern SQLAlchemy 2.0+ Context Manager with LIKE

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

my_conn = create_engine("sqlite:///my_db.db")
q = text("SELECT * FROM student WHERE name LIKE :pattern")

try:
    with my_conn.connect() as connection:
        r_set = connection.execute(q, {"pattern": "%John%"})
        for row in r_set:
            print(row)
except SQLAlchemyError as e:
    print("SQLAlchemy error:", e)
```
