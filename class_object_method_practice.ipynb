### 1. Defining Classes and Objects

```python
class Student:
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark

    def display(self):
        print(f"Student: {self.name}, Mark: {self.mark}")

s1 = Student("John Deo", 75)
s1.display()
```
