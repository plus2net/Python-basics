## 2. Traversing Downwards (.children vs .contents)

Inspect child nodes and filter out white-space text nodes.

```python
from bs4 import BeautifulSoup

html_doc = '''<div>
  <ul>
    <li>Item 1</li>
    <li>Item 2</li>
  </ul>
</div>'''

soup = BeautifulSoup(html_doc, 'html.parser')

print("Child Tag Names:", [child.name for child in soup.ul.children if child.name])
```

## 3. Traversing Upwards (.parent vs .parents)

Traverse ancestor nodes from an inner tag up to the document root.

```python
first_li = soup.find('li')
print("Immediate Parent:", first_li.parent.name)
print("All Ancestors:", [p.name for p in first_li.parents])
```
