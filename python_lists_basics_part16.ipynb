## 1. What Is a Python List?

A Python list stores multiple values in one variable.

Lists are:

- **Ordered**: items keep their position.
- **Mutable**: existing items can be changed.
- **Duplicate-friendly**: the same value can appear more than once.
- **Flexible**: a list can contain values of different data types.

````python
students = ["Alex", "Ronald", "John"]
print(students)
````

**Expected output**

```text
['Alex', 'Ronald', 'John']
```

The three names are stored in one list variable named `students`.

## 2. Creating Empty Lists

We can create an empty list using square brackets or the `list()` constructor.

````python
list_1 = []
list_2 = list()

print(list_1)
print(list_2)
````

**Expected output**

```text
[]
[]
```

Both statements create empty lists.

## 3. Lists Can Store Different Data Types and Duplicate Values

A list can contain strings, integers, floating-point values, Boolean values, and other objects.

````python
data = ["Python", 25, 9.5, True, "Python"]

print(data)
````

**Expected output**

```text
['Python', 25, 9.5, True, 'Python']
```

The value `"Python"` appears twice. Lists allow duplicate values.

## 4. Accessing List Items by Index

Python uses **zero-based indexing**.

For this list:

```text
Index:      0         1        2
Value:   'Alex'   'Ronald'   'John'
```

the first item is at index `0`.

````python
students = ["Alex", "Ronald", "John"]

print(students[0])
print(students[1])
print(students[2])
````

**Expected output**

```text
Alex
Ronald
John
```

## 5. Negative Indexing

Negative indexes access items from the end of the list.

- `-1` means the last item.
- `-2` means the second item from the end.

````python
students = ["Ronald", "John", "King", "Ravi", "Alex"]

print(students[-1])
print(students[-2])
````

**Expected output**

```text
Alex
Ravi
```

## 6. List Slicing

Slicing extracts a range of items.

The general form is:

```python
my_list[start:stop]
```

The `start` index is included. The `stop` index is not included.

````python
students = ["Ronald", "John", "King", "Ravi", "Alex"]

print(students[0:])
print(students[:3])
print(students[1:4])
````

**Expected output**

```text
['Ronald', 'John', 'King', 'Ravi', 'Alex']
['Ronald', 'John', 'King']
['John', 'King', 'Ravi']
```

`students[1:4]` starts at index `1` and stops before index `4`.

### Slicing Beyond the Last Position

A direct index that does not exist raises `IndexError`, but a slice can extend beyond the end of the list. Python simply returns the available items.

````python
students = ["Ronald", "John", "King", "Ravi", "Alex"]

print(students[1:8])
````

**Expected output**

```text
['John', 'King', 'Ravi', 'Alex']
```

## 7. Changing a List Item

Lists are mutable, so an existing item can be replaced by assigning a new value to its index.

````python
students = ["Alex", "Ronald", "John"]

students[1] = "Ravi"

print(students)
````

**Expected output**

```text
['Alex', 'Ravi', 'John']
```

The item at index `1` changed from `"Ronald"` to `"Ravi"`.

## 8. Displaying All List Items with a Loop

A `for` loop can read the values one at a time.

````python
students = ["Alex", "Ronald", "John"]

for name in students:
    print(name)
````

**Expected output**

```text
Alex
Ronald
John
```

The loop takes one value from the list during each iteration.

## 9. Creating a List from range()

`range()` produces a sequence of integers. Passing it to `list()` creates a list.

````python
numbers = list(range(5, 50, 10))

print(numbers)
````

**Expected output**

```text
[5, 15, 25, 35, 45]
```

The range starts at `5`, increases by `10`, and stops before `50`.

## 10. Unpacking List Elements

The `*` operator can unpack list items when calling `print()`.

````python
students = ["Alex", "Ronald", "John"]

print(students)
print(*students)
````

**Expected output**

```text
['Alex', 'Ronald', 'John']
Alex Ronald John
```

The first statement prints the list representation. The second prints the individual elements.

## 11. Searching for an Item with `in`

Use `in` to test whether a value is present in a list.

````python
students = ["Alex", "Ronald", "John"]

if "John" in students:
    print("Yes, included")
else:
    print("No, not included")
````

**Expected output**

```text
Yes, included
```

### Try It Yourself

## 12. Counting Items with len()

`len()` returns the number of items in a list.

````python
students = ["Alex", "Ronald", "John"]

print(len(students))
````

**Expected output**

```text
3
```

The list contains three items.

## 13. Finding Maximum and Minimum Values

For a numeric list, Python's `max()` function returns the largest value and `min()` returns the smallest value.

````python
my_list = [4, 2, 8, 6]

print("Maximum value:", max(my_list))
print("Minimum value:", min(my_list))
````

**Expected output**

```text
Maximum value: 8
Minimum value: 2
```

`max()` compares the numeric values and returns the highest one. `min()` returns the lowest one.

## 14. Finding the Mean of List Values

The mean is the sum of all numeric values divided by the number of values.

We can calculate it using `sum()` and `len()`.

````python
my_list = [4, 2, 8, 6]

mean_value = sum(my_list) / len(my_list)

print("Mean value:", mean_value)
````

**Expected output**

```text
Mean value: 5.0
```

The total is `20`, and there are `4` values, so the mean is `20 / 4 = 5.0`.

### Using `statistics.mean()`

Python also provides the `statistics` module. Its `mean()` function calculates the arithmetic mean directly.

````python
import statistics

my_list = [4, 2, 8, 6]

print("Mean value:", statistics.mean(my_list))
````

**Expected output**

```text
Mean value: 5
```

Both approaches are valid. `statistics.mean()` is convenient when you want a dedicated function for calculating the arithmetic mean.

## 15. A First Look at Common List Methods

Python provides methods for adding, removing, finding, copying, reversing, and sorting list items.

Some important methods are:

- `append(x)` - add one item at the end.
- `clear()` - remove all items.
- `copy()` - return a shallow copy.
- `count(x)` - count matching values.
- `extend(iterable)` - add items from another iterable.
- `index(x)` - return the position of a matching value.
- `insert(i, x)` - insert an item at a position.
- `pop([i])` - remove and return an item.
- `remove(x)` - remove the first matching value.
- `reverse()` - reverse the list in place.
- `sort()` - sort the list in place.

The Plus2net Lists tutorial contains dedicated pages and examples for these methods.

````python
languages = ["Python", "PHP"]

languages.append("JavaScript")

print(languages)
````

**Expected output**

```text
['Python', 'PHP', 'JavaScript']
```

This is only a first example. The individual list-method tutorials explain each method in more detail.

## Key Takeaways

- Python lists store multiple values in one variable.
- Lists are ordered and mutable.
- Indexing starts at `0`.
- Negative indexes start from the end of the list.
- A slice excludes its stop index.
- Slicing beyond the end of a list is allowed.
- Existing list items can be changed by index.
- Lists can contain duplicate values and mixed data types.
- `for` loops can process list values one by one.
- `in` checks membership and `len()` returns the number of items.
