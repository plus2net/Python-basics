### 1. Basic Invocability Checks
`callable(object)` returns `True` if an object can be called with `()`, otherwise `False`.

```python
def greet():
    return "Hello!"

num = 42
text = "Python"

print("greet is callable:", callable(greet))  # True
print("num is callable:  ", callable(num))    # False
print("text is callable: ", callable(text))   # False
```

### 2. Classes vs. Class Instances
Classes are callable constructors. Standard instances are not callable unless they define `__call__`.

```python
class User:
    def __init__(self, username):
        self.username = username

print("User class callable:", callable(User))  # True

u = User("alice")
print("User instance callable:", callable(u))  # False
```

### 3. Implementing __call__() for Stateful Callables
Make instances callable to maintain state across calls.

```python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor
    def __call__(self, value):
        return value * self.factor

double = Multiplier(2)
print("double instance is callable:", callable(double))  # True
print("double(15) =", double(15))  # 30
```

### 4. Bound Methods vs. Instance
Methods on instances are callable even when the instance itself is not.

```python
class MathService:
    def add(self, a, b):
        return a + b

svc = MathService()
print("Service instance callable:", callable(svc))      # False
print("Bound method add callable:", callable(svc.add))  # True
```

### 5. Production Recipe: Callback Registry
Accept flexible functions or callable objects in an event pipeline.

```python
def execute_pipeline(items, step_callable):
    if not callable(step_callable):
        raise TypeError("Expected a callable processing step")
    return [step_callable(x) for x in items]

names = ["  admin ", "user_1 ", " guest"]
print("Strip:", execute_pipeline(names, str.strip))
print("Upper:", execute_pipeline(names, lambda s: s.strip().upper()))
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying callable() mechanics.

```python
class CallableVerifier:
    @staticmethod
    def test_all():
        class A:
            pass
        class B:
            def __call__(self):
                return 1
                
        assert callable(A) is True
        assert callable(A()) is False
        assert callable(B) is True
        assert callable(B()) is True
        assert callable(lambda x: x) is True
        assert callable("hello") is False
        assert callable(42) is False
        
        print("All callable() unit assertions passed successfully! 100% verified.")

CallableVerifier.test_all()
```
