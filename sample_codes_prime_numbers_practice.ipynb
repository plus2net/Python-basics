## Example 1: First working example

Read the code, predict the output and run this cell.

```python
n = 50
for value in range(2, n + 1):
    prime = True
    for divisor in range(2, value):
        if value % divisor == 0:
            prime = False
            break
    if prime:
        print(value, end=" ")
```

## Example 2: Original half-range technique

Read the code, predict the output and run this cell.

```python
n = 50
for i in range(2, n + 1):
    flag = 0
    for j in range(2, i // 2 + 1):
        if i % j == 0:
            flag = 1
            break
    if flag == 0:
        print(i, end=", ")
```

## Example 3: Check if a single number is prime

Read the code, predict the output and run this cell.

```python
from unittest.mock import patch
with patch("builtins.input", side_effect=['1']):
    n = int(input("Enter number: "))
    if n < 2:
        print("Not prime")
    else:
        for i in range(2, n):
            if n % i == 0:
                print("Not prime")
                break
        else:
            print("Prime")
```

## Example 4: Faster primality test with math.isqrt()

Read the code, predict the output and run this cell.

```python
from math import isqrt
def is_prime(n):
    if n < 2:
        return False
    for divisor in range(2, isqrt(n) + 1):
        if n % divisor == 0:
            return False
    return True

for n in (0, 1, 2, 17, 25, 97):
    print(n, is_prime(n))
```

## Example 5: Print the first ten prime numbers

Read the code, predict the output and run this cell.

```python
from math import isqrt
def is_prime(n):
    return n >= 2 and all(n % d for d in range(2, isqrt(n) + 1))
count = 0
value = 2
while count < 10:
    if is_prime(value):
        print(value, end=" ")
        count += 1
    value += 1
```

## Example 6: Generate primes in a range

Read the code, predict the output and run this cell.

```python
from math import isqrt
def is_prime(n):
    return n >= 2 and all(n % d for d in range(2, isqrt(n) + 1))
primes = [n for n in range(20, 51) if is_prime(n)]
print(primes)
```

## Example 7: Use the Sieve of Eratosthenes

Read the code, predict the output and run this cell.

```python
limit = 40
sieve = [True] * (limit + 1)
sieve[0:2] = [False, False]
for p in range(2, int(limit ** 0.5) + 1):
    if sieve[p]:
        for multiple in range(p * p, limit + 1, p):
            sieve[multiple] = False
print([n for n, prime in enumerate(sieve) if prime])
```
