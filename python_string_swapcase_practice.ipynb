### 1. Basic str.swapcase() Operation
`str.swapcase()` converts uppercase to lowercase and lowercase to uppercase.

```python
s1 = "Hello World"
s2 = "Python 3.14 is AWESOME!"
print("s1.swapcase():", s1.swapcase())
print("s2.swapcase():", s2.swapcase())

assert s1.swapcase() == "hELLO wORLD"
assert s2.swapcase() == "pYTHON 3.14 IS awesome!"
```

### 2. The Unicode Non-Reversibility Gotcha
Observe why `s.swapcase().swapcase() == s` fails for German `ß`.

```python
word = "Straße"
step1 = word.swapcase()
step2 = step1.swapcase()

print("Original:       ", word)
print("1st swapcase:   ", step1)  # 'sTRASSE'
print("2nd swapcase:   ", step2)  # 'Strasse'
print("Is reversible?  ", word == step2)

assert word != step2
assert step1 == "sTRASSE"
assert step2 == "Strasse"
```

### 3. Comparison with upper(), lower(), and title()
Compare case transformation functions across inputs.

```python
samples = ["Hello World", "PyThOn", "Straße", "123"]
print(f"{'Input':<12} | {'swapcase':<12} | {'upper':<10} | {'lower':<10}")
print("-" * 50)
for s in samples:
    print(f"{s:<12} | {s.swapcase():<12} | {s.upper():<10} | {s.lower():<10}")

assert "PyThOn".swapcase() == "pYtHoN"
```

### 4. Accidental Caps Lock Error Correction
Fix words accidentally typed with Caps Lock enabled (e.g. `tHIS` -> `This`).

```python
def fix_caps_lock(text):
    words = text.split(' ')
    out = []
    for w in words:
        if len(w) > 1 and w[0].islower() and w[1:].isupper():
            out.append(w.swapcase())
        else:
            out.append(w)
    return ' '.join(out)

err = "tHIS IS a wELCOME MESSAGE"
corrected = fix_caps_lock(err)
print("Corrected text:", corrected)
assert corrected == "This IS a Welcome MESSAGE"
```

### 5. Practice Exercise: Case Correction Engine Suite
Run the complete engine with unit assertions.

```python
class CaseCorrectionEngine:
    @staticmethod
    def invert(text):
        return text.swapcase() if isinstance(text, str) else ""

    @staticmethod
    def is_reversible(text):
        return isinstance(text, str) and text.swapcase().swapcase() == text

    @staticmethod
    def fix_caps_lock(sentence):
        if not isinstance(sentence, str):
            return ""
        return " ".join(
            w.swapcase() if len(w) > 1 and w[0].islower() and w[1:].isupper() else w
            for w in sentence.split(" ")
        )

# Unit Assertions
eng = CaseCorrectionEngine()
assert eng.invert("Alpha") == "aLPHA"
assert eng.invert("123") == "123"
assert eng.is_reversible("Hello") is True
assert eng.is_reversible("Straße") is False
assert eng.fix_caps_lock("tHIS IS a wELCOME") == "This IS a Welcome"

print("All CaseCorrectionEngine unit assertions passed successfully!")
```
