## Example 5: Unpack the year, month and day

```python
import calendar
cal = calendar.Calendar()
for year, month, day in cal.itermonthdays3(2020, 7):
    if month == 7 and day <= 5:
        print(f'{year}-{month:02d}-{day:02d}')
```

## Example 6: Identify adjacent-month tuples

## Example 8: Organize tuples into complete weeks
Run the code below and inspect its printed output.

```python
import calendar
slots = list(calendar.Calendar().itermonthdays3(2020, 7))
weeks = [slots[i:i + 7] for i in range(0, len(slots), 7)]
for number, week in enumerate(weeks, start=1):
    print(f'Week {number}: {week[0]} to {week[-1]}')
```

## Example 11: Find weekday names

## Example 18: Make a simple weekday task plan
Run the code below and inspect its printed output.

```python
import calendar
from datetime import date
plan = []
for year, month, day in calendar.Calendar().itermonthdays3(2020, 7):
    if (year, month) == (2020, 7) and day <= 8:
        current = date(year, month, day)
        if current.weekday() < 5:
            plan.append(f'{current.isoformat()} ({current:%a}): Review tasks')
for item in plan:
    print(item)
```

## Exercise 7 — Count Monday–Friday workdays
Exclude adjacent-month dates and weekends. Do not account for public holidays.

**Challenge:** Edit the example, predict the result, then run it. The final `assert` lines check the expected answer.

```python
import calendar
from datetime import date
actual = [date(y, m, d)
          for y, m, d in calendar.Calendar().itermonthdays3(2020, 7)
          if (y, m) == (2020, 7)]
workdays = [d for d in actual if d.weekday() < 5]
print('Monday–Friday dates:', len(workdays))
assert len(workdays) == 23
```
