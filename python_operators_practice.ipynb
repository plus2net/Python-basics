### 1. Arithmetic Operators & Floor Division

```python
a, b = 29, 5
print("True division (/):  ", a / b)
print("Floor division (//):", a // b)
print("Modulus remainder:  ", a % b)
print("Negative floor:     ", -29 // 5)
```

### 2. Comparison & Chained Tests

```python
x = 20
print("Chained comparison (10 < x < 30):", 10 < x < 30)
print("Equality test (x == 20):         ", x == 20)
```

### 3. Logical Short-Circuiting

```python
env_host = None
active_host = env_host or "localhost"
print("Resolved host:", active_host)
```

### 4. Identity (is) vs Equality (==)

```python
list_a = [1, 2, 3]
list_b = [1, 2, 3]
print("list_a == list_b (Value equality):  ", list_a == list_b)
print("list_a is list_b (Memory identity):", list_a is list_b)
```

### 5. Bitwise Permission Flags

```python
READ = 1 << 0   # 1
WRITE = 1 << 1  # 2
EXEC = 1 << 2   # 4

perms = READ | WRITE
print("Has Read: ", bool(perms & READ))
print("Has Exec: ", bool(perms & EXEC))
```
