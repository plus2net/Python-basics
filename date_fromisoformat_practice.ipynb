# Python `date.fromisoformat()` — Practice Notebook

**Tutorial:** [Plus2Net date.fromisoformat()](https://www.plus2net.com/python/date-fromisoformat.php)  
**Run in:** Google Colab or Jupyter Notebook  
**Requirements:** Python 3.11+ for the compact and ISO-week examples; most examples also work on Python 3.7+. No third-party libraries.

This notebook teaches how to turn ISO date strings into real `date` objects, handle invalid input, compare dates, and choose between `fromisoformat()` and `strptime()`.

## Example 1: Convert an ISO date string to a Python date

Run the code and inspect the result.

```python
from datetime import date
dt = date.fromisoformat('2019-09-22')
print(dt)
```

## Example 2: Inspect the result type

Run the code and inspect the result.

```python
from datetime import date
parsed = date.fromisoformat('2024-02-29')
print(type(parsed).__name__)
print(parsed.year, parsed.month, parsed.day)
```

## Example 3: Original today example

Run the code and inspect the result.

*Note: the output depends on the day this cell is run.*

```python
from datetime import date
dt = date.fromisoformat(date.today().strftime('%Y-%m-%d'))
print(dt)
```

## Example 4: Format today only when needed

Run the code and inspect the result.

*Note: the output depends on the day this cell is run.*

```python
from datetime import date
now = date.today()
print(now.isoformat())
```

## Example 5: Read year/month/day

Run the code and inspect the result.

```python
from datetime import date
order_date = date.fromisoformat('2024-12-25')
print('Year:', order_date.year)
print('Month:', order_date.month)
print('Day:', order_date.day)
```

## Example 6: Original invalid-date example

Run the code and inspect the result.

```python
from datetime import date
try:
    dt = date.fromisoformat('2019-02-30')  # Invalid date
except ValueError as e:
    print(f"Error: {e}")
```

## Example 7: Leap year example

Run the code and inspect the result.

```python
from datetime import date
for value in ('2024-02-29', '2023-02-29'):
    try:
        print(value, '=>', date.fromisoformat(value))
    except ValueError:
        print(value, '=> invalid date')
```

## Example 8: Original method comparison

Run the code and inspect the result.

```python
from datetime import date, datetime

# Using fromisoformat
print(date.fromisoformat('2023-09-12'))

# Using strptime for the same result
print(datetime.strptime('2023-09-12', '%Y-%m-%d').date())
```

## Example 9: Parse a slash-separated date

Run the code and inspect the result.

```python
from datetime import datetime
text = '25/12/2024'
parsed = datetime.strptime(text, '%d/%m/%Y').date()
print(parsed)
```

## Example 10: ISO timestamp and timezone offset

Run the code and inspect the result.

```python
from datetime import datetime
stamp = datetime.fromisoformat('2024-06-17T14:30:00+05:30')
print(stamp.date())
print(stamp.time())
print(stamp.utcoffset())
```

## Example 11: Date round trip

Run the code and inspect the result.

```python
from datetime import date
original = date(2024, 3, 15)
text = original.isoformat()
restored = date.fromisoformat(text)
print(text)
print(original == restored)
```

## Example 14: Check unsupported inputs

Run the code and inspect the result.

```python
from datetime import date
for value in ('2024-06', '2024/06/17', '2024-169'):
    try:
        date.fromisoformat(value)
        print(value, ': accepted')
    except ValueError:
        print(value, ': unsupported')
```

## Example 15: Parse a submitted field

Run the code and inspect the result.

```python
from datetime import date
submitted_value = '2024-09-15'
try:
    chosen_date = date.fromisoformat(submitted_value)
    print('Accepted date:', chosen_date)
except ValueError:
    print('Please enter a valid date in YYYY-MM-DD format.')
```

## Example 16: Sort a list of ISO dates

Run the code and inspect the result.

```python
from datetime import date
values = ['2024-12-31', '2024-01-03', '2024-06-15']
ordered = sorted(date.fromisoformat(x) for x in values)
for value in ordered:
    print(value)
```

## Example 17: Filter event dates

Run the code and inspect the result.

```python
from datetime import date
cutoff = date.fromisoformat('2024-06-01')
events = ['2024-05-20', '2024-06-15', '2024-07-01']
for value in events:
    when = date.fromisoformat(value)
    if when >= cutoff:
        print(value)
```

## Example 18: Days between two parsed dates

Run the code and inspect the result.

```python
from datetime import date
start = date.fromisoformat('2024-01-01')
finish = date.fromisoformat('2024-01-10')
print((finish - start).days, 'days')
```

## Example 19: Validate dates from data rows

Run the code and inspect the result.

```python
from datetime import date
rows = [('A101', '2024-02-29'), ('A102', '2024-02-30')]
for reference, text in rows:
    try:
        parsed = date.fromisoformat(text)
        print(reference, parsed)
    except ValueError:
        print(reference, 'invalid date:', text)
```

## Example 20: Trim incidental surrounding whitespace

Run the code and inspect the result.

```python
from datetime import date
raw = '  2024-06-17  '
print(date.fromisoformat(raw.strip()))
```

The following assertions provide a quick check. Try solving the exercises independently first.

```python
from datetime import date, datetime

# Challenge 1
leap_day = date.fromisoformat('2024-02-29')
assert (leap_day.year, leap_day.month, leap_day.day) == (2024, 2, 29)

# Challenge 2
try:
    date.fromisoformat('2023-02-29')
except ValueError:
    invalid_was_rejected = True
else:
    invalid_was_rejected = False
assert invalid_was_rejected

# Challenge 3
new_years_eve = datetime.strptime('31/12/2024', '%d/%m/%Y').date()
assert new_years_eve.isoformat() == '2024-12-31'

# Challenge 4
days = (date.fromisoformat('2025-01-01') - date.fromisoformat('2024-12-25')).days
assert days == 7

# Challenge 5
values = ['2024-12-31', '2024-01-01', '2024-06-15']
ordered = sorted(date.fromisoformat(x) for x in values)
assert [x.isoformat() for x in ordered] == ['2024-01-01', '2024-06-15', '2024-12-31']

print('All five practice checks passed!')
```
