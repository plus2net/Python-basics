## 2. Create Database & Table with BLOB Column

```python
import sqlite3

my_conn = sqlite3.connect('student_blob_2.db')
try:
    my_conn.execute('''
    CREATE TABLE IF NOT EXISTS student_b(
        id INTEGER PRIMARY KEY,
        name TEXT,
        class TEXT,
        gender TEXT,
        photo BLOB
    );''')
    print("student_b Table created successfully")
except sqlite3.Error as my_error:
    print("error:", my_error)
my_conn.close()
```

## 3. Insert Records with BLOB Image Data

```python
import sqlite3

my_conn = sqlite3.connect('student_blob_2.db')
my_students = [
    [1, 'Alex J', 'Four', 'Female', '1.png'],
    [2, 'Ramana K', 'Three', 'Male', '2.png'],
    [3, 'Jack', 'Five', 'Female', '3.png'],
    [4, 'Ronne', 'Six', 'Female', '4.png']
]

for ids in my_students:
    with open(ids[4], 'rb') as fob:
        blob_data = fob.read()
    my_data = (None, ids[1], ids[2], ids[3], blob_data)
    q = "INSERT INTO student_b VALUES (?, ?, ?, ?, ?)"
    try:
        r_set = my_conn.execute(q, my_data)
        my_conn.commit()
        print(f"Record added for {ids[1]}, rows affected:", r_set.rowcount)
    except sqlite3.Error as my_error:
        print("error:", my_error)
my_conn.close()
```

## 4. Retrieve & Display Images using PIL and Matplotlib

```python
import sqlite3
from PIL import Image
import io
import matplotlib.pyplot as plt

my_conn = sqlite3.connect('student_blob_2.db')
r_set = my_conn.execute('SELECT * FROM student_b')
for row in r_set:
    image = Image.open(io.BytesIO(row[4]))
    plt.figure(figsize=(1, 1))
    plt.axis(False)
    plt.imshow(image)
    print(f"ID: {row[0]}, Name: {row[1]}, Class: {row[2]}, Gender: {row[3]}")
    plt.show()
my_conn.close()
```

## 5. Modern SQLAlchemy 2.0+ BLOB Storage Workflow

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

my_conn = create_engine("sqlite:///student_blob_2.db")
try:
    with my_conn.begin() as connection:
        connection.execute(text("""CREATE TABLE IF NOT EXISTS student_sa (
            id INTEGER PRIMARY KEY,
            name TEXT,
            photo BLOB
        )"""))
        print("SQLAlchemy BLOB table created successfully!")
except SQLAlchemyError as e:
    print("SQLAlchemy error:", e)
```
