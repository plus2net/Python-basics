### 1. Basic str.casefold() Transformation
For ASCII text, `casefold()` produces standard lowercase strings.

```python
s = "Welcome To Plus2Net Python"
print("Original: ", s)
print("Casefolded:", s.casefold())

assert s.casefold() == "welcome to plus2net python"
assert "PYTHON".casefold() == "python".casefold()
```

### 2. German Eszett: lower() vs. casefold()
`'ß'.lower()` remains `'ß'`, whereas `'ß'.casefold()` expands to `'ss'`.

```python
w1 = "Straße"
w2 = "STRASSE"

print("w1.lower():          ", w1.lower())
print("w2.lower():          ", w2.lower())
print("lower() comparison:  ", w1.lower() == w2.lower())

print("w1.casefold():       ", w1.casefold())
print("w2.casefold():       ", w2.casefold())
print("casefold() comparison:", w1.casefold() == w2.casefold())

assert (w1.lower() == w2.lower()) is False
assert (w1.casefold() == w2.casefold()) is True
```

### 3. Greek Sigma Normalization
Both Greek lowercase sigmas ('ς' final and 'σ' middle) fold to 'σ'.

```python
final_sigma = 'ς'
middle_sigma = 'σ'

print(f"final_sigma.casefold():  {final_sigma.casefold()}")
print(f"middle_sigma.casefold(): {middle_sigma.casefold()}")

assert final_sigma.casefold() == middle_sigma.casefold() == 'σ'
```

### 4. Case-Insensitive Dictionary Implementation
A lookup dictionary using casefolded keys.

```python
class CaselessDict:
    def __init__(self):
        self._map = {}
    def __setitem__(self, k, v):
        self._map[k.casefold()] = v
    def __getitem__(self, k):
        return self._map[k.casefold()]
    def __contains__(self, k):
        return k.casefold() in self._map

headers = CaselessDict()
headers['Content-Type'] = 'application/json'
print("content-type lookup:", headers['content-type'])
assert headers['content-type'] == 'application/json'
assert 'CONTENT-TYPE' in headers
```

### 5. Practice Exercise: Multilingual Search Index
Run the international search index with self-checking assertions.

```python
class MultilingualSearchIndex:
    def __init__(self):
        self._documents = []

    def add_document(self, doc_id, text):
        self._documents.append({'id': doc_id, 'text': text, 'folded': text.casefold()})

    def search(self, query):
        if not query:
            return []
        q_folded = query.casefold()
        return [d['id'] for d in self._documents if q_folded in d['folded']]

# Unit Assertions
idx = MultilingualSearchIndex()
idx.add_document(1, 'Goethestraße Bakery, Munich')
idx.add_document(2, 'Greek Portal: Τεχνολογία')
idx.add_document(3, 'Python Standard Library')

# German sharp S search
assert idx.search('strasse') == [1]
assert idx.search('STRASSE') == [1]
assert idx.search('Straße') == [1]

# Greek search
assert idx.search('τεχνολογία') == [2]

# English search
assert idx.search('PYTHON') == [3]
assert idx.search('unknown') == []

print('All MultilingualSearchIndex assertions passed successfully!')
```
