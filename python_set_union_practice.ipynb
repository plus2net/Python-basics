### 1. Basic Set Union
`A.union(B)` and `A | B` return a brand-new set containing all distinct elements from both inputs without mutating the originals.

```python
A = {1, 2, 3}
B = {3, 4, 5}

# Using | operator
union_op = A | B
print("Union using | operator:", sorted(union_op))

# Using union() method
C = {'a', 'b', 'c'}
D = {'a', 'y', 'z'}
union_method = C.union(D)
print("Union using union():   ", sorted(union_method))
print("Original A untouched:   ", A == {1, 2, 3})
```

### 2. Multi-Set Chaining
Combine three or more collections in a single expressive call.

```python
s1 = {'a', 'b', 'c'}
s2 = {'a', 'y', 'z'}
s3 = {'a', 'k', 'l'}

merged = s1.union(s2, s3)
print("Merged 3 sets:", sorted(merged))

chained_op = s1 | s2 | s3
assert merged == chained_op
```

### 3. Heterogeneous Collections: Lists and Strings
`union()` accepts any iterable. The `|` operator requires set operands.

```python
primary = {'a', 'b', 'c'}

# union() with list
with_list = primary.union(['a', 'x', 'y'])
print("Union with list:", sorted(with_list))

# String character unpacking
with_str = primary.union('Alex')
print("Union with string:", sorted(with_str))

# | operator raises TypeError on list
try:
    err = primary | ['a', 'x']
except TypeError as e:
    print("Operator raised TypeError:", e)
```

### 4. Real-World Recipe: Role Entitlement Consolidation
Compute effective user permissions across hierarchical roles without altering underlying definitions.

```python
def get_effective_permissions(user_scopes: set, *groups) -> set:
    return user_scopes.union(*groups)

user_grants = {"view_dashboard", "export_csv"}
team_grants = ["edit_wiki", "export_csv"]
org_grants = ("access_vpn", "sso_login")

effective = get_effective_permissions(user_grants, team_grants, org_grants)
print("Effective permissions:", sorted(effective))
assert effective == {"view_dashboard", "export_csv", "edit_wiki", "access_vpn", "sso_login"}
```

### 5. Unit Assertion Suite
Validate deduplication, non-mutating behavior, and empty set union invariants.

```python
# Assertion 1: Deduplication and immutability of inputs
s_a = {1, 2, 3}
s_b = {3, 4, 5}
res = s_a.union(s_b)
assert res == {1, 2, 3, 4, 5}
assert s_a == {1, 2, 3}

# Assertion 2: Multi-collection union
res_multi = s_a.union([4, 5], (6, 7))
assert res_multi == {1, 2, 3, 4, 5, 6, 7}

# Assertion 3: Union with empty set is copy
clone = s_a.union(set())
assert clone == s_a
assert clone is not s_a

print("All set.union assertions passed successfully!")
```
