## Example 1: First working example

Read the code, predict the output and run this cell.

```python
for row in range(1, 6):
    for col in range(1, 6):
        print(f"{row * col:3}", end=" ")
    print()
```

## Example 2: Original 10 × 10 nested for-loop table

Read the code, predict the output and run this cell.

```python
rows = 10
columns = 10
for i in range(1, rows + 1):
    for j in range(1, columns + 1):
        c = i * j
        print("{:3d}".format(c), end=" ")
    print()
```

## Example 3: Create a table using while loops

Read the code, predict the output and run this cell.

```python
rows = 3
columns = 5
i = 1
while i <= rows:
    j = 1
    while j <= columns:
        print(f"{i*j:3}", end=" ")
        j += 1
    print()
    i += 1
```

## Example 5: Reusable table function

Read the code, predict the output and run this cell.

```python
def multiplication_table(number, up_to=5):
    return [number * i for i in range(1, up_to + 1)]
print(multiplication_table(9, 6))
```

## Example 6: Format a grid with aligned columns

Read the code, predict the output and run this cell.

```python
for i in range(1, 6):
    print(" ".join(f"{i*j:3d}" for j in range(1, 6)))
```

## Example 7: Validate row and column values

Read the code, predict the output and run this cell.

```python
rows, columns = 4, 0
if rows <= 0 or columns <= 0:
    print("Both dimensions must be positive")
else:
    print([[i * j for j in range(1, columns+1)] for i in range(1, rows+1)])
```
