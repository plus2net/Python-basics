### 1. Basic str.isalpha() Validation
`str.isalpha()` returns `True` if all characters are letters and length > 0.

```python
print("'Python'.isalpha():     ", 'Python'.isalpha())
print("'Welcome'.isalpha():    ", 'Welcome'.isalpha())
print("'plus2net'.isalpha():   ", 'plus2net'.isalpha())
print("'Hello World'.isalpha():", 'Hello World'.isalpha())
print("''.isalpha():           ", ''.isalpha())

assert 'Python'.isalpha() is True
assert 'plus2net'.isalpha() is False
assert 'Hello World'.isalpha() is False
assert ''.isalpha() is False
```

### 2. Unicode and International Alphabets
`str.isalpha()` supports international alphabetic writing systems (Latin accents, Cyrillic, Greek, CJK ideographs).

```python
international_samples = ['Café', 'München', 'España', 'Москва', 'Αθήνα', '日本語', '한국어']

for s in international_samples:
    print(f"{s:<10}: {s.isalpha()}")
    assert s.isalpha() is True
```

### 3. Comparison with isalnum() and isascii()
Examine behavior across alphanumeric strings, underscores, and non-ASCII letters.

```python
test_cases = ['Hello', 'Hello2', '2026', 'my_var', 'Café', 'Hello World']
print(f"{'String':<15} | {'isalpha':<8} | {'isalnum':<8} | {'isascii':<8}")
print("-" * 45)
for t in test_cases:
    print(f"{t:<15} | {str(t.isalpha()):<8} | {str(t.isalnum()):<8} | {str(t.isascii()):<8}")

assert 'Hello'.isalpha() and 'Hello'.isascii()
assert 'Café'.isalpha() and not 'Café'.isascii()
```

### 4. NLP Word Tokenization Filter
Filter raw sentence tokens into pure alphabetic vocabulary words.

```python
sentence = "In 2026, Python 3.14 delivers 25% faster asynchronous execution!"
words = [w.strip(".,!?:;") for w in sentence.split() if w.strip(".,!?:;").isalpha()]
print("Filtered words:", words)
assert 'Python' in words
assert '2026' not in words
```

### 5. Practice Exercise: Human Name & Lexical Token Validator
Run the complete validator testing multi-part names and token classification.

```python
class LexicalValidator:
    def validate_human_name(self, full_name):
        if not isinstance(full_name, str):
            return False
        trimmed = full_name.strip()
        if not trimmed:
            return False
        clean_letters = trimmed.replace(" ", "").replace("-", "").replace("'", "")
        return len(clean_letters) > 0 and clean_letters.isalpha()

    def classify_token(self, token):
        if not token:
            return "EMPTY"
        if token.isalpha():
            return "WORD"
        if token.isdigit():
            return "NUMBER"
        if token.isalnum():
            return "ALPHANUMERIC_CODE"
        return "MIXED_OR_PUNCTUATION"

# Unit Assertions
val = LexicalValidator()
assert val.validate_human_name("Alexander") is True
assert val.validate_human_name("Mary Jane O'Connor") is True
assert val.validate_human_name("Jean-Claude") is True
assert val.validate_human_name("User_123") is False
assert val.validate_human_name("") is False

assert val.classify_token("Algorithm") == "WORD"
assert val.classify_token("2026") == "NUMBER"
assert val.classify_token("Python3") == "ALPHANUMERIC_CODE"
assert val.classify_token("hello_world") == "MIXED_OR_PUNCTUATION"

print("All name validation and token classification assertions passed successfully!")
```
