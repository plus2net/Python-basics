# Python math.frexp(): Decompose Float into Mantissa and Exponent

### 1. Basic Decomposition with math.frexp()
`math.frexp(x)` decomposes any numeric value `x` into a tuple `(m, e)` where `x == m * (2 ** e)` and `0.5 <= abs(m) < 1.0`.


```python
import math

print("frexp(1.0): ", math.frexp(1.0))   # (0.5, 1)
print("frexp(4.0): ", math.frexp(4.0))   # (0.5, 3)
print("frexp(8.0): ", math.frexp(8.0))   # (0.5, 4)
print("frexp(0.125):", math.frexp(0.125)) # (0.5, -2)
print("frexp(10.0): ", math.frexp(10.0))  # (0.625, 4)
print("frexp(-5.0): ", math.frexp(-5.0))  # (-0.625, 3)
```

### 2. Mantissa Normalization Range [0.5, 1.0) & Log2 Relationship
Because the mantissa is constrained to `[0.5, 1.0)`, the exponent `e` is directly related to the base-2 logarithm: for `x > 0`, `e == math.floor(math.log2(x)) + 1`.


```python
import math

numbers = [0.25, 0.7, 1.0, 7.5, 1024.0]
print(f"{'x':>8} | {'frexp exponent':>15} | {'floor(log2(x)) + 1':>18} | Match?")
print("-" * 55)
for n in numbers:
    _, e = math.frexp(n)
    expected_e = math.floor(math.log2(n)) + 1
    assert e == expected_e
    print(f"{n:8.2f} | {e:15d} | {expected_e:18d} | {e == expected_e}")
```

### 3. Reconstructing Floats via math.ldexp()
`math.ldexp()` is the exact inverse of `math.frexp()`. Combining them allows bit-exact serialization and reconstruction without floating-point rounding error.


```python
import math

test_values = [0.125, 1.0, 13.375, 98765.4321, 1e12, -42.0]
for val in test_values:
    m, e = math.frexp(val)
    reconstructed = math.ldexp(m, e)
    assert reconstructed == val, f"Mismatch for {val}"
    print(f"Original: {val:12.4f} -> (m={m:10.6f}, e={e:3d}) -> Reconstructed: {reconstructed:12.4f}")
print("All round-trip reconstructions matched perfectly!")
```

### 4. Special Values: Zero, Signed Zero, Inf, and NaN
Examine how `math.frexp()` handles edge cases defined by the IEEE 754 standard.


```python
import math

print("frexp(0.0):  ", math.frexp(0.0))
print("frexp(-0.0): ", math.frexp(-0.0))
print("frexp(inf):  ", math.frexp(float("inf")))
print("frexp(nan):  ", math.frexp(float("nan")))
```

### 5. Hands-on Practice Exercise: Machine Precision & ULP Calculator
Unit in the Last Place (ULP) measures the resolution/spacing between adjacent 64-bit IEEE 754 floats. Given 53 bits of precision, `ULP(x) == 2**(e - 53) == math.ldexp(1.0, e - 53)`.

Implement `calculate_ulp(x)` and verify with self-checking assertions.


```python
import math

def calculate_ulp(x):
    """
    Calculates the Unit in the Last Place (ULP) for a 64-bit IEEE 754 float x.
    """
    if not isinstance(x, (int, float)):
        raise TypeError("Input must be a numeric value.")
    if math.isnan(x) or math.isinf(x):
        return float("nan")
    if x == 0.0:
        return math.ldexp(1.0, -1074)
    _, e = math.frexp(x)
    return math.ldexp(1.0, e - 53)

# Unit Assertions
eps = 2.220446049250313e-16  # 2**-52
assert math.isclose(calculate_ulp(1.0), eps)
assert math.isclose(calculate_ulp(2.0), eps * 2)
assert math.isclose(calculate_ulp(0.5), eps / 2)
assert math.isclose(calculate_ulp(-1.0), eps)

# Type verification
try:
    calculate_ulp("invalid")
    assert False, "Should raise TypeError"
except TypeError:
    pass

print("ULP at x = 1.0 (Machine Epsilon):", calculate_ulp(1.0))
print("ULP at x = 1,000,000.0:          ", calculate_ulp(1_000_000.0))
print("All ULP machine precision assertions passed successfully!")
```
