## 2. Run Webpage Auditor Script (Localhost or Live URL)

Fetch a target webpage and inspect its SEO metadata, GA4 presence, alt coverage, and schema flags.

```python
import re
import requests
from urllib.parse import urlparse, urljoin
from bs4 import BeautifulSoup

url_to_audit = "https://www.plus2net.com/python/pandas.php"
UA = "Mozilla/5.0 (compatible; AI-Ready-Auditor/1.1)"

response = requests.get(url_to_audit, headers={"User-Agent": UA}, timeout=30)
soup = BeautifulSoup(response.text, "lxml")

title = soup.title.string.strip() if soup.title and soup.title.string else "Missing"
print(f"Title: {title}")
print(f"HTTP Status: {response.status_code}")
```

## 3. Detailed Internal vs External Links Extraction

Extract and classify all hyperlinked URLs into internal and external destinations.

```python
parsed_base = urlparse(url_to_audit)
base_host = parsed_base.netloc

internal_links = []
external_links = []

for a in soup.find_all("a", href=True):
    href = a["href"].strip()
    if not href or href.startswith("#") or href.startswith("javascript:"):
        continue
    
    full_url = urljoin(url_to_audit, href)
    parsed_href = urlparse(full_url)
    
    if parsed_href.netloc == base_host:
        internal_links.append(full_url)
    else:
        external_links.append(full_url)

print(f"Internal Links Count: {len(set(internal_links))}")
print(f"External Links Count: {len(set(external_links))}")
```
