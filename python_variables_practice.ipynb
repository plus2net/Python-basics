### 1. Variable Declaration & Multiple Assignment

```python
a = 45
my_name = "Raju"
_mark = 65
address = "World"
print(address, my_name, _mark, a)

width, height = 710, 710
print(f"Width: {width}, Height: {height}")
```

### 2. Dynamic Typing in Python

```python
data = "plus2net"
print("Type of data:", type(data))

data = 420
print("Type of data after reassignment:", type(data))
```

### 3. Object Memory Pointer Inspection

```python
x = [1, 2, 3]
y = x
print("id(x) == id(y):", id(x) == id(y))
print("x is y:        ", x is y)

y.append(4)
print("x after mutating y:", x)
```

### 4. Extended Sequence Unpacking

```python
x, y = 5, 10
x, y = y, x
print(f"Swapped: x={x}, y={y}")

head, *tail = [10, 20, 30, 40]
print("Head:", head)
print("Tail:", tail)
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Swapping
val1, val2 = 1, 2
val1, val2 = val2, val1
assert val1 == 2 and val2 == 1

# Test 2: Unpacking
first, *rest = "python"
assert first == "p" and "".join(rest) == "ython"

# Test 3: Type conversion
num_str = "123"
assert int(num_str) + 7 == 130

print("All variable unit assertions successfully verified! 100% passed.")
```
