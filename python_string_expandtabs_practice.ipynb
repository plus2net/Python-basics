### 1. Basic str.expandtabs() and Tab Stop Mechanics
`str.expandtabs()` advances text to the next tab stop multiple.

```python
s1 = "a\tb"
s2 = "abcd\te"
print("s1 expandtabs(8):", repr(s1.expandtabs(8)))
print("s2 expandtabs(8):", repr(s2.expandtabs(8)))

assert s1.expandtabs(8) == "a       b"
assert s2.expandtabs(8) == "abcd    e"
```

### 2. Custom Tab Sizes (4, 8, 12, 0)
Explore how different `tabsize` values impact spacing.

```python
text = "Name\tAge\tRole"
print("tabsize=4: ", repr(text.expandtabs(4)))
print("tabsize=8: ", repr(text.expandtabs(8)))
print("tabsize=12:", repr(text.expandtabs(12)))
print("tabsize=0: ", repr(text.expandtabs(0)))

assert text.expandtabs(4) == "Name    Age Role"
assert text.expandtabs(0) == "NameAgeRole"
```

### 3. Multi-line Reset with Newlines
Each newline `\n` resets the column counter back to 0.

```python
table = "Item\tPrice\nApple\t1.50\nBanana\t0.50"
formatted = table.expandtabs(10)
print(formatted)

lines = formatted.splitlines()
assert lines[0].index("Price") == 10
assert lines[1].index("1.50") == 10
```

### 4. PEP 8 Tab-to-Space Normalizer
Convert tabbed code into 4-space indented Python code.

```python
raw_code = "def run():\n\tx = 10\n\tif x > 5:\n\t\tprint('ok')"
pep8_code = raw_code.expandtabs(4)
print(pep8_code)

assert "    x = 10" in pep8_code
assert "        print('ok')" in pep8_code
```

### 5. Practice Exercise: TSV Grid Formatter Suite
Run the complete tabular grid formatter with unit assertions.

```python
class TabularFormatter:
    @staticmethod
    def calculate_spaces(current_col, tabsize=8):
        if tabsize <= 0:
            return 0
        return tabsize - (current_col % tabsize)

    @staticmethod
    def format_tsv(rows, col_width=12):
        tsv = "\n".join("\t".join(str(c) for c in row) for row in rows)
        return tsv.expandtabs(col_width)

# Unit Assertions
fmt = TabularFormatter()
assert fmt.calculate_spaces(1, tabsize=8) == 7
assert fmt.calculate_spaces(4, tabsize=8) == 4
assert fmt.calculate_spaces(8, tabsize=8) == 8
assert fmt.calculate_spaces(3, tabsize=4) == 1
assert fmt.calculate_spaces(5, tabsize=0) == 0

data = [
    ["SKU", "Title", "Stock"],
    ["A1", "Laptop", 12],
    ["B22", "Mouse", 50]
]
grid = fmt.format_tsv(data, col_width=10)
print(grid)

lines = grid.splitlines()
assert lines[0].index("Title") == 10
assert lines[1].index("Laptop") == 10

print("All TabularFormatter assertions passed successfully!")
```
