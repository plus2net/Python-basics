## Example 1: First working example

Read the code, predict the output and run this cell.

```python
from math import factorial
n = 145
total = sum(factorial(int(ch)) for ch in str(n))
print("Sum of digit factorials:", total)
print("Strong number:", total == n)
```

## Example 2: Original manual factorial helper

Read the code, predict the output and run this cell.

```python
def my_factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
print(my_factorial(5))
```

## Example 3: Check a number entered by the user

Read the code, predict the output and run this cell.

```python
from unittest.mock import patch
with patch("builtins.input", side_effect=['145']):
    from math import factorial
    raw = input("Enter a nonnegative whole number: ").strip()
    if not raw.isdecimal():
        print("Enter digits only")
    else:
        n = int(raw)
        total = sum(factorial(int(digit)) for digit in str(n))
        print("Sum:", total)
        print("Strong number:", total == n)
```

## Example 4: Reusable strong-number function

Read the code, predict the output and run this cell.

```python
from math import factorial
def is_strong(n):
    return n >= 0 and sum(factorial(int(ch)) for ch in str(n)) == n
for number in (1, 2, 145, 146, 40585):
    print(number, is_strong(number))
```

## Example 5: List strong numbers in a range

Read the code, predict the output and run this cell.

```python
from math import factorial
facts = [factorial(i) for i in range(10)]
def strong(n):
    return n >= 0 and sum(facts[int(ch)] for ch in str(n)) == n
print([n for n in range(1, 100000) if strong(n)])
```

## Example 6: Why 0 is not strong

Read the code, predict the output and run this cell.

```python
from math import factorial
print("0!:", factorial(0))
print("0 is strong:", factorial(0) == 0)
```

## Example 7: Show the calculation

Read the code, predict the output and run this cell.

```python
from math import factorial
n = 145
terms = [factorial(int(d)) for d in str(n)]
print(" + ".join(map(str, terms)), "=", sum(terms))
```
