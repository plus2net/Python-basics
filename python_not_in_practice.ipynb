# Python in and not in Membership Operators Practice

**Tutorial & Full Documentation:** [https://www.plus2net.com/python/not-in.php](https://www.plus2net.com/python/not-in.php)

Run cells in order in Google Colab. This notebook demonstrates membership checks across lists, strings, dictionaries, and tuples. Save a copy in Google Drive to keep your edits.

### 1. List Membership Checks

```python
my_list = [3, 4, 5, 8, 3]
print(5 in my_list)
print(9 not in my_list)
assert (5 in my_list) == True
assert (9 not in my_list) == True
print("List membership tests passed!")
```

### 2. String Substring Checks

```python
url = 'plus2net.com'
print('2' in url)
print('net' in url)
print('org' not in url)
assert ('2' in url) == True
assert ('org' not in url) == True
print("String membership tests passed!")
```

### 3. Dictionary Key Checks

```python
my_dict = {0: 'A', 1: 'B'}
print(1 in my_dict)
print('A' in my_dict)  # False, because 'A' is a value, not a key
assert (1 in my_dict) == True
assert ('A' in my_dict) == False
print("Dictionary membership tests passed!")
```
