# Python `date.resolution` and `datetime.resolution` — Practice

**Tutorial:** [Plus2Net: Python date.resolution](https://www.plus2net.com/python/date-resolution.php)  
**Repository:** [Plus2Net Python Basics](https://github.com/plus2net/Python-basics)

Learn how small a difference the `date`, `datetime`, `time`, and `timedelta` types can represent. All examples use Python's standard library. Run each cell, experiment, then check the assertions.

## 1. Original date example
`date.resolution` is **one day**. It is an attribute, not a function.

```python
from datetime import date
dt = date.today().resolution
print(dt)
assert dt.days == 1
assert dt.total_seconds() == 86400
print("Check passed")
```

## 2. Original datetime example
A `datetime` object can represent microseconds.

```python
from datetime import datetime
dt = datetime.now().resolution
print(dt)
assert dt.total_seconds() == 0.000001
print("Check passed")
```

## 3. Use the historical example from the page

```python
import datetime
# Year, Month, date, Hour, Minute, Second, microsecond
dt = datetime.datetime(2019,12,31,23,59,59,345234)
print(dt.resolution)
assert dt.resolution == datetime.timedelta(microseconds=1)
```

## 4. Compare all four related types

```python
from datetime import date, datetime, time, timedelta
for type_name, cls in [("date",date),("datetime",datetime),("time",time),("timedelta",timedelta)]:
    print(type_name, cls.resolution)
assert date.resolution == timedelta(days=1)
assert time.resolution == timedelta(microseconds=1)
assert timedelta.resolution == timedelta(microseconds=1)
```

## 5. Exercise: add one day to a leap-year date
**Challenge:** Predict the date after 2024-02-28, then run the code.

```python
from datetime import date
start = date(2024,2,28)
result = start + date.resolution
print(result)
assert result == date(2024,2,29)
```

## 6. Exercise: cross into a new year

```python
from datetime import date
last_day = date(2024,12,31)
next_day = last_day + date.resolution
print(next_day)
assert next_day == date(2025,1,1)
```

## 7. Exercise: microseconds across midnight

```python
from datetime import datetime
last_microsecond = datetime(2024,2,29,23,59,59,999999)
following = last_microsecond + datetime.resolution
print(following)
assert following == datetime(2024,3,1,0,0,0)
```

## 8. Convert a resolution into seconds

```python
from datetime import date, datetime
print(date.resolution.total_seconds())
print(datetime.resolution.total_seconds())
assert date.resolution.total_seconds() == 86400
assert datetime.resolution.total_seconds() == 1e-6
```

## 9. Access the same attribute from a class and an instance

```python
from datetime import date, datetime
print(date.resolution == date.today().resolution)
print(datetime.resolution == datetime.now().resolution)
assert date.resolution == date.today().resolution
assert datetime.resolution == datetime.now().resolution
```

## 10. Resolution is a duration, not a function
Do not write `date.resolution()`! This cell catches the expected `TypeError`.

```python
from datetime import date
try:
    date.resolution()
except TypeError as error:
    print("Expected error:", type(error).__name__)
else:
    raise AssertionError("Expected a TypeError")
```

## 11. Difference between two fixed datetimes
This **observed difference** is not the same idea as the type's minimum representable increment.

```python
from datetime import datetime, timedelta
first = datetime(2024,7,1,10,0,0,100000)
second = datetime(2024,7,1,10,0,0,104000)
difference = second - first
print("Type resolution:", datetime.resolution)
print("Observed difference:", difference)
assert difference == timedelta(milliseconds=4)
```

## 12. Inspect the computer clock
The actual result depends on the Colab runtime or your system. Don't hardcode a particular value.

```python
import time
info = time.get_clock_info("time")
print("Clock resolution:", info.resolution, "seconds")
print("Monotonic:", info.monotonic)
assert info.resolution > 0
```

## 13. Capture an aware UTC log timestamp
A timestamp's microsecond formatting does **not** guarantee actual microsecond accuracy.

```python
from datetime import datetime, timezone
stamp = datetime.now(timezone.utc)
print(stamp.isoformat(timespec="microseconds"))
assert stamp.tzinfo is not None
```

## 14. Create a distinct log event ID
If your system requires guaranteed record uniqueness, enforce it in your database. UUIDs are useful identifiers, but no timestamp format alone is a uniqueness constraint.

```python
from uuid import uuid4
from datetime import datetime, timezone
event = {"event_id": str(uuid4()), "timestamp": datetime.now(timezone.utc).isoformat(), "kind": "login"}
print(event)
assert len(event["event_id"]) == 36
assert event["kind"] == "login"
```

## 15. Exercise: compare timedelta increments
**Question:** Which expression matches `datetime.resolution` exactly?

```python
from datetime import datetime, timedelta
one_microsecond = timedelta(microseconds=1)
one_millisecond = timedelta(milliseconds=1)
assert datetime.resolution == one_microsecond
assert datetime.resolution != one_millisecond
print("1 millisecond contains", int(one_millisecond/one_microsecond), "microseconds")
```

## 16. Exercise: build a sequence of 3 calendar dates

```python
from datetime import date
days = [date(2024,2,28) + i * date.resolution for i in range(3)]
print(days)
assert days == [date(2024,2,28), date(2024,2,29), date(2024,3,1)]
```

## 17. Exercise: calculate microsecond difference

```python
from datetime import datetime
start = datetime(2024,6,1,12,0,0,5)
end = datetime(2024,6,1,12,0,0,8)
steps = (end-start)/datetime.resolution
print("Representable one-microsecond steps:", steps)
assert steps == 3
```
