### 1. Basic Truth Value Testing
Convert numbers, strings, and containers to booleans.

```python
print("bool(10):  ", bool(10))
print("bool(0):   ", bool(0))
print("bool('hi'):", bool('hi'))
print("bool(''):  ", bool(''))
print("bool([]):  ", bool([]))
```

### 2. Booleans are Subclasses of int
Verify numeric inheritance and arithmetic with True and False.

```python
print("issubclass(bool, int):", issubclass(bool, int))
print("True + True:          ", True + True)
print("True == 1:            ", True == 1)
```

### 3. Custom Class __bool__() and __len__() Protocols
Define custom truth testing on user classes.

```python
class Account:
    def __init__(self, balance):
        self.balance = balance
    def __bool__(self):
        return self.balance > 0

print("Funded:  ", bool(Account(100)))
print("Unfunded:", bool(Account(0)))
```

### 4. Production Recipe: Config Flag Normalization
Parse boolean configuration strings safely.

```python
def parse_flag(val):
    if isinstance(val, str):
        return val.strip().lower() in {'true', '1', 'yes', 'on'}
    return bool(val)

print("Parse 'YES':  ", parse_flag("YES"))
print("Parse 'false':", parse_flag("false"))
print("Parse '0':    ", parse_flag("0"))
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying bool() mechanics.

```python
assert bool() is False
assert bool(1) is True
assert bool(0) is False
assert bool("False") is True
assert issubclass(bool, int)

print("All bool() unit assertions passed! 100% verified.")
```
