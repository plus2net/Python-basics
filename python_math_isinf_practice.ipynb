### 1. Basic Infinity Detection with math.isinf()
`math.isinf(x)` returns `True` if `x` is positive infinity (`+inf`) or negative infinity (`-inf`). It returns `False` for all finite numbers, zeros, and NaNs.


```python
import math

print("isinf(float('inf')): ", math.isinf(float("inf")))   # True
print("isinf(float('-inf')):", math.isinf(float("-inf")))  # True
print("isinf(1e308 * 2.0):    ", math.isinf(1e308 * 2.0))    # True (overflow)
print("isinf(0.0):            ", math.isinf(0.0))            # False
print("isinf(1e308):          ", math.isinf(1e308))          # False
print("isinf(float('nan')):   ", math.isinf(float("nan")))   # False
```

### 2. Distinguishing Positive vs. Negative Infinity
`math.isinf(x)` returns `True` for both directions. Test `x > 0` for positive infinity and `x < 0` for negative infinity.


```python
import math

def classify_infinity(val):
    if not math.isinf(val):
        return "Finite or NaN"
    return "Positive Infinity (+inf)" if val > 0 else "Negative Infinity (-inf)"

print("+inf: ", classify_infinity(float("inf")))
print("-inf: ", classify_infinity(float("-inf")))
print("100.0:", classify_infinity(100.0))
```

### 3. IEEE 754 Arithmetic Rules with Infinity
Observe how operations with infinity behave according to the IEEE 754 standard.


```python
import math

inf = float("inf")
print("inf + 50.0:  ", inf + 50.0)
print("inf * -2.0:  ", inf * -2.0)
print("1.0 / inf:   ", 1.0 / inf)
print("1.0 / -inf:  ", 1.0 / -inf)
print("inf - inf:   ", inf - inf)   # NaN (indeterminate)
print("inf / inf:   ", inf / inf)   # NaN (indeterminate)
```

### 4. The IEEE 754 Triad: isinf vs. isnan vs. isfinite
Validate that exactly one category holds true for any valid float.


```python
import math

test_samples = [42.5, 0.0, float("inf"), float("-inf"), float("nan")]
print(f"{'Value':>12} | {'isinf':>8} | {'isnan':>8} | {'isfinite':>8}")
print("-" * 44)
for val in test_samples:
    inf = math.isinf(val)
    nan = math.isnan(val)
    fin = math.isfinite(val)
    assert fin == (not (inf or nan))
    print(f"{str(val):>12} | {str(inf):>8} | {str(nan):>8} | {str(fin):>8}")
```

### 5. Hands-on Practice Exercise: Gradient Explosion Guard & Clipper
In deep learning, exploding gradients occur when numerical calculations diverge to `inf` or `-inf`.
Implement `sanitize_and_clip_gradients(gradients, clip_threshold)` and verify with self-checking assertions.


```python
import math

def sanitize_and_clip_gradients(gradients, clip_threshold=100.0):
    """
    Inspects gradients for infinite values.
    Clips infinite or out-of-bound values to [-clip_threshold, +clip_threshold].
    Returns (clipped_gradients, had_infinities).
    """
    if not isinstance(gradients, (list, tuple)):
        raise TypeError("Gradients must be a list or tuple.")
    
    sanitized = []
    found_inf = False
    
    for g in gradients:
        if math.isnan(g):
            sanitized.append(0.0)  # Reset NaN to zero
        elif math.isinf(g):
            found_inf = True
            # Clip to +/- clip_threshold based on direction
            clipped = math.copysign(clip_threshold, g)
            sanitized.append(clipped)
        else:
            clamped = max(-clip_threshold, min(clip_threshold, float(g)))
            sanitized.append(clamped)
            
    return (sanitized, found_inf)

# Test Data
raw_grads = [2.5, float("inf"), -15.0, float("-inf"), 120.0]
clean_grads, diverged = sanitize_and_clip_gradients(raw_grads, clip_threshold=50.0)

# Unit Assertions
assert diverged is True
assert clean_grads[0] == 2.5
assert clean_grads[1] == 50.0   # +inf clipped to +50.0
assert clean_grads[2] == -15.0
assert clean_grads[3] == -50.0  # -inf clipped to -50.0
assert clean_grads[4] == 50.0   # 120.0 clamped to +50.0

print("Original gradients: ", raw_grads)
print("Clipped gradients:  ", clean_grads)
print("All gradient sanitizer assertions passed successfully!")
```
