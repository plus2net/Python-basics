### 1. Basic str.isprintable() Evaluation
`str.isprintable()` returns `True` if all characters are printable, or if the string is empty.

```python
print("'Welcome to plus2net'.isprintable():", 'Welcome to plus2net'.isprintable())
print("'Price: $99.95!'.isprintable():       ", 'Price: $99.95!'.isprintable())
print("'   '.isprintable():                  ", '   '.isprintable())
print("''.isprintable():                     ", ''.isprintable())
print("'Hello\\tWorld'.isprintable():         ", 'Hello\tWorld'.isprintable())
print("'Line 1\\nLine 2'.isprintable():       ", 'Line 1\nLine 2'.isprintable())
print("'Null\\x00Byte'.isprintable():         ", 'Null\x00Byte'.isprintable())

assert 'Welcome to plus2net'.isprintable() is True
assert ''.isprintable() is True
assert 'Hello\tWorld'.isprintable() is False
assert 'Line 1\nLine 2'.isprintable() is False
```

### 2. Unicode General Categories and Non-Printable Classification
Inspect character categories using `unicodedata.category()`.

```python
import unicodedata

chars = ['A', ' ', '\t', '\n', '\x00', '€', '中']
print(f"{'Char':<6} | {'Repr':<8} | {'Category':<10} | {'isprintable':<12}")
print("-" * 45)
for c in chars:
    cat = unicodedata.category(c)
    print(f"{c:<6} | {repr(c):<8} | {cat:<10} | {str(c.isprintable()):<12}")

assert ' '.isprintable() is True
assert '\t'.isprintable() is False
```

### 3. Detecting Invisible Zero-Width Spaces and ANSI Escapes
Flag zero-width spaces (`\u200b`) and terminal ANSI color escape sequences.

```python
clean_text = "admin"
spoofed_text = "ad\u200bmin"
ansi_text = "\x1b[31mERROR\x1b[0m"

print("Clean text isprintable:  ", clean_text.isprintable())
print("Spoofed text isprintable:", spoofed_text.isprintable())
print("ANSI text isprintable:   ", ansi_text.isprintable())

assert clean_text.isprintable() is True
assert spoofed_text.isprintable() is False
assert ansi_text.isprintable() is False
```

### 4. Log Message Sanitization
Neutralize log injection by escaping non-printable characters.

```python
def sanitize_log(msg):
    if msg.isprintable():
        return msg
    return "".join(c if c.isprintable() else repr(c)[1:-1] for c in msg)

raw_log = "Payment error\r[ALERT] Fake message"
safe_log = sanitize_log(raw_log)
print("Safe log:", safe_log)
assert safe_log.isprintable() is True
```

### 5. Practice Exercise: Text Security Auditor Suite
Run the complete verification suite with self-checking assertions.

```python
class TextSecurityAuditor:
    @staticmethod
    def is_clean_single_line(text):
        if not isinstance(text, str):
            return False
        return text.isprintable()

    @staticmethod
    def sanitize_for_display(text, replacement="?"):
        if not isinstance(text, str):
            return ""
        return "".join(ch if ch.isprintable() else replacement for ch in text)

    @staticmethod
    def find_bad_codepoints(text):
        return [hex(ord(ch)) for ch in text if not ch.isprintable()]

# Unit Assertions
auditor = TextSecurityAuditor()
assert auditor.is_clean_single_line("Valid message from API") is True
assert auditor.is_clean_single_line("") is True
assert auditor.is_clean_single_line("Hello\nWorld") is False
assert auditor.is_clean_single_line("Null\x00Byte") is False

raw = "Login\tSuccess\x00!"
assert auditor.sanitize_for_display(raw, replacement="_") == "Login_Success_!"
assert auditor.find_bad_codepoints(raw) == ['0x9', '0x0']

print("All str.isprintable() audit assertions passed successfully!")
```
