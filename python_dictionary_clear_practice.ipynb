### 1. Basic In-Place Reset Demonstration
`dict.clear()` empties all elements from the dictionary and returns `None`.

```python
student = {"name": "Alex", "roll": 24, "grade": "A"}
print("Before clear:", student)

ret = student.clear()
print("After clear: ", student)  # {}
print("Return value:", ret)      # None
print("Type:        ", type(student))
```

### 2. The Critical Difference: clear() vs. Reassignment (d = {})
See how `clear()` mutates shared references in-place, while `d = {}` merely points one variable to a new object.

```python
# Case A: clear() affects all aliases
cache_a = {"session": "tok_991", "auth": True}
alias_a = cache_a
cache_a.clear()
print("Case A - alias_a:", alias_a)  # {} (also cleared!)

# Case B: Reassigning cache_b = {} breaks the alias
cache_b = {"session": "tok_992", "auth": True}
alias_b = cache_b
cache_b = {}
print("Case B - alias_b:", alias_b)  # {'session': 'tok_992', 'auth': True}
```

### 3. Preserving Memory Identity
`clear()` keeps the underlying memory address identical (`id(d)` unchanged).

```python
data = {i: i * 2 for i in range(100)}
orig_id = id(data)

data.clear()
print("Memory ID preserved?", id(data) == orig_id)  # True
print("Length after clear: ", len(data))            # 0
```

### 4. Real-World Recipe: Cache Invalidation in Service Modules
Purge an in-memory cache without breaking references held by other parts of the application.

```python
class CacheManager:
    def __init__(self):
        self._cache = {}

    def set(self, k: str, v) -> None:
        self._cache[k] = v

    def flush(self) -> int:
        count = len(self._cache)
        self._cache.clear()
        return count

cache = CacheManager()
cache.set("q1", {"data": 1})
cache.set("q2", {"data": 2})
print("Flushed entries:", cache.flush())  # 2
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying in-place clearing, memory reuse, and alias propagation.

```python
class ClearVerifier:
    @staticmethod
    def verify_reset(target: dict) -> bool:
        saved = id(target)
        target.clear()
        return len(target) == 0 and id(target) == saved

v = ClearVerifier()
d = {"a": 1, "b": 2}
ref = d

# Test 1: In-place clearing & address preservation
assert v.verify_reset(d) is True
assert len(ref) == 0
assert ref is d

# Test 2: Double clearing is safe
assert d.clear() is None

print("All dict.clear() unit assertions passed successfully! 100% verified.")
```
