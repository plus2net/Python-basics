### 1. Basic Parallel Sequence Pairing
Pair elements from multiple lists into tuples.

```python
names = ['Ravi', 'Raju', 'Alex']
ids = [1, 2, 3]
scores = [30, 40, 50]

paired = list(zip(names, ids, scores))
print("Paired elements:", paired)
```

### 2. Unzipping Sequences with zip(*zipped)
Separate grouped tuples back into independent series.

```python
pairs = [('Ravi', 1, 30), ('Raju', 2, 40), ('Alex', 3, 50)]
names_out, ids_out, scores_out = zip(*pairs)

print("Names: ", names_out)
print("IDs:   ", ids_out)
print("Scores:", scores_out)
```

### 3. Rapid Dictionary Construction
Build dictionaries by combining key and value lists.

```python
keys = ["name", "age", "city"]
values = ["Alice", 25, "New York"]
person_dict = dict(zip(keys, values))
print("Person dict:", person_dict)
```

### 4. Unequal Lengths: strict=True vs. zip_longest
Compare strict length validation with default padding.

```python
from itertools import zip_longest

a = [1, 2]
b = ['x', 'y', 'z']

print("zip_longest:", list(zip_longest(a, b, fillvalue="N/A")))

try:
    list(zip(a, b, strict=True))
except ValueError as e:
    print("strict=True caught error:", e)
```

### 5. Production Recipe: Matrix Transposition
Transpose a 2D matrix using `zip(*matrix)`.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
transposed = [list(col) for col in zip(*matrix)]
print("Transposed matrix:", transposed)
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying zip() mechanics.

```python
assert list(zip([1, 2], ['a', 'b'])) == [(1, 'a'), (2, 'b')]
assert list(zip(*[(1, 'a'), (2, 'b')])) == [(1, 2), ('a', 'b')]
assert dict(zip(['k1', 'k2'], [10, 20])) == {'k1': 10, 'k2': 20}
assert list(zip([1, 2, 3], ['x', 'y'])) == [(1, 'x'), (2, 'y')]

print("All zip() unit assertions passed! 100% verified.")
```
