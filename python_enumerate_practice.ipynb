### 1. Basic List Enumeration
Pair each element with an automatically incremented index counter.

```python
my_list = ['One', 'Two', 'Three', 'Four']
en = enumerate(my_list)

print("Type of enumerate object:", type(en))
print("Unpacked pairs:           ", list(en))
```

### 2. Custom Starting Offsets & String Traversal
Use the `start` parameter to begin counting from 1 or any custom offset.

```python
items = ('apple', 'banana', 'cherry')
for idx, fruit in enumerate(items, start=100):
    print(f"Fruit #{idx}: {fruit}")

text = "plus2net"
for idx, ch in enumerate(text):
    print(f"{idx}->{ch}", end=" ")
print()
```

### 3. enumerate() vs. range(len())
Compare idiomatic sequence traversal with manual index lookups.

```python
languages = ['Python', 'C++', 'Java', 'Ruby']
for idx, lang in enumerate(languages, start=1):
    print(f"{idx}. {lang}")
```

### 4. Nested Tuple and Dictionary Unpacking
Unpack compound data structures like lists of tuples and dictionary `.items()`.

```python
students = [('Alice', 85), ('Bob', 90), ('Charlie', 78)]
for rank, (name, score) in enumerate(students, start=1):
    print(f"Rank {rank}: {name} ({score} pts)")

config = {'host': 'localhost', 'port': 8080, 'debug': True}
for step, (k, v) in enumerate(config.items(), start=1):
    print(f"Setting #{step}: {k} = {v}")
```

### 5. Production Recipe: Line Scanning & Matching
Track 1-based line numbers when inspecting text records.

```python
log_entries = [
    "2026-10-14 08:00:00 INFO Service started",
    "2026-10-14 08:01:10 WARN High latency",
    "2026-10-14 08:02:15 ERROR Database disconnect",
    "2026-10-14 08:03:00 ERROR Retry attempt failed",
]

for line_no, content in enumerate(log_entries, start=1):
    if "ERROR" in content:
        print(f"Error found on Line {line_no}: {content.split('ERROR')[-1].strip()}")
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying enumerate() mechanics.

```python
sample = ['alpha', 'beta', 'gamma']
assert list(enumerate(sample)) == [(0, 'alpha'), (1, 'beta'), (2, 'gamma')]
assert list(enumerate(sample, start=5)) == [(5, 'alpha'), (6, 'beta'), (7, 'gamma')]
assert list(enumerate("hi")) == [(0, 'h'), (1, 'i')]

# Verify iterator single-pass exhaustion
it = enumerate([10, 20])
assert list(it) == [(0, 10), (1, 20)]
assert list(it) == []

print("All enumerate() unit assertions passed! 100% verified.")
```
