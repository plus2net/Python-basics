### 1. Basic Floor and Ceiling Operations
- `math.floor(x)` rounds downwards towards $-\infty$.
- `math.ceil(x)` rounds upwards towards $+\infty$.

```python
import math

print("floor(3.2): ", math.floor(3.2))
print("ceil(3.2):  ", math.ceil(3.2))
print("floor(3.0): ", math.floor(3.0))
print("ceil(3.0):  ", math.ceil(3.0))

assert math.floor(3.2) == 3
assert math.ceil(3.2) == 4
assert math.floor(3.0) == 3
assert math.ceil(3.0) == 3
```

### 2. Negative Number Behaviors (Crucial Edge Cases)
Notice how floor and ceiling move along the real number line for negative floats.

```python
import math

print("floor(-3.2):", math.floor(-3.2))
print("ceil(-3.2): ", math.ceil(-3.2))
print("floor(-3.8):", math.floor(-3.8))
print("ceil(-3.8): ", math.ceil(-3.8))

assert math.floor(-3.2) == -4
assert math.ceil(-3.2) == -3
assert math.floor(-3.8) == -4
assert math.ceil(-3.8) == -3
```

### 3. Comparison with trunc() and round()
Compare `floor`, `ceil`, `trunc` (truncation towards 0), and `round` (Banker's round-half-to-even).

```python
import math

test_vals = [2.3, 2.5, 2.7, -2.3, -2.5, -2.7]
print(f"{'x':>6} | {'floor':>6} | {'ceil':>6} | {'trunc':>6} | {'round':>6}")
print("-" * 40)
for v in test_vals:
    print(f"{v:>6.1f} | {math.floor(v):>6} | {math.ceil(v):>6} | {math.trunc(v):>6} | {round(v):>6}")

# Truncation matches floor for positive, ceil for negative
assert math.trunc(2.7) == math.floor(2.7)
assert math.trunc(-2.7) == math.ceil(-2.7)
```

### 4. Pure Integer Ceiling Division Idioms
Compare `math.ceil(a / b)` against pure integer arithmetic `(a + b - 1) // b` and `-(-a // b)`.

```python
import math

pairs = [(10, 3), (12, 3), (1, 5), (100, 7)]
for a, b in pairs:
    c1 = math.ceil(a / b)
    c2 = (a + b - 1) // b
    c3 = -(-a // b)
    print(f"{a} / {b}: ceil={c1}, formula={c2}, inverted={c3}")
    assert c1 == c2 == c3
```

### 5. Practice Exercise: Parking Rate Billing & Pagination Engine
Run the verification suite testing real-world pagination and hourly interval billing.

```python
import math

class ParkingRateCalculator:
    def __init__(self, hourly_rate=5.0, grace_period_minutes=15):
        self.hourly_rate = hourly_rate
        self.grace_period_minutes = grace_period_minutes

    def calculate_bill(self, parked_minutes):
        if parked_minutes <= 0 or parked_minutes <= self.grace_period_minutes:
            return 0.0
        billable_hours = math.ceil(parked_minutes / 60.0)
        return billable_hours * self.hourly_rate

def paginate(total_items, page_size, current_page=1):
    if total_items <= 0 or page_size <= 0:
        return {'total_pages': 1, 'start_index': 0, 'end_index': 0}
    total_pages = math.ceil(total_items / page_size)
    page = min(max(1, current_page), total_pages)
    start_idx = (page - 1) * page_size
    end_idx = min(start_idx + page_size, total_items)
    return {'total_pages': total_pages, 'page': page, 'start_index': start_idx, 'end_index': end_idx}

# Run Unit Assertions
calc = ParkingRateCalculator(hourly_rate=5.0, grace_period_minutes=15)
assert calc.calculate_bill(10) == 0.0
assert calc.calculate_bill(15) == 0.0
assert calc.calculate_bill(16) == 5.0
assert calc.calculate_bill(60) == 5.0
assert calc.calculate_bill(61) == 10.0

p = paginate(total_items=73, page_size=10, current_page=1)
assert p['total_pages'] == 8
assert p['start_index'] == 0
assert p['end_index'] == 10

print("Parking bill for 75 mins ($5/hr):", calc.calculate_bill(75))
print("Total pages for 73 items (10/pg):", p['total_pages'])
print("All math.floor() and math.ceil() assertions passed successfully!")
```
