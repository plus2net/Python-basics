### 1. Single-Line & Inline Comments

```python
# Block comment explaining setup
item_price = 45.0  # Unit price in USD (2 leading spaces before #)
tax_rate = 0.08    # 8% sales tax

total = item_price * (1 + tax_rate)
print(f"Total: USD {total:.2f}")
```

### 2. Multiline Comments vs. Docstrings

```python
def calculate_perimeter(length: float, width: float) -> float:
    """Calculates the perimeter of a rectangle.
    
    Args:
        length (float): Length dimension.
        width (float): Width dimension.
    """
    # Formula: 2 * (l + w)
    return 2 * (length + width)

print("Perimeter:", calculate_perimeter(5.0, 3.0))
print("\nDocstring:")
print(calculate_perimeter.__doc__)
```

### 3. Bytecode Disassembly

```python
import dis

def sample():
    # Comment is stripped
    x = 10
    return x

print("Bytecode for sample():")
dis.dis(sample)
```

### 4. Self-Checking Unit Verification Suite
