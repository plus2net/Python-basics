### 1. Basic Invocations & Conversions

```python
# range(stop)
print("range(5):       ", list(range(5)))

# range(start, stop)
print("range(2, 7):    ", list(range(2, 7)))

# range(start, stop, step)
print("range(10, 30, 5):", list(range(10, 30, 5)))
```

### 2. Negative Stepping

```python
# Decrementing from 10 down to 2
countdown = range(10, 0, -2)
print("Countdown: ", list(countdown))

# Negative numbers
neg_prog = range(-10, 0, 3)
print("Neg range: ", list(neg_prog))
```

### 3. O(1) Memory & Arithmetic Operations

```python
import sys

small_r = range(5)
huge_r = range(10_000_000_000)

print("Memory small_r:", sys.getsizeof(small_r), "bytes")
print("Memory huge_r: ", sys.getsizeof(huge_r), "bytes")
print("5_000 in huge_r:", 5_000 in huge_r)
print("Index 500:      ", huge_r[500])
```

### 4. Dataset Batching Recipe

```python
def make_chunks(lst, n):
    for i in range(0, len(lst), n):
        yield lst[i : i + n]

data = [f"Item_{i}" for i in range(1, 11)]
batches = list(make_chunks(data, 3))
for i, b in enumerate(batches, 1):
    print(f"Batch {i}: {b}")
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Sequence matching
assert list(range(4)) == [0, 1, 2, 3]
assert list(range(1, 4)) == [1, 2, 3]
assert list(range(0, 10, 2)) == [0, 2, 4, 6, 8]

# Test 2: Negative stepping
assert list(range(3, 0, -1)) == [3, 2, 1]

# Test 3: Slicing and indexing
r = range(10, 50, 5)
assert r[0] == 10
assert r[-1] == 45
assert list(r[1:4]) == [15, 20, 25]

# Test 4: Range equality
assert range(0) == range(2, 2)
assert range(0, 3, 2) == range(0, 4, 2)

print("All range() unit assertions successfully verified! 100% passed.")
```
