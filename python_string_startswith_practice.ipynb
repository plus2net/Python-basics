## 1. Basic Prefix Matching & Case Sensitivity

`str.startswith(prefix)` tests if the string begins with the specified substring.
Matching is strictly case-sensitive.

```python
banner = "Welcome to plus2net.com tutorials"

print("Starts with 'Welcome':", banner.startswith("Welcome"))
print("Starts with 'welcome':", banner.startswith("welcome"))

assert banner.startswith("Welcome") is True
assert banner.startswith("welcome") is False

# Empty string prefix is always True
assert banner.startswith("") is True
assert "".startswith("") is True

print("\nAll basic prefix assertions passed!")
```

## 2. The Tuple-of-Prefixes Power Feature

Passing a **tuple** tests multiple candidates at once.
Passing a list raises a `TypeError`.

```python
valid_schemes = ("https://", "http://", "ftp://", "sftp://")

assert "https://api.plus2net.com".startswith(valid_schemes) is True
assert "ftp://files.host.net".startswith(valid_schemes) is True
assert "mailto:admin@host.net".startswith(valid_schemes) is False

# The List TypeError Trap
try:
    "https://example.com".startswith(["https://", "http://"])
    assert False, "Expected TypeError not raised"
except TypeError as e:
    assert "must be str or a tuple" in str(e)

print("Tuple prefix assertions passed!")
```

## 3. Offset Matching: `start` and `end` Parameters

Testing prefixes at specified index positions avoids allocating temporary slice strings.

```python
text = "Data Science and Machine Learning"

# Offset 5 matches 'Science'
assert text.startswith("Science", 5) is True
assert text.startswith("Science", 5, 12) is True
assert text.startswith("Learning", -8) is True

print("Offset prefix assertions passed!")
```

## 4. Real-World Engineering Recipes

Explore three common software engineering patterns utilizing `str.startswith()`:
1. Path Traversal Guard (Security Defense)
2. Multi-Syntax Comment & Directive Filter
3. Digital Asset Filename Classifier

```python
import os

# Recipe 1: Safe path guard
def is_safe_path(target: str, base_dir: str = "/var/www/uploads") -> bool:
    real_target = os.path.realpath(target)
    real_base = os.path.realpath(base_dir)
    return real_target.startswith(real_base + os.sep)

assert is_safe_path("/var/www/uploads/avatar.png") is True
assert is_safe_path("/var/www/uploads/../../etc/passwd") is False

# Recipe 2: Multi-syntax comment filter
def is_comment(line: str) -> bool:
    return line.strip().startswith(("#", "//", ";", "--"))

assert is_comment("  # Python comment") is True
assert is_comment("  // JS comment") is True
assert is_comment("SELECT * FROM tbl;") is False

# Recipe 3: Asset classifier
def classify_asset(name: str) -> str:
    up = name.upper()
    if up.startswith(("IMG_", "PHOTO_")):
        return "IMAGE"
    elif up.startswith(("VID_", "CLIP_")):
        return "VIDEO"
    return "DOCUMENT"

assert classify_asset("img_2026.png") == "IMAGE"
assert classify_asset("VID_001.mp4") == "VIDEO"
assert classify_asset("report.pdf") == "DOCUMENT"

print("\nAll engineering recipes passed!")
```

## 5. Production Implementation & Self-Checking Unit Assertions

This production-grade `PrefixMatcher` class encapsulates protocol validation, comment checking, and offset matching.

```python
class PrefixMatcher:
    @staticmethod
    def has_valid_protocol(url: str) -> bool:
        """Validates web or transport protocol prefixes."""
        valid_schemes = ("https://", "http://", "ftp://", "sftp://")
        return url.lower().startswith(valid_schemes)

    @staticmethod
    def is_code_comment(line: str) -> bool:
        """Tests if a line starts with common comment characters."""
        markers = ("#", "//", "--", ";")
        return line.strip().startswith(markers)

    @staticmethod
    def match_at_index(text: str, prefix: str, offset: int) -> bool:
        """Checks prefix at designated offset."""
        return text.startswith(prefix, offset)


# ==========================================
# UNIT ASSERTION SUITE
# ==========================================
matcher = PrefixMatcher()

# 1. Single string matching
assert "Python".startswith("Py") is True
assert "Python".startswith("py") is False

# 2. Tuple matching
assert matcher.has_valid_protocol("https://plus2net.com") is True
assert matcher.has_valid_protocol("sftp://files.host.net") is True
assert matcher.has_valid_protocol("gopher://old.net") is False

# 3. Comment line filtering
assert matcher.is_code_comment("  # This is a comment") is True
assert matcher.is_code_comment("  -- SQL comment") is True
assert matcher.is_code_comment("var x = 10;") is False

# 4. Offset matching
assert matcher.match_at_index("ABCDEF", "CD", 2) is True
assert matcher.match_at_index("ABCDEF", "CD", 3) is False

# 5. Empty string prefix is always True
assert "anything".startswith("") is True
assert "".startswith("") is True

# 6. TypeError validation when passing a list
try:
    "https://example.com".startswith(["https://", "http://"])
    assert False, "Expected TypeError not raised"
except TypeError as e:
    assert "must be str or a tuple" in str(e)

print("All str.startswith() production unit assertions passed successfully!")
```
