## 1. Print July 2020

`TextCalendar.prmonth(year, month, w, l)` prints a formatted calendar. Watch how the spaces align the columns.

## 2. Change the first weekday

Weekdays always have fixed values: Monday `0` through Sunday `6`. `firstweekday` only changes their order on the calendar.

```python
import calendar
cal = calendar.TextCalendar(firstweekday=calendar.FRIDAY)
cal.prmonth(2020, 7, w=3, l=0)
```

```python
assert cal.getfirstweekday() == 4
assert cal.formatmonth(2020, 7, 3, 0).splitlines()[1].startswith("Fri")
print("Check passed: Friday-first calendar")
```

## 3. Width and lines per week

`w=2` gives the compact default date columns. `w=5` spreads them out. `l=2` adds a blank line between week rows.

```python
cal = calendar.TextCalendar()
small = cal.formatmonth(2020, 7, w=2, l=1)
wide = cal.formatmonth(2020, 7, w=5, l=1)
spaced = cal.formatmonth(2020, 7, w=2, l=2)
print("Compact:\n", small)
print("Wide:\n", wide)
print("Spaced:\n", spaced)
```

```python
assert len(wide.splitlines()[1]) > len(small.splitlines()[1])
assert spaced.count("\n\n") > small.count("\n\n")
assert cal.formatmonth(2020, 7, w=0, l=0) == cal.formatmonth(2020, 7, w=2, l=1)
print("Check passed: width and line spacing")
```

## 5. Capture printed output

Existing code that calls `prmonth()` can be captured using `StringIO`. This is mainly useful when you cannot change that code to call `formatmonth()`.

```python
from contextlib import redirect_stdout
from io import StringIO

buffer = StringIO()
with redirect_stdout(buffer):
    calendar.prmonth(2020, 7)
output_text = buffer.getvalue()
assert "July 2020" in output_text
print("Captured", len(output_text), "characters")
```

## 7. Print three month titles

Use a `for` loop to print calendars for several months. This example shows their titles without the full grids.

```python
my_cal = calendar.TextCalendar()
for month in (6, 7, 8):
    print(my_cal.formatmonth(2020, month).splitlines()[0].strip())

assert [calendar.month_name[n] for n in (6, 7, 8)] == ["June", "July", "August"]
print("Check passed: 3 months")
```

## 8. Leap-year checks and invalid inputs

The `calendar` module calculates leap years automatically. Check the number of days in February in two different years.

```python
assert calendar.monthrange(2024, 2)[1] == 29
assert calendar.monthrange(2023, 2)[1] == 28
assert calendar.weekday(2020, 7, 1) == 2
print("Check passed: leap-year and weekday calculations")
```

```python
year = 2024
month = 13
if isinstance(month, int) and 1 <= month <= 12:
    calendar.prmonth(year, month)
else:
    print("Please enter a month from 1 to 12.")

assert not (1 <= month <= 12)
```
