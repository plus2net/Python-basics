### 1. Read and Modify XML Configuration Files

```python
import xml.etree.ElementTree as ET

# Create sample config.xml
config_data = """<config>
    <server>
        <host>127.0.0.1</host>
        <port>8080</port>
    </server>
    <logging>
        <level>DEBUG</level>
        <file>app.log</file>
    </logging>
</config>"""

with open("config.xml", "w") as f:
    f.write(config_data)

tree = ET.parse("config.xml")
root = tree.getroot()

print("Host:", root.find("./server/host").text)
print("Port:", root.find("./server/port").text)
print("Log Level:", root.find("./logging/level").text)
```
