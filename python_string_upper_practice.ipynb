## 1. Basic Uppercase Conversion & Non-Cased Characters

`str.upper()` converts all cased characters to uppercase.
Digits, whitespace, punctuation, and emojis pass through unaltered.

```python
sample = "Welcome to plus2net! Status: 200 OK & verified ✅"
uppercased = sample.upper()

print("Original:  ", sample)
print("Uppercase: ", uppercased)

assert uppercased == "WELCOME TO PLUS2NET! STATUS: 200 OK & VERIFIED ✅"
assert "".upper() == ""

print("\nAll basic uppercase assertions passed!")
```

## 2. Unicode Expansions & String Length Changes

String length is **not** invariant under uppercase conversion.
German `ß` expands to `SS`, increasing string length by one.

```python
german = "straße"
german_upper = german.upper()

print("Original:", german, "| Length:", len(german))
print("Upper:   ", german_upper, "| Length:", len(german_upper))

assert german_upper == "STRASSE"
assert len(german) == 6
assert len(german_upper) == 7

# Greek sigmas (regular σ and final ς both map to capital Σ)
assert "σ".upper() == "ς".upper() == "Σ"

print("\nUnicode expansion assertions passed!")
```

## 3. Caseless Matching: `upper()` vs. `casefold()`

While `upper()` can match German `"groß"` and `"GROSS"`, standard caseless comparison in Python should use `casefold()`.

```python
s1 = "groß"
s2 = "GROSS"

# Lowercase comparison fails
assert (s1.lower() == s2.lower()) is False

# Uppercase succeeds
assert (s1.upper() == s2.upper()) is True

# casefold() is the canonical Python solution
assert (s1.casefold() == s2.casefold()) is True

print("Caseless matching assertions passed!")
```

## 4. Real-World Engineering Recipes

Explore three common software engineering patterns utilizing `str.upper()`:
1. HTTP Method Validator
2. Environment Variable Configuration Canonicalizer
3. ISO 4217 Currency Code Sanitizer

```python
# Recipe 1: HTTP Method Normalization
VALID_METHODS = {"GET", "POST", "PUT", "DELETE", "PATCH"}

def normalize_method(verb: str) -> str:
    clean = verb.strip().upper()
    if clean not in VALID_METHODS:
        raise ValueError(f"Invalid verb: {verb}")
    return clean

assert normalize_method("  get  ") == "GET"
assert normalize_method("post") == "POST"

# Recipe 2: Environment Config Keys
def canonicalize_env(config: dict) -> dict:
    return {k.strip().replace("-", "_").upper(): str(v).strip() for k, v in config.items()}

env_in = {"db-host": "localhost", "db_port": "5432"}
env_out = canonicalize_env(env_in)
print("Recipe 2 ENV:", env_out)
assert env_out == {"DB_HOST": "localhost", "DB_PORT": "5432"}

# Recipe 3: Currency code sanitizer
def sanitize_currency(code: str) -> str:
    clean = code.strip().upper()
    if len(clean) != 3 or not clean.isalpha():
        raise ValueError(f"Invalid code: {code}")
    return clean

assert sanitize_currency("usd") == "USD"
assert sanitize_currency(" inr ") == "INR"

print("\nAll engineering recipes passed!")
```

## 5. Production Implementation & Self-Checking Unit Assertions

This production-grade `UppercaseFormatter` class encapsulates HTTP method checking, ISO currency normalization, and case equality testing.

```python
class UppercaseFormatter:
    @staticmethod
    def normalize_http_verb(verb: str) -> str:
        """Uppercases and validates standard HTTP method."""
        valid = {"GET", "POST", "PUT", "DELETE", "PATCH", "HEAD", "OPTIONS"}
        v = verb.strip().upper()
        if v not in valid:
            raise ValueError(f"Invalid verb: {verb}")
        return v

    @staticmethod
    def iso_currency(code: str) -> str:
        """Normalizes 3-letter currency codes."""
        c = code.strip().upper()
        if len(c) != 3 or not c.isalpha():
            raise ValueError(f"Invalid code: {code}")
        return c

    @staticmethod
    def is_equal_uppercase(s1: str, s2: str) -> bool:
        """Case-insensitive comparison via upper()."""
        return s1.upper() == s2.upper()


# ==========================================
# UNIT ASSERTION SUITE
# ==========================================
formatter = UppercaseFormatter()

# 1. Basic conversions
assert "hello".upper() == "HELLO"
assert "Welcome to Plus2Net 2026".upper() == "WELCOME TO PLUS2NET 2026"
assert "".upper() == ""

# 2. Unicode expansion check (length increases from 6 to 7)
assert "straße".upper() == "STRASSE"
assert len("straße") == 6
assert len("straße".upper()) == 7

# 3. HTTP Verb normalization
assert formatter.normalize_http_verb("  get ") == "GET"
assert formatter.normalize_http_verb("post") == "POST"

# 4. Currency code sanitization
assert formatter.iso_currency("usd") == "USD"
assert formatter.iso_currency(" inr ") == "INR"

# 5. Equality check
assert formatter.is_equal_uppercase("python", "PYTHON") is True
assert formatter.is_equal_uppercase("groß", "GROSS") is True

print("All str.upper() production unit assertions passed successfully!")
```
