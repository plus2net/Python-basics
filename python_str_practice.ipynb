### 1. Basic String Conversions
Convert numbers, booleans, and collections into readable strings.

```python
print("str(3.4):     ", str(3.4))
print("str([1, 2]):  ", str([1, 2]))
print("str(True):    ", str(True))
print("str(None):    ", str(None))
```

### 2. Byte Decoding with Encodings and Error Handlers
Decode byte arrays using strict, ignore, and replace policies.

```python
b = b"plus2net.com \xff \xfe"
print("errors='ignore': ", str(b, encoding='ascii', errors='ignore'))
print("errors='replace':", str(b, encoding='utf-8', errors='replace'))
```

### 3. str() vs. repr()
Compare human-readable output against unambiguous technical representation.

```python
import datetime
dt = datetime.datetime(2026, 10, 15, 8, 30)
print("str(): ", str(dt))
print("repr():", repr(dt))
```

### 4. Custom Class __str__() Implementation
Implement readable string representations for user objects.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def __str__(self):
        return f"{self.name} for {self.price}"

p = Product("Laptop", 999.00)
print("str(p):", str(p))
assert str(p) == "Laptop for 999.0"
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying str() mechanics.

```python
assert str(42) == "42"
assert str(3.14) == "3.14"
assert str(b"test", encoding="utf-8") == "test"
assert str(True) == "True"

print("All str() unit assertions passed! 100% verified.")
```
