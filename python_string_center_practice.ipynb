## 1. Basic Centering & Field Width Mechanics

`str.center(width[, fillchar])` centers a string within a field of total length `width`.
If `width <= len(s)`, Python returns the string unaltered without truncation.

```python
title = "plus2net.com"
centered = title.center(20)

print("Original: ", repr(title))
print("Centered: ", repr(centered))
print("Length:   ", len(centered))

assert len(centered) == 20
assert centered == "    plus2net.com    "

# Width smaller than length: never truncates
long_word = "PythonProgramming"
assert long_word.center(5) == "PythonProgramming"

print("\nAll basic centering assertions passed!")
```

## 2. The Odd-Padding Asymmetry Rule

When `(width - len(s))` is odd, the extra padding character is placed on the **right margin**:
`left_pad = (width - len(s)) // 2`
`right_pad = (width - len(s)) - left_pad`

```python
# Even padding: 3 chars in 'cat', width 7 -> 4 padding (2 left, 2 right)
even = "cat".center(7, "-")
print("Even padding:", repr(even))
assert even == "--cat--"

# Odd padding: 3 chars in 'cat', width 6 -> 3 padding (1 left, 2 right)
odd = "cat".center(6, "-")
print("Odd padding: ", repr(odd))
assert odd == "-cat--"

print("\nOdd-margin asymmetry assertions passed!")
```

## 3. Custom Fill Characters & Single-Character Constraint

`fillchar` must be exactly one character. Passing empty or multi-character strings raises `TypeError`.

```python
header = " SYSTEM REBOOT "
assert header.center(30, "=") == "======= SYSTEM REBOOT ========"
assert header.center(30, "*") == "******* SYSTEM REBOOT ********"

# TypeError on invalid fillchar length
try:
    header.center(30, "==")
    assert False, "Expected TypeError not raised"
except TypeError as e:
    assert "must be exactly one character" in str(e)

print("Fill character assertions passed!")
```

## 4. Alignment Comparison: `center()` vs. f-Strings

In f-strings, `^` represents center alignment: `f"{s:{fill}^{width}}"`.

```python
word = "DATA"
width = 10
fill = "-"

via_method = word.center(width, fill)
via_fstring = f"{word:{fill}^{width}}"

print("Method:  ", via_method)
print("f-string:", via_fstring)
assert via_method == via_fstring == "---DATA---"

print("Alignment equivalence assertions passed!")
```

## 5. Real-World Engineering Recipes

Explore three common software engineering patterns utilizing `str.center()`:
1. CLI Terminal Section Banners
2. Receipt / Invoice Section Dividers
3. ASCII Table Row Cells

```python
# Recipe 1: CLI Banner
def cli_banner(title: str, total_width: int = 40, char: str = "=") -> str:
    return f" {title.strip().upper()} ".center(total_width, char)

b1 = cli_banner("deploy started", 36)
print("Recipe 1 Banner:\n", b1)
assert b1 == "========== DEPLOY STARTED =========="

# Recipe 2: Receipt divider
def receipt_divider(title: str, width: int = 30) -> str:
    return title.center(width, "-")

r1 = receipt_divider(" TOTAL DUE ")
print("Recipe 2 Divider:\n", r1)
assert r1 == "--------- TOTAL DUE ----------"

# Recipe 3: Table row cells
def table_row(cols: list[str], widths: list[int]) -> str:
    return "|" + "|".join(c.center(w) for c, w in zip(cols, widths)) + "|"

row = table_row(["101", "Auth API", "OK"], [6, 12, 6])
print("Recipe 3 Table Row:\n", row)
assert row == "| 101  |  Auth API  |  OK  |"

print("\nAll engineering recipes passed!")
```

## 6. Production Implementation & Self-Checking Unit Assertions

This production-grade `TextCenterer` class encapsulates banner generation, table cell centering, and automated unit assertions.

```python
class TextCenterer:
    @staticmethod
    def make_banner(title: str, width: int = 40, border: str = "=") -> str:
        """Formats a bordered banner with spaces around title."""
        padded = f" {title.strip()} "
        return padded.center(width, border)

    @staticmethod
    def center_cell(content: str, column_width: int) -> str:
        """Centers text within a table column."""
        return content.center(column_width)


# ==========================================
# UNIT ASSERTION SUITE
# ==========================================
centerer = TextCenterer()

# 1. Basic space centering
assert "test".center(8) == "  test  "
assert "test".center(4) == "test"
assert "test".center(2) == "test"

# 2. Odd-padding asymmetry check
assert "cat".center(6, "-") == "-cat--"
assert "cat".center(7, "-") == "--cat--"

# 3. Custom fillchar
assert "ABC".center(9, "*") == "***ABC***"

# 4. Banner generation
banner = centerer.make_banner("READY", 20, "=")
assert banner == "====== READY ======="
assert len(banner) == 20

print("All str.center() production unit assertions passed successfully!")
```
