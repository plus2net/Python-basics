### 1. Basic Exponential Calculations
`math.exp(x)` computes $e^x$ directly using compiled C platform libraries.

```python
import math

print("e^0: ", math.exp(0))
print("e^1: ", math.exp(1))
print("e^5: ", math.exp(5))
print("e^-2:", math.exp(-2))
```

### 2. Logarithmic Inverse Verification
`math.exp()` is the exact inverse of `math.log()`.

```python
import math

x = 5.0
print("Original x:           ", x)
print("exp(log(x)):          ", math.exp(math.log(x)))
print("Is close to original? ", math.isclose(math.exp(math.log(x)), x))
```

### 3. OverflowError vs. Negative Underflow
For $x > 709.78$, $e^x$ exceeds standard 64-bit float representation and raises `OverflowError`. Extreme negative values cleanly underflow to `0.0`.

```python
import math

try:
    math.exp(710)
except OverflowError as err:
    print("Caught expected OverflowError:", err)

print("Extreme negative underflow exp(-800):", math.exp(-800))
```

### 4. Continuous Compounding Financial Growth
Calculate continuous compounding growth using $A = P \cdot e^{rt}$.

```python
import math

def continuous_compound(principal, annual_rate_pct, years):
    r = annual_rate_pct / 100.0
    return principal * math.exp(r * years)

# $1,000 invested at 5% for 3 years
growth = continuous_compound(1000, 5, 3)
print("Compounded Balance: $", round(growth, 2))
```

### 5. Hands-on Practice Exercise: Sigmoid Activation Function
Run the self-checking unit assertions below to verify a numerically stable logistic sigmoid function.

```python
def sigmoid(x):
    if x < -700:
        return 0.0
    elif x > 700:
        return 1.0
    return 1.0 / (1.0 + math.exp(-x))

# Unit Assertions
assert sigmoid(0) == 0.5
assert round(sigmoid(2), 4) == 0.8808
assert round(sigmoid(-2), 4) == 0.1192
assert sigmoid(-800) == 0.0
assert sigmoid(800) == 1.0

print("Sigmoid(0):  ", sigmoid(0))
print("Sigmoid(2):  ", round(sigmoid(2), 4))
print("Sigmoid(-2): ", round(sigmoid(-2), 4))
print("All sigmoid unit assertions passed successfully!")
```
