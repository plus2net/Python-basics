# Python math.fmod(): Float Modulo vs. % Operator

### 1. Basic math.fmod() Calculations
`math.fmod(x, y)` returns the floating-point remainder of `x / y` as defined by the C standard library.

```python
import math

print("fmod(10, 2):  ", math.fmod(10, 2))
print("fmod(10, 3):  ", math.fmod(10, 3))
print("fmod(10.5, 4):", math.fmod(10.5, 4))
```

### 2. Sign Differences: fmod() vs. % Operator
`math.fmod()` retains the sign of `x` (dividend), whereas `%` retains the sign of `y` (divisor).

```python
import math

x, y = -25, 3
print("math.fmod(-25, 3):", math.fmod(x, y))  # -1.0
print("-25 % 3:         ", x % y)          # 2

a, b = 25, -3
print("math.fmod(25, -3): ", math.fmod(a, b))  # 1.0
print("25 % -3:          ", a % b)          # -2
```

### 3. Floating-Point Precision with math.isclose()
Because binary floats can introduce tiny rounding residues, compare floating-point results with `math.isclose()`.

```python
import math

res = math.fmod(10.5, 3.2)
print("Raw result:", res)
print("Is close to 0.9?", math.isclose(res, 0.9))
```

### 4. Zero Division Error Handling
`math.fmod(x, 0)` raises `ValueError: math domain error`, whereas `x % 0` raises `ZeroDivisionError`.

```python
import math

try:
    math.fmod(10, 0)
except ValueError as err:
    print("Caught expected domain error:", err)
```

### 5. Hands-on Practice Exercise: Symmetric Angle Normalizer
Run the self-checking unit assertions below to verify that angles are normalized into the range `[-180, 180)`.

```python
def normalize_heading(angle):
    wrapped = math.fmod(angle + 180, 360)
    if wrapped < 0:
        wrapped += 360
    return wrapped - 180

# Unit Assertions
assert normalize_heading(0) == 0.0
assert normalize_heading(190) == -170.0
assert normalize_heading(-200) == 160.0
assert normalize_heading(360) == 0.0
assert normalize_heading(725) == 5.0

print("Heading for 190 deg: ", normalize_heading(190))
print("Heading for -200 deg:", normalize_heading(-200))
print("All heading normalizer assertions passed successfully!")
```
