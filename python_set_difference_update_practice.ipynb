### 1. Basic In-Place Set Subtraction
`A.difference_update(B)` removes all elements of `B` from `A` directly in-place.

```python
A = {"a", "b", "c", "d"}
B = {"a", "c", "z"}

# Remove elements of B from A in-place
A.difference_update(B)
print("Set A after subtraction:", A)  # {'b', 'd'}
print("Set B untouched:        ", B)  # {'a', 'c', 'z'}
```

### 2. Multi-Iterable Subtraction
Pass multiple iterables simultaneously to strip elements in a single operation.

```python
pool = {1, 2, 3, 4, 5, 6, 7, 8}
blacklist = {2, 4}
filter_list = [6, 8]
single_tup = (1,)

pool.difference_update(blacklist, filter_list, single_tup)
print("Filtered pool:", pool)  # {3, 5, 7}
```

### 3. The String Character Unpacking Trap
Strings are iterables of characters. Wrap strings in a list when removing full words.

```python
# Trap: Passing a bare string
chars = {"a", "b", "c", "x", "y"}
chars.difference_update("Alex")  # Removes 'A', 'l', 'e', 'x'
print("After bare string:", chars)  # {'a', 'b', 'c', 'y'}

# Correct: Wrap in a list
words = {"admin", "moderator", "user"}
words.difference_update(["admin"])
print("Proper word removal:", words)  # {'moderator', 'user'}
```

### 4. Real-World Recipe: Stop-Words Filtration in NLP
Purge common grammatical stop-words from an in-memory document vocabulary.

```python
def clean_vocabulary(raw_words: set, stop_words: list) -> set:
    raw_words.difference_update(stop_words)
    return raw_words

vocab = {"python", "the", "is", "great", "language", "and"}
stops = ["the", "is", "and", "of", "to"]

cleaned = clean_vocabulary(vocab, stops)
print("Meaningful vocabulary:", cleaned)  # {'python', 'great', 'language'}
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying multi-iterable purging and in-place mutation.

```python
class DiffUpdateVerifier:
    @staticmethod
    def purge(target: set, *iterables) -> set:
        target.difference_update(*iterables)
        return target

v = DiffUpdateVerifier()
s = {"a", "b", "c", "d"}
assert v.purge(s, ["a"], ("c",), {"x"}) == {"b", "d"}
assert s == {"b", "d"}

num_set = {1, 2, 3}
ret = num_set.difference_update({2})
assert ret is None
assert num_set == {1, 3}

print("All set.difference_update() unit assertions passed successfully! 100% verified.")
```
