### 1. Basic Calculations & Odd Function Symmetry
`math.sinh(x)` computes the hyperbolic sine of `x`. It satisfies odd symmetry: `sinh(-x) == -sinh(x)`, passing through the origin at `sinh(0.0) == 0.0`.


```python
import math

print("sinh(0.0): ", math.sinh(0.0))   # 0.0
print("sinh(1.0): ", round(math.sinh(1.0), 6))   # 1.175201
print("sinh(-1.0):", round(math.sinh(-1.0), 6))  # -1.175201
print("sinh(2.0): ", round(math.sinh(2.0), 6))   # 3.626860
print("sinh(-2.0):", round(math.sinh(-2.0), 6))  # -3.626860
```

### 2. Exponential Equivalence: (e^x - e^-x) / 2
Verify that `math.sinh(x)` yields results identical to the exponential formulation using `math.exp()`.


```python
import math

def formula_sinh(x):
    return (math.exp(x) - math.exp(-x)) / 2.0

test_values = [0.0, 0.5, 1.5, 3.5, 10.0]
for val in test_values:
    b = math.sinh(val)
    f = formula_sinh(val)
    assert math.isclose(b, f)
    print(f"x = {val:4.1f} | math.sinh = {b:12.6f} | formula = {f:12.6f} | Match: True")
```

### 3. Fundamental Hyperbolic Identities
Verify the Pythagorean hyperbolic identity and the double-angle identity: `sinh(2x) == 2 * sinh(x) * cosh(x)`.


```python
import math

x = 1.5
sinh_2x = math.sinh(2 * x)
double_angle = 2 * math.sinh(x) * math.cosh(x)
print("sinh(2x):     ", sinh_2x)
print("2*sinh*cosh:  ", double_angle)
assert math.isclose(sinh_2x, double_angle)
print("Identity hold: True")
```

### 4. Overflow Handling with Large Inputs
Because `sinh(x)` grows exponentially, values exceeding `|x| ~ 710.475` raise `OverflowError: math range error`.


```python
import math

print("sinh(100):", math.sinh(100))
print("sinh(710):", math.sinh(710))

try:
    math.sinh(711)
except OverflowError as err:
    print("Caught expected OverflowError on sinh(711):", err)
```

### 5. Hands-on Practice Exercise: Ocean Wave Particle Velocity Model
In coastal engineering (Airy wave theory), vertical wave particle velocity is modeled as:
`w(z) = W_0 * sinh(k * (z + d)) / sinh(k * d)`.
Implement `vertical_particle_velocity()` and verify with self-checking assertions.


```python
import math

def vertical_particle_velocity(surface_velocity, depth_z, water_depth_d, wave_number_k):
    """
    Calculates vertical velocity of water particles at elevation z:
    w(z) = W_0 * sinh(k * (z + d)) / sinh(k * d)
    """
    if water_depth_d <= 0 or wave_number_k <= 0:
        raise ValueError("Water depth and wave number must be strictly positive.")
    if depth_z > 0.0 or depth_z < -water_depth_d:
        raise ValueError("Elevation z must lie within [-water_depth_d, 0.0].")
        
    numerator = math.sinh(wave_number_k * (depth_z + water_depth_d))
    denominator = math.sinh(wave_number_k * water_depth_d)
    return round(surface_velocity * (numerator / denominator), 4)

# Unit Assertions
assert vertical_particle_velocity(1.5, 0.0, 20.0, 0.1) == 1.5   # Surface
assert vertical_particle_velocity(1.5, -20.0, 20.0, 0.1) == 0.0  # Seabed
assert vertical_particle_velocity(1.5, -10.0, 20.0, 0.1) == 0.4859 # Mid-depth

# Out of bounds check
try:
    vertical_particle_velocity(1.5, 5.0, 20.0, 0.1)
    assert False, "Should raise ValueError"
except ValueError:
    pass

print("Surface velocity (z=0m): ", vertical_particle_velocity(2.0, 0.0, 30.0, 0.08), "m/s")
print("Seabed velocity (z=-30m):", vertical_particle_velocity(2.0, -30.0, 30.0, 0.08), "m/s")
print("All wave mechanics velocity assertions passed successfully!")
```
