### 1. Binary Representations and Formatted Bitstrings
Use `bin()`, `int(..., 2)`, and f-string formatting (`08b`) to inspect numbers in binary.

```python
val = 120
print(f"Decimal: {val}")
print(f"bin():   {bin(val)}")
print(f"Sliced:  {bin(val)[2:]}")
print(f"Padded:  {val:08b}")

# Binary literal back to integer
parsed = int('1111000', 2)
print(f"Parsed int: {parsed}")
assert parsed == 120
```

### 2. Bitwise AND (&) and Bitwise OR (|)
- Bitwise AND (`&`): Sets bit to `1` only if both bits are `1`.
- Bitwise OR (`|`): Sets bit to `1` if at least one bit is `1`.

```python
a = 12  # 00001100
b = 15  # 00001111

and_res = a & b
or_res = a | b

print(f"{a} & {b} = {and_res} (binary: {and_res:08b})")
print(f"{a} | {b} = {or_res} (binary: {or_res:08b})")

assert and_res == 12
assert or_res == 15

# Parity test via Bitwise AND
is_even = lambda n: (n & 1) == 0
assert is_even(42) is True
assert is_even(37) is False
```

### 3. Bitwise XOR (^) & Reversible Properties
- Bitwise XOR (`^`): Sets bit to `1` if bits are different.
- Properties: `x ^ x == 0`, `x ^ 0 == x`, `x ^ y ^ y == x`.

```python
a = 12  # 00001100
b = 15  # 00001111
xor_res = a ^ b
print(f"{a} ^ {b} = {xor_res} (binary: {xor_res:08b})")
assert xor_res == 3

# Variable swap without temporary variable
x, y = 77, 188
x ^= y
y ^= x
x ^= y
print(f"Swapped: x={x}, y={y}")
assert x == 188 and y == 77

# Find single unique number among pairs
items = [9, 3, 5, 3, 9]
unique = 0
for item in items:
    unique ^= item
print(f"Unique item: {unique}")
assert unique == 5
```

### 4. Bitwise NOT (~) & Two's Complement Arithmetic
Python represents negative numbers using two's complement: `~x = -(x + 1)`.

```python
print(f"~15  = {~15}")
print(f"~4   = {~4}")
print(f"~0   = {~0}")
print(f"~-22 = {~-22}")

assert ~15 == -16
assert ~4 == -5
assert ~0 == -1
assert ~-22 == 21

# Fixed 8-bit unsigned NOT emulation
val = 0b00001111
not_8bit = ~val & 0xFF
print(f"8-bit NOT of {val:08b}: {not_8bit:08b} ({not_8bit})")
assert not_8bit == 240
```

### 5. Shift Operators (<< and >>) for Fast Arithmetic
- Left shift (`<< n`): Multiplies by 2^n.
- Right shift (`>> n`): Floor divides by 2^n.

```python
val = 14
print(f"{val} << 1 = {val << 1} (14 * 2)")
print(f"{val} << 2 = {val << 2} (14 * 4)")
print(f"{val} >> 1 = {val >> 1} (14 // 2)")
print(f"{val} >> 2 = {val >> 2} (14 // 4)")

assert (14 << 1) == 28
assert (14 << 2) == 56
assert (14 >> 1) == 7
assert (14 >> 2) == 3
assert (90 >> 3) == 11
```

### 6. Common Bit Manipulation Recipes
- Check power of 2: `(n > 0) and (n & (n - 1) == 0)`
- 24-bit RGB Color Packing and Unpacking

```python
def is_power_of_two(n):
    return n > 0 and (n & (n - 1)) == 0

assert is_power_of_two(1) is True
assert is_power_of_two(16) is True
assert is_power_of_two(18) is False

# Color packing
def pack_rgb(r, g, b):
    return (r << 16) | (g << 8) | b

def unpack_rgb(hex_val):
    return (hex_val >> 16) & 0xFF, (hex_val >> 8) & 0xFF, hex_val & 0xFF

hex_color = pack_rgb(255, 128, 64)
print(f"Packed: 0x{hex_color:06X}")
assert hex_color == 0xFF8040
assert unpack_rgb(hex_color) == (255, 128, 64)
```

### 7. Exercise: Unix-Style File Permission Bitmask Engine
Run and test the complete `PermissionManager` with self-checking assertions.

```python
class PermissionManager:
    READ    = 0b100  # 4
    WRITE   = 0b010  # 2
    EXECUTE = 0b001  # 1

    def __init__(self, initial_perms=0):
        self.perms = initial_perms

    def grant(self, flag):
        self.perms |= flag

    def revoke(self, flag):
        self.perms &= ~flag

    def has_permission(self, flag):
        return (self.perms & flag) == flag

    def toggle(self, flag):
        self.perms ^= flag

    def to_rwx(self):
        r = 'r' if self.has_permission(self.READ) else '-'
        w = 'w' if self.has_permission(self.WRITE) else '-'
        x = 'x' if self.has_permission(self.EXECUTE) else '-'
        return f"{r}{w}{x}"

# Run comprehensive unit tests
pm = PermissionManager()
assert pm.to_rwx() == "---"

pm.grant(PermissionManager.READ | PermissionManager.WRITE)
assert pm.perms == 6
assert pm.to_rwx() == "rw-"

pm.grant(PermissionManager.EXECUTE)
assert pm.perms == 7
assert pm.to_rwx() == "rwx"

pm.revoke(PermissionManager.WRITE)
assert pm.perms == 5
assert pm.to_rwx() == "r-x"

pm.toggle(PermissionManager.READ)
assert pm.to_rwx() == "--x"

pm.toggle(PermissionManager.READ)
assert pm.to_rwx() == "r-x"

print("Final permission status:", pm.to_rwx(), f"(binary: {pm.perms:03b})")
print("All bitwise operator tests and assertions passed successfully!")
```
