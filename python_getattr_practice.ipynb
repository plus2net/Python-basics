### 1. Basic Attribute Retrieval & Default Values
`getattr(obj, name, default)` retrieves attribute values dynamically without crashing on missing keys.

```python
class Animal:
    species = "carnivores"
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

tiger = Animal("Ronald", 5)

# Read existing attributes
print("Age:", getattr(tiger, "age"))
print("Species:", getattr(tiger, "species"))

# Missing attribute with fallback
height = getattr(tiger, "height", "No such attribute")
print("Height with default:", height)
```

### 2. Dynamic Method Dispatch
Retrieve bound methods dynamically and execute them.

```python
class MathOps:
    def add(self, x, y):
        return x + y
    def multiply(self, x, y):
        return x * y

ops = MathOps()
action = "multiply"

func = getattr(ops, action, None)
if callable(func):
    print(f"{action}(6, 7) =", func(6, 7))
```

### 3. Non-Identifier Attribute Retrieval
Retrieve keys with hyphens that are illegal under dot notation.

```python
class Headers:
    pass

h = Headers()
setattr(h, "content-type", "application/json")
setattr(h, "cache-control", "no-cache")

print("Content-Type:", getattr(h, "content-type"))
print("Cache-Control:", getattr(h, "cache-control"))
```

### 4. Chained & Nested Path Lookup
Traverse nested object structures dynamically using reduce.

```python
import functools

class Engine:
    hp = 450

class Vehicle:
    def __init__(self):
        self.engine = Engine()

def resolve_path(obj, path, default=None):
    try:
        return functools.reduce(getattr, path.split("."), obj)
    except AttributeError:
        return default

v = Vehicle()
print("Nested HP:", resolve_path(v, "engine.hp"))
print("Missing nested:", resolve_path(v, "engine.turbo.psi", "N/A"))
```

### 5. Production Recipe: Event Dispatcher
Route actions to handler methods cleanly.

```python
class Dispatcher:
    def on_login(self, user):
        return f"User {user} authenticated."
    def on_logout(self, user):
        return f"User {user} signed out."
    def dispatch(self, event_name, user):
        handler = getattr(self, f"on_{event_name}", None)
        if handler:
            return handler(user)
        return f"No handler for event: {event_name}"

d = Dispatcher()
print(d.dispatch("login", "alice"))
print(d.dispatch("unknown", "alice"))
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying getattr() mechanics.

```python
class GetattrVerifier:
    @staticmethod
    def test_all():
        class Item:
            tag = "gadget"
            def __init__(self, price):
                self.price = price
                
        it = Item(100)
        assert getattr(it, "price") == 100
        assert getattr(it, "tag") == "gadget"
        assert getattr(it, "sku", "DEFAULT_SKU") == "DEFAULT_SKU"
        
        try:
            getattr(it, "nonexistent")
            assert False
        except AttributeError:
            pass
            
        print("All getattr() unit assertions passed successfully! 100% verified.")

GetattrVerifier.test_all()
```
