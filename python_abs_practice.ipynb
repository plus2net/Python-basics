### 1. Basic Absolute Values and Type Preservation
Observe that abs() preserves int and float types.

```python
print("abs(-4):  ", abs(-4), type(abs(-4)))
print("abs(-4.6):", abs(-4.6), type(abs(-4.6)))
```

### 2. Complex Numbers Modulus
Compute the Euclidean length of complex vectors.

```python
z = 3 + 4j
print("Magnitude of 3 + 4j:", abs(z))
assert abs(z) == 5.0
```

### 3. Built-in abs() vs. math.fabs()
Compare type preservation and complex number support.

```python
import math
print("abs(-10):       ", abs(-10), type(abs(-10)))
print("math.fabs(-10): ", math.fabs(-10), type(math.fabs(-10)))
```

### 4. Custom Class __abs__() Implementation
Define custom vector magnitude logic using dunder methods.

```python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __abs__(self):
        return (self.x**2 + self.y**2) ** 0.5

v = Vector2D(6, 8)
print("Vector magnitude:", abs(v))
assert abs(v) == 10.0
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying abs() mechanics.

```python
assert abs(-5) == 5
assert abs(-5.0) == 5.0
assert abs(3 + 4j) == 5.0
assert isinstance(abs(-10), int)
assert isinstance(abs(-10.0), float)

print("All abs() unit assertions passed! 100% verified.")
```
