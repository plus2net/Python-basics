### 1. Basic Expression Evaluation

```python
# Math calculation
math_val = eval("(25 * 4) + (100 // 5) - 15")
print("Math result:      ", math_val)

# Boolean comparison
bool_val = eval("10 > 5 and 'py' in 'python'")
print("Boolean result:   ", bool_val)

# String operations
str_val = eval("'plus2net'.upper() + ' tutorial'")
print("String result:    ", str_val)
```

### 2. Custom Scopes & Builtins Isolation

```python
import math

# Explicitly exposed functions and variables
custom_scope = {"fsum": math.fsum, "factor": 2.5}
res = eval("fsum([10, 20, 30]) * factor", custom_scope)
print("Calculation with custom scope:", res)

# Suppressing builtins entirely
strict_scope = {"__builtins__": {}}
try:
    eval("sum([1, 2, 3])", strict_scope)
except NameError as err:
    print("Correctly caught suppressed builtin:", err)
```

### 3. Precompiled Bytecode in Loops

```python
import time

expr = "x ** 2 + 2 * x + 1"
iterations = 50000

# Raw string evaluation
t0 = time.perf_counter()
for x in range(iterations):
    _ = eval(expr)
raw_duration = time.perf_counter() - t0

# Precompiled evaluation
code_obj = compile(expr, "<benchmark>", "eval")
t1 = time.perf_counter()
for x in range(iterations):
    _ = eval(code_obj)
precompiled_duration = time.perf_counter() - t1

print(f"Raw string eval:     {raw_duration:.4f} s")
print(f"Precompiled bytecode:{precompiled_duration:.4f} s")
print(f"Performance speedup: {raw_duration / precompiled_duration:.2f}x faster")
```

### 4. Safe Parsing with ast.literal_eval()

```python
import ast

serialized = "{'topic': 'Python eval()', 'tags': ['builtins', 'metaprogramming'], 'rating': 5}"
parsed_obj = ast.literal_eval(serialized)

print("Parsed dictionary:", parsed_obj)
print("Dictionary type:  ", type(parsed_obj).__name__)

# Disallowed code triggers ValueError safely
try:
    ast.literal_eval("__import__('os').system('echo unsafe')")
except ValueError as err:
    print("Safely rejected code execution:", err)
```

### 5. AST-Validated Safe Formula Calculator

```python
import ast

def safe_evaluator(expr_str: str, context: dict) -> float:
    tree = ast.parse(expr_str, mode='eval')
    for node in ast.walk(tree):
        if not isinstance(node, (ast.Expression, ast.BinOp, ast.UnaryOp, 
                                ast.Constant, ast.Name, ast.Load,
                                ast.Add, ast.Sub, ast.Mult, ast.Div, ast.Pow)):
            raise ValueError(f"Disallowed expression element: {type(node).__name__}")
    compiled_expr = compile(tree, "<safe_calc>", "eval")
    return eval(compiled_expr, {"__builtins__": {}}, context)

# Test legitimate formula
formula = "base_price * quantity * (1 - discount)"
params = {"base_price": 120.0, "quantity": 5, "discount": 0.15}
result = safe_evaluator(formula, params)
print("Safe formula evaluation result:", result)
```
