### 1. Memory-Efficient Parsing with iterparse

```python
import xml.etree.ElementTree as ET

# Create sample large XML file
xml_data = """<Root>
    <Row>
        <id>1</id>
        <name>John Deo</name>
        <class>Four</class>
        <mark>75</mark>
        <gender>female</gender>
    </Row>
    <Row>
        <id>2</id>
        <name>Max Ruin</name>
        <class>Three</class>
        <mark>85</mark>
        <gender>male</gender>
    </Row>
</Root>"""

with open("large_file.xml", "w") as f:
    f.write(xml_data)

data = []
for event, elem in ET.iterparse("large_file.xml", events=("start", "end")):
    if event == "start" and elem.tag == "Row":
        student_data = {}
    elif event == "end" and elem.tag == "Row":
        data.append(student_data)
        elem.clear()
    elif event == "end" and elem.tag in ("id", "name", "class", "mark", "gender"):
        student_data[elem.tag] = elem.text

print(f"Processed {len(data)} rows successfully using iterparse!")
print(data)
```
