### 1. Basic In-Place Symmetric Difference
`A.symmetric_difference_update(B)` mutates `A` in-place, keeping only elements present in either `A` or `B`, but not in both.

```python
A = {'a', 'b', 'c'}
B = {'a', 'y', 'z'}

initial_id = id(A)

# In-place symmetric difference mutation
ret = A.symmetric_difference_update(B)

print("Mutated Set A:", sorted(A))
print("Unchanged Set B:", sorted(B))
print("Return Value:", ret)  # None
print("Memory ID Preserved:", id(A) == initial_id)  # True
```

### 2. Heterogeneous Collections: Lists and Strings
Passing a list operates cleanly. Passing a string decomposes it into individual character tokens.

```python
# List input
set_data = {'a', 'b', 'c'}
list_input = ['a', 'x', 'y']
set_data.symmetric_difference_update(list_input)
print("After list input:", sorted(set_data))

# String unpacking nuance
char_set = {'a', 'b', 'c', 'x', 'y'}
char_set.symmetric_difference_update('Alex')
print("After string input:", sorted(char_set))

# Proper word-level symmetric difference (wrap in collection)
tags = {"python", "flask"}
tags.symmetric_difference_update(["flask", "fastapi"])
print("Word-level tags:", sorted(tags))
```

### 3. Method vs Operator (`^=`)
The `^=` operator requires the right-hand operand to be a set; `symmetric_difference_update()` accepts any iterable.

```python
s = {1, 2, 3}

# Method accepts list:
s.symmetric_difference_update([3, 4])
print("Method result with list:", s)

# Operator with list raises TypeError:
try:
    s ^= [4, 5]
except TypeError as e:
    print("Operator raised TypeError:", e)
```

### 4. Real-World Recipe: State Toggling (Reversibility)
Symmetric difference acts as a toggle. Applying the same delta twice restores the exact original state.

```python
active_flags = {"beta_ui", "dark_mode", "telemetry"}
toggle_batch = {"beta_ui", "experimental_ai"}

# 1st sync toggles flags
active_flags.symmetric_difference_update(toggle_batch)
print("Toggled flags:", sorted(active_flags))

# 2nd sync restores original state
active_flags.symmetric_difference_update(toggle_batch)
print("Restored flags:", sorted(active_flags))
assert active_flags == {"beta_ui", "dark_mode", "telemetry"}
```

### 5. Real-World Recipe: Cache vs Database Delta Auditing
Audit drift between Redis cache keys and database primary keys.

```python
cached_ids = {101, 102, 103, 104}
db_active_ids = [102, 103, 105]

sync_delta = cached_ids.copy()
sync_delta.symmetric_difference_update(db_active_ids)
print("Desynchronized IDs:", sorted(sync_delta))
assert sync_delta == {101, 104, 105}
```
