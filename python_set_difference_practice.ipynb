### 1. Basic Two-Set Subtraction
`A.difference(B)` returns a new set containing elements in `A` that are not in `B`. Neither original set is mutated.

```python
A = {1, 2, 3}
B = {3, 4, 5}

# Using arithmetic subtraction operator
result_op = A - B
print("A - B:", result_op)  # {1, 2}

# Using difference() method
result_meth = A.difference(B)
print("A.difference(B):", result_meth)  # {1, 2}

# Verify originals remain unmodified
print("Original A:", A)  # {1, 2, 3}
print("Original B:", B)  # {3, 4, 5}
```

### 2. Method vs Operator Flexibility
The `difference()` method accepts any iterable (list, tuple, string), whereas the `-` operator strictly requires set operands.

```python
base_set = {"a", "b", "c"}
filter_list = ["a", "x", "y"]

# Method accepts a list cleanly
clean_diff = base_set.difference(filter_list)
print("difference() with list:", clean_diff)  # {'b', 'c'}

# Operator with a list raises TypeError
try:
    invalid = base_set - filter_list
except TypeError as err:
    print("Caught expected error:", err)
```

### 3. Multi-Iterable Subtraction
Pass multiple iterables simultaneously to strip elements in a single atomic call.

```python
A = {"a", "b", "c", "d", "e"}
B = {"a", "y"}
C_list = ["b", "z"]
D_tuple = ("c",)

result = A.difference(B, C_list, D_tuple)
print("Filtered set:", result)  # {'d', 'e'}
```

### 4. String Character Unpacking Trap
Strings are iterables of characters. When filtering words, wrap them in a list or set.

```python
usernames = {"Alex", "Maria", "David"}

# Trap: bare string decomposes into 'A', 'l', 'e', 'x'
trap_res = usernames.difference("Alex")
print("Trap (bare string):", trap_res)  # {'Alex', 'Maria', 'David'}

# Correct: wrap word in list
correct_res = usernames.difference(["Alex"])
print("Correct (list of words):", correct_res)  # {'Maria', 'David'}
```

### 5. Production Recipe: RBAC Permission Auditing
Audit assigned permissions against required security compliance baselines.

```python
def audit_permissions(required: set[str], granted: set[str]) -> set[str]:
    """Returns permissions needed for compliance that are missing."""
    return required.difference(granted)

compliance_baseline = {"view_logs", "export_csv", "manage_roles", "2fa_enforced"}
analyst_role = {"view_logs", "export_csv"}

missing = audit_permissions(compliance_baseline, analyst_role)
print("Missing compliance permissions:", sorted(missing))  # ['2fa_enforced', 'manage_roles']
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying difference properties and immutability.

```python
class SetDifferenceVerifier:
    @staticmethod
    def test_all():
        s1 = {1, 2, 3, 4, 5}
        s2 = {4, 5, 6, 7}
        
        # Basic difference
        assert s1.difference(s2) == {1, 2, 3}
        assert s1 - s2 == {1, 2, 3}
        
        # Caller unmodified
        assert s1 == {1, 2, 3, 4, 5}
        
        # Heterogeneous iterables
        assert s1.difference([1], (2,), {3}) == {4, 5}
        
        # Disjoint subtraction returns identical elements
        assert s1.difference({99, 100}) == s1
        
        # Superset subtraction returns empty set
        assert s1.difference({1, 2, 3, 4, 5, 6}) == set()
        
        # Zero argument returns copy
        copy_s = s1.difference()
        assert copy_s == s1
        assert copy_s is not s1
        
        print("All set.difference() unit assertions passed successfully! 100% verified.")

SetDifferenceVerifier.test_all()
```
