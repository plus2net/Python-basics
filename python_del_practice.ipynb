# Python del Statement Practice

Run cells in order in Google Colab. This notebook demonstrates how to delete variables, remove list items by index or slice, and delete dictionary keys using the `del` statement. Save a copy in Google Drive to keep your edits.

### 1. Deleting Variables

```python
x = 100
print("Before deletion, x =", x)
del x
try:
    print(x)
except NameError as e:
    print("Successfully deleted x. Caught expected NameError:", e)
assert 'x' not in locals()
print("Variable deletion test passed!")
```

### 2. Deleting List Items and Slices

```python
numbers = [10, 20, 30, 40, 50]
del numbers[0]
print("After deleting index 0:", numbers)
assert numbers == [20, 30, 40, 50]

del numbers[1:3]
print("After deleting slice 1:3:", numbers)
assert numbers == [20, 50]
print("List deletion tests passed!")
```

### 3. Deleting Dictionary Keys

```python
student = {'id': 1, 'name': 'John', 'mark': 85}
del student['mark']
print("After deleting key 'mark':", student)
assert 'mark' not in student
assert student == {'id': 1, 'name': 'John'}
print("Dictionary key deletion test passed!")
```
