## Example 1: First working example

Read the code, predict the output and run this cell.

```python
n = "549"
my_sum = 0
for digit in n:
    my_sum += int(digit)
print("Sum of digits:", my_sum)
```

## Example 3: Use a while loop and remainder

Read the code, predict the output and run this cell.

```python
n = 1251
my_sum = 0
while n > 0:
    digit = n % 10
    my_sum += digit
    n //= 10
print("Sum of digits:", my_sum)
```

## Example 4: Write a reusable function

Read the code, predict the output and run this cell.

```python
def digit_sum(n):
    n = abs(n)
    total = 0
    while n:
        total += n % 10
        n //= 10
    return total

for value in (0, 549, -1251):
    print(value, digit_sum(value))
```

## Example 5: Find the sum recursively

Read the code, predict the output and run this cell.

```python
def recursive_digit_sum(n):
    n = abs(n)
    if n < 10:
        return n
    return n % 10 + recursive_digit_sum(n // 10)

print(recursive_digit_sum(1251))
```

## Example 6: One-line approach using sum()

Read the code, predict the output and run this cell.

```python
n = -7092
answer = sum(int(ch) for ch in str(abs(n)))
print(answer)
```

## Example 7: Handle invalid input

Read the code, predict the output and run this cell.

```python
raw = "12x"
try:
    n = int(raw)
    print(sum(map(int, str(abs(n)))))
except ValueError:
    print("Enter a whole number")
```
