### 1. Basic Two-Set Intersection
`A.intersection(B)` returns a new set containing elements present in both `A` and `B`. Neither input set is modified.

```python
A = {1, 2, 3}
B = {3, 4, 5}

# Using & operator
result_op = A & B
print("A & B:", result_op)  # {3}

# Using intersection() method
result_meth = A.intersection(B)
print("A.intersection(B):", result_meth)  # {3}

# Verify originals remain unmodified
print("Original A:", A)  # {1, 2, 3}
print("Original B:", B)  # {3, 4, 5}
```

### 2. Method vs Operator Flexibility
The `intersection()` method accepts any iterable (list, tuple, string), whereas the `&` operator strictly requires set operands.

```python
base_set = {"a", "b", "c"}
filter_list = ["a", "x", "y"]

# Method accepts a list cleanly
clean_intersection = base_set.intersection(filter_list)
print("intersection() with list:", clean_intersection)  # {'a'}

# Operator with a list raises TypeError
try:
    invalid = base_set & filter_list
except TypeError as err:
    print("Caught expected error:", err)
```

### 3. Multi-Iterable Intersection
Find elements common across multiple collections simultaneously.

```python
A = {"a", "b", "c"}
B = {"a", "y", "z"}
C_list = ["a", "k", "l"]
D_tuple = ("a", "m")

result = A.intersection(B, C_list, D_tuple)
print("Common elements:", result)  # {'a'}
```

### 4. String Character Unpacking Trap
Strings are iterables of characters. When matching full words, wrap them in a list or set.

```python
usernames = {"Alex", "Maria", "David"}

# Trap: bare string decomposes into 'A', 'l', 'e', 'x'
trap_res = usernames.intersection("Alex")
print("Trap (bare string):", trap_res)  # set()

# Correct: wrap word in list
correct_res = usernames.intersection(["Alex"])
print("Correct (list of words):", correct_res)  # {'Alex'}
```

### 5. Production Recipe: Shared Interest Matching
Find mutual interests across user profiles for content recommendations.

```python
def mutual_interests(user1: set[str], user2: list[str]) -> set[str]:
    """Returns topics shared by both users."""
    return user1.intersection(user2)

profile_1 = {"python", "docker", "machine-learning", "cloud"}
profile_2 = ["python", "docker", "go", "kubernetes"]

shared = mutual_interests(profile_1, profile_2)
print("Shared interest tags:", sorted(shared))  # ['docker', 'python']
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying intersection properties and immutability.

```python
class SetIntersectionVerifier:
    @staticmethod
    def test_all():
        s1 = {1, 2, 3, 4, 5}
        s2 = {4, 5, 6, 7}
        
        # Basic intersection
        assert s1.intersection(s2) == {4, 5}
        assert s1 & s2 == {4, 5}
        
        # Caller unmodified
        assert s1 == {1, 2, 3, 4, 5}
        
        # Heterogeneous iterables
        assert s1.intersection([4, 5, 99], (5, 6)) == {5}
        
        # Disjoint intersection yields empty set
        assert s1.intersection({99, 100}) == set()
        
        # Zero argument returns copy
        copy_s = s1.intersection()
        assert copy_s == s1
        assert copy_s is not s1
        
        print("All set.intersection() unit assertions passed successfully! 100% verified.")

SetIntersectionVerifier.test_all()
```
