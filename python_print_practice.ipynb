### 1. Custom Separators and Line Endings

```python
# Custom separators
print("Apple", "Banana", "Cherry", sep=" | ")

# Custom end parameter
print("Loading", end="...")
print("Ready!")
```

### 2. Collection Unpacking into print()

```python
data = ["alpha", "beta", "gamma", "delta"]
# Splat unpacking directly prints elements separated by commas
print(*data, sep=", ")
```

### 3. Stream Redirection with io.StringIO

```python
import io

stream = io.StringIO()
print("System initialized.", file=stream)
print("Port 8080 listening.", file=stream)

# Read back the captured output
output_log = stream.getvalue()
print("Captured in memory:\n" + output_log)
```

### 4. Dynamic In-Place Progress Bar

```python
import time

total_steps = 10
for step in range(1, total_steps + 1):
    pct = int((step / total_steps) * 100)
    print(f"\rProgress: {pct}% [{'#' * step}{'.' * (total_steps - step)}]", end="", flush=True)
    time.sleep(0.05)
print("\nTask complete!")
```

### 5. Self-Checking Unit Verification Suite

```python
import io

# Test 1: Separator test
b1 = io.StringIO()
print(1, 2, 3, sep=":", file=b1)
assert b1.getvalue().strip() == "1:2:3"

# Test 2: End character test
b2 = io.StringIO()
print("Hello", end="!", file=b2)
assert b2.getvalue() == "Hello!"

# Test 3: Unpacking test
b3 = io.StringIO()
words = ["plus2net", "python"]
print(*words, sep="_", file=b3)
assert b3.getvalue().strip() == "plus2net_python"

# Test 4: print() return value is None
assert print("test", file=io.StringIO()) is None

print("All print() unit assertions successfully verified! 100% passed.")
```
