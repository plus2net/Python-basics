### 1. Basic Iterator Traversal with next()
Extract iterators from sequences and step forward.

```python
text = "plus2net"
str_it = iter(text)

print("1st:", next(str_it))
print("2nd:", next(str_it))
print("3rd:", next(str_it))
```

### 2. Handling Exhaustion and Defaults
Observe StopIteration and suppress it using `next(it, default)`.

```python
items = iter([10, 20])
print(next(items, "EMPTY"))
print(next(items, "EMPTY"))
print(next(items, "EMPTY"))  # Safely returns default
```

### 3. Custom Class Iterator Protocol
Implement a custom countdown class supporting `__iter__` and `__next__`.

```python
class CountDown:
    def __init__(self, start):
        self.current = start
    def __iter__(self):
        return self
    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        val = self.current
        self.current -= 1
        return val

print("CountDown:", list(CountDown(4)))
```

### 4. Sentinel Callable Iteration
Use `iter(callable, sentinel)` to halt when a target value is reached.

```python
stream = [1, 2, 3, 99, 4, 5]
# Halts upon hitting sentinel 99
reader = iter(lambda: stream.pop(0), 99)
print("Consumed before 99:", list(reader))
```

### 5. Production Recipe: Stream Chunking
Chunk strings into fixed 2-character blocks.

```python
import io
buf = io.StringIO("ABCDEFGH")
chunks = list(iter(lambda: buf.read(2), ""))
print("Chunks:", chunks)
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying iter() protocol mechanics.

```python
it = iter("py")
assert next(it) == "p"
assert next(it) == "y"
assert next(it, "END") == "END"

vals = [10, 20, 0]
assert list(iter(lambda: vals.pop(0), 0)) == [10, 20]

print("All iter() unit assertions passed! 100% verified.")
```
