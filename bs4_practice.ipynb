## 5. Extract titles, headings, parent tags and H2 text

BeautifulSoup can find elements, inspect parent tags and extract text. The first sample uses `.string`; the next one uses `get_text()` for more reliable reading when tags contain nested elements. The collected H2 results depend on the current source webpage.

**Tutorial example 8 of 24.** Print the HTML title, H1, H1 string, and title parent name. **Expected:** page-specific HTML; the parent is usually `head`.

````python
import requests
link = "https://www.plus2net.com/html_tutorial/html-canvas.php"
content = requests.get(link)

from bs4 import BeautifulSoup
soup = BeautifulSoup(content.text, 'html.parser')
print(soup.title) # gets title tag
print(soup.h1)    # gets H1 tag
print(soup.h1.string)
#print(soup.title.parent) # full string within the parent tag
print(soup.title.parent.name)
````

**Tutorial example 9 of 24.** Find title and H1 safely before extracting text and the parent tag name.

````python
title_tag = soup.find("title")
h1_tag = soup.find("h1")
print(title_tag.get_text(strip=True) if title_tag else "No title")
print(h1_tag.get_text(" ", strip=True) if h1_tag else "No H1")
print(title_tag.parent.name if title_tag else "No title parent")
````

**Tutorial example 10 of 24.** Collect H2 elements and their direct `.string` values.

````python
h2_tags=soup.find_all('h2') # List of all h2 tags
h2_strings = [tag.string for tag in h2_tags] # remove the h2 tag and keep only string part
````

**Tutorial example 11 of 24.** Use `get_text(" ", strip=True)` to include nested heading text.

````python
h2_titles = [tag.get_text(" ", strip=True) for tag in soup.find_all("h2")]
for title in h2_titles:
    print(title)
````

## 9. Collect links and convert relative URLs to absolute URLs

The first example extracts raw `href` values; the second converts relative addresses to absolute URLs, keeps HTTP(S) links, and removes duplicates.

**Tutorial example 19 of 24.** Use `findAll("a")` to retrieve raw link targets. **Expected:** a Python list of href values.

````python
import requests
link="https://www.plus2net.com"
content = requests.get(link)
from bs4 import BeautifulSoup
soup = BeautifulSoup(content.text, 'html.parser')

my_links = []
for ln in soup.findAll('a'):
    my_links.append(ln.get('href'))
print(my_links)
````

**Tutorial example 20 of 24.** Use `find_all("a", href=True)` with `urljoin` to print unique absolute links.

````python
from urllib.parse import urljoin
from bs4 import BeautifulSoup
import requests

url = "https://www.plus2net.com/"
response = requests.get(url, timeout=15)
response.raise_for_status()
soup = BeautifulSoup(response.text, "html.parser")

links = []
for tag in soup.find_all("a", href=True):
    absolute_url = urljoin(response.url, tag["href"])
    if absolute_url.startswith(("http://", "https://")):
        links.append(absolute_url)

for link in sorted(set(links)):
    print(link)
````

### Practice canonical matching without MySQL

This standalone exercise demonstrates the canonical comparison without requiring a database, credentials, or external requests.

````python
from bs4 import BeautifulSoup
from urllib.parse import urljoin

examples = [
    {
        "url": "https://www.plus2net.com/python/bs4.php",
        "html": '<html><head><link rel="canonical" href="https://www.plus2net.com/python/bs4.php"></head></html>',
    },
    {
        "url": "https://www.plus2net.com/python/other.php",
        "html": '<html><head><link rel="canonical" href="https://www.plus2net.com/python/bs4.php"></head></html>',
    },
]

mismatches = []
for record in examples:
    soup_test = BeautifulSoup(record["html"], "html.parser")
    tag = soup_test.find("link", rel="canonical")
    actual = urljoin(record["url"], tag.get("href", "")) if tag else ""
    if record["url"] != actual:
        mismatches.append((record["url"], actual))

print("Canonical mismatches:", len(mismatches))
for expected, actual in mismatches:
    print("Expected:", expected, "Found:", actual)
assert len(mismatches) == 1
````

**Expected output:** `Canonical mismatches: 1` followed by the URL whose canonical points to a different page.

## 11. Generate a table of contents — clipboard is optional

**Tutorial example 24 of 24.** Generate unique heading IDs, build a `<nav>` with links, and copy it to the clipboard on a compatible local desktop. The webpage address uses the live public URL.

**Optional source example (requires your own database or clipboard setup):**

```python
import re
import requests
import pyperclip
from bs4 import BeautifulSoup

url = "https://www.plus2net.com/php_tutorial/sqlite-paging.php"
response = requests.get(url, timeout=15)
response.raise_for_status()
soup = BeautifulSoup(response.text, "html.parser")

headings = soup.find_all("h2")
links = []
used_ids = set()
for index, heading in enumerate(headings, start=1):
    title = heading.get_text(" ", strip=True)
    base = re.sub(r"[^a-z0-9]+", "-", title.lower()).strip("-")
    base = base or "section"
    anchor = base
    suffix = 2
    while anchor in used_ids:
        anchor = f"{base}-{suffix}"
        suffix += 1
    used_ids.add(anchor)
    heading["id"] = anchor
    item = soup.new_tag("li")
    link = soup.new_tag("a", href="#" + anchor)
    link.string = title
    item.append(link)
    links.append(item)

nav = soup.new_tag("nav")
nav["aria-label"] = "Table of contents"
ul = soup.new_tag("ul")
for item in links:
    ul.append(item)
nav.append(ul)

result = str(nav)
pyperclip.copy(result)
print(result)

# Save str(soup) if you also want the generated heading IDs in the HTML.
```

### Colab-friendly table of contents exercise

This local example builds a table of contents from sample headings without network access or clipboard requirements.

````python
import re
from bs4 import BeautifulSoup

page_html = '''<html><body>
<h2>Getting Started</h2>
<h2>Parsing HTML</h2>
<h2>Parsing HTML</h2>
</body></html>'''
soup_toc = BeautifulSoup(page_html, "html.parser")
used_ids = set()
nav = soup_toc.new_tag("nav")
nav["aria-label"] = "Table of contents"
ul = soup_toc.new_tag("ul")

for heading in soup_toc.find_all("h2"):
    title = heading.get_text(" ", strip=True)
    base = re.sub(r"[^a-z0-9]+", "-", title.lower()).strip("-") or "section"
    anchor = base
    n = 2
    while anchor in used_ids:
        anchor = f"{base}-{n}"
        n += 1
    used_ids.add(anchor)
    heading["id"] = anchor
    item = soup_toc.new_tag("li")
    a = soup_toc.new_tag("a", href="#" + anchor)
    a.string = title
    item.append(a)
    ul.append(item)
nav.append(ul)

print(nav.prettify())
assert [h["id"] for h in soup_toc.find_all("h2")] == [
    "getting-started", "parsing-html", "parsing-html-2"
]
assert len(nav.find_all("a")) == 3
````

**Expected:** three navigation links, including `#parsing-html` and `#parsing-html-2`. The assertions check unique anchors.

## 12. Self-validating BeautifulSoup practice

These offline exercises consolidate the key techniques from the tutorial: `find()`, `find_all()`, attributes, CSS selection, text extraction, and link resolution. Run the code and confirm that the assertions pass.

````python
from bs4 import BeautifulSoup
from urllib.parse import urljoin

practice_url = "https://www.plus2net.com/python/bs4.php"
practice_html = """<html><head>
<title>BeautifulSoup Practice</title>
<meta name="description" content="A simple practice webpage">
<link rel="canonical" href="https://www.plus2net.com/python/bs4.php">
</head><body>
<h1>Web <span>Scraping</span> with Python</h1>
<h2>Install</h2><h2>Examples</h2>
<a href="pillow.php">Pillow</a><a href="https://www.plus2net.com/python/sqlite.php">SQLite</a>
</body></html>"""
practice_soup = BeautifulSoup(practice_html, "html.parser")

assert practice_soup.title.get_text(strip=True) == "BeautifulSoup Practice"
assert practice_soup.h1.get_text(" ", strip=True) == "Web Scraping with Python"
assert len(practice_soup.find_all("h2")) == 2
assert practice_soup.select_one('meta[name="description"]')["content"] == "A simple practice webpage"
assert practice_soup.find("link", rel="canonical")["href"] == practice_url

all_links = [urljoin(practice_url, a["href"]) for a in practice_soup.find_all("a", href=True)]
assert len(all_links) == 2
assert all_links[0] == "https://www.plus2net.com/python/pillow.php"
print("All six BeautifulSoup checks passed.")
````

**Expected output:** `All six BeautifulSoup checks passed.`
