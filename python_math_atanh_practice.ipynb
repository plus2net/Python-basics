### 1. Basic Inverse Hyperbolic Tangent Calculations
`math.atanh(x)` takes a numeric ratio in the open interval `(-1.0, 1.0)` and returns a real number in `(-inf, +inf)`.

```python
import math

print("atanh(0):   ", math.atanh(0))
print("atanh(0.5): ", math.atanh(0.5))
print("atanh(-0.5):", math.atanh(-0.5))
print("atanh(0.95):", math.atanh(0.95))
```

### 2. Logarithmic Formula Equivalence
Verify that `math.atanh(x)` equals `0.5 * math.log((1 + x) / (1 - x))`.

```python
import math

x = 0.6
builtin_val = math.atanh(x)
formula_val = 0.5 * math.log((1 + x) / (1 - x))

print("math.atanh(0.6):", builtin_val)
print("Log formula:    ", formula_val)
print("Is close?       ", math.isclose(builtin_val, formula_val))
```

### 3. Open Domain Bounds: ValueError on |x| >= 1.0
Values of $x \ge 1.0$ or $x \le -1.0$ are outside the domain and raise `ValueError: math domain error`.

```python
import math

for invalid_val in [-1.5, -1.0, 1.0, 2.0]:
    try:
        math.atanh(invalid_val)
    except ValueError as err:
        print(f"Caught expected error for {invalid_val}: {err}")
```

### 4. math.atan() vs. math.atanh()
Circular arc tangent produces an angle in radians, whereas hyperbolic arc tangent produces a real number on the hyperbola.

```python
import math

val = 0.5
print("Circular atan(0.5):   ", math.atan(val), "radians")
print("Hyperbolic atanh(0.5):", math.atanh(val), "(pure real value)")
```

### 5. Hands-on Practice Exercise: Fisher z-Transformation
Run the self-checking unit assertions below to verify statistical transformation of Pearson correlation coefficients: $z = \text{atanh}(r)$.

```python
def fisher_z_transform(correlation_r):
    if abs(correlation_r) >= 1.0:
        raise ValueError("Pearson r must be strictly between -1.0 and 1.0.")
    z_score = math.atanh(correlation_r)
    return round(z_score, 4)

# Unit Assertions
assert fisher_z_transform(0.0) == 0.0
assert fisher_z_transform(0.5) == 0.5493
assert fisher_z_transform(-0.5) == -0.5493
assert fisher_z_transform(0.75) == 0.9730
assert fisher_z_transform(0.90) == 1.4722

print("Fisher z for r=0.75:", fisher_z_transform(0.75))
print("Fisher z for r=0.90:", fisher_z_transform(0.90))
print("All Fisher z-transformation assertions passed successfully!")
```
