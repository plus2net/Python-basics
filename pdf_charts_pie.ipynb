## 2. Pie Chart with Custom Slice Popouts & Styling

Create a pie chart with custom dimensions, slice popouts, and label radii using ReportLab.

```python
from pathlib import Path
from reportlab.lib import colors
from reportlab.graphics.shapes import Drawing
from reportlab.graphics.charts.piecharts import Pie
from reportlab.graphics import renderPDF

output_dir = Path('output')
output_dir.mkdir(exist_ok=True)
pdf_pie_options = output_dir / 'pie_chart_options.pdf'

d = Drawing(800, 800)
pc = Pie()
pc.x = 120
pc.y = 300
pc.width = 400
pc.height = 400
pc.data = [10, 20, 30, 40, 50, 60]
pc.labels = ['a', 'b', 'c', 'd', 'e', 'f']
pc.sideLabels = 0
pc.simpleLabels = 0
pc.slices.strokeWidth = 0.5
pc.slices[3].popout = 10
pc.slices[3].strokeWidth = 2
pc.slices[3].strokeDashArray = [2, 2]
pc.slices[3].labelRadius = 1.3
pc.slices[3].fontColor = colors.red
pc.slices[3].fontSize = 26
d.add(pc)

renderPDF.drawToFile(d, str(pdf_pie_options), '')
print(f"Pie chart options PDF created: {pdf_pie_options}")
```

## 3. Pie Chart from SQLite Database Query

Demonstrate querying an SQLite database using SQLAlchemy to populate pie chart data dynamically.

```python
from sqlalchemy import create_engine, Table, Column, Integer, String, MetaData

# Create an in-memory SQLite database and sample student table for demonstration
engine = create_engine('sqlite:///:memory:')
metadata = MetaData()
student_table = Table('student', metadata,
    Column('id', Integer, primary_key=True),
    Column('name', String),
    Column('class', String)
)
metadata.create_all(engine)

# Insert sample data
with engine.connect() as conn:
    conn.execute(student_table.insert(), [
        {'name': 'Alice', 'class': 'Class A'},
        {'name': 'Bob', 'class': 'Class A'},
        {'name': 'Charlie', 'class': 'Class B'},
        {'name': 'David', 'class': 'Class B'},
        {'name': 'Eve', 'class': 'Class B'},
        {'name': 'Frank', 'class': 'Class C'}
    ])
    conn.commit()

    # Query grouped data
    q = "SELECT class, count(*) as no FROM student GROUP BY class"
    my_cursor = conn.execute(q)
    my_result = my_cursor.fetchall()
    my_classes = [row[0] for row in my_result]
    my_Nos = [row[1] for row in my_result]

pdf_pie_db = output_dir / 'pie_chart_database.pdf'
d = Drawing(800, 800)
pc = Pie()
pc.x = 120
pc.y = 300
pc.width = 400
pc.height = 400
pc.data = my_Nos
pc.labels = my_classes
pc.sideLabels = 0
pc.simpleLabels = 0
pc.slices.strokeWidth = 0.5
pc.slices[3].popout = 3 if len(my_Nos) > 3 else 1
d.add(pc)

renderPDF.drawToFile(d, str(pdf_pie_db), '')
print(f"Database pie chart PDF created: {pdf_pie_db}")
```
