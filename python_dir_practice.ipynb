### 1. Scope Inspection Without Arguments

```python
sample_var = "plus2net"
record_count = 100

current_scope = dir()
user_vars = [name for name in current_scope if not name.startswith("_")]
print("User variables in scope:", user_vars)
```

### 2. Inspecting Modules & Built-in Types

```python
import math

math_items = dir(math)
print("Total math attributes:", len(math_items))
print("Math functions starting with 'f':", [x for x in math_items if x.startswith('f')])

list_public = [m for m in dir(list) if not m.startswith("_")]
print("List public methods:", list_public)
```

### 3. Custom Class with __dir__() Override

```python
class APIClient:
    def __init__(self, endpoint):
        self.endpoint = endpoint
        self._api_key = "secret_token_123"
        
    def get_data(self):
        return {}
        
    def __dir__(self):
        # Only expose public interface in dir()
        return ["endpoint", "get_data"]

client = APIClient("https://api.plus2net.com")
print("Customized dir(client):", dir(client))
```

### 4. Dynamic Plugin Hook Discovery

```python
class PaymentHooks:
    def on_success(self, tx_id):
        return f"Tx {tx_id} completed successfully"
    def on_failure(self, tx_id):
        return f"Tx {tx_id} failed"

def execute_hook(plugin, event_name, *args):
    method_name = f"on_{event_name}"
    if method_name in dir(plugin):
        method = getattr(plugin, method_name)
        return method(*args)
    return "Hook not supported"

hooks = PaymentHooks()
print(execute_hook(hooks, "success", "TX-9901"))
print(execute_hook(hooks, "refund", "TX-9901"))
```

### 5. Self-Checking Unit Verification Suite

```python
import math

# Test 1: dir returns list
assert isinstance(dir(), list)

# Test 2: List methods presence
l_dir = dir([])
assert "append" in l_dir and "pop" in l_dir

# Test 3: Math module attributes
assert "sqrt" in dir(math) and "pi" in dir(math)

# Test 4: Custom class dir override
class Mock:
    def __dir__(self): return ["foo", "bar"]
assert dir(Mock()) == ['bar', 'foo']

# Test 5: Alphabetical sort guarantee
attrs = dir(str)
assert attrs == sorted(attrs)

print("All dir() unit assertions successfully verified! 100% passed.")
```
