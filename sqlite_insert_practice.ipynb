## 1. Setup Database & Table

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_conn.execute('''CREATE TABLE IF NOT EXISTS student(
    id INTEGER PRIMARY KEY,
    name TEXT,
    class TEXT,
    mark INTEGER,
    gender TEXT
)''')
print("Student table ready!")
my_conn.close()
```

## 2. Adding Single Record & Getting `lastrowid` (`sqlite3`)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_query = "INSERT INTO student VALUES (NULL, 'New Name', 'Four', 65, 'female')"
r_set = my_conn.execute(my_query)
my_conn.commit()
print("Inserted record ID:", r_set.lastrowid)
my_conn.close()
```

## 3. Parameterized Single Insert (Preventing SQL Injection)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = (None, 'Second Name', 'Five', 75, 'male')
my_query = "INSERT INTO student VALUES (?, ?, ?, ?, ?)"
r_set = my_conn.execute(my_query, my_data)
my_conn.commit()
print("ID of last record added:", r_set.lastrowid)
my_conn.close()
```

## 4. Bulk Insert using `executemany()` & `rowcount`

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = [
    (9, 'Tes Qry', 'Six', 78, 'male'),
    (10, 'Big John', 'Four', 55, 'female'),
    (11, 'Ronald', 'Six', 89, 'female'),
    (12, 'Recky', 'Six', 94, 'female'),
    (13, 'Kty', 'Seven', 88, 'female')
]
my_query = "INSERT INTO student VALUES (?, ?, ?, ?, ?)"
curs = my_conn.executemany(my_query, my_data)
my_conn.commit()
print("Number of records added:", curs.rowcount)
my_conn.close()
```

## 5. Handling UNIQUE Constraint Errors with `try-except`

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = [
    (9, 'Duplicate Name', 'Six', 78, 'male')
]
my_query = "INSERT INTO student VALUES (?, ?, ?, ?, ?)"
try:
    curs = my_conn.executemany(my_query, my_data)
    my_conn.commit()
    print("Records added:", curs.rowcount)
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 6. Modern SQLAlchemy 2.0+ Insert Standard

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

my_conn = create_engine("sqlite:///my_db.db")
q = text("INSERT INTO student VALUES (NULL, :name, :cls, :mark, :gender)")
my_data = {"name": "SQLAlchemy User", "cls": "Five", "mark": 85, "gender": "male"}

try:
    with my_conn.begin() as connection:
        r_set = connection.execute(q, my_data)
        print("No of records added:", r_set.rowcount)
        print("Id of last record added:", r_set.inserted_primary_key[0] if r_set.inserted_primary_key else "Inserted")
except SQLAlchemyError as e:
    print("SQLAlchemy Error:", e)
```

## 7. Understanding AUTOINCREMENT Table Creation

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_conn.execute('''CREATE TABLE IF NOT EXISTS student_auto(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    class TEXT,
    mark INTEGER,
    gender TEXT
)''')

my_query = "INSERT INTO student_auto (name, class, mark, gender) VALUES ('Auto Name', 'Four', 65, 'Female')"
r_set = my_conn.execute(my_query)
my_conn.commit()
print("AUTOINCREMENT Table inserted ID:", r_set.lastrowid)
my_conn.close()
```
