### Part 1: Creation & Slicing

```python
# Q1-Q3: Creation and range
blank_list = []
blank_list.extend(['Alex', 'Ronn', 'Geek'])
print("Q1:", blank_list)

mixed = ['Alex', 4, 'Ronn', 6.54]
print("Q2 index 2:", mixed[2])

range_lst = list(range(5, 50, 10))
print("Q3 range list:", range_lst)
```

### Part 2: Mutation & Methods

```python
names = ["Alex", "Ronald", "John", "Sarah", "Michael"]
print("Index of John:", names.index("John"))
names.insert(2, "Ronn")
print("After insert:", names)
popped = names.pop(2)
print("Popped item:", popped)
```

### Part 3: Deduplication & Sorting

```python
# Deduplicate preserving order
dups = ['apple', 'banana', 'apple', 'cherry']
unique = list(dict.fromkeys(dups))
print("Deduplicated:", unique)

# Sort by length
words = ["Alexander", "Sam", "Christopher", "Jo"]
words.sort(key=len)
print("Sorted by length:", words)
```

### Part 4: Self-Checking Unit Verification Suite

```python
# Verify solutions
assert list(range(5, 50, 10))[0] == 5
assert list(range(5, 50, 10))[-1] == 45
assert len(list(range(5, 50, 10))) == 5

sample = [2, 4, 1, 5, 3]
assert sample[-1] == 3
assert sample[-2] == 5
assert sample[2:4] == [1, 5]

common = list(set([1, 2, 3]) & set([2, 3, 4]))
assert sorted(common) == [2, 3]

print("All list question unit assertions successfully verified! 100% passed.")
```
