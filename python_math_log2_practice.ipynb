### 1. Basic Calculations & Exact Powers of Two
`math.log2(x)` computes the base-2 logarithm of `x`. For exact powers of two, it returns exact integral floating-point numbers.


```python
import math

print("log2(1):    ", math.log2(1))       # 0.0
print("log2(2):    ", math.log2(2))       # 1.0
print("log2(8):    ", math.log2(8))       # 3.0
print("log2(1024): ", math.log2(1024))    # 10.0
print("log2(65536):", math.log2(65536))   # 16.0
print("log2(0.5):  ", math.log2(0.5))     # -1.0
print("log2(0.25): ", math.log2(0.25))    # -2.0
print("log2(10):   ", round(math.log2(10), 6)) # 3.321928
```

### 2. Precision Advantage: math.log2(x) vs. math.log(x, 2)
`math.log(x, 2)` calculates `ln(x) / ln(2)`, which introduces division rounding noise. `math.log2(x)` delegates to hardware C99 instructions, ensuring exact results.


```python
import math

powers = [8, 64, 1024, 32768, 1048576]
print(f"{'x':>8} | {'math.log2(x)':>14} | {'math.log(x, 2)':>16} | Difference")
print("-" * 55)
for p in powers:
    res_log2 = math.log2(p)
    res_log_base = math.log(p, 2)
    diff = abs(res_log2 - res_log_base)
    print(f"{p:8d} | {res_log2:14.1f} | {res_log_base:16.14f} | {diff:.2e}")
```

### 3. Domain Constraints (x > 0)
Logarithms are undefined for zero and negative real numbers, raising `ValueError: math domain error`.


```python
import math

for bad in [0.0, -4.0]:
    try:
        math.log2(bad)
    except ValueError as err:
        print(f"Caught expected error on {bad}: {err}")

print("log2(inf):", math.log2(float("inf")))
print("log2(nan):", math.log2(float("nan")))
```

### 4. Computer Science: Bit-Length & Binary Search Depth
The minimum bits required to represent an integer `N` is `math.floor(math.log2(N)) + 1`.


```python
import math

test_ints = [1, 2, 7, 255, 1023, 65535]
for n in test_ints:
    bits = math.floor(math.log2(n)) + 1
    assert bits == n.bit_length()
    print(f"Integer: {n:5d} -> Bits: {bits:2d} (matches bit_length: {n.bit_length():2d})")
```

### 5. Hands-on Practice Exercise: Shannon Entropy & Password Strength
Calculate Shannon Information Entropy `H(X) = -sum(p_i * log2(p_i))` and password search space entropy `length * log2(pool_size)`.


```python
import math
from collections import Counter

def shannon_entropy(text):
    """Computes Shannon Information Entropy of a string in bits per character."""
    if not text:
        return 0.0
    length = len(text)
    counts = Counter(text)
    entropy = 0.0
    for count in counts.values():
        prob = count / length
        entropy -= prob * math.log2(prob)
    return round(entropy, 4)

def password_search_entropy(password_length, charset_size):
    """Calculates total brute-force entropy in bits: length * log2(charset_size)."""
    if password_length <= 0 or charset_size <= 1:
        raise ValueError("Length must be positive and charset size must be > 1.")
    return round(password_length * math.log2(charset_size), 2)

# Unit Assertions
assert shannon_entropy("aaaaaa") == 0.0
assert shannon_entropy("ababab") == 1.0
assert shannon_entropy("abcd") == 2.0
assert password_search_entropy(8, 10) == 26.58
assert password_search_entropy(12, 62) == 71.45

print("Shannon Entropy of 'plus2net':", shannon_entropy("plus2net"), "bits/char")
print("Search entropy for 16-char password:", password_search_entropy(16, 94), "bits")
print("All Shannon entropy and password assertions passed successfully!")
```
