### 1. Basic frozenset Creation
Construct immutable sets and verify duplicate elimination.

```python
values = [1, 2, 3, 2, 1]
frozen = frozenset(values)
print("frozenset:", frozen)
assert len(frozen) == 3
```

### 2. Set Operations Return New frozensets
Perform union, intersection, and difference operations.

```python
A = frozenset({1, 2, 3})
B = frozenset({3, 4, 5})
print("Union:       ", A | B)
print("Intersection:", A & B)
print("Difference:  ", A - B)
assert A & B == frozenset({3})
```

### 3. frozenset as Dictionary Keys
Use order-independent composite keys in dictionaries.

```python
roles = {
    frozenset({'read', 'write'}): 'editor',
    frozenset({'read'}): 'viewer'
}
print("Lookup role:", roles[frozenset({'write', 'read'})])
assert roles[frozenset({'write', 'read'})] == 'editor'
```

### 4. Production Recipe: LRU Cache with frozensets
Memoize functions where collection argument order does not matter.

```python
from functools import lru_cache

@lru_cache(maxsize=32)
def calculate_total(items_set: frozenset) -> int:
    price_map = {'book': 15, 'pen': 3, 'notebook': 7}
    return sum(price_map.get(item, 0) for item in items_set)

order1 = frozenset(['book', 'pen'])
order2 = frozenset(['pen', 'book'])
print("Order 1:", calculate_total(order1))
print("Order 2:", calculate_total(order2))
assert calculate_total.cache_info().hits == 1
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying frozenset() mechanics.

```python
assert frozenset() == frozenset([])
assert len(frozenset([1, 2, 3, 3])) == 3
assert not hasattr(frozenset(), 'add')
assert hash(frozenset([1, 2])) == hash(frozenset([2, 1]))

print("All frozenset() unit assertions passed! 100% verified.")
```
