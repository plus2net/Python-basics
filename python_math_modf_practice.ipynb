# Python math.modf(): Separate Fractional and Integer Parts of a Float

### 1. Basic Float Splitting with math.modf()
`math.modf(x)` returns a 2-element tuple of floats: `(fractional_part, integer_part)`. Notice that the fractional part comes first (`index 0`) and the integer part comes second (`index 1`).


```python
import math

# Basic float
parts = math.modf(23.45)
print("modf(23.45):       ", parts)
print("Fraction (index 0):", parts[0])
print("Integer  (index 1):", parts[1])

# Integer input returns 0.0 fractional part
print("modf(42):          ", math.modf(42))

# Unpacking syntax
frac, whole = math.modf(math.pi)
print(f"Pi: whole = {int(whole)}, fraction = {frac:.6f}")
```

### 2. Sign Inheritance with Negative Numbers
Both components inherit the sign of `x`. If `x` is negative, both parts are negative, ensuring `frac + whole == x` exactly.


```python
import math

val = -12.34
frac, whole = math.modf(val)
print("Value:     ", val)
print("Fractional:", frac)
print("Integer:   ", whole)
print("Sum check: ", frac + whole)
print("Comparison with % 1:     ", val % 1.0)     # Always positive (0.66)
print("Comparison with divmod:  ", divmod(val, 1)) # (-13.0, 0.66)
```

### 3. IEEE 754 Floating-Point Precision Management
Binary floating-point cannot represent certain decimal fractions (like 0.45) with infinite precision. Use `round()` when presenting numbers to human users.


```python
import math

frac, whole = math.modf(23.45)
print("Raw fractional part:   ", frac)
print("Clean rounded fraction:", round(frac, 4))
```

### 4. Special Values: Zero, Inf, and NaN
Examine how `math.modf()` processes special IEEE 754 floating-point edge cases.


```python
import math

print("modf(0.0):  ", math.modf(0.0))
print("modf(-0.0): ", math.modf(-0.0))
print("modf(inf):  ", math.modf(float("inf")))
print("modf(nan):  ", math.modf(float("nan")))
```

### 5. Hands-on Practice Exercise: GPS Decimal Degrees to DMS Converter
In Geographic Information Systems (GIS), coordinates in Decimal Degrees (DD) are converted to Degrees, Minutes, and Seconds (DMS) using repeated fractional extraction with `math.modf()`.

Implement `dd_to_dms(decimal_degrees, is_latitude=True)` and verify with self-checking unit assertions.


```python
import math

def dd_to_dms(decimal_degrees, is_latitude=True):
    """
    Converts decimal degrees coordinate to (degrees, minutes, seconds, compass_direction).
    """
    if not isinstance(decimal_degrees, (int, float)):
        raise TypeError("Coordinate must be a numeric value.")
    
    is_negative = decimal_degrees < 0
    abs_dd = abs(decimal_degrees)
    
    # 1. Separate degrees and fractional degree
    frac_deg, deg = math.modf(abs_dd)
    
    # 2. Convert fractional degree to minutes
    total_minutes = frac_deg * 60.0
    frac_min, minutes = math.modf(total_minutes)
    
    # 3. Convert fractional minutes to seconds
    seconds = round(frac_min * 60.0, 2)
    
    if is_latitude:
        direction = "S" if is_negative else "N"
    else:
        direction = "W" if is_negative else "E"
        
    return (int(deg), int(minutes), seconds, direction)

# Unit Assertions
# San Francisco Latitude: 37.7749 deg N -> 37 deg 46' 29.64" N
assert dd_to_dms(37.7749, is_latitude=True) == (37, 46, 29.64, "N")

# San Francisco Longitude: -122.4194 deg W -> 122 deg 25' 9.84" W
assert dd_to_dms(-122.4194, is_latitude=False) == (122, 25, 9.84, "W")

# Equator / Prime Meridian origin
assert dd_to_dms(0.0) == (0, 0, 0.0, "N")

# Type verification
try:
    dd_to_dms("invalid")
    assert False, "Should raise TypeError"
except TypeError:
    pass

lat_dms = dd_to_dms(37.7749, True)
lon_dms = dd_to_dms(-122.4194, False)
print(f"San Francisco Coordinates: {lat_dms[0]} deg {lat_dms[1]}' {lat_dms[2]}\" {lat_dms[3]}, {lon_dms[0]} deg {lon_dms[1]}' {lon_dms[2]}\" {lon_dms[3]}")
print("All GPS DMS conversion assertions passed successfully!")
```
