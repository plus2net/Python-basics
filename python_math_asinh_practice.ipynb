### 1. Basic Inverse Hyperbolic Sine Calculations
`math.asinh(x)` returns the inverse hyperbolic sine of `x`. Because the hyperbolic sine function is strictly monotonic across all real numbers, `math.asinh(x)` accepts any real numeric value from negative infinity to positive infinity.


```python
import math

print("asinh(0.0):   ", math.asinh(0.0))
print("asinh(1.0):   ", round(math.asinh(1.0), 6))
print("asinh(-1.0):  ", round(math.asinh(-1.0), 6))
print("asinh(10.0):  ", round(math.asinh(10.0), 6))
print("asinh(-10.0): ", round(math.asinh(-10.0), 6))
```

### 2. Logarithmic Equivalence Verification
Mathematically, the inverse hyperbolic sine is defined in terms of natural logarithms:
`asinh(x) == ln(x + sqrt(x^2 + 1))`

We can verify this equivalence numerically using `math.isclose()`.


```python
import math

def manual_asinh(x):
    return math.log(x + math.sqrt(x**2 + 1))

test_values = [-25.0, -2.5, 0.0, 1.0, 7.5, 50.0]
print(f"{'x':>8} | {'math.asinh(x)':>14} | {'manual_asinh(x)':>15} | Match?")
print("-" * 50)
for val in test_values:
    built_in = math.asinh(val)
    formula = manual_asinh(val)
    matches = math.isclose(built_in, formula, rel_tol=1e-12)
    print(f"{val:8.2f} | {built_in:14.6f} | {formula:15.6f} | {matches}")
```

### 3. Odd Function Symmetry: asinh(-x) == -asinh(x)
`math.asinh` is an odd function symmetric about the origin. Passing negative inputs yields the exact negated value of positive inputs, with `asinh(0.0) == 0.0`.


```python
import math

samples = [0.5, 2.0, 8.4, 15.0]
for s in samples:
    pos = math.asinh(s)
    neg = math.asinh(-s)
    print(f"asinh({s:4.1f}) = {pos:8.4f} | asinh({-s:4.1f}) = {neg:8.4f} | Sum: {pos + neg}")
```

### 4. Circular (math.asin) vs Hyperbolic (math.asinh)
Circular `math.asin(x)` requires `-1.0 <= x <= 1.0` and represents an angle (converted via `math.degrees()`).
Hyperbolic `math.asinh(x)` accepts any real number and represents dimensionless hyperbolic growth.


```python
import math

val = 5.0
print("Input x =", val)
print("math.asinh(5.0):", math.asinh(val))

try:
    math.asin(val)
except ValueError as err:
    print("math.asin(5.0) raises ValueError as expected:", err)
```

### 5. Hands-on Practice Exercise: IHS Financial Data Transformer
In econometric modelling, standard log transformation `log(x)` fails on zeros or negative values (such as business losses or cash deficits). The Inverse Hyperbolic Sine (IHS) transformation handles positive, negative, and zero values while retaining logarithmic properties for large magnitudes.

Implement `ihs_transform(values, scale=1.0)` and verify with self-checking assertions.


```python
import math

def ihs_transform(values, scale=1.0):
    """
    Transforms a list of numeric values using the Inverse Hyperbolic Sine (IHS) formula:
    asinh(x / scale)
    """
    if scale <= 0:
        raise ValueError("Scale parameter must be strictly positive.")
    return [round(math.asinh(x / scale), 4) for x in values]

# Unit Assertions
assert ihs_transform([0.0]) == [0.0]
assert ihs_transform([0.0, 1.0, -1.0]) == [0.0, 0.8814, -0.8814]
assert ihs_transform([100.0], scale=10.0) == [2.9982]
assert ihs_transform([-50.0], scale=1.0) == [-4.6053]

# Edge case: scale <= 0 raises ValueError
try:
    ihs_transform([10.0], scale=-2.0)
    assert False, "Should have raised ValueError"
except ValueError:
    pass

raw_financials = [-15000.0, -2500.0, 0.0, 4800.0, 62000.0]
transformed = ihs_transform(raw_financials, scale=1000.0)
print("Raw financial cash flows: ", raw_financials)
print("IHS transformed (scale=1k):", transformed)
print("All unit assertions passed successfully!")
```
