### 1. Multi-Statement Execution & Dynamic Function Declaration

```python
script = """
def format_records(items):
    return [item.strip().title() for item in items]

raw_data = ['  apple ', 'banana  ', ' cherry']
cleaned = format_records(raw_data)
print('Cleaned records:', cleaned)
"""

exec(script)
```

### 2. Namespace Scoping & Function Extraction

```python
module_source = """
def calculate_discount(price, tier):
    rates = {'gold': 0.20, 'silver': 0.10, 'bronze': 0.05}
    return price * (1 - rates.get(tier, 0.0))

version = '2.4.0'
"""

custom_scope = {}
exec(module_source, custom_scope)

# Extract the generated function from the dictionary
calc_fn = custom_scope['calculate_discount']
print("Gold discount price:  ", calc_fn(100.0, 'gold'))
print("Silver discount price:", calc_fn(100.0, 'silver'))
print("Module version:       ", custom_scope['version'])
```

### 3. Loop Performance: Precompiled Bytecode with compile()

```python
import time

code_text = """
total = 0
for i in range(50):
    total += i * 2
"""
iterations = 10000

# Raw string execution
t0 = time.perf_counter()
scope_raw = {}
for _ in range(iterations):
    exec(code_text, scope_raw)
raw_time = time.perf_counter() - t0

# Precompiled execution
compiled_obj = compile(code_text, '<loop>', 'exec')
t1 = time.perf_counter()
scope_comp = {}
for _ in range(iterations):
    exec(compiled_obj, scope_comp)
comp_time = time.perf_counter() - t1

print(f"Raw string exec:     {raw_time:.4f} s")
print(f"Precompiled exec:    {comp_time:.4f} s")
print(f"Speedup factor:      {raw_time / comp_time:.2f}x faster")
```

### 4. Production Recipe: Dynamic Policy Rule Evaluator

```python
rules = """
def classify_risk(score, debt_ratio):
    if score >= 720 and debt_ratio < 0.35:
        return 'Low Risk'
    elif score >= 640:
        return 'Moderate Risk'
    return 'High Risk'
"""

engine_scope = {}
exec(compile(rules, '<policy>', 'exec'), engine_scope)
classifier = engine_scope['classify_risk']

print("Score 750, ratio 0.25:", classifier(750, 0.25))
print("Score 660, ratio 0.40:", classifier(660, 0.40))
print("Score 590, ratio 0.50:", classifier(590, 0.50))
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Statement execution
ns = {}
exec("x = 10\ny = 20\nz = x + y", ns)
assert ns['z'] == 30

# Test 2: exec always returns None
assert exec("val = 42") is None

# Test 3: Dynamic function creation
fn_ns = {}
exec("def double(n): return n * 2", fn_ns)
assert fn_ns['double'](7) == 14

# Test 4: Precompiled execution
c_obj = compile("res = 5 * 5", '<test>', 'exec')
c_ns = {}
exec(c_obj, c_ns)
assert c_ns['res'] == 25

print("All exec() unit assertions successfully verified! 100% passed.")
```
