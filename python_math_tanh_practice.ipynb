### 1. Basic Calculations & Odd Function Symmetry
`math.tanh(x)` computes the hyperbolic tangent of `x`. It satisfies odd symmetry: `tanh(-x) == -tanh(x)`, passing through the origin at `tanh(0.0) == 0.0`.


```python
import math

print("tanh(0.0): ", math.tanh(0.0))    # 0.0
print("tanh(1.0): ", round(math.tanh(1.0), 6))    # 0.761594
print("tanh(-1.0):", round(math.tanh(-1.0), 6))   # -0.761594
print("tanh(2.0): ", round(math.tanh(2.0), 6))    # 0.964028
print("tanh(-2.0):", round(math.tanh(-2.0), 6))   # -0.964028
```

### 2. Ratio Equivalence: sinh(x) / cosh(x)
Verify that `math.tanh(x)` matches the ratio of hyperbolic sine to hyperbolic cosine.


```python
import math

val = 2.5
from_tanh = math.tanh(val)
from_ratio = math.sinh(val) / math.cosh(val)
print("math.tanh(2.5):   ", from_tanh)
print("sinh(2.5)/cosh(2.5):", from_ratio)
assert math.isclose(from_tanh, from_ratio)
print("Matches? True")
```

### 3. Machine Learning: Zero-Centered Tanh Activation & Derivative
In neural networks, `tanh` provides zero-centered outputs in `(-1, 1)`. The derivative is `tanh'(z) = 1 - tanh(z)^2`.


```python
import math

def tanh_activation_and_derivative(z):
    activated = math.tanh(z)
    derivative = 1.0 - activated**2
    return (activated, derivative)

inputs = [-3.0, -1.0, 0.0, 1.0, 3.0]
print(f"{'Input z':>8} | {'tanh(z)':>10} | {'tanh\'(z)':>10}")
print("-" * 34)
for z in inputs:
    act, der = tanh_activation_and_derivative(z)
    print(f"{z:8.1f} | {act:10.4f} | {der:10.4f}")
```

### 4. Asymptotic Saturation & Special Values
Unlike `sinh` and `cosh`, `tanh` never overflows. It approaches `+1.0` or `-1.0` as $|x| \to \infty$.


```python
import math

print("tanh(5):    ", math.tanh(5))
print("tanh(20):   ", math.tanh(20))
print("tanh(1000): ", math.tanh(1000)) # Never overflows!
print("tanh(inf):  ", math.tanh(float("inf")))
print("tanh(-inf): ", math.tanh(float("-inf")))
```

### 5. Hands-on Practice Exercise: Relativistic Velocity Addition Engine
In Special Relativity, velocities add via rapidities: `v_total = c * tanh(atanh(v1/c) + atanh(v2/c))`.
Implement `relativistic_velocity_addition()` and verify with self-checking assertions.


```python
import math

SPEED_OF_LIGHT = 299792458.0  # c in m/s

def relativistic_velocity_addition(v1, v2, c=SPEED_OF_LIGHT):
    """Combines two velocities v1 and v2 relativistically: v_total = c * tanh(atanh(v1/c) + atanh(v2/c))."""
    if abs(v1) >= c or abs(v2) >= c:
        raise ValueError("Subluminal velocities must be strictly less than c.")
    rapidity1 = math.atanh(v1 / c)
    rapidity2 = math.atanh(v2 / c)
    combined_v = c * math.tanh(rapidity1 + rapidity2)
    return round(combined_v, 2)

# Unit Assertions
assert relativistic_velocity_addition(0.0, 100000.0) == 100000.0
v_sub = 0.75 * SPEED_OF_LIGHT
expected_c_fraction = round(0.96 * SPEED_OF_LIGHT, 2)
assert relativistic_velocity_addition(v_sub, v_sub) == expected_c_fraction
assert relativistic_velocity_addition(v_sub, -v_sub) == 0.0

# Velocity limit check
try:
    relativistic_velocity_addition(SPEED_OF_LIGHT, 500.0)
    assert False, "Should raise ValueError"
except ValueError:
    pass

print("Two 0.75c velocities combined:", relativistic_velocity_addition(v_sub, v_sub), "m/s")
print("All relativistic velocity assertions passed successfully!")
```
