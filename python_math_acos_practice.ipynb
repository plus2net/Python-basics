### 1. Basic Arc Cosine Calculations
`math.acos(x)` takes a numeric ratio in $[-1.0, 1.0]$ and returns the principal angle in radians $[0, \pi]$.

```python
import math

print("acos(1.0): ", math.acos(1.0))
print("acos(0.5): ", math.acos(0.5))
print("acos(0.0): ", math.acos(0.0))
print("acos(-1.0):", math.acos(-1.0))
```

### 2. Converting Output Radians to Degrees
Use `math.degrees()` to convert output angles from radians to degrees ($0^\circ$ to $180^\circ$).

```python
import math

ratios = [1.0, 0.8660254, 0.70710678, 0.5, 0.0, -0.5, -1.0]
for r in ratios:
    rad = math.acos(r)
    deg = math.degrees(rad)
    print(f"acos({r:5.2f}) -> {rad:6.4f} rad == {deg:5.1f} deg")
```

### 3. Handling Domain Errors Outside [-1.0, 1.0]
Passing values outside $[-1.0, 1.0]$ raises `ValueError: math domain error`.

```python
import math

for invalid_ratio in [-1.05, 1.5]:
    try:
        math.acos(invalid_ratio)
    except ValueError as err:
        print(f"Caught expected domain error for {invalid_ratio}: {err}")
```

### 4. Triangle Angles via the Law of Cosines
Solve for angle $C$ in an arbitrary triangle with sides $a$, $b$, and $c$: $C = \arccos(\frac{a^2 + b^2 - c^2}{2ab})$.

```python
import math

a, b, c = 3, 4, 5
cos_C = (a**2 + b**2 - c**2) / (2 * a * b)
angle_C = math.degrees(math.acos(cos_C))
print("Triangle 3-4-5 Angle C:", angle_C, "degrees")
```

### 5. Hands-on Practice Exercise: Vector Angle & Orientation Calculator
Run the self-checking unit assertions below to verify vector angle calculation using the dot product formula: $\theta = \arccos(\frac{\vec{u} \cdot \vec{v}}{\|\vec{u}\| \|\vec{v}\|})$.

```python
def calculate_vector_angle(v1, v2):
    dot_product = sum(a * b for a, b in zip(v1, v2))
    mag1 = math.sqrt(sum(a**2 for a in v1))
    mag2 = math.sqrt(sum(b**2 for b in v2))
    if mag1 == 0 or mag2 == 0:
        raise ValueError("Zero-length vectors do not possess an angle.")
    cos_theta = max(-1.0, min(1.0, dot_product / (mag1 * mag2)))
    angle_rad = math.acos(cos_theta)
    return round(math.degrees(angle_rad), 2)

# Unit Assertions
assert calculate_vector_angle([1, 0], [5, 0]) == 0.0
assert calculate_vector_angle([1, 0], [0, 1]) == 90.0
assert calculate_vector_angle([1, 0], [-1, 0]) == 180.0
assert calculate_vector_angle([1, 1, 0], [1, 0, 0]) == 45.0

print("Angle [1, 0] & [0, 1]:    ", calculate_vector_angle([1, 0], [0, 1]), "deg")
print("Angle [1, 1, 0] & [1, 0, 0]:", calculate_vector_angle([1, 1, 0], [1, 0, 0]), "deg")
print("All vector angle assertions passed successfully!")
```
