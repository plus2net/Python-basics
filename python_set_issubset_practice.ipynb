### 1. Basic Subset Testing
`B.issubset(A)` returns `True` if every element of `B` is inside `A`; `False` otherwise.

```python
A = {'x', 'y', 'z', 'a', 'b', 'c'}
B = {'x', 'y', 'z'}

# Full containment -> True
print("Is B a subset of A?", B.issubset(A))

C = {'x', 'y', 'k'}
# Partial overlap -> False ('k' is missing)
print("Is C a subset of A?", C.issubset(A))
```

### 2. Subset (`<=`) vs. Proper Subset (`<`)
Equal sets are subsets (`<=`), but NOT proper subsets (`<`). The `<` operator requires the caller to be strictly smaller.

```python
s1 = {1, 2}
s2 = {1, 2}
s3 = {1, 2, 3}

print("s1 <= s2 (Equal sets are subsets):       ", s1 <= s2)  # True
print("s1 < s2  (Equal sets are NOT proper subsets):", s1 < s2)   # False
print("s1 < s3  (s1 is strictly smaller than s3):   ", s1 < s3)   # True
```

### 3. Heterogeneous Collections: Lists and Strings
Passing a list checks if set items are present in the list. Passing a string decomposes into individual characters.

```python
# List target
required_flags = {'x', 'y'}
system_flags = ['a', 'b', 'x', 'y']
print("Flags in list:", required_flags.issubset(system_flags))

# String character unpacking
char_set = {'y', 't', 'W', 'o', 'e'}
print("Characters in 'Welcome to Python':", char_set.issubset('Welcome to Python'))

# Word-level check (wrap in list or set)
needed_terms = {"python", "django"}
available_tags = ["python", "django", "docker"]
print("Words in tags:", needed_terms.issubset(available_tags))
```

### 4. Real-World Recipe: RBAC Entitlement Verification
Validate that requested operations do not exceed user role entitlements.

```python
def validate_api_action(requested: set, allowed: set) -> bool:
    return requested.issubset(allowed)

dev_permissions = {"read_code", "push_branch", "view_ci_logs"}

action_1 = {"read_code", "view_ci_logs"}
action_2 = {"push_branch", "deploy_production"}

print("Action 1 authorized:", validate_api_action(action_1, dev_permissions))
print("Action 2 authorized:", validate_api_action(action_2, dev_permissions))
assert validate_api_action(action_1, dev_permissions) is True
assert validate_api_action(action_2, dev_permissions) is False
```

### 5. Unit Assertion Suite
Validate empty set invariants, reflexive subset rules, and dictionary key handling.

```python
# Assertion 1: Reflexive property
s = {10, 20, 30}
assert s.issubset(s) is True
assert (s <= s) is True
assert (s < s) is False

# Assertion 2: Empty set containment
assert set().issubset(s) is True
assert (set() < s) is True

# Assertion 3: Heterogeneous iterables
assert {'a', 'b'}.issubset(['a', 'b', 'c']) is True
assert {'a', 'z'}.issubset(['a', 'b', 'c']) is False

# Assertion 4: Dictionary key evaluation
d = {"key1": 1, "key2": 2}
assert {"key1"}.issubset(d) is True
assert {1}.issubset(d) is False  # Checks keys, not values

print("All set.issubset assertions passed successfully!")
```
