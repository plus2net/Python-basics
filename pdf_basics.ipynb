## 2. Basic Canvas Operations & Text Rendering

Create a simple canvas PDF, position strings using `drawString()`, and adjust fonts and colors.

```python
from pathlib import Path
from reportlab.pdfgen import canvas

output_dir = Path('output')
my_path = output_dir / 'my_pdf.pdf'

c = canvas.Canvas(str(my_path), bottomup=0)
c.setFont("Helvetica", 20)
c.setFillColor('lightblue')
c.drawString(200, 200, "Hello World")
c.showPage()
c.save()
print(f"Canvas PDF created: {my_path}")
```

## 3. Lines, Grids, and Watermarks

Draw line elements, vector rectangles, coordinate grids, and 45-degree watermarks.

```python
from reportlab.lib.units import inch
from reportlab.lib.pagesizes import letter

wm_path = output_dir / 'watermarked.pdf'
c = canvas.Canvas(str(wm_path), bottomup=1, pagesize=letter)
c.translate(inch, inch)

# Draw Rectangle
c.setLineWidth(10)
c.setStrokeColor('yellow')
c.setFillColor('lightgreen')
c.rect(1 * inch, 2 * inch, 3.5 * inch, 5 * inch, fill=1)

# Watermark
c.rotate(45)
c.setFillColorCMYK(0, 0, 0, 0.08)
c.setFont("Helvetica", 100)
c.drawString(2 * inch, 1 * inch, "SAMPLE")
c.rotate(-45)

c.showPage()
c.save()
print(f"Watermarked PDF created: {wm_path}")
```
