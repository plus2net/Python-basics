### 1. Basic Shallow Copy

```python
original = ["Alex", "Ronald", "John"]
copied = original.copy()

copied.append("Sarah")
print("Original: ", original)
print("Copied:   ", copied)
```

### 2. Assignment (=) vs. Copying

```python
items = [10, 20, 30]
alias = items
clone = items.copy()

print("id(items) == id(alias):", id(items) == id(alias))
print("id(items) == id(clone):", id(items) == id(clone))
```

### 3. Shallow vs. Deepcopy on Nested Lists

```python
import copy

matrix = [1, [2, 3], 4]
shallow_m = matrix.copy()
deep_m = copy.deepcopy(matrix)

matrix[1][0] = 999
print("Original modified:", matrix)
print("Shallow affected: ", shallow_m)
print("Deep isolated:    ", deep_m)
```

### 4. Transactional Snapshot Recipe

```python
state = ["Init", "Loaded"]
snapshot = state.copy()

# Speculative modification
state.append("Unstable_Feature")
print("Modified:", state)

# Rollback
state = snapshot.copy()
print("Restored:", state)
```

### 5. Self-Checking Unit Verification Suite

```python
import copy

# Test 1: Independence of top-level lists
a = [1, 2, 3]
b = a.copy()
assert a == b and a is not b
b.append(4)
assert len(a) == 3 and len(b) == 4

# Test 2: Slicing creates shallow copy
c = a[:]
assert c == a and c is not a

# Test 3: Deep copy independence
nested = [["x", "y"]]
deep_cl = copy.deepcopy(nested)
nested[0].append("z")
assert len(deep_cl[0]) == 2

print("All list.copy() unit assertions successfully verified! 100% passed.")
```
