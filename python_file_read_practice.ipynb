### 1. Setup: Creating Sample Files for Practice
Create sample text and binary data files directly in the Colab / Jupyter environment.

```python
sample_text = """Line 1: Python File I/O with plus2net
Line 2: Context managers ensure safe resource cleanup
Line 3: Buffered chunking handles multi-gigabyte files
Line 4: Modern Python stream processing
"""

with open("demo_data.txt", "w", encoding="utf-8") as f:
    f.write(sample_text)

print("Created demo_data.txt successfully.")
```

### 2. Reading Entire Files via Context Manager (with open)
The `with` statement ensures the file is automatically closed when execution leaves the block.

```python
with open("demo_data.txt", "r", encoding="utf-8") as fob:
    full_content = fob.read()
    print("File length in characters:", len(full_content))
    print("--- File Content ---")
    print(full_content)

print("Is file handle closed?", fob.closed)  # True
```

### 3. Reading Incremental Slices with the size Argument
Passing a positive integer to `read(size)` advances the internal file cursor by at most `size` units.

```python
with open("demo_data.txt", "r", encoding="utf-8") as fob:
    header = fob.read(15)  # Read first 15 characters
    body = fob.read(25)    # Read next 25 characters
    
    print("Header slice:", repr(header))
    print("Body slice:  ", repr(body))
    print("Cursor position:", fob.tell())  # 40
```

### 4. Memory-Efficient Chunked Reading for Large Files
Process files in fixed-size blocks (e.g. 64 KB) to prevent `MemoryError` on massive files.

```python
def count_file_characters(filepath: str, chunk_size: int = 32) -> int:
    total_characters = 0
    with open(filepath, "r", encoding="utf-8") as stream:
        while chunk := stream.read(chunk_size):
            total_characters += len(chunk)
    return total_characters

total = count_file_characters("demo_data.txt")
print(f"Total characters counted in chunks: {total}")
```

### 5. Binary Reading Mode ('rb') & Header Inspection
Read raw bytes without automatic newline translations. Useful for image headers and binary protocols.

```python
with open("demo_data.txt", "rb") as fob:
    first_bytes = fob.read(10)
    print("Raw bytes:  ", first_bytes)
    print("Data type:  ", type(first_bytes))
    print("Hex values: ", first_bytes.hex())
```

### 6. Pitfall Demonstration: End-of-File Cursor & seek(0) Rewind
Once EOF is reached, subsequent `read()` calls return empty string `""` until rewound.

```python
with open("demo_data.txt", "r", encoding="utf-8") as f:
    first_read = f.read()
    second_read = f.read()
    print("Second read returns empty:", repr(second_read))
    
    # Rewind cursor back to offset 0
    f.seek(0)
    rewound_read = f.read(20)
    print("After f.seek(0):", repr(rewound_read))
```

### 7. Self-Checking Verification Suite
Run unit assertions verifying file reading correctness across all modes.

```python
class FileReaderSuite:
    @staticmethod
    def read_full_text(path: str) -> str:
        with open(path, "r", encoding="utf-8") as stream:
            return stream.read()

    @staticmethod
    def read_in_chunks(path: str, chunk_size: int = 16) -> list:
        chunks = []
        with open(path, "r", encoding="utf-8") as stream:
            while chunk := stream.read(chunk_size):
                chunks.append(chunk)
        return chunks

suite = FileReaderSuite()
assert suite.read_full_text("demo_data.txt") == sample_text
assert "".join(suite.read_in_chunks("demo_data.txt", 8)) == sample_text

with open("demo_data.txt", "r", encoding="utf-8") as f:
    assert f.read(6) == "Line 1"

print("All file.read() unit assertions passed successfully! 100% verified.")
```
