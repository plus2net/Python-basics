### 1. Basic & Aliased Imports

```python
import math as m

val = 144
root = m.sqrt(val)
print(f"Square root of {val}: {root}")
print(f"Pi constant: {m.pi:.4f}")
```

### 2. Module Introspection with dir() and vars()

```python
import json

# Inspect available public functions
public_members = [name for name in dir(json) if not name.startswith('_')]
print("Public json functions:", public_members[:6])
print("Module name:", json.__name__)
```

### 3. Graceful ModuleNotFoundError Handling

```python
try:
    import non_existent_library
except ModuleNotFoundError as e:
    print("Safely caught missing module:", e.name)
```

### 4. Self-Checking Unit Verification Suite

```python
import sys
import math

# Test 1: Math module in sys.modules
assert 'math' in sys.modules

# Test 2: Function availability
assert hasattr(math, 'sqrt') and callable(math.sqrt)
assert math.sqrt(25) == 5.0

# Test 3: ModuleNotFoundError handling
caught = False
try:
    import dummy_test_package_xyz
except ModuleNotFoundError:
    caught = True
assert caught is True

print("All module practice assertions successfully verified! 100% passed.")
```
