## 2. Using CSS Selectors for Classes, IDs, and Attributes

Parse HTML strings and select tags using standard CSS syntax.

```python
from bs4 import BeautifulSoup

content = '''<div class="container">
  <h2 class="title">CSS Selectors Demo</h2>
  <p id="p1">First Paragraph</p>
  <p class="highlight">Second Paragraph</p>
</div>'''

soup = BeautifulSoup(content, 'html.parser')

print("By Class:", [p.text for p in soup.select('.highlight')])
print("By ID:", soup.select_one('#p1').text)
```
