### 1. Small Integer Caching & Identity vs Equality

```python
a = 30
b = 30
print("a is b (30):", a is b)
assert (a is b) == True

x = 300
y = 300
print("x is y (300):", x is y)
print("x == y (300):", x == y)
assert (x == y) == True
print("Small integer caching tests passed!")
```

### 2. Type Checking with is and type()

```python
x = 500
print("type(x) is int:", type(x) is int)
assert (type(x) is int) == True

x_float = 100.34
print("type(x_float) is float:", type(x_float) is float)
assert (type(x_float) is float) == True
print("Type checking tests passed!")
```

### 3. Using is not Operator

```python
x = 56.78
y = 100
print("x is not y:", x is not y)
assert (x is not y) == True
print("is not tests passed!")
```
