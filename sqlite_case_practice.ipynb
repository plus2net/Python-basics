## 2. Basic CASE Statement (`sqlite3`)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = """SELECT id, name, class, mark, gender,
CASE class
    WHEN 'Four' THEN '1st floor'
    WHEN 'Five' THEN '2nd floor'
    WHEN 'Three' THEN '2nd floor'
    WHEN 'Two' THEN '1st floor'
    ELSE 'Ground floor'
END AS location
FROM student LIMIT 0,5"""

try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row[0], row[1], row[2], row[3], row[4], '-', row[5])
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 3. Allotting Grades Based on Marks (`CASE WHEN mark >= ...`)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = """SELECT id, name, class, mark, gender,
CASE
    WHEN mark >= 90 THEN 'A'
    WHEN mark >= 80 THEN 'B'
    WHEN mark >= 70 THEN 'C'
    ELSE 'FAIL'
END AS grade
FROM student LIMIT 0,5"""

try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row[0], row[1], row[2], row[3], row[4], '-', row[5])
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 4. Using CASE with BETWEEN Ranges

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
q = """SELECT id, name, class, mark, gender,
CASE
    WHEN mark BETWEEN 90 AND 100 THEN 'A'
    WHEN mark BETWEEN 80 AND 89 THEN 'B'
    WHEN mark BETWEEN 70 AND 79 THEN 'C'
    ELSE 'FAIL'
END AS grade
FROM student LIMIT 0,5"""

try:
    my_cursor = my_conn.execute(q)
    for row in my_cursor:
        print(row[0], row[1], row[2], row[3], row[4], '-', row[5])
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 5. Parameterized CASE Query

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = (90, 100, 80, 89, 70, 79)
q = """SELECT id, name, class, mark, gender,
CASE
    WHEN mark BETWEEN ? AND ? THEN 'A'
    WHEN mark BETWEEN ? AND ? THEN 'B'
    WHEN mark BETWEEN ? AND ? THEN 'C'
    ELSE 'FAIL'
END AS grade
FROM student LIMIT 0,5"""

try:
    my_cursor = my_conn.execute(q, my_data)
    for row in my_cursor:
        print(row[0], row[1], row[2], row[3], row[4], '-', row[5])
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 6. Financial Year Quarter Grouping using Date Ranges

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_data = (
    '2020-04-01', '2020-06-30',
    '2020-07-01', '2020-09-30',
    '2020-10-01', '2020-12-31',
    '2021-01-01', '2021-03-31'
)
q = """SELECT id, name, mark, class, dot,
CASE
    WHEN dot BETWEEN ? AND ? THEN 'Q1'
    WHEN dot BETWEEN ? AND ? THEN 'Q2'
    WHEN dot BETWEEN ? AND ? THEN 'Q3'
    WHEN dot BETWEEN ? AND ? THEN 'Q4'
    ELSE 'Outside FY'
END AS Date
FROM student LIMIT 0,5"""

try:
    my_cursor = my_conn.execute(q, my_data)
    for row in my_cursor:
        print(row[0], row[1], row[2], row[3], row[4], '-', row[5])
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 7. Modern SQLAlchemy 2.0+ Context Manager with CASE

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

my_conn = create_engine("sqlite:///my_db.db")
q = text("""SELECT id, name, mark, class,
CASE
    WHEN mark >= 90 THEN 'A'
    WHEN mark >= 80 THEN 'B'
    WHEN mark >= 70 THEN 'C'
    ELSE 'FAIL'
END AS grade
FROM student LIMIT 0,5""")

try:
    with my_conn.connect() as connection:
        r_set = connection.execute(q)
        for row in r_set:
            print(row)
except SQLAlchemyError as e:
    print("SQLAlchemy error:", e)
```
