### 1. Basic str.isspace() Validation
`str.isspace()` returns `True` if all characters in the string are whitespace, and length > 0.

```python
print("'   '.isspace():      ", '   '.isspace())
print("'\\t'.isspace():       ", '\t'.isspace())
print("'\\n'.isspace():       ", '\n'.isspace())
print("' \\t\\r\\n '.isspace(): ", ' \t\r\n '.isspace())
print("'  hello  '.isspace(): ", '  hello  '.isspace())
print("''.isspace():         ", ''.isspace())

assert '   '.isspace() is True
assert '\t'.isspace() is True
assert '\n'.isspace() is True
assert '  hello  '.isspace() is False
assert ''.isspace() is False
```

### 2. The Empty String Trap: isspace() vs. not string.strip()
Empty strings return `False` under `isspace()`. Use `not s.strip()` to detect blank inputs.

```python
def is_blank(text):
    return not text or text.isspace()

print("Empty string:     ", is_blank(''))
print("Spaces only:      ", is_blank('   '))
print("Non-empty text:   ", is_blank('data'))

assert is_blank('') is True
assert is_blank('   ') is True
assert is_blank('data') is False
```

### 3. Unicode Whitespace (NBSP, Em-Space, CJK Spaces)
`str.isspace()` recognizes international typographic spaces.

```python
nbsp = '\u00a0'        # Non-breaking space
em_space = '\u2003'    # Em-space
cjk_space = '\u3000'   # Ideographic CJK full-width space
zw_space = '\u200b'    # Zero-width space (category Cf, NOT space separator)

print("NBSP isspace:      ", nbsp.isspace())
print("Em-space isspace:  ", em_space.isspace())
print("CJK space isspace: ", cjk_space.isspace())
print("Zero-width isspace:", zw_space.isspace())

assert nbsp.isspace() is True
assert em_space.isspace() is True
assert cjk_space.isspace() is True
assert zw_space.isspace() is False
```

### 4. Skipping Blank Lines Without Modifying Indentation
Filter blank lines while preserving leading whitespace on content lines.

```python
code_lines = [
    "def greet():\n",
    "    \n",               # 4 spaces
    "    print('Hello')\n",
    "\t\n",                # Tab
    "    return True\n"
]

filtered = [l for l in code_lines if not l.isspace()]
print(f"Retained {len(filtered)} of {len(code_lines)} lines.")
assert len(filtered) == 3
```

### 5. Practice Exercise: Paragraph Extractor & Form Auditor
Run the complete document cleaner and paragraph extractor with unit assertions.

```python
class DocumentCleaner:
    @staticmethod
    def is_blank_line(line):
        if not isinstance(line, str):
            return True
        return len(line) == 0 or line.isspace()

    @staticmethod
    def extract_paragraphs(text_lines):
        paragraphs = []
        current = []
        for line in text_lines:
            if DocumentCleaner.is_blank_line(line):
                if current:
                    paragraphs.append('\n'.join(current))
                    current = []
            else:
                current.append(line.strip())
        if current:
            paragraphs.append('\n'.join(current))
        return paragraphs

# Unit Assertions
cleaner = DocumentCleaner()
assert cleaner.is_blank_line('') is True
assert cleaner.is_blank_line('   \t\n') is True
assert cleaner.is_blank_line('\u00a0\u3000') is True
assert cleaner.is_blank_line('Data content') is False

doc = [
    'First Paragraph Title',
    'Body of paragraph one.',
    '   ',
    'Second Paragraph Title',
    'Body of paragraph two.'
]
paras = cleaner.extract_paragraphs(doc)
assert len(paras) == 2
assert 'First Paragraph Title' in paras[0]
assert 'Second Paragraph Title' in paras[1]

print('All DocumentCleaner assertions passed successfully!')
```
