### 1. Basic Element Insertion & Duplicate Handling
`set.add(item)` modifies the set in-place and returns `None`. Duplicate additions are silent no-ops.

```python
names = {"Alex", "Ronald", "John"}

# Insert new element
names.add("Lewis")
print("After adding Lewis:", names)

# Duplicate insertion
names.add("Alex")
print("After duplicate add:", names)
```

### 2. set.add() vs set.update()
`add()` embeds the object as a single item; `update()` iterates and unpacks containers.

```python
s1 = {1, 2}
s2 = {1, 2}

# add() inserts tuple intact
s1.add((3, 4))
print("add((3, 4)):", s1)  # {1, 2, (3, 4)}

# update() unpacks tuple members
s2.update((3, 4))
print("update((3, 4)):", s2)  # {1, 2, 3, 4}
```

### 3. Hashability Constraints: Unhashable Types
Sets require hashable elements. Adding mutable lists or dicts raises TypeError.

```python
valid_set = {10, 20}

# Adding immutable frozenset works
valid_set.add(frozenset({30, 40}))
print("With frozenset:", valid_set)

# Adding mutable list raises TypeError
try:
    valid_set.add([50, 60])
except TypeError as err:
    print("Caught expected TypeError:", err)
```

### 4. Hash Equality Nuance: 1 == 1.0 == True
Equal values with identical hash codes collide and are not added as duplicates.

```python
flags = {1}
flags.add(1.0)
flags.add(True)
print("Flags set (only one member):", flags)  # {1}

zeros = {0}
zeros.add(False)
print("Zeros set (only one member):", zeros)  # {0}
```

### 5. Production Recipe: Streaming Visitor Deduplication
Collect unique entities dynamically from a high-throughput event stream.

```python
events = ["usr_1", "usr_2", "usr_1", "usr_3", "usr_2", "usr_4"]
unique_users = set()

for ev in events:
    unique_users.add(ev)

print("Unique users count:", len(unique_users))  # 4
print("Unique users:", sorted(unique_users))  # ['usr_1', 'usr_2', 'usr_3', 'usr_4']
```
