### 1. Basic Disjoint Evaluation
`A.isdisjoint(B)` returns `True` if no items overlap; `False` if at least one item is shared.

```python
# Overlapping sets -> False
A = {'x', 'y', 'z'}
B = {'x', 'b', 'c'}
print("Overlapping check (shares 'x'):", A.isdisjoint(B))

# Disjoint sets -> True
C = {'a', 'b', 'c'}
print("Disjoint check (no shared elements):", A.isdisjoint(C))
```

### 2. Heterogeneous Iterables & String Character Splitting
Passing a list tests list items. Passing a string decomposes into individual characters.

```python
# List evaluation
allowed_ports = {80, 443, 8080}
blocked_ports = [21, 22, 23, 25]
print("List test:", allowed_ports.isdisjoint(blocked_ports))

# String character trap
target_chars = {'x', 'y', 'z'}
print("String test ('Alex' contains 'x'):", target_chars.isdisjoint('Alex'))

# Word-level check (wrap in list or set)
roles = {"admin", "editor"}
print("Word-level test:", roles.isdisjoint(["guest"]))
```

### 3. Early Short-Circuit Demonstration
`isdisjoint()` exits on the first collision, avoiding full collection scans.

```python
huge_set = set(range(1_000_000))
candidate = [0, 2_000_000, 3_000_000]

# Immediate short-circuit on 0:
res = huge_set.isdisjoint(candidate)
print("Short-circuit result:", res)
assert res is False
```

### 4. Real-World Recipe: Separation of Duties (SoD)
Enforce compliance where certain roles must remain mutually exclusive.

```python
def check_sod_compliance(user_roles: set, conflict_rules: dict) -> bool:
    for role, prohibited in conflict_rules.items():
        if role in user_roles and not user_roles.isdisjoint(prohibited):
            return False
    return True

rules = {
    "invoice_creator": {"invoice_approver", "payment_dispatcher"}
}

user_ok = {"invoice_creator", "report_viewer"}
user_bad = {"invoice_creator", "payment_dispatcher"}

print("User OK compliant?", check_sod_compliance(user_ok, rules))
print("User Bad compliant?", check_sod_compliance(user_bad, rules))
assert check_sod_compliance(user_ok, rules) is True
assert check_sod_compliance(user_bad, rules) is False
```

### 5. Unit Assertion Suite
Verify edge cases including empty set invariants and dictionary key handling.

```python
# Assertion 1: Empty set invariants
assert set().isdisjoint({1, 2, 3}) is True
assert set().isdisjoint(set()) is True

# Assertion 2: Disjoint vs overlapping
s = {1, 2, 3}
assert s.isdisjoint({4, 5}) is True
assert s.isdisjoint({3, 4}) is False

# Assertion 3: Dictionary keys only
d_keys = {"a": 1, "b": 2}
assert {1, 2}.isdisjoint(d_keys) is True  # 1 and 2 not in {'a', 'b'}
assert {"a"}.isdisjoint(d_keys) is False

print("All assertions passed successfully!")
```
