### 1. Basic str.rindex() Usage: Rightmost 0-Based Offsets
`str.rindex()` returns the highest (rightmost) 0-based index where the target substring begins.

```python
sentence = "Find the last apple in the apple orchard"

# 'apple' appears at index 14 and index 27
first_pos = sentence.index("apple")
last_pos = sentence.rindex("apple")

print("Full string:                  ", sentence)
print("First occurrence (str.index): ", first_pos)  # 14
print("Last occurrence  (str.rindex):", last_pos)   # 27
```

### 2. Strict Exception Guarantee: Handling ValueError
Unlike `rfind()`, which returns `-1`, `rindex()` raises a `ValueError` when the target is missing.

```python
def locate_tail_token(payload: str, target: str) -> int | None:
    try:
        return payload.rindex(target)
    except ValueError:
        print(f"Target '{target}' not present in payload.")
        return None

pos1 = locate_tail_token("config.yaml", ".")
print("Found dot at index:", pos1)

pos2 = locate_tail_token("config_yaml", ".")
print("Missing dot returned:", pos2)
```

### 3. Constraining Search Windows with start and end Offsets
Constrain searches to specific substrings without copying or slicing memory.

```python
log_entry = "error1: info; error2: critical; error3: debug"

# Search within first 25 characters [0:25]
range_pos = log_entry.rindex("error", 0, 25)
print("Last error in [0:25]:", range_pos)  # Index 14 (error2)

# Search with negative start boundary (last 15 characters)
tail_pos = log_entry.rindex("error", -15)
print("Last error in tail:  ", tail_pos)   # Index 32 (error3)
```

### 4. Real-World Recipe: Strict File Extension Extraction
Enforce validation by requiring that the dot separator exists.

```python
def strict_extract_extension(filepath: str) -> str:
    try:
        dot_idx = filepath.rindex(".")
        if dot_idx == 0 or filepath[dot_idx - 1] == "/":
            raise ValueError("Hidden file without extension")
        return filepath[dot_idx + 1:]
    except ValueError as err:
        raise ValueError(f"Malformed filename '{filepath}': {err}")

ext = strict_extract_extension("reports/financial_2026.pdf")
print("Extracted file extension:", ext)
```

### 5. Real-World Recipe: Slicing Text After Last Delimiter (Basename)
Extract the filename from a deeply nested path using `rindex()`.

```python
def get_basename(path: str) -> str:
    try:
        slash_pos = path.rindex("/")
        return path[slash_pos + 1:]
    except ValueError:
        return path

sample_path = "user/documents/reports/summary.pdf"
print("Full path: ", sample_path)
print("Base name: ", get_basename(sample_path))
```

### 6. Real-World Recipe: Parsing Trailing URL Endpoint Identifiers
Extract resource IDs from REST API paths by locating the rightmost forward slash.

```python
endpoint = "https://api.plus2net.com/v2/users/84920"
last_slash = endpoint.rindex("/")
user_id = endpoint[last_slash + 1:]
print("Endpoint:        ", endpoint)
print("Extracted User ID:", user_id)
```

### 7. Pitfalls: Case Sensitivity & Normalization
`rindex()` is strictly case-sensitive. Use `.casefold()` for case-insensitive rightmost lookups.

```python
data = "Apple and apple"

# Exact match
print("Lower match:", data.rindex("apple"))  # 10

# Casefold match
ci_index = data.casefold().rindex("APPLE".casefold())
print("Case-insensitive match:", ci_index)    # 10
```

### 8. Self-Checking Verification Suite
Run complete unit assertions verifying all `rindex` behaviors and edge cases.

```python
class RightIndexVerifier:
    @staticmethod
    def safe_rindex(source: str, target: str) -> int | None:
        try:
            return source.rindex(target)
        except ValueError:
            return None

    @staticmethod
    def extract_tail(source: str, sep: str) -> str:
        try:
            idx = source.rindex(sep)
            return source[idx + len(sep):]
        except ValueError:
            return source

# Verification assertions
verifier = RightIndexVerifier()
assert verifier.safe_rindex("banana", "a") == 5
assert verifier.safe_rindex("banana", "na") == 4
assert verifier.safe_rindex("banana", "z") is None

try:
    "hello world".rindex("python")
    assert False, "ValueError should have been raised"
except ValueError:
    pass

assert verifier.extract_tail("/usr/local/bin/python3", "/") == "python3"
assert verifier.extract_tail("standalone", "/") == "standalone"

print("All str.rindex() unit assertions passed successfully! 100% verified.")
```
