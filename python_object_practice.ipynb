### 1. Root Hierarchy Verification

```python
class Widget:
    pass

print("Widget is subclass of object:", issubclass(Widget, object))
print("int is subclass of object:   ", issubclass(int, object))
print("list is subclass of object:  ", issubclass(list, object))
print("lambda is instance of object:", isinstance(lambda x: x, object))
```

### 2. Featureless Instances without __dict__

```python
bare = object()
print("Type:      ", type(bare))
print("Memory ID: ", hex(id(bare)))

try:
    bare.value = 10
except AttributeError as err:
    print("AttributeError caught:", err)
```

### 3. The Unique Sentinel Pattern

```python
_SENTINEL = object()

def get_setting(settings_map, key, default=_SENTINEL):
    if key in settings_map:
        return settings_map[key]
    if default is _SENTINEL:
        raise KeyError(f"Key '{key}' not set!")
    return default

store = {"auth_token": None, "retries": 3}
print("Auth token found: ", get_setting(store, "auth_token"))
print("Default fallback:  ", get_setting(store, "timeout", default=60))
```

### 4. Callable Object Protocol

```python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor
    def __call__(self, val):
        return self.factor * val

double = Multiplier(2)
triple = Multiplier(3)

print("double(15):", double(15))
print("triple(15):", triple(15))
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Root hierarchy
assert issubclass(bool, object)
assert isinstance(42, object)

# Test 2: Sentinel uniqueness
sentinel_a = object()
sentinel_b = object()
assert sentinel_a is not sentinel_b

# Test 3: AttributeError on bare object assignment
try:
    sentinel_a.data = 1
    assert False
except AttributeError:
    pass

# Test 4: Callable protocol
class Adder:
    def __call__(self, a, b): return a + b
assert Adder()(10, 20) == 30

print("All object() unit assertions successfully verified! 100% passed.")
```
