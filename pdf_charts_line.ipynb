## 2. Horizontal Line Chart with Categorical Data

Build a `HorizontalLineChart` with custom stroke colors, stroke widths, and category axis labels.

```python
from pathlib import Path
from reportlab.lib import colors
from reportlab.graphics.shapes import Drawing
from reportlab.graphics.charts.linecharts import HorizontalLineChart
from reportlab.graphics import renderPDF

output_dir = Path('output')
output_dir.mkdir(exist_ok=True)
pdf_line_chart = output_dir / 'line_chart.pdf'

d = Drawing(1000, 500)
lc = HorizontalLineChart()

data = [
    (13, 5, 20, 22, 37, 45, 19, 4),
    (5, 20, 46, 38, 23, 21, 6, 14)
]
lc.x = 50
lc.y = 50
lc.height = 400
lc.width = 930
lc.data = data
lc.lines[0].strokeWidth = 2
lc.lines[1].strokeWidth = 4
lc.lines[0].strokeColor = colors.blue
lc.lines[1].strokeColor = colors.red
lc.valueAxis.valueMin = 0
lc.valueAxis.valueMax = 50
lc.valueAxis.valueStep = 10
lc.categoryAxis.labels.boxAnchor = 'ne'
lc.categoryAxis.labels.dx = 5
lc.categoryAxis.labels.dy = -2
lc.categoryAxis.labels.angle = 30
lc.categoryAxis.categoryNames = ['Jan-22', 'Feb-22', 'Mar-22', 'Apr-22', 'May-22', 'Jun-22', 'Jul-22', 'Aug-22']
d.add(lc, '')

renderPDF.drawToFile(d, str(pdf_line_chart), '')
print(f"Horizontal line chart PDF created: {pdf_line_chart}")
```

## 3. Scientific Line Plot with Coordinate Markers

Build a `LinePlot` with numerical $(x, y)$ coordinate tuples, custom markers (`FilledCircle`, `Circle`), and label formatting.

```python
from reportlab.graphics.charts.lineplots import LinePlot
from reportlab.graphics.widgets.markers import makeMarker

pdf_line_plot = output_dir / 'line_plot.pdf'
d = Drawing(1000, 500)

data = [
    ((1, 4), (2, 2), (2.5, 1), (3, 3), (4, 5)),
    ((1, 3), (2, 3), (2.5, 2), (3.5, 5), (4, 6))
]
lp = LinePlot()
lp.x = 40
lp.y = 50
lp.height = 445
lp.width = 945
lp.data = data
lp.joinedLines = 1
lp.lines[0].symbol = makeMarker('FilledCircle')
lp.lines[1].symbol = makeMarker('Circle')
lp.lineLabelFormat = '%2.0f'
lp.strokeColor = colors.black
lp.xValueAxis.valueMin = 0
lp.xValueAxis.valueMax = 5
lp.xValueAxis.valueSteps = [1, 2, 2.5, 3, 4, 5]
lp.xValueAxis.labelTextFormat = '%2.1f'
lp.yValueAxis.valueMin = 0
lp.yValueAxis.valueMax = 7
lp.yValueAxis.valueSteps = [1, 2, 3, 5, 6]
d.add(lp)

renderPDF.drawToFile(d, str(pdf_line_plot), '')
print(f"Line plot PDF created: {pdf_line_plot}")
```
