### 1. Basic Instance Namespace Inspection
`vars(instance)` returns the instance `__dict__` mapping.

```python
class Animal:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

tiger = Animal("Ronald", 5)
print("Instance vars:", vars(tiger))

# Mutate in-place via vars dict
vars(tiger)["name"] = "Leo"
print("Updated name:", tiger.name)
```

### 2. Using vars() Without Arguments
`vars()` with no arguments returns the local symbol table dictionary.

```python
def test_local_namespace():
    x = 10
    y = "example"
    return vars()

print("Local namespace:", test_local_namespace())
```

### 3. Class vs. Instance: mappingproxy
Classes return a read-only `mappingproxy` while instances return a mutable `dict`.

```python
class SimpleClass:
    species = "mammal"

cls_ns = vars(SimpleClass)
print("Class namespace type:", type(cls_ns))

try:
    cls_ns["species"] = "bird"
except TypeError as err:
    print("Caught mappingproxy error:", err)
```

### 4. Slotted Classes and Primitives Lacking __dict__
Objects without `__dict__` raise TypeError.

```python
try:
    vars([1, 2, 3])
except TypeError as err:
    print("List TypeError:", err)

class SlottedPoint:
    __slots__ = ("x", "y")

pt = SlottedPoint()
try:
    vars(pt)
except TypeError as err:
    print("Slotted TypeError:", err)
```

### 5. Production Recipe: Object State Snapshot & Rollback
Capture and restore instance state dictionaries.

```python
class Record:
    def __init__(self, val, status):
        self.val = val
        self.status = status

rec = Record(100, "active")
backup = vars(rec).copy()

# Mutate
rec.status = "cancelled"
rec.val = 0
print("Mutated:", vars(rec))

# Restore
vars(rec).clear()
vars(rec).update(backup)
print("Restored:", vars(rec))
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying vars() mechanics.

```python
class VarsVerifier:
    @staticmethod
    def test_all():
        class Person:
            def __init__(self, name):
                self.name = name
                
        p = Person("Alice")
        assert vars(p) == {"name": "Alice"}
        vars(p)["name"] = "Bob"
        assert p.name == "Bob"
        
        try:
            vars(42)
            assert False
        except TypeError:
            pass
            
        print("All vars() unit assertions passed successfully! 100% verified.")

VarsVerifier.test_all()
```
