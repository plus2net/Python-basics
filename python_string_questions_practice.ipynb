## 1. Multi-line String Length & Line Splitting

**Challenge**: Calculate total character count and total line count of a multi-line string.


```python
doc = """Welcome to Python
String Tutorials
at plus2net.com"""

total_chars = len(doc)
total_lines = len(doc.splitlines())
print(f"Chars: {total_chars}, Lines: {total_lines}")

# Unit Assertions
doc = """Welcome to Python
String Tutorials
at plus2net.com"""
assert len(doc) == 50
assert len(doc.splitlines()) == 3
print('Verified!')
```

## 2. Initial Capitalization without title() Side Effects

**Challenge**: Capitalize only the first letter while keeping subsequent uppercase characters intact.


```python
text = "welcome to Plus2Net Python"
result = text[0].upper() + text[1:] if text else ""
print(result)

# Unit Assertions
t = "welcome to Plus2Net Python"
assert t[0].upper() + t[1:] == "Welcome to Plus2Net Python" 
print('Verified!')
```

## 3. Symmetrical Centering with Asterisks

**Challenge**: Center a string and fill both sides with asterisks equal to half of its length.


```python
name = "plus2net"
pad = len(name) // 2
centered = name.center(len(name) + pad * 2, "*")
print(centered)

# Unit Assertions
n = "plus2net"
assert n.center(len(n) + (len(n)//2)*2, "*") == "****plus2net****" 
print('Verified!')
```

## 4. International Caseless Matching (casefold vs. lower)

**Challenge**: Demonstrate why casefold() matches German sharp 'ß' while lower() fails.


```python
s1 = "dDcßeF"
s2 = "DDCSSef"
print("lower() equal:   ", s1.lower() == s2.lower())
print("casefold() equal:", s1.casefold() == s2.casefold())

# Unit Assertions
s1, s2 = "dDcßeF", "DDCSSef"
assert (s1.lower() == s2.lower()) is False
assert (s1.casefold() == s2.casefold()) is True
print('Verified!')
```

## 5. Bounded Substring Occurrence Counting

**Challenge**: Count occurrences of 'co' and 'o' across specific slice index offsets.


```python
s = "Welcome to Python"
print("Count 'co':       ", s.count('co'))
print("Count 'o' [4:10]: ", s.count('o', 4, 10))

# Unit Assertions
s = "Welcome to Python"
assert s.count('co') == 1
assert s.count('o', 4, 10) == 2
print('Verified!')
```

## 7. Symmetrical Palindrome Verification

**Challenge**: Verify whether a phrase reads the same backward after sanitizing punctuation.


```python
def is_palindrome(text: str) -> bool:
    clean = [c.casefold() for c in text if c.isalnum()]
    return clean == clean[::-1]

print("'racecar': ", is_palindrome("racecar"))
print("'Panama':  ", is_palindrome("A man, a plan, a canal: Panama"))

# Unit Assertions
def is_p(t):
    c = [ch.casefold() for ch in t if ch.isalnum()]
    return c == c[::-1]
assert is_p("racecar") is True
assert is_p("A man, a plan, a canal: Panama") is True
print('Verified!')
```

## 8. Safe Key-Value Parsing (partition vs. split)

**Challenge**: Use partition() to prevent unpacking ValueError crashes when delimiters are missing.


```python
line = "TIMEOUT"
key, sep, val = line.partition("=")
val = val if sep else "30"
print(f"Key: {key}, Value: {val}")

# Unit Assertions
k, s, v = "TIMEOUT".partition("=")
assert k == "TIMEOUT" and s == "" and v == "" 
print('Verified!')
```

## 9. Universal Newline Splitting with keepends

**Challenge**: Split multiline text across Windows CRLF and Unix LF while keeping line breaks.


```python
mixed = "Line1\r\nLine2\nLine3"
kept = mixed.splitlines(keepends=True)
print("Preserved:", kept)

# Unit Assertions
m = "Line1\r\nLine2\nLine3"
assert m.splitlines(keepends=True) == ['Line1\r\n', 'Line2\n', 'Line3']
print('Verified!')
```

## 10. Simultaneous Character Swapping with translate()

**Challenge**: Swap characters 'A' and 'B' in 'ABBA' in a single pass without rule collision.


```python
table = str.maketrans("AB", "BA")
swapped = "ABBA".translate(table)
print("Swapped:", swapped)

# Unit Assertions
t = str.maketrans("AB", "BA")
assert "ABBA".translate(t) == "BAAB" 
print('Verified!')
```

## 11. High-Speed Mass Punctuation Deletion

**Challenge**: Delete all standard ASCII punctuation from text in native C using maketrans and translate.


```python
import string

raw = "Hello, world! Welcome to Python: fast & robust."
table = str.maketrans("", "", string.punctuation)
clean = raw.translate(table)
print(clean)

# Unit Assertions
import string
t = str.maketrans("", "", string.punctuation)
assert "Hello, world!".translate(t) == "Hello world" 
print('Verified!')
```

## 12. Sign-Aware Leading Zero Padding (zfill)

**Challenge**: Pad signed negative and positive numbers to 6 digits preserving sign placement.


```python
pos = "+42"
neg = "-42"
print(pos.zfill(6))
print(neg.zfill(6))

# Unit Assertions
assert "+42".zfill(6) == "+00042"
assert "-42".zfill(6) == "-00042" 
print('Verified!')
```

## 13. Date Extraction & ISO Conversion with Regex

**Challenge**: Extract DD-MM-YYYY dates from text and rearrange into ISO YYYY-MM-DD.


```python
import re

text = "Birthday: 24-12-2002"
iso = re.sub(r"\b(\d{2})-(\d{2})-(\d{4})\b", r"\3-\2-\1", text)
print(iso)

# Unit Assertions
import re
assert re.sub(r"\b(\d{2})-(\d{2})-(\d{4})\b", r"\3-\2-\1", "24-12-2002") == "2002-12-24" 
print('Verified!')
```
