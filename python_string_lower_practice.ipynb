# Python str.lower(): Convert Strings to Lowercase

### 1. Basic Lowercase Conversion
`str.lower()` converts all uppercase letters to lowercase, returning a new string while leaving the original string unchanged.

```python
text1 = "WELCOME TO PLUS2NET"
text2 = "Python 3.12 Programming!"
text3 = "already lowercase"

print("1:", text1.lower())
print("2:", text2.lower())
print("3:", text3.lower())
```

### 2. Strings with Numbers, Symbols & Emojis
Non-alphabetic characters remain completely unaffected by `lower()`.

```python
data = "Order #4892-XYZ © 2026 🚀"
print("Cleaned:", data.lower())
```

### 3. Case-Insensitive User Input Normalization
Normalize user inputs with `.strip().lower()` before comparison to eliminate casing variations.

```python
user_input = "  YES  "

if user_input.strip().lower() == "yes":
    print("Action confirmed!")
else:
    print("Action rejected.")
```

### 4. lower() vs. casefold() for Unicode Caseless Matching
German sharp s ('ß') stays 'ß' under `lower()`, but converts to 'ss' under `casefold()`.

```python
german_upper = "STRASSE"
german_sharp = "straße"

print("lower() equal?   ", german_upper.lower() == german_sharp.lower())        # False
print("casefold() equal?", german_upper.casefold() == german_sharp.casefold())  # True
```

### 5. Hands-on Practice Exercise: User Registration Normalizer
Run the self-checking unit assertions below to verify that duplicate emails in different casings are prevented.

```python
def register_user(accounts_store, email, display_name):
    clean_email = email.strip().lower()
    clean_name = display_name.strip()
    
    if clean_email in accounts_store:
        return False, "Duplicate account: email already registered."
        
    accounts_store[clean_email] = clean_name
    return True, "Registration successful."

db = {}

# Test cases with assertions
ok1, _ = register_user(db, "Alice@Plus2Net.com", "Alice Cooper")
assert ok1 is True
assert "alice@plus2net.com" in db

ok2, msg = register_user(db, "  ALICE@plus2net.COM  ", "Alice Clone")
assert ok2 is False
assert msg == "Duplicate account: email already registered."

ok3, _ = register_user(db, "Bob@Plus2Net.com", "Bob Builder")
assert ok3 is True
assert len(db) == 2

print("Accounts Database:", db)
print("All account normalizer assertions passed successfully!")
```
