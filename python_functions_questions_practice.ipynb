### Part 1: Parameters, *args & **kwargs

```python
# Q1 - Q4: Parameters & Arguments
def sum_all(*numbers):
    return sum(numbers)

def build_profile(first, last, **info):
    res = {'name': f"{first} {last}"}
    res.update(info)
    return res

print("Sum all:", sum_all(10, 20, 30, 40))
print("Profile:", build_profile("Alex", "Ronald", role="Admin", dept="IT"))
```

### Part 2: Return Values & Mutability

```python
# Q5 - Q8: Data flow and pure functions
def get_min_max(nums):
    return min(nums), max(nums)

low, high = get_min_max([12, 45, 2, 89, 23])
print(f"Min: {low}, Max: {high}")

def remove_negatives(lst):
    return [x for x in lst if x >= 0]

data = [10, -5, 20, -1, 30]
print("Cleaned:", remove_negatives(data))
```

### Part 3: Scope, Closures & Default Argument Trap

```python
# Q9 - Q12: Scope and Closures
def make_counter():
    count = 0
    def step():
        nonlocal count
        count += 1
        return count
    return step

c = make_counter()
print("Step 1:", c())
print("Step 2:", c())

# Safe default list
def safe_append(item, target_list=None):
    if target_list is None:
        target_list = []
    target_list.append(item)
    return target_list

print("Call 1:", safe_append("A"))
print("Call 2:", safe_append("B"))
```

### Part 4: Recursion & Higher-Order Functions

```python
# Q13 - Q15: Recursion and flattening
def factorial(n):
    return 1 if n <= 1 else n * factorial(n - 1)

def flatten(nested):
    flat = []
    for item in nested:
        if isinstance(item, list):
            flat.extend(flatten(item))
        else:
            flat.append(item)
    return flat

print("Factorial 6:", factorial(6))
print("Flattened:  ", flatten([1, [2, [3, 4], 5], [6]]))
```
