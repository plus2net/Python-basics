### 1. Basic Exponential Subtraction
`math.expm1(x)` computes $e^x - 1$.

```python
import math

print("expm1(1.0): ", math.expm1(1.0))
print("expm1(0.0): ", math.expm1(0.0))
print("expm1(-1.0):", math.expm1(-1.0))

assert math.expm1(0.0) == 0.0
assert abs(math.expm1(1.0) - (math.e - 1)) < 1e-15
```

### 2. Catastrophic Cancellation Benchmark
Compare `math.expm1(x)` against naive `math.exp(x) - 1` for small $x$.

```python
import math

test_powers = [1e-4, 1e-8, 1e-12, 1e-15, 1e-17]

print(f"{'x':>10} | {'math.exp(x) - 1':>22} | {'math.expm1(x)':>22}")
print("-" * 60)
for x in test_powers:
    naive = math.exp(x) - 1.0
    accurate = math.expm1(x)
    print(f"{x:>10.1e} | {naive:>22.16e} | {accurate:>22.16e}")

# At 1e-17, naive exp(x) - 1 evaluates to 0.0
assert (math.exp(1e-17) - 1.0) == 0.0
assert math.expm1(1e-17) == 1e-17
```

### 3. Inverse Duality with math.log1p()
`math.expm1(x)` is the exact inverse of `math.log1p(y)`.

```python
import math

x_vals = [1e-14, 1e-6, 0.25, 2.0, -0.5]

for x in x_vals:
    round_trip = math.log1p(math.expm1(x))
    print(f"Original: {x:>10.2e}, Reconstructed: {round_trip:>10.2e}")
    assert abs(round_trip - x) / abs(x) < 1e-14

print("All round-trip log1p(expm1(x)) tests passed!")
```

### 4. Continuous Compounding Financial Calculation
Compute accrued interest over micro-intervals: $I = P \cdot (e^{rt} - 1)$.

```python
import math

def accrued_interest(principal, annual_rate, hours):
    t_years = hours / (365.25 * 24)
    return principal * math.expm1(annual_rate * t_years)

p = 5_000_000.0  # $5 Million
r = 0.06         # 6%
interest_1hr = accrued_interest(p, r, 1)
print(f"1-Hour Accrued Interest on $5M: ${interest_1hr:.4f}")
assert interest_1hr > 0.0
```

### 5. Practice Exercise: High-Precision Rate & Radioactive Decay
Run the complete verification suite with self-checking assertions.

```python
import math

def effective_rate(nominal_rate, duration_seconds):
    t_years = duration_seconds / (365.25 * 86400)
    return math.expm1(nominal_rate * t_years)

def safe_inverse(eff_rate, duration_seconds):
    t_years = duration_seconds / (365.25 * 86400)
    return math.log1p(eff_rate) / t_years

# Unit Assertions
assert effective_rate(0.05, 0) == 0.0

r_10s = effective_rate(0.045, 10)
assert r_10s > 0.0
assert abs(safe_inverse(r_10s, 10) - 0.045) < 1e-14

# Check asymptotic behavior as x -> -inf
assert math.expm1(-600) == -1.0

print("10-second effective rate:", r_10s)
print("All math.expm1() practice assertions passed successfully!")
```
