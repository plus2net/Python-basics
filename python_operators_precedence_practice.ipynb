### 1. PEMDAS Grouping

```python
print("4 + 2 * 3   = ", 4 + 2 * 3)
print("(4 + 2) * 3 = ", (4 + 2) * 3)
```

### 2. Right-to-Left Exponentiation

```python
# 2 ** 3 ** 2 is 2 ** (3 ** 2) = 2 ** 9 = 512
print("2 ** 3 ** 2   =", 2 ** 3 ** 2)
print("(2 ** 3) ** 2 =", (2 ** 3) ** 2)
```

### 3. Logical and vs or Traps

```python
print("True or False and False:   ", True or False and False)
print("(True or False) and False: ", (True or False) and False)
```

### 4. Bitwise vs Comparison Precedence

```python
flags = 4
# flags & 1 == 0 evaluates as flags & (1 == 0) -> 0
print("flags & 1 == 0:  ", flags & 1 == 0)
print("(flags & 1) == 0:", (flags & 1) == 0)
```

### 5. Self-Checking Unit Verification Suite
