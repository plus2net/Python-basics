### 1. In-Place List Inversion

```python
letters = ['a', 'b', 'c', 'd', 'e']
print("Original: ", letters)

res = letters.reverse()
print("Reversed: ", letters)
print("Return val:", res)  # Always None
```

### 2. reverse() vs. Slicing vs. reversed()

```python
data = [10, 20, 30, 40]

# Slicing creates new list
sliced = data[::-1]
print("Sliced:  ", sliced)

# reversed() returns iterator
rev_iter = reversed(data)
print("Iterated:", list(rev_iter))
```

### 3. Palindrome Checker Recipe

```python
def is_palindrome(lst):
    return lst == lst[::-1]

print("radar palindrome:  ", is_palindrome(list('radar')))
print("python palindrome: ", is_palindrome(list('python')))
```

### 4. Recursive Deep Reversal

```python
def deep_rev(lst):
    lst.reverse()
    for x in lst:
        if isinstance(x, list):
            deep_rev(x)

nested = [[1, 2], [3, [4, 5]]]
deep_rev(nested)
print("Deep reversed:", nested)
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: In-place reversal
a = [1, 2, 3]
assert a.reverse() is None
assert a == [3, 2, 1]

# Test 2: Double reversal idempotency
a.reverse()
assert a == [1, 2, 3]

# Test 3: Empty and single item list
e = []
e.reverse()
assert e == []

s = [99]
s.reverse()
assert s == [99]

print("All list.reverse() unit assertions successfully verified! 100% passed.")
```
