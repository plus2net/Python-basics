### 1. Basic Summation
Sum collections of integers, floats, and sets.

```python
numbers = [10, 20, 30, 40]
print("List sum: ", sum(numbers))
print("Tuple sum:", sum((2.5, 4.0, 1.5)))
print("Set sum:  ", sum({5, 10, 15, 10}))
```

### 2. Using the start Parameter
Provide initial numeric offsets or sequence accumulators.

```python
print("Sum with start=50:", sum([15, 25, 10], start=50))
print("Flatten nested:   ", sum([[1, 2], [3, 4]], start=[]))
```

### 3. Precision: sum() vs. math.fsum()
Observe how math.fsum() prevents floating-point rounding drift.

```python
import math
vals = [0.1] * 10
print("Standard sum: ", sum(vals))
print("math.fsum():   ", math.fsum(vals))
```

### 4. Production Recipe: Weighted Average
Calculate weighted averages across paired lists.

```python
grades = [85, 90, 78]
weights = [0.2, 0.5, 0.3]
weighted = sum(g * w for g, w in zip(grades, weights))
print(f"Weighted average: {weighted:.2f}")
assert round(weighted, 2) == 85.40
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying sum() mechanics.

```python
assert sum([10, 20, 30]) == 60
assert sum([10, 20], start=5) == 35
assert sum([]) == 0
assert sum(1 for x in [5, 12, 18, 3] if x > 10) == 2
assert sum([[1], [2, 3]], []) == [1, 2, 3]

print("All sum() unit assertions passed! 100% verified.")
```
