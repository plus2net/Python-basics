### Part 1: Ranges & Traversal

```python
# Q1: Sum of first 10 natural numbers
total = 0
for i in range(1, 11):
    total += i
print("Sum of 1..10:", total)

# Q2: String traversal
word = 'Python'
chars = [f'Char: {ch}' for ch in word]
print("First two chars:", chars[:2])

# Q3: Enumerate with 1-based indexing
frameworks = ['Django', 'Flask', 'FastAPI']
indexed = [f'{idx}. {item}' for idx, item in enumerate(frameworks, start=1)]
print("Frameworks:", indexed)
```

### Part 2: Step Increments & Mathematical Loops

```python
# Q4: Even numbers & countdown
evens = list(range(2, 21, 2))
countdown = list(range(10, 0, -1))
print("Evens:    ", evens)
print("Countdown:", countdown)

# Q5: Factorial
num = 5
factorial = 1
for i in range(1, num + 1):
    factorial *= i
print(f"{num}! = {factorial}")
```

### Part 3: Break, Continue & for...else

```python
# Q7: Search with break
items = [10, 45, 88, 32, 99]
target = 88
found_pos = None
for pos, x in enumerate(items):
    if x == target:
        found_pos = pos
        break
print(f"Target {target} found at index:", found_pos)

# Q8: Skip multiples of 3
filtered = [n for n in range(1, 16) if n % 3 != 0]
print("Filtered numbers:", filtered)

# Q9: Prime check with for...else
def is_prime(candidate):
    if candidate < 2:
        return False
    for d in range(2, int(candidate ** 0.5) + 1):
        if candidate % d == 0:
            return False
    return True

print("Is 17 prime?", is_prime(17))
print("Is 25 prime?", is_prime(25))
```

### Part 4: Nested Loops & Matrix Operations

```python
# Q11: Matrix transposition
orig = [[1, 2], [3, 4], [5, 6]]
transposed = []
for c in range(len(orig[0])):
    transposed.append([orig[r][c] for r in range(len(orig))])
print("Transposed matrix:", transposed)

# Q12: Fibonacci series
fib = [0, 1]
for _ in range(2, 8):
    fib.append(fib[-1] + fib[-2])
print("Fibonacci (8 terms):", fib)
```
