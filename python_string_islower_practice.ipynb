### 1. Basic str.islower() Validation
`str.islower()` returns `True` if all cased characters are lowercase and length of cased characters > 0.

```python
print("'welcome to python'.islower(): ", 'welcome to python'.islower())
print("'python_3.14'.islower():       ", 'python_3.14'.islower())
print("'Welcome to python'.islower(): ", 'Welcome to python'.islower())
print("'12345'.islower():             ", '12345'.islower())
print("''.islower():                  ", ''.islower())

assert 'welcome to python'.islower() is True
assert 'python_3.14'.islower() is True
assert 'Welcome to python'.islower() is False
assert '12345'.islower() is False
assert ''.islower() is False
```

### 2. Unicode and International Lowercase Scripts
`str.islower()` checks Unicode letter categories across Latin, Greek, Cyrillic, and German `ß`.

```python
intl = ['café', 'münchen', 'straße', 'москва', 'αθήνα']
for s in intl:
    print(f"{s:<10}: {s.islower()}")
    assert s.islower() is True

# Uncased scripts (Chinese) lack case
assert '中文'.islower() is False
assert 'python 中文'.islower() is True
```

### 3. Password Policy: islower() vs. any(c.islower())
Demonstrate why `password.islower()` is incorrect for checking required lowercase characters.

```python
pwd = "SecretPass123!"
all_lower = pwd.islower()
has_lower = any(c.islower() for c in pwd)

print(f"pwd.islower():               {all_lower} (requires all letters to be lower)")
print(f"any(c.islower() for c in pwd): {has_lower} (satisfies password policy)")

assert all_lower is False
assert has_lower is True
```

### 4. SEO URL Slug and PEP 8 Module Validation
Verify web slug formatting and lowercase module names.

```python
def is_valid_slug(slug):
    s = slug.strip()
    return s.islower() and all(c.isalnum() or c == '-' for c in s)

print("Slug valid:  ", is_valid_slug('python-islower-guide'))
print("Slug invalid:", is_valid_slug('Python-Islower-Guide'))

assert is_valid_slug('python-islower-guide') is True
assert is_valid_slug('Python-Islower-Guide') is False
```

### 5. Practice Exercise: Naming & Security Auditor Suite
Run the complete auditing class with self-checking assertions.

```python
class NamingAndSecurityAuditor:
    @staticmethod
    def is_clean_url_slug(slug):
        if not isinstance(slug, str) or not slug:
            return False
        return slug.islower() and all(c.isalnum() or c == '-' for c in slug)

    @staticmethod
    def audit_password_lowercase(password):
        if not isinstance(password, str):
            return False, "Invalid input type."
        if not any(c.islower() for c in password):
            return False, "Password must include at least one lowercase character."
        return True, "Valid lowercase compliance."

# Unit Assertions
auditor = NamingAndSecurityAuditor()
assert auditor.is_clean_url_slug("python-islower-guide") is True
assert auditor.is_clean_url_slug("release-v3-14") is True
assert auditor.is_clean_url_slug("Python-Guide") is False
assert auditor.is_clean_url_slug("python_guide") is False
assert auditor.is_clean_url_slug("123-456") is False
assert auditor.is_clean_url_slug("") is False

assert auditor.audit_password_lowercase("ADMIN_123") == (False, "Password must include at least one lowercase character.")
assert auditor.audit_password_lowercase("Admin_123") == (True, "Valid lowercase compliance.")

print("All str.islower() assertions passed successfully!")
```
