## 1. Basic Right Alignment & Width Overflow Mechanics

`str.rjust(width[, fillchar])` places text on the right and pads the remaining space on the left.
If `width <= len(s)`, Python returns the original string unaltered without truncation.

```python
price = "$149.50"
aligned = price.rjust(12)

print("Original: ", repr(price))
print("Aligned:  ", repr(aligned))
print("Length:   ", len(aligned))

assert len(aligned) == 12
assert aligned == "     $149.50"

# Width smaller than length: never truncates
headline = "Enterprise Annual Report"
assert headline.rjust(10) == "Enterprise Annual Report"

print("\nAll basic right alignment assertions passed!")
```

## 2. The Signed Number Trap: `rjust()` vs. `zfill()`

`rjust(width, '0')` places zeros before the negative sign (`'00-42'`), corrupting valid numeric syntax.
`zfill(width)` places zeros after the sign (`'-0042'`).

```python
num = "-45.00"

# Flawed: rjust() corrupts sign
bad = num.rjust(10, "0")
print("rjust(10, '0'):", bad)  # '0000-45.00'
assert bad == "0000-45.00"

# Correct: zfill() preserves sign
good = num.zfill(10)
print("zfill(10):     ", good) # '-000045.00'
assert good == "-000045.00"

print("\nrjust vs zfill sign contrast passed!")
```

## 3. Directional Comparison: `rjust()` vs. `ljust()` vs. f-Strings

- `s.rjust(w, fill)` &harr; `f"{s:{fill}>{w}}"`
- `s.ljust(w, fill)` &harr; `f"{s:{fill}<{w}}"`

```python
token = "TOTAL"
w = 10
fill = "."

right = token.rjust(w, fill)
left = token.ljust(w, fill)

print("Right:", right)
print("Left: ", left)
assert right == ".....TOTAL"
assert left == "TOTAL....."

# Equivalence with f-strings
assert right == f"{token:{fill}>{w}}"
assert left == f"{token:{fill}<{w}}"

print("\nDirectional comparison assertions passed!")
```

## 4. Real-World Engineering Recipes

Explore four common software engineering patterns utilizing `str.rjust()`:
1. Financial Accounting Ledger Alignment
2. High-Precision Benchmark Telemetry Display
3. Table of Contents Page Leaders
4. Terminal Process Stat Table

```python
# Recipe 1: Financial ledger columns
def ledger_entry(name: str, debit: float, credit: float) -> str:
    d_str = f"${debit:,.2f}".rjust(12) if debit > 0 else "".rjust(12)
    c_str = f"${credit:,.2f}".rjust(12) if credit > 0 else "".rjust(12)
    return f"{name.ljust(20)} | {d_str} | {c_str}"

e1 = ledger_entry("Operating Cash", 14500.50, 0.0)
e2 = ledger_entry("Accounts Payable", 0.0, 3200.75)
print("Recipe 1 Ledger:\n" + e1 + "\n" + e2)
assert "$14,500.50" in e1
assert "$3,200.75" in e2

# Recipe 2: Benchmark telemetry
def benchmark_line(op: str, ms: float, width: int = 40) -> str:
    ms_str = f"{ms:>7.3f} ms"
    return (op + " ").ljust(width - len(ms_str), ".") + ms_str

b_line = benchmark_line("TLS Handshake", 24.152)
print("\nRecipe 2 Benchmark:\n", b_line)
assert b_line.endswith(" 24.152 ms")

# Recipe 3: TOC Page leader
def toc_line(title: str, pg: int, width: int = 35) -> str:
    return title.ljust(width - len(str(pg)), ".") + str(pg)

t_line = toc_line("1. Python Syntax", 5)
print("\nRecipe 3 TOC:\n", t_line)
assert t_line == "1. Python Syntax..................5"

print("\nAll engineering recipes passed!")
```

## 5. Production Implementation & Self-Checking Unit Assertions

This production-grade `RightAligner` class encapsulates currency formatting, TOC leaders, and automated unit assertions.

```python
class RightAligner:
    @staticmethod
    def align_currency(amount: float, field_width: int = 12) -> str:
        """Right-aligns currency strings for tabular alignment."""
        formatted = f"${amount:,.2f}"
        return formatted.rjust(field_width)

    @staticmethod
    def align_preceding_leader(title: str, page: int, total_width: int = 35) -> str:
        """Formats chapter and page with dot leaders."""
        pg_str = str(page)
        return title.ljust(total_width - len(pg_str), ".") + pg_str


# ==========================================
# UNIT ASSERTION SUITE
# ==========================================
aligner = RightAligner()

# 1. Basic space right alignment
assert "test".rjust(8) == "    test"
assert "test".rjust(4) == "test"
assert "test".rjust(2) == "test"

# 2. Custom fillchar
assert "50".rjust(6, "0") == "000050"
assert "End".rjust(8, "*") == "*****End"

# 3. Currency alignment
c1 = aligner.align_currency(1250.50, 12)
c2 = aligner.align_currency(45.00, 12)
assert c1 == "   $1,250.50"
assert c2 == "      $45.00"
assert len(c1) == 12
assert len(c2) == 12

# 4. Dot leader TOC line
toc = aligner.align_preceding_leader("Chapter 1", 5, 25)
assert toc == "Chapter 1...............5"
assert len(toc) == 25

# 5. Contrast with zfill on negative numbers
assert "-42".rjust(6, "0") == "000-42"
assert "-42".zfill(6) == "-00042"

# 6. TypeError validation on invalid fillchar
try:
    "sample".rjust(10, "--")
    assert False, "Expected TypeError not raised"
except TypeError as e:
    assert "must be exactly one character" in str(e)

print("All str.rjust() production unit assertions passed successfully!")
```
