### 1. Basic Square Root Calculations with math.sqrt()
`math.sqrt(x)` returns the principal non-negative square root of `x` as a 64-bit IEEE 754 float.


```python
import math

print("sqrt(4):   ", math.sqrt(4))      # 2.0
print("sqrt(16):  ", math.sqrt(16))     # 4.0
print("sqrt(100): ", math.sqrt(100))    # 10.0
print("sqrt(2):   ", round(math.sqrt(2), 6))   # 1.414214
print("sqrt(20):  ", round(math.sqrt(20), 6))  # 4.472136
print("sqrt(0.25):", math.sqrt(0.25))   # 0.5
print("sqrt(0.0): ", math.sqrt(0.0))    # 0.0
print("sqrt(-0.0):", math.sqrt(-0.0))   # -0.0
```

### 2. Comparison: math.sqrt() vs. ** 0.5 vs. cmath.sqrt()
`math.sqrt()` enforces strict real-number domain protection (raises `ValueError` on negative inputs), whereas `** 0.5` and `cmath.sqrt()` return complex numbers.


```python
import math
import cmath

try:
    math.sqrt(-16)
except ValueError as err:
    print("math.sqrt(-16):  ", err)

print("cmath.sqrt(-16): ", cmath.sqrt(-16))
print("(-16) ** 0.5:    ", (-16) ** 0.5)
```

### 3. Exact Integer Square Roots: math.isqrt() (Python 3.8+)
For massive integers beyond 53 bits of precision, `math.isqrt(n)` computes exact integer floor square roots without float conversion errors.


```python
import math

huge_int = 10**40
exact_root = math.isqrt(huge_int)
print("huge_int:          ", huge_int)
print("isqrt(10**40):     ", exact_root)
print("exact_root**2 == n?", exact_root**2 == huge_int)
```

### 4. Euclidean 2D Distance Calculation
Calculate straight-line distance between 2D Cartesian coordinates using the Pythagorean theorem.


```python
import math

def calculate_2d_distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    return math.sqrt(dx**2 + dy**2)

p1 = (13639.4, 1554.8)
p2 = (13637.5, 1570.6)
dist = calculate_2d_distance(p1[0], p1[1], p2[0], p2[1])
print(f"Distance between points: {dist:.2f} meters")
```

### 5. Hands-on Practice Exercise: Quadratic Equation Solver
Solve quadratic polynomial equations `a*x^2 + b*x + c = 0` for real roots using `math.sqrt()`.
Implement `solve_quadratic(a, b, c)` and verify with self-checking assertions.


```python
import math

def solve_quadratic(a, b, c):
    """
    Solves a*x^2 + b*x + c = 0 for real roots.
    Returns:
      - (root1, root2) tuple for 2 distinct real roots
      - (root,) tuple for 1 repeated root
      - () empty tuple if roots are complex (no real roots)
    """
    if a == 0:
        raise ValueError("Coefficient 'a' cannot be zero in a quadratic equation.")
        
    discriminant = b**2 - 4 * a * c
    
    if discriminant < 0:
        return ()
    elif discriminant == 0:
        root = -b / (2.0 * a)
        return (round(root, 4),)
    else:
        sqrt_disc = math.sqrt(discriminant)
        r1 = (-b + sqrt_disc) / (2.0 * a)
        r2 = (-b - sqrt_disc) / (2.0 * a)
        return (round(r1, 4), round(r2, 4))

# Unit Assertions
assert solve_quadratic(1, -5, 6) == (3.0, 2.0)
assert solve_quadratic(1, -4, 4) == (2.0,)
assert solve_quadratic(1, 0, 1) == ()

try:
    solve_quadratic(0, 2, 3)
    assert False, "Should raise ValueError"
except ValueError:
    pass

print("Roots of 2x^2 - 7x + 3 = 0:", solve_quadratic(2, -7, 3))
print("All quadratic equation assertions passed successfully!")
```
