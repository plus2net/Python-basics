# Python math.copysign(): Copy Sign Bit Between Numbers

### 1. Basic Sign Transfer with math.copysign()
`math.copysign(x, y)` returns a float with the magnitude (absolute value) of `x` and the sign bit of `y`.


```python
import math

print("copysign(3, -2): ", math.copysign(3, -2))   # -3.0
print("copysign(-3, -2):", math.copysign(-3, -2))  # -3.0
print("copysign(-3, 2): ", math.copysign(-3, 2))   # 3.0
print("copysign(3, 2):  ", math.copysign(3, 2))    # 3.0
print("Always float:    ", type(math.copysign(10, 1)))
```

### 2. Detecting IEEE 754 Negative Zero (-0.0)
Ordinary numerical comparisons (`0.0 == -0.0`) evaluate to True, and `-0.0 < 0` evaluates to False. `math.copysign()` inspects the underlying hardware sign bit directly, allowing robust negative zero detection.


```python
import math

pos_zero = 0.0
neg_zero = -0.0

print("pos_zero == neg_zero: ", pos_zero == neg_zero)  # True
print("neg_zero < 0.0:        ", neg_zero < 0.0)       # False
print("copysign(1.0, pos_zero):", math.copysign(1.0, pos_zero)) # 1.0
print("copysign(1.0, neg_zero):", math.copysign(1.0, neg_zero)) # -1.0

def is_negative_zero(val):
    return val == 0.0 and math.copysign(1.0, val) < 0.0

print("is_negative_zero(0.0): ", is_negative_zero(pos_zero))
print("is_negative_zero(-0.0):", is_negative_zero(neg_zero))
```

### 3. Comparing with Manual Ternary Patterns
The manual pattern `abs(x) if y >= 0 else -abs(x)` fails on `-0.0` because `-0.0 >= 0` is True. `math.copysign()` correctly copies the sign bit.


```python
import math

x = 5.0
y = -0.0

flawed_result = abs(x) if y >= 0 else -abs(x)
correct_result = math.copysign(x, y)

print("Flawed ternary result on -0.0:", flawed_result)  # 5.0 (Incorrect!)
print("math.copysign result on -0.0: ", correct_result) # -5.0 (Correct!)
```

### 4. Special Values: Infinities and NaNs
Observe how `math.copysign()` applies signs across infinities and NaNs.


```python
import math

print("copysign(inf, -1): ", math.copysign(float("inf"), -1))
print("copysign(5.0, -inf):", math.copysign(5.0, float("-inf")))
print("copysign(0.0, -1):  ", math.copysign(0.0, -1))
```

### 5. Hands-on Practice Exercise: Kinematic Physics Vector Engine
In physics simulations, scalar speeds (magnitudes) are computed from energy or acceleration, while directions are derived from displacement vectors.

Implement `calculate_directed_velocity(speed, target_x, current_x)` and `simulate_rebound(velocity, restitution)` with self-checking unit assertions.


```python
import math

def calculate_directed_velocity(speed, target_x, current_x):
    """Applies the direction of displacement (target - current) to a scalar speed."""
    if speed < 0:
        raise ValueError("Speed magnitude must be non-negative.")
    displacement = target_x - current_x
    return math.copysign(speed, displacement)

def simulate_rebound(velocity, restitution=0.8):
    """Reverses velocity direction after striking a barrier, applying restitution damping."""
    if not (0.0 <= restitution <= 1.0):
        raise ValueError("Restitution coefficient must be between 0.0 and 1.0.")
    rebounded_speed = abs(velocity) * restitution
    return math.copysign(rebounded_speed, -velocity)

# Unit Assertions
assert calculate_directed_velocity(25.0, 100.0, 40.0) == 25.0
assert calculate_directed_velocity(25.0, 10.0, 40.0) == -25.0

assert simulate_rebound(50.0, 0.8) == -40.0
assert simulate_rebound(-40.0, 0.5) == 20.0
assert simulate_rebound(0.0, 0.8) == -0.0

print("Directed velocity towards x=10m:", calculate_directed_velocity(15.0, 10.0, 50.0))
print("Rebound from +30 m/s barrier:   ", simulate_rebound(30.0, 0.75))
print("All kinematic physics assertions passed successfully!")
```
