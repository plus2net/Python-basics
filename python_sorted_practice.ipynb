### 1. Basic Sorting Across Collections
Sort numbers, strings, and collections non-destructively.

```python
numbers = [12, 3, 8, 7, 14]
sorted_nums = sorted(numbers)
print("Sorted copy:  ", sorted_nums)
print("Original list:", numbers)

fruits = ["Orange", "Banana", "Apple", "Mango"]
print("Sorted fruits: ", sorted(fruits))
```

### 2. Descending Sort with reverse=True
Sort elements in descending order directly.

```python
desc_numbers = sorted(numbers, reverse=True)
print("Descending numbers:", desc_numbers)

desc_fruits = sorted(fruits, reverse=True)
print("Descending fruits: ", desc_fruits)
```

### 3. Custom Key Functions & Case-Insensitive Sort
Use `key=len` and `key=str.lower` to customize comparison logic.

```python
words = ["Three", "One", "Seven", "thirtyfive", "Two"]
print("Sorted by length:   ", sorted(words, key=len))

mixed_cases = ["banana", "Apple", "orange", "Cherry"]
print("Case-insensitive:   ", sorted(mixed_cases, key=str.lower))
```

### 4. sorted() vs. list.sort()
Compare non-destructive copying with in-place mutation.

```python
orig = [5, 2, 9, 1]
new_copy = sorted(orig)
print("sorted() result:", new_copy)
print("original intact:", orig)

ret = orig.sort()
print("sort() in-place:", orig)
print("sort() return:  ", ret)
```

### 5. Multi-Criteria Sorting on Complex Records
Sort dictionaries by multiple attributes (e.g. department ascending, salary descending).

```python
employees = [
    {"name": "Alice", "dept": "HR", "salary": 75000},
    {"name": "Bob", "dept": "IT", "salary": 92000},
    {"name": "Charlie", "dept": "IT", "salary": 85000},
    {"name": "Diana", "dept": "HR", "salary": 80000},
]

sorted_emps = sorted(employees, key=lambda e: (e["dept"], -e["salary"]))
for emp in sorted_emps:
    print(f"{emp['dept']} | {emp['salary']} | {emp['name']}")
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying sorted() behaviors.

```python
assert sorted([5, 1, 3]) == [1, 3, 5]
assert sorted([5, 1, 3], reverse=True) == [5, 3, 1]
assert sorted(["bb", "a", "ccc"], key=len) == ["a", "bb", "ccc"]
assert sorted([(1, "b"), (2, "a"), (1, "a")]) == [(1, "a"), (1, "b"), (2, "a")]

print("All sorted() unit assertions passed! 100% verified.")
```
