### 1. Sizing Standard Collections

```python
sample_text = "plus2net.com"
items = [10, 20, 30, 40]
profile = {"name": "Alex", "age": 25, "active": True}
tags = {"python", "coding", "tutorial"}

print("String length:    ", len(sample_text))
print("List length:      ", len(items))
print("Dictionary length:", len(profile))
print("Set length:       ", len(tags))
```

### 2. O(1) Sizing of Massive Ranges

```python
# Ranges compute length mathematically in O(1) time without allocating RAM
massive_range = range(0, 100_000_000_000, 25)
print("Massive range calculated length:", len(massive_range))
```

### 3. Custom Class Implementing __len__()

```python
class ShoppingCart:
    def __init__(self):
        self.items = []
    def add(self, item):
        self.items.append(item)
    def __len__(self):
        return len(self.items)

cart = ShoppingCart()
cart.add("Keyboard")
cart.add("Mouse")

print("Items in cart via len():", len(cart))
print("Cart has items (bool):  ", bool(cart))
```

### 4. Recursive Deep Item Counter

```python
def count_deep(data):
    if isinstance(data, (list, tuple, set)):
        return sum(count_deep(x) for x in data)
    if isinstance(data, dict):
        return sum(count_deep(v) for v in data.values())
    return 1

nested_structure = [1, [2, 3], [4, [5, 6, [7, 8]]]]
print("Shallow len():", len(nested_structure))
print("Deep count:   ", count_deep(nested_structure))
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Sequence lengths
assert len("python") == 6
assert len([1, 2, 3]) == 3
assert len({'a': 1, 'b': 2}) == 2

# Test 2: Custom __len__
class Box:
    def __len__(self): return 99
assert len(Box()) == 99

# Test 3: Truthiness fallback
class Sized:
    def __init__(self, n): self.n = n
    def __len__(self): return self.n
assert bool(Sized(10)) is True
assert bool(Sized(0)) is False

# Test 4: Range length calculation
assert len(range(10, 50, 5)) == 8

print("All len() unit assertions successfully verified! 100% passed.")
```
