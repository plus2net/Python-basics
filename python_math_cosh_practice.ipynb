### 1. Basic Calculations & Even Symmetry
`math.cosh(x)` computes the hyperbolic cosine of `x`. It has a global minimum of 1.0 at `x = 0.0` and satisfies even symmetry: `cosh(-x) == cosh(x)`.


```python
import math

print("cosh(0.0): ", math.cosh(0.0))   # 1.0
print("cosh(1.0): ", round(math.cosh(1.0), 6))   # 1.543081
print("cosh(-1.0):", round(math.cosh(-1.0), 6))  # 1.543081
print("cosh(2.0): ", round(math.cosh(2.0), 6))   # 3.762196
print("cosh(-2.0):", round(math.cosh(-2.0), 6))  # 3.762196
```

### 2. Exponential Equivalence: (e^x + e^-x) / 2
Verify that `math.cosh(x)` yields results identical to the exponential formulation using `math.exp()`.


```python
import math

def formula_cosh(x):
    return (math.exp(x) + math.exp(-x)) / 2.0

test_values = [0.0, 0.5, 1.5, 3.5, 10.0]
for val in test_values:
    b = math.cosh(val)
    f = formula_cosh(val)
    assert math.isclose(b, f)
    print(f"x = {val:4.1f} | math.cosh = {b:12.6f} | formula = {f:12.6f} | Match: True")
```

### 3. Fundamental Identity: cosh^2(x) - sinh^2(x) == 1
On the unit hyperbola, hyperbolic cosine and sine satisfy the fundamental identity `cosh^2 - sinh^2 == 1`.


```python
import math

test_points = [0.0, 0.5, 1.8, 5.0, -3.2]
for x in test_points:
    c = math.cosh(x)
    s = math.sinh(x)
    diff = c**2 - s**2
    assert math.isclose(diff, 1.0)
    print(f"x = {x:4.1f} | cosh^2 - sinh^2 = {diff:.12f} | Valid: True")
```

### 4. Overflow Handling with Large Inputs
Because `cosh(x)` grows exponentially, values exceeding `|x| ~ 710.475` raise `OverflowError: math range error`.


```python
import math

print("cosh(100):", math.cosh(100))
print("cosh(710):", math.cosh(710))

try:
    math.cosh(711)
except OverflowError as err:
    print("Caught expected OverflowError on cosh(711):", err)
```

### 5. Hands-on Practice Exercise: Power Line Catenary Sag Calculator
High-voltage cables suspended between two utility towers form a catenary curve: `y(x) = a * cosh(x / a)`.
The mid-span sag for span length `L` is: `Sag = a * (cosh(L / (2 * a)) - 1)`.

Implement `calculate_catenary_sag(span_meters, catenary_param_a)` and verify with self-checking assertions.


```python
import math

def calculate_catenary_sag(span_meters, catenary_param_a):
    """
    Calculates cable sag at mid-span:
    Sag = a * (cosh(span / (2 * a)) - 1)
    """
    if span_meters < 0 or catenary_param_a <= 0:
        raise ValueError("Span must be non-negative and parameter 'a' must be positive.")
    
    half_span = span_meters / 2.0
    sag = catenary_param_a * (math.cosh(half_span / catenary_param_a) - 1.0)
    return round(sag, 4)

# Unit Assertions
assert calculate_catenary_sag(0.0, 500.0) == 0.0
assert calculate_catenary_sag(200.0, 1000.0) == 5.0042
assert calculate_catenary_sag(400.0, 800.0) == 25.1307

# Validation error check
try:
    calculate_catenary_sag(100.0, -50.0)
    assert False, "Should raise ValueError"
except ValueError:
    pass

print("Sag for 300m span (a = 1200m):", calculate_catenary_sag(300.0, 1200.0), "meters")
print("All catenary sag assertions passed successfully!")
```
