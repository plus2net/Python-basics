### 1. Basic Iterator Traversal and StopIteration
Advance an iterator position one element at a time.

```python
numbers = [18, 5, 15]
num_iter = iter(numbers)

print("1st:", next(num_iter))
print("2nd:", next(num_iter))
print("3rd:", next(num_iter))
```

### 2. Preventing Exceptions with Default Fallback Values
Use the `default` parameter to suppress StopIteration exceptions.

```python
colors = iter(['red', 'green'])
print(next(colors, 'empty'))
print(next(colors, 'empty'))
print(next(colors, 'empty'))  # Safely returns default
```

### 3. Finding the First Matching Item in a Collection
Use `next()` with a generator expression to find matching items efficiently.

```python
users = [
    {"id": 1, "name": "Alice", "role": "editor"},
    {"id": 2, "name": "Bob", "role": "admin"},
    {"id": 3, "name": "Charlie", "role": "admin"},
]

first_admin = next((u for u in users if u["role"] == "admin"), None)
print("Found admin:", first_admin["name"])
```

### 4. Skipping Header Rows in Data Streams
Consume and isolate header lines before streaming records.

```python
import io
stream = io.StringIO("id,name\n1,Alice\n2,Bob\n")
header = next(stream).strip()
print("Header line:", header)
print("First data: ", next(stream).strip())
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying next() mechanics.

```python
it = iter([1, 2])
assert next(it) == 1
assert next(it) == 2
assert next(it, "END") == "END"

vals = [10, 25, 30]
assert next(x for x in vals if x > 20) == 25
assert next((x for x in vals if x > 100), -1) == -1

print("All next() unit assertions passed! 100% verified.")
```
