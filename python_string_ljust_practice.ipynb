## 1. Basic Left Alignment & Overflow Mechanics

`str.ljust(width[, fillchar])` aligns text to the left and pads on the right.
If `width <= len(s)`, Python returns the original string unaltered without truncation.

```python
label = "CPU Load"
padded = label.ljust(15)

print("Original: ", repr(label))
print("Padded:   ", repr(padded))
print("Length:   ", len(padded))

assert len(padded) == 15
assert padded == "CPU Load       "

# Width smaller than length: never truncates
long_title = "DatabaseMigration"
assert long_title.ljust(8) == "DatabaseMigration"

print("\nAll basic left alignment assertions passed!")
```

## 2. Custom Fill Characters & Dot Leaders

Passing a custom character fills the remaining right margin.

```python
item = "Authentication Service"
leader = item.ljust(35, ".") + " [ONLINE]"

print("Leader line:", leader)
assert leader == "Authentication Service............. [ONLINE]"

# Single character constraint validation
try:
    item.ljust(35, "..")
    assert False, "Expected TypeError not raised"
except TypeError as e:
    assert "must be exactly one character" in str(e)

print("Fill character assertions passed!")
```

## 3. Directional Comparison: `ljust()` vs. `rjust()` vs. f-Strings

- `s.ljust(w, fill)` &harr; `f"{s:{fill}<{w}}"`
- `s.rjust(w, fill)` &harr; `f"{s:{fill}>{w}}"`

```python
text = "LOG"
w = 8
fill = "-"

left = text.ljust(w, fill)
right = text.rjust(w, fill)

print("Left: ", left)
print("Right:", right)
assert left == "LOG-----"
assert right == "-----LOG"

# Equivalence with f-strings
assert left == f"{text:{fill}<{w}}"
assert right == f"{text:{fill}>{w}}"

print("\nDirectional comparison assertions passed!")
```

## 4. Real-World Engineering Recipes

Explore three common software engineering patterns utilizing `str.ljust()`:
1. Columnar Inventory Table Generator
2. Fixed-Width Flat File Record Exporter
3. CLI Diagnostic Dot-Leader Formatter

```python
# Recipe 1: Columnar table
def format_table(rows: list[tuple[str, str, float]]) -> str:
    lines = ["Item Name".ljust(20) + "Category".ljust(15) + "Price".rjust(10), "-" * 45]
    for name, cat, price in rows:
        lines.append(name.ljust(20) + cat.ljust(15) + f"${price:,.2f}".rjust(10))
    return "\n".join(lines)

inv = [
    ("Keyboard", "Hardware", 129.99),
    ("Mouse Pad", "Accessories", 19.50)
]
tbl = format_table(inv)
print("Recipe 1 Table:\n" + tbl)
assert "Keyboard            Hardware" in tbl

# Recipe 2: Flat file record exporter
def flat_record(acc: str, name: str, code: str) -> str:
    return f"{acc.ljust(10)[:10]}{name.ljust(20)[:20]}{code.ljust(5)[:5]}"

rec = flat_record("ACC-101", "Alice Smith", "DEP")
print("Recipe 2 Flat Record:", repr(rec), "Length:", len(rec))
assert len(rec) == 35

# Recipe 3: CLI status dot leader
def cli_status(label: str, status: str, width: int = 35) -> str:
    prefix = (label + " ").ljust(width - len(status) - 1, ".")
    return f"{prefix} {status}"

stat = cli_status("Database Migration", "[OK]", 30)
print("Recipe 3 Status:", stat)
assert stat.endswith(" [OK]")
assert len(stat) == 30

print("\nAll engineering recipes passed!")
```

## 5. Production Implementation & Self-Checking Unit Assertions

This production-grade `ColumnFormatter` class encapsulates safe column padding with truncation, dot-leader generation, and automated unit assertions.

```python
class ColumnFormatter:
    @staticmethod
    def format_column(val: str, col_width: int, max_len: int = 0) -> str:
        """Left-aligns text with optional safety truncation."""
        s = str(val)
        if max_len > 0 and len(s) > max_len:
            s = s[:max_len - 3] + "..."
        return s.ljust(col_width)

    @staticmethod
    def format_dot_leader(title: str, status: str, total_width: int = 40) -> str:
        """Creates dot leaders between label and status."""
        prefix = f"{title} "
        leader = prefix.ljust(total_width - len(status) - 1, ".")
        return f"{leader} {status}"


# ==========================================
# UNIT ASSERTION SUITE
# ==========================================
formatter = ColumnFormatter()

# 1. Basic space padding
assert "Python".ljust(10) == "Python    "
assert "Python".ljust(6) == "Python"
assert "Python".ljust(3) == "Python"

# 2. Custom fillchar
assert "Item".ljust(8, "-") == "Item----"
assert "Data".ljust(10, ".") == "Data......"

# 3. Column formatter with safety truncation
assert formatter.format_column("Short", 10) == "Short     "
assert formatter.format_column("VeryLongEnterpriseName", 12, max_len=12) == "VeryLongE..."

# 4. Dot leader formatting
dl = formatter.format_dot_leader("Network Check", "[OK]", 30)
assert dl.startswith("Network Check ")
assert dl.endswith(" [OK]")
assert len(dl) == 30

print("All str.ljust() production unit assertions passed successfully!")
```
