### 1. Basic Bytearray Allocations
Construct mutable buffers from sizes, strings, and lists of integers.

```python
print("bytearray(4):           ", bytearray(4))
print("bytearray('hi', 'utf8'):", bytearray('hi', 'utf-8'))
print("bytearray([65, 66, 67]):", bytearray([65, 66, 67]))
```

### 2. In-Place Mutation: Indexing & Slicing
Modify bytes in-place without memory reallocations.

```python
ba = bytearray(b"Hello!")
ba[0] = ord('J')
print("After ba[0]:", ba)
assert ba == bytearray(b"Jello!")
```

### 3. Hexadecimal Conversions: fromhex() & .hex()
Parse hex strings into mutable byte buffers and export them.

```python
ba_hex = bytearray.fromhex('F1 F1')
print("fromhex:   ", ba_hex)
ba_hex[0] = 0xAA
print("Export hex:", ba_hex.hex())
assert ba_hex.hex() == "aaf1"
```

### 4. Production Recipe: XOR Buffer Masking
Mask and unmask a binary packet buffer in-place.

```python
data = bytearray(b"TopSecretPacket")
key = 0x42
# In-place masking
for i in range(len(data)):
    data[i] ^= key
print("Masked:   ", data)
# In-place unmasking
for i in range(len(data)):
    data[i] ^= key
print("Unmasked: ", data)
assert data == bytearray(b"TopSecretPacket")
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying bytearray() mechanics.

```python
assert bytearray() == bytearray(b"")
assert bytearray(2) == bytearray(b"\x00\x00")
assert bytearray([65, 66]) == bytearray(b"AB")
assert bytearray.fromhex("ff00") == bytearray(b"\xff\x00")

print("All bytearray() unit assertions passed! 100% verified.")
```
