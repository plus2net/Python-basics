### 1. Basic str.isupper() Validation
`str.isupper()` returns `True` if all cased characters are uppercase and length of cased characters > 0.

```python
print("'WELCOME TO PYTHON'.isupper(): ", 'WELCOME TO PYTHON'.isupper())
print("'PYTHON 3.14'.isupper():       ", 'PYTHON 3.14'.isupper())
print("'Welcome to Python'.isupper(): ", 'Welcome to Python'.isupper())
print("'12345'.isupper():             ", '12345'.isupper())
print("''.isupper():                  ", ''.isupper())

assert 'WELCOME TO PYTHON'.isupper() is True
assert 'PYTHON 3.14'.isupper() is True
assert 'Welcome to Python'.isupper() is False
assert '12345'.isupper() is False
assert ''.isupper() is False
```

### 2. Unicode and International Uppercase Scripts
`str.isupper()` checks Unicode letter categories across Latin, Greek, and Cyrillic.

```python
intl = ['CAFÉ', 'MÜNCHEN', 'МОСКВА', 'ΑΘΗΝΑ']
for s in intl:
    print(f"{s:<10}: {s.isupper()}")
    assert s.isupper() is True

# Uncased scripts (Chinese) lack case
assert '中文'.isupper() is False
assert 'PYTHON 中文'.isupper() is True
```

### 3. Password Policy: isupper() vs. any(c.isupper())
Demonstrate why `password.isupper()` is incorrect for checking required uppercase characters.

```python
pwd = "SecretPass123!"
all_upper = pwd.isupper()
has_upper = any(c.isupper() for c in pwd)

print(f"pwd.isupper():               {all_upper} (requires all letters to be upper)")
print(f"any(c.isupper() for c in pwd): {has_upper} (satisfies password policy)")

assert all_upper is False
assert has_upper is True
```

### 4. Financial Ticker Validation
Verify stock market symbols (e.g. 'AAPL', 'NVDA', 'BRK.A').

```python
def is_valid_ticker(symbol):
    s = symbol.strip()
    return 1 <= len(s) <= 5 and s.isupper() and all(c.isalnum() or c == '.' for c in s)

print("AAPL valid: ", is_valid_ticker('AAPL'))
print("BRK.B valid:", is_valid_ticker('BRK.B'))
print("aapl valid: ", is_valid_ticker('aapl'))

assert is_valid_ticker('AAPL') is True
assert is_valid_ticker('BRK.B') is True
assert is_valid_ticker('aapl') is False
```

### 5. Practice Exercise: Security & Ticker Auditor Suite
Run the complete auditing class with self-checking assertions.

```python
class SecurityAndTickerAuditor:
    @staticmethod
    def is_valid_ticker(symbol):
        if not isinstance(symbol, str):
            return False
        s = symbol.strip()
        if not (1 <= len(s) <= 5):
            return False
        return s.isupper() and all(c.isalnum() or c == '.' for c in s)

    @staticmethod
    def audit_password_casing(password):
        if not isinstance(password, str):
            return False, "Invalid type."
        has_upper = any(c.isupper() for c in password)
        has_lower = any(c.islower() for c in password)
        if not has_upper:
            return False, "Password missing uppercase character."
        if not has_lower:
            return False, "Password missing lowercase character."
        return True, "Valid casing."

# Unit Assertions
auditor = SecurityAndTickerAuditor()
assert auditor.is_valid_ticker("MSFT") is True
assert auditor.is_valid_ticker("BRK.A") is True
assert auditor.is_valid_ticker("msft") is False
assert auditor.is_valid_ticker("TOOLONG") is False
assert auditor.is_valid_ticker("123") is False

assert auditor.audit_password_casing("Alpha123!") == (True, "Valid casing.")
assert auditor.audit_password_casing("alllower123!") == (False, "Password missing uppercase character.")
assert auditor.audit_password_casing("ALLUPPER123!") == (False, "Password missing lowercase character.")

print("All str.isupper() assertions passed successfully!")
```
