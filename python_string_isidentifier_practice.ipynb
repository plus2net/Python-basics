### 1. Basic str.isidentifier() Syntax Checking
`str.isidentifier()` returns `True` if a string follows valid Python identifier grammar.

```python
print("'user_name'.isidentifier():    ", 'user_name'.isidentifier())
print("'_private'.isidentifier():      ", '_private'.isidentifier())
print("'__init__'.isidentifier():      ", '__init__'.isidentifier())
print("'item2'.isidentifier():         ", 'item2'.isidentifier())
print("'2item'.isidentifier():         ", '2item'.isidentifier())
print("'user-name'.isidentifier():     ", 'user-name'.isidentifier())
print("''.isidentifier():              ", ''.isidentifier())

assert 'user_name'.isidentifier() is True
assert '_private'.isidentifier() is True
assert '2item'.isidentifier() is False
assert 'user-name'.isidentifier() is False
assert ''.isidentifier() is False
```

### 2. The Keyword Trap & The keyword Module
Keywords return `True` under `isidentifier()`, but cannot be bound as variable names.

```python
import keyword

keywords = ['class', 'def', 'for', 'return', 'import']
for kw in keywords:
    print(f"{kw:<8} | isidentifier: {str(kw.isidentifier()):<5} | iskeyword: {str(keyword.iskeyword(kw)):<5}")
    assert kw.isidentifier() is True
    assert keyword.iskeyword(kw) is True

def is_valid_variable(name):
    return name.isidentifier() and not keyword.iskeyword(name)

assert is_valid_variable('class') is False
assert is_valid_variable('valid_var') is True
```

### 3. PEP 3131 Unicode Non-ASCII Identifiers
Python 3 natively supports international character sets in identifiers.

```python
unicode_vars = ['θ', 'α_beta', 'скорость', '数据_1', 'résumé']
for uv in unicode_vars:
    print(f"{uv:<12}: {uv.isidentifier()}")
    assert uv.isidentifier() is True

# Emojis and symbols are rejected
assert '🚀'.isidentifier() is False
```

### 4. Dynamic Attribute Validation (setattr)
Validate incoming dictionary keys before binding them as object attributes.

```python
import keyword

class DynamicConfig:
    def set_attr(self, key, val):
        if not key.isidentifier() or keyword.iskeyword(key):
            raise ValueError(f"Invalid attribute name: {key}")
        setattr(self, key, val)

cfg = DynamicConfig()
cfg.set_attr('timeout', 30)
print("Timeout:", cfg.timeout)
assert cfg.timeout == 30

try:
    cfg.set_attr('123_invalid', 'fail')
except ValueError as e:
    print("Caught invalid attr:", e)
```

### 5. Practice Exercise: Schema Field Sanitizer & Validator
Run the complete validator testing keyword avoidance and column sanitization.

```python
import re
import keyword

class IdentifierManager:
    @staticmethod
    def is_valid_variable(name):
        if not isinstance(name, str):
            return False
        return name.isidentifier() and not keyword.iskeyword(name)

    @staticmethod
    def sanitize(raw_name):
        if not isinstance(raw_name, str):
            return "_var"
        clean = re.sub(r'[^a-zA-Z0-9_]+', '_', raw_name).strip('_')
        if not clean:
            clean = "_var"
        elif clean[0].isdigit():
            clean = f"_{clean}"
        if keyword.iskeyword(clean):
            clean = f"{clean}_"
        return clean

# Unit Assertions
mgr = IdentifierManager()
assert mgr.is_valid_variable("user_id") is True
assert mgr.is_valid_variable("_private") is True
assert mgr.is_valid_variable("class") is False
assert mgr.is_valid_variable("2nd_place") is False
assert mgr.is_valid_variable("user-name") is False

assert mgr.sanitize("user-name") == "user_name"
assert mgr.sanitize("100_percent") == "_100_percent"
assert mgr.sanitize("def") == "def_"
assert mgr.sanitize("Order Amount ($)") == "Order_Amount"
assert mgr.is_valid_variable(mgr.sanitize("100_percent")) is True

print("All identifier management assertions passed successfully!")
```
