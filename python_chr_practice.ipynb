### 1. Basic ASCII & Unicode Conversions

```python
# Basic ASCII characters
print("Code 65: ", chr(65))
print("Code 97: ", chr(97))
print("Code 48: ", chr(48))

# Extended Unicode symbols
print("Euro currency: ", chr(8364))
print("Rocket emoji:  ", chr(128640))
```

### 2. Bidirectional Symmetry: chr() vs. ord()

```python
characters = ['A', 'z', 'ñ', '€', '🐍', 'Ω']

for ch in characters:
    code = ord(ch)
    restored = chr(code)
    print(f"Char: {ch} | Code: {code:<7} | Restored: {restored} | Matches: {ch == restored}")
```

### 3. Generating Character Sequences

```python
# Generate alphabet ranges using list comprehensions
upper_alpha = [chr(i) for i in range(ord('A'), ord('Z') + 1)]
lower_alpha = [chr(i) for i in range(ord('a'), ord('z') + 1)]

print("Uppercase alphabet:", "".join(upper_alpha))
print("Lowercase alphabet:", "".join(lower_alpha))
```

### 4. Production Recipe: Reversible Caesar Cipher

```python
def caesar_encrypt(text: str, shift: int = 3) -> str:
    result = []
    for char in text:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            result.append(chr((ord(char) - base + shift) % 26 + base))
        else:
            result.append(char)
    return "".join(result)

msg = "Python plus2net 2026"
encrypted = caesar_encrypt(msg, 4)
decrypted = caesar_encrypt(encrypted, -4)

print("Original: ", msg)
print("Encrypted:", encrypted)
print("Decrypted:", decrypted)
```
