## 1. Basic Titlecasing & Lowercase Normalization

`str.title()` capitalizes the first character of each word and converts all subsequent characters to lowercase.

```python
headline = "PYTHON PROGRAMMING AND DATA STRUCTURES"
mixed = "wELcOmE tO pLuS2nEt"

print("From All-Caps:   ", headline.title())
print("From Mixed-Case: ", mixed.title())

assert headline.title() == "Python Programming And Data Structures"
assert mixed.title() == "Welcome To Plus2Net"

print("\nAll basic titlecasing assertions passed!")
```

## 2. The Apostrophe & Digits Word-Boundary Traps

In Python, *any non-letter character* is considered a word boundary.
Apostrophes and digits cause unexpected capitalizations: `"they're" -> "They'Re"`, `"3rd" -> "3Rd"`.

```python
# Reproducing the apostrophe and digit traps
assert "they're".title() == "They'Re"
assert "don't".title() == "Don'T"
assert "john's".title() == "John'S"
assert "3rd edition".title() == "3Rd Edition"
assert "plus2net".title() == "Plus2Net"

print("Word boundary trap reproductions verified!")
```

## 3. The Clean Alternative: `string.capwords()`

`string.capwords()` splits exclusively on whitespace, avoiding apostrophe word-splits.

```python
import string

phrase = "they're reading 3rd edition books"
cw = string.capwords(phrase)
print("string.capwords():", cw)

assert cw == "They're Reading 3rd Edition Books"

print("string.capwords() assertions passed!")
```

## 4. Real-World Engineering Recipes

Explore three common software engineering patterns utilizing title casing:
1. Contraction-Safe Natural Headline Capitalizer (Regex)
2. Editorial Headline Formatter (Preserving Minor Words)
3. User Profile Name Normalization

```python
import re

# Recipe 1: Regex contraction-safe titlecase
def safe_title(text: str) -> str:
    return re.sub(r"(?<![a-zA-Z'])\b[a-z]", lambda m: m.group(0).upper(), text.lower())

r1 = safe_title("don't miss the 3rd annual O'connor lecture!")
print("Recipe 1 Regex Title:", r1)
assert r1 == "Don't Miss The 3rd Annual O'connor Lecture!"

# Recipe 2: Editorial headline (APA/Chicago style)
def editorial_title(title: str) -> str:
    minor = {"a", "an", "the", "and", "but", "or", "in", "on", "at", "to", "for", "of"}
    words = title.strip().split()
    if not words:
        return ""
    res = []
    for i, w in enumerate(words):
        lw = w.lower()
        if i == 0 or i == len(words) - 1 or lw not in minor:
            res.append(lw.capitalize())
        else:
            res.append(lw)
    return " ".join(res)

r2 = editorial_title("the art of clean code and the joy of testing")
print("Recipe 2 Editorial:", r2)
assert r2 == "The Art of Clean Code and the Joy of Testing"

# Recipe 3: User name normalizer (hyphens supported)
def clean_name(raw: str) -> str:
    parts = []
    for p in raw.strip().split():
        parts.append("-".join(sub.capitalize() for sub in p.split("-")))
    return " ".join(parts)

assert clean_name("mary-jane watson") == "Mary-Jane Watson"
assert clean_name("ALICE SMITH") == "Alice Smith"

print("\nAll engineering recipes passed!")
```

## 5. Production Implementation & Self-Checking Unit Assertions

This production-grade `TitleFormatter` class encapsulates standard title casing, whitespace capwords, and editorial headline styling.

```python
import string

class TitleFormatter:
    @staticmethod
    def standard_title(text: str) -> str:
        """Standard CPython str.title() wrapper."""
        return text.title()

    @staticmethod
    def whitespace_capwords(text: str) -> str:
        """Uses string.capwords to avoid apostrophe capitalization."""
        return string.capwords(text)

    @staticmethod
    def editorial_headline(title: str) -> str:
        """Applies editorial style keeping minor prepositions in lowercase."""
        minor = {"a", "an", "the", "and", "but", "or", "in", "on", "at", "to", "for", "of"}
        words = title.strip().split()
        if not words:
            return ""
        res = []
        for i, w in enumerate(words):
            lw = w.lower()
            if i == 0 or i == len(words) - 1 or lw not in minor:
                res.append(lw.capitalize())
            else:
                res.append(lw)
        return " ".join(res)


# ==========================================
# UNIT ASSERTION SUITE
# ==========================================
formatter = TitleFormatter()

# 1. Standard titlecasing
assert formatter.standard_title("hello world") == "Hello World"
assert formatter.standard_title("PYTHON PROGRAMMING") == "Python Programming"

# 2. Reproduction of the apostrophe trap
assert formatter.standard_title("they're") == "They'Re"
assert formatter.standard_title("3rd place") == "3Rd Place"

# 3. Safe alternative with string.capwords()
assert formatter.whitespace_capwords("they're here") == "They're Here"
assert formatter.whitespace_capwords("3rd place") == "3rd Place"

# 4. Editorial headline styling
assert formatter.editorial_headline("the guide to python") == "The Guide to Python"
assert formatter.editorial_headline("a tale of two cities") == "A Tale of Two Cities"
assert formatter.editorial_headline("") == ""

print("All str.title() production unit assertions passed successfully!")
```
