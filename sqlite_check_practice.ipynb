## 1. Setup Database & Environment

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_conn.execute("DROP TABLE IF EXISTS students")
print("Database initialized and clean!")
my_conn.close()
```

## 2. Creating Table with Basic Column-Level CHECK Constraint

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_conn.execute("DROP TABLE IF EXISTS students")

create_query = '''CREATE TABLE students (
student_id INTEGER PRIMARY KEY,
name TEXT NOT NULL,
age INTEGER CHECK(age >= 5),
email TEXT UNIQUE,
class_id INTEGER)'''

my_conn.execute(create_query)

# Test failing CHECK constraint (age < 5)
insert_invalid = '''INSERT INTO students (student_id, name, age, email, class_id)
VALUES (1, 'abc', 4, 'abc1@example.com', 9)'''
try:
    my_conn.execute(insert_invalid)
except sqlite3.Error as my_error:
    print("error:", my_error)

# Test valid insert (age >= 5)
insert_valid = '''INSERT INTO students (student_id, name, age, email, class_id)
VALUES (1, 'abc', 8, 'abc1@example.com', 9)'''
r_set = my_conn.execute(insert_valid)
my_conn.commit()
print("Valid record added successfully, rows affected:", r_set.rowcount)
my_conn.close()
```

## 3. Restricting Values with Range (`age BETWEEN / AND`) & List Options (`class_id IN (...)`)

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_conn.execute("DROP TABLE IF EXISTS students")

create_query = '''CREATE TABLE students (
student_id INTEGER PRIMARY KEY,
name TEXT NOT NULL,
age INTEGER,
email TEXT UNIQUE,
class_id INTEGER,
CONSTRAINT my_check1 CHECK(age >= 5 AND age <= 20),
CONSTRAINT my_check2 CHECK(class_id IN (1, 2, 3)))'''

my_conn.execute(create_query)

# Test invalid class_id option
insert_invalid = '''INSERT INTO students (student_id, name, age, email, class_id)
VALUES (7, 'abc', 8, 'abc7@example.com', 4)'''
try:
    my_conn.execute(insert_invalid)
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 4. Enforcing Strict Data Types using `typeof()`

```python
import sqlite3

my_conn = sqlite3.connect('my_db.db')
my_conn.execute("DROP TABLE IF EXISTS students")

create_query = '''CREATE TABLE students (
student_id INTEGER PRIMARY KEY,
name TEXT NOT NULL,
age INTEGER,
email TEXT UNIQUE,
class_id INTEGER CHECK(typeof(class_id) = 'integer'))'''

my_conn.execute(create_query)

# Attempt to insert a string into class_id
insert_string = '''INSERT INTO students (student_id, name, age, email, class_id)
VALUES (7, 'abc', 8, 'abc7@example.com', 'a')'''
try:
    my_conn.execute(insert_string)
except sqlite3.Error as my_error:
    print("error:", my_error)
finally:
    my_conn.close()
```

## 5. Modern SQLAlchemy 2.0+ Context Manager with Table Constraints

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

my_conn = create_engine("sqlite:///my_db.db")

try:
    with my_conn.begin() as connection:
        connection.execute(text("DROP TABLE IF EXISTS students"))
        create_query = text('''CREATE TABLE students (
            student_id INTEGER PRIMARY KEY,
            name TEXT NOT NULL,
            age INTEGER CHECK(age >= 5 AND age <= 20),
            email TEXT UNIQUE,
            class_id INTEGER CHECK(typeof(class_id) = 'integer') CHECK(class_id IN (1, 2, 3))
        )''')
        connection.execute(create_query)
        print("Table created via SQLAlchemy 2.0+ successfully!")
except SQLAlchemyError as e:
    print("SQLAlchemy error:", e)
```
