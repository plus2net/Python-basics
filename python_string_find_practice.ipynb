### 1. Basic str.find() Usage
`str.find()` returns the lowest index if found, or `-1` if not found.

```python
phrase = "Welcome to plus2net.com"
print("Index of 'Welcome':", phrase.find("Welcome"))
print("Index of 'plus':   ", phrase.find("plus"))
print("Index of 'co':     ", phrase.find("co"))
print("Index of 'Python': ", phrase.find("Python"))

assert phrase.find("Welcome") == 0
assert phrase.find("plus") == 11
assert phrase.find("Python") == -1
```

### 2. Searching with start and end Offsets
Restrict search range without creating new slice objects.

```python
text = "apple pie, apple cider, apple crisp"
print("First occurrence:      ", text.find("apple"))
print("Second occurrence (start=5):", text.find("apple", 5))
print("Between [5:15]:        ", text.find("apple", 5, 15))

assert text.find("apple") == 0
assert text.find("apple", 5) == 11
assert text.find("apple", 5, 15) == -1
```

### 3. Case-Insensitive Search with casefold()
Pair `find()` with `casefold()` for international case-insensitive search.

```python
haystack = "Learn Python at Plus2Net"
pos_lower = haystack.casefold().find("python".casefold())
pos_upper = haystack.casefold().find("PYTHON".casefold())

print("Offset for 'python':", pos_lower)
print("Offset for 'PYTHON':", pos_upper)

assert pos_lower == 6
assert pos_upper == 6
```

### 4. Finding All Occurrences
Advance search offset in a loop to find all indices.

```python
def find_all(s, sub):
    indices = []
    start = 0
    while True:
        pos = s.find(sub, start)
        if pos == -1:
            break
        indices.append(pos)
        start = pos + len(sub)
    return indices

seq = "abc-abc-abc-abc"
matches = find_all(seq, "abc")
print("All 'abc' matches:", matches)
assert matches == [0, 4, 8, 12]
```

### 5. Practice Exercise: Delimited Tag Extractor Suite
Run the complete search engine with self-checking assertions.

```python
class StringSearchEngine:
    @staticmethod
    def find_all(text, sub):
        if not sub:
            return []
        indices = []
        start = 0
        while True:
            pos = text.find(sub, start)
            if pos == -1:
                break
            indices.append(pos)
            start = pos + len(sub)
        return indices

    @staticmethod
    def extract_between(text, start_tag, end_tag):
        p1 = text.find(start_tag)
        if p1 == -1:
            return None
        content_start = p1 + len(start_tag)
        p2 = text.find(end_tag, content_start)
        if p2 == -1:
            return None
        return text[content_start:p2]

# Unit Assertions
engine = StringSearchEngine()
sample = "<title>Python str.find() Guide</title>"
assert engine.extract_between(sample, "<title>", "</title>") == "Python str.find() Guide"
assert engine.extract_between(sample, "<body>", "</body>") is None

haystack = "The quick brown fox jumps over the lazy dog"
assert haystack.find("quick") == 4
assert haystack.find("elephant") == -1

print("Extracted title:", engine.extract_between(sample, "<title>", "</title>"))
print("All str.find() practice assertions passed successfully!")
```
