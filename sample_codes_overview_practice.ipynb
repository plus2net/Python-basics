## Example 4: Convert decimal to binary

Read the code, predict the output and run this cell.

```python
n = 156
print(bin(n)[2:])

def binary_digits(n):
    if n == 0:
        return "0"
    digits = ""
    while n > 0:
        digits = str(n % 2) + digits
        n //= 2
    return digits
print(binary_digits(156))
```

## Example 8: Vowel or consonant checker

Read the code, predict the output and run this cell.

```python
letter = "E".lower()
if len(letter) != 1 or not letter.isalpha():
    print("Enter one letter")
elif letter in "aeiou":
    print("Vowel")
else:
    print("Consonant")
```
