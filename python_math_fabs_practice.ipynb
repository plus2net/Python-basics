# Python math.fabs(): Float Absolute Value Function

### 1. Basic Float Absolute Values with math.fabs()
`math.fabs(x)` always returns the magnitude of `x` as a 64-bit float.


```python
import math

print("fabs(-4):   ", math.fabs(-4))       # 4.0 (float)
print("fabs(-4.6): ", math.fabs(-4.6))     # 4.6
print("fabs(4):    ", math.fabs(4))        # 4.0
print("Type:       ", type(math.fabs(-4))) # <class 'float'>
```

### 2. Deep Comparison: math.fabs() vs. Built-in abs()
`abs()` preserves integer types and returns complex modulus, whereas `math.fabs()` always returns `float` and rejects complex numbers.


```python
import math

print("abs(-10):      ", abs(-10), type(abs(-10)))          # 10 (int)
print("math.fabs(-10):", math.fabs(-10), type(math.fabs(-10))) # 10.0 (float)

c_num = 3 + 4j
print("abs(3+4j):     ", abs(c_num))  # 5.0
try:
    math.fabs(c_num)
except TypeError as err:
    print("math.fabs(3+4j):", err)    # TypeError
```

### 3. Signed Zero & Special Values
`math.fabs()` clears the sign bit of negative zero (`-0.0`), converting it into positive zero (`+0.0`).


```python
import math

neg_zero = -0.0
abs_zero = math.fabs(neg_zero)
print("fabs(-0.0):             ", abs_zero)
print("Sign is positive?       ", math.copysign(1.0, abs_zero) > 0)
print("fabs(-inf):             ", math.fabs(float("-inf")))
print("fabs(nan):              ", math.fabs(float("nan")))
```

### 4. Algorithmic Applications: Tolerance & Manhattan Distance
Check numerical solver convergence and calculate L1 taxi-cab grid distances.


```python
import math

def has_converged(curr, prev, tol=1e-6):
    return math.fabs(curr - prev) < tol

def manhattan_distance(p1, p2):
    return math.fabs(p1[0] - p2[0]) + math.fabs(p1[1] - p2[1])

print("Converged? (1.0000002 vs 1.0):", has_converged(1.0000002, 1.0))
print("Manhattan dist between (2,3) & (7,9):", manhattan_distance((2, 3), (7, 9)))
```

### 5. Hands-on Practice Exercise: Mean Absolute Percentage Error (MAPE)
Calculate forecasting accuracy MAPE: `(1/n) * sum(|Actual - Forecast| / |Actual|) * 100%`.
Implement `calculate_mape(actuals, forecasts)` and verify with self-checking assertions.


```python
import math

def calculate_mape(actuals, forecasts):
    """Computes Mean Absolute Percentage Error (MAPE) as a percentage."""
    if len(actuals) != len(forecasts):
        raise ValueError("Actuals and forecasts lists must have identical lengths.")
    if not actuals:
        return 0.0
        
    percentage_errors = []
    for act, fcst in zip(actuals, forecasts):
        if act == 0:
            raise ZeroDivisionError("Actual value cannot be zero in MAPE calculations.")
        abs_error = math.fabs(act - fcst)
        pct_err = abs_error / math.fabs(act)
        percentage_errors.append(pct_err)
        
    mean_err = math.fsum(percentage_errors) / len(percentage_errors)
    return round(mean_err * 100.0, 2)

# Unit Assertions
assert calculate_mape([100, 200, 300], [100, 200, 300]) == 0.0
assert calculate_mape([100.0], [110.0]) == 10.0
assert calculate_mape([100.0], [90.0]) == 10.0

actual_sales = [120.0, 150.0, 210.0, 95.0]
model_preds = [126.0, 141.0, 220.5, 104.5]
mape_score = calculate_mape(actual_sales, model_preds)
assert mape_score == 6.5

print(f"Model Forecast MAPE: {mape_score}%")
print("All MAPE evaluation assertions passed successfully!")
```
