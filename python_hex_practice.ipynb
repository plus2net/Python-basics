# Python hex(): Base-16 Integer Conversion Practice

### 1. Basic Hexadecimal Conversions
Convert positive and negative integers to base-16 strings.

```python
my_num = 27
print("hex(27): ", hex(my_num))
print("hex(-27):", hex(-my_num))
print("hex(0):  ", hex(0))
```

### 2. Binary and Octal Conversions
Pass binary `0b` and octal `0o` literals directly to `hex()`.

```python
bin_val = 0b11011  # 27
oct_val = 0o33     # 27

print("Binary to Hex:", hex(bin_val))
print("Octal to Hex: ", hex(oct_val))
```

### 3. Formatting Without Prefix & Case Control
Use format specifiers `'x'` (lowercase) and `'X'` (uppercase).

```python
val = 255
print("format lowercase:", format(val, 'x'))
print("format uppercase:", format(val, 'X'))
print("f-string padded: ", f"{val:04x}")
```

### 4. Floating-Point Numbers: float.hex()
Inspect exact IEEE 754 representations without decimal rounding.

```python
pi = 3.141592653589793
h = pi.hex()
print("pi.hex():", h)
print("Lossless restored:", float.fromhex(h))
assert float.fromhex(h) == pi
```

### 5. Production Recipe: Hex to RGB Color Converter
Parse 6-digit hex color strings into (R, G, B) integer tuples.

```python
def hex_to_rgb(hex_code):
    clean = hex_code.lstrip('#')
    return tuple(int(clean[i:i+2], 16) for i in (0, 2, 4))

print("#FF65AA to RGB:", hex_to_rgb("#FF65AA"))
assert hex_to_rgb("#FF65AA") == (255, 101, 170)
```

### 6. Self-Checking Verification Suite
Run unit assertions verifying hex() mechanics.

```python
assert hex(27) == "0x1b"
assert hex(-27) == "-0x1b"
assert hex(0) == "0x0"
assert int("0x1b", 16) == 27

class Pointer:
    def __index__(self):
        return 16
assert hex(Pointer()) == "0x10"

print("All hex() unit assertions passed! 100% verified.")
```
