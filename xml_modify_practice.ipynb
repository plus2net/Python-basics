# Python Modify XML: ElementTree Practice

**Tutorial & Full Documentation:** [https://www.plus2net.com/python/xml-modify.php](https://www.plus2net.com/python/xml-modify.php)

Run cells in order in Google Colab. This notebook demonstrates how to modify XML files programmatically using Python's `xml.etree.ElementTree` module (updating nodes, adding sub-elements, and removing nodes). Save a copy in Google Drive to keep your edits.

### 1. Setup Sample XML and Modify Nodes

```python
import xml.etree.ElementTree as ET

# Create sample XML file for demonstration
xml_data = """<students>
    <student>
        <id>1</id>
        <name>John Deo</name>
        <mark>75</mark>
    </student>
    <student>
        <id>2</id>
        <name>Max Ruin</name>
        <mark>85</mark>
    </student>
</students>"""

with open("students.xml", "w") as f:
    f.write(xml_data)

# Load XML file
tree = ET.parse("students.xml")
root = tree.getroot()

# Update text of a specific node
for student in root.findall("student"):
    sid = student.find("id").text
    if sid == "2":
        student.find("name").text = "Max Updated"

# Add a new child element
for student in root.findall("student"):
    grade = ET.SubElement(student, "grade")
    grade.text = "A"

# Remove a specific element
for student in root.findall("student"):
    mark = student.find("mark")
    if mark is not None:
        student.remove(mark)

# Save changes
tree.write("updated_sample.xml", encoding="utf-8", xml_declaration=True)
print("XML file modified and saved successfully!")
```
