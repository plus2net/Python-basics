### 1. Basic Type Verification
`isinstance(obj, classinfo)` returns `True` if the object belongs to the class or its ancestors.

```python
num = 42
text = "plus2net"
items = [1, 2, 3]

print("num is int:  ", isinstance(num, int))   # True
print("text is str:  ", isinstance(text, str))  # True
print("items is list:", isinstance(items, list)) # True
```

### 2. Checking Multiple Types with Tuples
Pass a tuple of candidate types for logical OR validation.

```python
val = 3.14
print("val is int or float:", isinstance(val, (int, float)))  # True
print("val is int or str:  ", isinstance(val, (int, str)))    # False
```

### 3. Inheritance & Polymorphism
A child instance is an instance of all its parent classes.

```python
class Animal:
    pass
class Dog(Animal):
    pass

d = Dog()
print("d is Dog:   ", isinstance(d, Dog))     # True
print("d is Animal:", isinstance(d, Animal))  # True
print("type(d) is Animal:", type(d) is Animal) # False
```

### 4. Abstract Base Classes
Verify interface compliance with collections.abc.

```python
import collections.abc

data = [10, 20, 30]
print("data is Sequence:", isinstance(data, collections.abc.Sequence))
print("data is Iterable:", isinstance(data, collections.abc.Iterable))
```

### 5. Production Recipe: Polymorphic Normalizer
Handle single items vs collections uniformly.

```python
def normalize_list(raw_val):
    if isinstance(raw_val, (list, tuple, set)):
        return list(raw_val)
    return [raw_val]

print(normalize_list("single"))
print(normalize_list(["a", "b"]))
```

Run unit assertions verifying isinstance() mechanics.
