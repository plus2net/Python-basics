### 1. Basic Octal Conversion
Convert integers into octal string literals prefixed with `0o`.

```python
my_num = 27
print("oct(27): ", oct(my_num))
print("oct(-27):", oct(-my_num))
print("oct(0):  ", oct(0))
```

### 2. Binary and Hexadecimal Inter-Base Translations
Pass binary `0b` and hex `0x` literals directly into `oct()`.

```python
bin_val = 0b11011  # 27
hex_val = 0x1b     # 27

print("Binary 0b11011 to octal: ", oct(bin_val))
print("Hex 0x1b to octal:        ", oct(hex_val))
```

### 3. Formatting Without Prefix
Use format specifier `'o'` to output octal digits without `'0o'`.

```python
val = 27
print("format(27, 'o'):", format(val, 'o'))
print("f'{27:o}':      ", f"{val:o}")
```

### 4. Production Recipe: Linux File Permission Decoder
Decode POSIX file permission octals into symbolic strings.

```python
def decode_permissions(octal_mode):
    perms = ["---", "--x", "-w-", "-wx", "r--", "r-x", "rw-", "rwx"]
    digits = [int(d) for d in f"{octal_mode:o}"[-3:]]
    return "".join(perms[d] for d in digits)

print("0o755 permission string:", decode_permissions(0o755))
print("0o644 permission string:", decode_permissions(0o644))
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying oct() mechanics.

```python
assert oct(27) == "0o33"
assert oct(-27) == "-0o33"
assert oct(0) == "0o0"
assert int("0o755", 8) == 493

class Mode:
    def __index__(self):
        return 64
assert oct(Mode()) == "0o100"

print("All oct() unit assertions passed! 100% verified.")
```
