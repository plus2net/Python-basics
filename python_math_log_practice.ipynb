# Python math.log(): Calculate Logarithms

**Tutorial & Full Documentation:** [https://www.plus2net.com/python/math-log.php](https://www.plus2net.com/python/math-log.php)

### 1. Natural Logarithms (Base e)
When called with a single argument, `math.log(x)` computes the natural logarithm $\ln(x)$.

```python
import math

print("ln(1): ", math.log(1))
print("ln(e): ", math.log(math.e))
print("ln(2): ", math.log(2))
print("ln(10):", math.log(10))
```

### 2. Custom Base Logarithms
Pass the desired base as the second argument: `math.log(x, base)`.

```python
import math

print("log2(8):   ", math.log(8, 2))
print("log10(100):", math.log(100, 10))
print("log5(625): ", math.log(625, 5))
```

### 3. Specialized Hardware Logarithms: log2, log10 & log1p
Specialized functions provide greater numerical accuracy and speed than general change-of-base calculations.

```python
import math

print("log2(1024):  ", math.log2(1024))
print("log10(10000):", math.log10(10000))

# log1p for values close to zero
tiny = 1e-15
print("log1p(1e-15):", math.log1p(tiny))
```

### 4. Handling Domain Errors (Non-Positive Numbers)
Logarithms are only defined for $x > 0$. Evaluating non-positive inputs raises `ValueError: math domain error`.

```python
import math

for test_val in [-5, 0]:
    try:
        math.log(test_val)
    except ValueError as err:
        print(f"Caught expected error for {test_val}: {err}")
```

### 5. Hands-on Practice Exercise: Investment Doubling Time Calculator
Run the self-checking unit assertions below to verify exact doubling times using the logarithmic formula.

```python
def calculate_doubling_time(annual_rate_percent):
    if annual_rate_percent <= 0:
        raise ValueError("Growth rate must be strictly positive.")
    r = annual_rate_percent / 100.0
    return math.log(2) / math.log(1 + r)

years_7 = calculate_doubling_time(7)
years_10 = calculate_doubling_time(10)
years_12 = calculate_doubling_time(12)

# Unit Assertions
assert round(years_7, 2) == 10.24
assert round(years_10, 2) == 7.27
assert round(years_12, 2) == 6.12

print("7%  Annual Doubling Time:", round(years_7, 2), "years")
print("10% Annual Doubling Time:", round(years_10, 2), "years")
print("All doubling time assertions passed successfully!")
```
