# Python `calendar.itermonthdates()` — Interactive Practice

**Main tutorial:** [Python calendar.itermonthdates() with examples](https://www.plus2net.com/python/calendar-itermonthdates.php)

Learn how Python builds a month calendar with complete seven-day weeks using `datetime.date` values. Follow each demo, change a value, and run its verification cell. Only Python's standard library is needed.

## 1. Print every date in July 2020

The iterator covers five complete Monday–Sunday weeks. You will see two June dates and two August dates besides July.

```python
import calendar
my_cal = calendar.Calendar()
for x in my_cal.itermonthdates(2020, 7):
    print(x)
```

```python
import calendar
from datetime import date
july = list(calendar.Calendar().itermonthdates(2020, 7))
assert len(july) == 35
assert july[0] == date(2020, 6, 29)
assert july[-1] == date(2020, 8, 2)
assert all(isinstance(day, date) for day in july)
print('Check passed: 35 real date objects, June 29 to August 2')
```

## 2. Start on Saturday

Weekday numbering is fixed at Monday = 0 through Sunday = 6. `firstweekday` changes the order, not the underlying date.

```python
import calendar
my_cal = calendar.Calendar(firstweekday=5)
for x in my_cal.itermonthdates(2020, 7):
    print(x)
```

```python
saturday_dates = list(calendar.Calendar(firstweekday=5).itermonthdates(2020, 7))
assert (saturday_dates[0].isoformat(), saturday_dates[-1].isoformat()) == ('2020-06-27', '2020-07-31')
assert len(saturday_dates) % 7 == 0
print('Check passed: Saturday-first boundaries')
```

## 3. Read date attributes

A yielded value is not text—it has `.year`, `.month` and `.day` properties.

```python
import calendar
my_cal = calendar.Calendar()
for x in my_cal.itermonthdates(2020, 7):
    print("Year:", x.year, ", Month:", x.month, ", Day :", x.day)
```

```python
first = next(calendar.Calendar().itermonthdates(2020, 7))
assert (first.year, first.month, first.day) == (2020, 6, 29)
assert first.weekday() == 0
print('Check passed: first date is Monday June 29')
```

## 4. Count days and display the boundaries

Convert an iterator to a list to use `len()`, indexing and repeated loops.

```python
import calendar
all_days = list(calendar.Calendar().itermonthdates(2020, 7))
print('Count:', len(all_days))
print('First:', all_days[0], '| Last:', all_days[-1])
```

```python
assert len(all_days) == 5 * 7
print('Check passed: exactly five complete weeks')
```

## 5. Filter out dates from the previous and following months

Keep the requested year as well as month; otherwise December/January boundaries may be mishandled.

```python
import calendar
month_only = [day for day in calendar.Calendar().itermonthdates(2020, 7)
              if (day.year, day.month) == (2020, 7)]
print('July dates:', len(month_only))
print('First:', month_only[0], '| Last:', month_only[-1])
```

```python
assert len(month_only) == 31
assert all((d.year, d.month) == (2020, 7) for d in month_only)
print('Check passed: all 31 dates belong to July')
```

## 6. Find dates from adjacent months

The padding dates make full calendar rows. You can label or dim them when displaying a grid.

```python
import calendar
adjacent = [d for d in calendar.Calendar().itermonthdates(2020, 7)
            if (d.year, d.month) != (2020, 7)]
print('Adjacent dates:', adjacent)
```

```python
assert [str(d) for d in adjacent] == ['2020-06-29', '2020-06-30', '2020-08-01', '2020-08-02']
print('Check passed: 4 adjacent-month dates')
```

## 7. Change the first weekday to Sunday

Use `calendar.SUNDAY` instead of remembering the number 6.

```python
import calendar
sun_cal = calendar.Calendar(firstweekday=calendar.SUNDAY)
sun_dates = list(sun_cal.itermonthdates(2020, 7))
print(sun_dates[0], 'to', sun_dates[-1])
```

```python
assert sun_cal.getfirstweekday() == 6
assert sun_dates[0].weekday() == 6
assert len(sun_dates) % 7 == 0
print('Check passed: weeks begin on Sunday')
```

## 8. Group into weeks

Split the complete sequence into seven-date rows or use `monthdatescalendar()` directly.

```python
import calendar
cal = calendar.Calendar()
dates = list(cal.itermonthdates(2020, 7))
weeks = [dates[i:i + 7] for i in range(0, len(dates), 7)]
for week in weeks:
    print(' '.join(f'{d.day:02d}' for d in week))
```

```python
assert len(weeks) == 5
assert all(len(week) == 7 for week in weeks)
assert weeks == cal.monthdatescalendar(2020, 7)
print('Check passed: five seven-day week lists')
```

## 9. Display weekday names

Date objects support `.weekday()`, `.strftime()` and `.isoformat()`.

```python
import calendar
week = list(calendar.Calendar().itermonthdates(2020, 7))[:7]
for d in week:
    print(d.isoformat(), d.strftime('%A'))
```

```python
assert week[0].strftime('%A') == 'Monday'
assert week[-1].strftime('%A') == 'Sunday'
print('Check passed: weekday labels correspond to dates')
```

## 10. Build a Monday–Friday schedule

This excludes weekends, but does not account for public holidays.

```python
import calendar
workdays = [d for d in calendar.Calendar().itermonthdates(2020, 7)
            if (d.year, d.month) == (2020, 7) and d.weekday() < 5]
print('July 2020 workdays:', len(workdays))
print('First:', workdays[0], '| Last:', workdays[-1])
```

```python
assert len(workdays) == 23
assert all(d.weekday() < 5 for d in workdays)
print('Check passed: 23 weekdays in July 2020')
```

## 11. Test a leap-year February

February 2024 contains 29 actual month dates.

```python
import calendar
feb = [d for d in calendar.Calendar().itermonthdates(2024, 2)
       if (d.year, d.month) == (2024, 2)]
print('February 2024:', len(feb), 'dates; last:', feb[-1])
```

```python
assert len(feb) == 29
assert feb[-1].day == 29
print('Check passed: leap day included')
```

## 12. Date calculations

Use `datetime.timedelta` rather than adding integers to formatted strings.

```python
from datetime import date, timedelta
meeting = date(2020, 7, 1)
print('Meeting:', meeting)
print('Follow-up:', meeting + timedelta(days=7))
```

```python
assert meeting + timedelta(days=7) == date(2020, 7, 8)
print('Check passed: a week is seven days')
```

## 13. One-time iterator consumption

Once you call `list()` on the remaining elements, that same iterator is exhausted.

```python
import calendar
it = calendar.Calendar().itermonthdates(2020, 7)
print('First:', next(it))
print('Remaining:', len(list(it)))
print('After exhaustion:', len(list(it)))
```

```python
assert len(list(calendar.Calendar().itermonthdates(2020, 7))) == 35
print('Check passed: calling the method again creates a fresh iterator')
```

## 14. Compare month iterators

Notice how the return type changes when you use `itermonthdays()`, `itermonthdays2()` or `itermonthdays3()`.

```python
import calendar
cal = calendar.Calendar()
print('date objects:', list(cal.itermonthdates(2020, 7))[:2])
print('day numbers:', list(cal.itermonthdays(2020, 7))[:2])
print('(day, weekday):', list(cal.itermonthdays2(2020, 7))[:2])
print('(year, month, day):', list(cal.itermonthdays3(2020, 7))[:2])
```

```python
from datetime import date
assert isinstance(next(cal.itermonthdates(2020, 7)), date)
assert isinstance(next(cal.itermonthdays3(2020, 7)), tuple)
print('Check passed: date objects vs tuple data')
```

## 15. Explore the year boundary safely

`datetime.date` supports only years 1–9999. `itermonthdays3()` can produce year/month/day integer tuples outside that range.

```python
import calendar
values = list(calendar.Calendar().itermonthdays3(9999, 12))
print('First:', values[0], '| Last:', values[-1])
```

```python
assert len(values) % 7 == 0
assert values[-1][0] >= 9999
print('Check passed: tuple generator handles year boundary')
```

## 16. Handle invalid input

An out-of-range month raises `calendar.IllegalMonthError` when you iterate.

```python
import calendar
try:
    values = list(calendar.Calendar().itermonthdates(2020, 13))
except calendar.IllegalMonthError as exc:
    print('Invalid month:', exc)
```

```python
assert issubclass(calendar.IllegalMonthError, ValueError)
print('Check passed: invalid month can be handled as a ValueError')
```
