### 1. Basic Two-Argument Exponentiation
Calculate powers with integer and floating-point inputs.

```python
print("pow(2, 3):  ", pow(2, 3))
print("pow(2.5, 3):", pow(2.5, 3))
print("pow(2, -2): ", pow(2, -2))
```

### 2. High-Speed Three-Argument Modular Exponentiation
Compute (base ** exp) % mod without intermediate numeric overflow.

```python
base, exp, mod = 7, 256, 13
res = pow(base, exp, mod)
print(f"pow({base}, {exp}, {mod}) = {res}")
assert res == 9
```

### 3. Modular Multiplicative Inverse (Python 3.8+)
Compute modular inverse using `pow(base, -1, mod)`.

```python
a, m = 3, 11
inv = pow(a, -1, m)
print(f"Inverse of {a} modulo {m}: {inv}")
assert (a * inv) % m == 1
```

### 4. Production Recipe: Diffie-Hellman Key Exchange
Compute public keys and shared secrets using modular exponentiation.

```python
p, g = 23, 5
a_priv, b_priv = 6, 15
A = pow(g, a_priv, p)
B = pow(g, b_priv, p)

s_alice = pow(B, a_priv, p)
s_bob = pow(A, b_priv, p)
assert s_alice == s_bob
print("Shared secret verified:", s_alice)
```

### 5. Self-Checking Verification Suite
Run unit assertions verifying pow() mechanics.

```python
assert pow(2, 4) == 16
assert pow(2, -1) == 0.5
assert pow(7, 256, 13) == 9
assert pow(3, -1, 11) == 4
assert round(5000 * pow(1.07, 5), 2) == 7012.76

print("All pow() unit assertions passed! 100% verified.")
```
