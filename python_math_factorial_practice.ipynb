### 1. Basic Factorials & 0! Identity
`math.factorial(n)` computes $n! = n \times (n-1) \times \dots \times 1$, with $0! = 1$.

```python
import math

print("0! =", math.factorial(0))
print("1! =", math.factorial(1))
print("5! =", math.factorial(5))
print("10! =", math.factorial(10))

assert math.factorial(0) == 1
assert math.factorial(5) == 120
assert math.factorial(10) == 3628800
```

### 2. Large Number Computation (Arbitrary-Precision Integers)
Python computes arbitrarily large factorials without integer overflow.

```python
import math

fact_50 = math.factorial(50)
print(f"50! = {fact_50}")
print(f"Length of 50!: {len(str(fact_50))} digits")

fact_100 = math.factorial(100)
print(f"Length of 100!: {len(str(fact_100))} digits")
assert len(str(fact_100)) == 158
```

### 3. Comparing math.factorial vs. Loop vs. Recursion
Benchmark loop and recursion against `math.factorial()`, and observe the recursion depth limit.

```python
import math

def factorial_iterative(n):
    res = 1
    for i in range(2, n + 1):
        res *= i
    return res

def factorial_recursive(n):
    if n <= 1:
        return 1
    return n * factorial_recursive(n - 1)

assert math.factorial(15) == factorial_iterative(15) == factorial_recursive(15)
print("15! =", math.factorial(15))
```

### 4. Permutations and Combinations (math.perm and math.comb)
Explore $P(n, k)$ and $C(n, k)$ formulas.

```python
import math

n, k = 10, 3
perm_val = math.perm(n, k)
comb_val = math.comb(n, k)

print(f"P(10, 3) = {perm_val} (arrangements where order matters)")
print(f"C(10, 3) = {comb_val} (selections where order does not matter)")

assert perm_val == 720
assert comb_val == 120
assert perm_val == math.factorial(n) // math.factorial(n - k)
assert comb_val == math.factorial(n) // (math.factorial(k) * math.factorial(n - k))
```

### 5. Practice Exercise: Lottery Odds & Anagram Permutations
Run the verification suite computing lottery jackpot probabilities and anagram permutations.

```python
import math
from collections import Counter

def lottery_odds(pool, pick, bonus=1):
    return math.comb(pool, pick) * bonus

def anagram_arrangements(word):
    counts = Counter(word.upper())
    denom = 1
    for c in counts.values():
        denom *= math.factorial(c)
    return math.factorial(len(word)) // denom

# Unit Assertions
odds_6_49 = lottery_odds(49, 6)
assert odds_6_49 == 13_983_816

odds_powerball = lottery_odds(69, 5, bonus=26)
assert odds_powerball == 292_201_338

assert anagram_arrangements("BOOK") == 12
assert anagram_arrangements("MISSISSIPPI") == 34650

print("Lottery 6/49 odds: 1 in", f"{odds_6_49:,}")
print("Powerball odds:    1 in", f"{odds_powerball:,}")
print("Anagrams of 'MISSISSIPPI':", anagram_arrangements("MISSISSIPPI"))
print("All combinatorial assertions passed successfully!")
```
