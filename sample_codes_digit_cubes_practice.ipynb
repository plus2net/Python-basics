## Example 1: First working example

Read the code, predict the output and run this cell.

```python
n = 371
total = sum(int(digit) ** 3 for digit in str(n))
print("Sum of cubes:", total)
print("Matches:", total == n)
```

## Example 2: Original digit-by-digit loop

Read the code, predict the output and run this cell.

```python
from unittest.mock import patch
with patch("builtins.input", side_effect=['371']):
    n = input("Enter a number: ").strip()
    if n.isdecimal():
        my_sum = 0
        for i in range(len(n)):
            my_sum += pow(int(n[i]), 3)
        print("Sum of cubes:", my_sum)
        print("Digits match:", my_sum == int(n))
    else:
        print("Use nonnegative whole numbers")
```

## Example 3: Find all matching multi-digit numbers

Read the code, predict the output and run this cell.

```python
matches = []
for n in range(10, 100000):
    total = sum(int(d) ** 3 for d in str(n))
    if total == n:
        matches.append(n)
print(matches)
print("Smallest:", min(matches))
print("Largest:", max(matches))
```

## Example 4: One-digit values also match

Read the code, predict the output and run this cell.

```python
print([n for n in range(10) if n == n ** 3])
```

## Example 5: Reusable cube-sum function

Read the code, predict the output and run this cell.

```python
def cube_sum(n):
    return sum(int(d) ** 3 for d in str(n))
for n in (153, 370, 371, 407, 1634):
    print(n, cube_sum(n), n == cube_sum(n))
```

## Example 6: Cube sum versus Armstrong numbers

Read the code, predict the output and run this cell.

```python
n = 1634
digits = str(n)
fixed_cube = sum(int(d) ** 3 for d in digits)
armstrong_power = sum(int(d) ** len(digits) for d in digits)
print("Fixed cube:", fixed_cube)
print("Armstrong:", armstrong_power)
```
